> **North Star:** The LLM orchestrates. Deterministic telemetry supplies network truth. Standards knowledge provides grounding. ML supplies probabilistic evidence. Policy controls actions. Verification closes the loop.


##### Introduction — Goal

The goal of M05_03 is to create a deterministic, auditable and diagnosable set of **5G RAN Accessibility investigation challenges** without modifying the authoritative M05_01 healthy telemetry or the formal M05_02 KPI products.

The notebook will:

- restore the validated M05_01/M05_02 lineage without rebuilding it;
- define a bounded first incident bank within the existing Accessibility causal domain;
- create incident **overlays on copied healthy observation windows**, never on the canonical source;
- preserve mathematically valid standardized PM hierarchies;
- include positive evidence, exclusion evidence, topology context and limited non-causal distractors;
- recompute the two formal Accessibility KPIs from the modified PM counters;
- separate **agent-visible evidence** from **hidden benchmark truth**;
- reserve a held-out evaluation subset;
- and persist immutable benchmark artifacts for the Telemetry MCP and later agent evaluation.

> **Can a controlled synthetic benchmark create realistic 5G RAN Accessibility incidents whose root causes are discoverable from observable evidence, distinguishable from plausible competing hypotheses, mathematically consistent with the formal KPI layer, and hidden from the future agent until independent evaluation?**

M05_03 does **not** evaluate an LLM yet. Its job is to create the challenges the later agent must solve.

##### Introduction — Benchmark Design Principles

The incident bank follows the causal-observability rules already established in Module 5:

1. **Root cause is hidden.**
2. **Every admitted fault has observable positive evidence.**
3. **Every admitted fault has exclusion evidence** that helps reject nearby hypotheses.
4. **Topology scope matters** — cell, site, CU domain or RAN/Core boundary.
5. **Chronology matters** — pre-incident, active incident and recovery intervals are retained.
6. **Distractors are non-causal** and must not become a second hidden fault.
7. **Formal KPI values are recomputed from PM counters**; they are never directly injected.
8. **Operational thresholds are synthetic design choices**, not represented as 3GPP requirements.
9. **Healthy controls are included** so the future agent can be penalized for inventing incidents.
10. **Evaluation cases are held out** from the development-label artifact.


##### Introduction — Initial Formal Incident Bank

The first benchmark deliberately uses a **bounded subset** of the broader causal catalogue rather than exercising all 37 declared fault concepts at once.

| Family | Primary scope | Main discriminating idea |
|---|---|---|
| `HEALTHY_CONTROL` | full network | natural healthy variability only |
| `RADIO_INTERFERENCE` | cell | SINR/BLER degrade while RSRP/load remain broadly normal |
| `RADIO_RESOURCE_CONGESTION` | site | PRB/load/admission pressure increase while radio quality remains healthy |
| `N2_PACKET_LOSS_LATENCY` | CU domain / N2 path | NG stage degrades with N2 loss/latency evidence while radio/CU-CP remain healthy |
| `CU_CP_PROCESSING_PRESSURE` | CU domain | NG stage degrades with CU-CP CPU/queue/delay evidence while N2 remains healthy |
| `QOS_CONFIGURATION` | cell + 5QI 1 | 5QI-specific DRB degradation correlated with a QoS configuration change |
| `RRC_CONFIGURATION` | cell | RRC establishment degrades after an RRC-profile change while radio/load remain healthy |
| `AMF_FACING_BOUNDARY` | CU-domain boundary | NG stage degrades with RAN-observed AMF reachability/peer evidence while internal RAN/N2 evidence remains healthy |

The AMF case is intentionally **boundary-scoped**. The benchmark must not claim an internal 5GC root cause that the RAN evidence cannot prove.


##### Introduction — Episode and Split Policy

The V1 benchmark contains:

```text
8 case families
× 4 deterministic episodes per family
= 32 benchmark cases

Per family:
3 DEVELOPMENT cases
1 held-out EVAL case
```

Each case contains a **4-hour observation window**:

```text
1 hour healthy pre-window
2 hours active incident window
1 hour recovery window

16 × 15-minute intervals per case
```

The future agent receives the complete 4-hour observation window but does **not** receive the phase boundaries.

##### Cell 04 — Topology Inventory and Deterministic Target Maps

This cell reconstructs only the mappings required by the benchmark:

- NRCellCU ↔ NRCellDU within each site;
- site ↔ CU domain;
- CU domain ↔ CU-CP / N2 path / AMF-facing peer.

The mapping is derived from canonical observations rather than recreated from memory.


````python
if not globals().get("M05_03_TELEMETRY_RUNTIME_PASS", False):
    raise RuntimeError("Cell 04 requires Cell 03.")

con.execute("""
CREATE OR REPLACE TABLE m05_03_cell_map AS
WITH
cu AS (
    SELECT DISTINCT entity_id AS nr_cell_cu_id, site_id, cu_domain_id
    FROM healthy_telemetry
    WHERE entity_type = 'NR_CELL_CU'
),
du AS (
    SELECT DISTINCT entity_id AS nr_cell_du_id, site_id, cu_domain_id
    FROM healthy_telemetry
    WHERE entity_type = 'NR_CELL_DU'
),
cu_ranked AS (
    SELECT *,
           ROW_NUMBER() OVER (
               PARTITION BY site_id ORDER BY nr_cell_cu_id
           ) AS sector_rank
    FROM cu
),
du_ranked AS (
    SELECT *,
           ROW_NUMBER() OVER (
               PARTITION BY site_id ORDER BY nr_cell_du_id
           ) AS sector_rank
    FROM du
)
SELECT
    c.site_id, c.cu_domain_id, c.sector_rank,
    c.nr_cell_cu_id, d.nr_cell_du_id
FROM cu_ranked AS c
JOIN du_ranked AS d
  ON c.site_id = d.site_id
 AND c.cu_domain_id = d.cu_domain_id
 AND c.sector_rank = d.sector_rank
ORDER BY c.site_id, c.sector_rank
""")

con.execute("""
CREATE OR REPLACE TABLE m05_03_domain_map AS
SELECT
    cu_domain_id,
    MAX(CASE WHEN operational_scope = 'CU_CP' THEN entity_id END) AS cu_cp_id,
    MAX(CASE WHEN operational_scope = 'N2_PATH' THEN entity_id END) AS n2_path_id,
    MAX(CASE WHEN operational_scope = 'AMF_PEER' THEN entity_id END) AS amf_peer_id
FROM healthy_telemetry
WHERE cu_domain_id IS NOT NULL
GROUP BY cu_domain_id
ORDER BY cu_domain_id
""")

CELL_MAP_DF = con.execute("SELECT * FROM m05_03_cell_map").df()
DOMAIN_MAP_DF = con.execute("SELECT * FROM m05_03_domain_map").df()

topology_diag = con.execute("""
SELECT
    COUNT(DISTINCT site_id) AS sites,
    COUNT(DISTINCT nr_cell_cu_id) AS nr_cell_cu,
    COUNT(DISTINCT nr_cell_du_id) AS nr_cell_du,
    COUNT(DISTINCT cu_domain_id) AS cu_domains,
    MIN(sector_rank) AS min_sector_rank,
    MAX(sector_rank) AS max_sector_rank
FROM m05_03_cell_map
""").df()

domain_nulls = int(
    DOMAIN_MAP_DF[["cu_cp_id", "n2_path_id", "amf_peer_id"]]
    .isna().sum().sum()
)

checks = {
    "sites_10": int(topology_diag.loc[0, "sites"]) == 10,
    "nr_cell_cu_30": int(topology_diag.loc[0, "nr_cell_cu"]) == 30,
    "nr_cell_du_30": int(topology_diag.loc[0, "nr_cell_du"]) == 30,
    "cu_domains_2": int(topology_diag.loc[0, "cu_domains"]) == 2,
    "three_sectors_per_site":
        int(topology_diag.loc[0, "min_sector_rank"]) == 1
        and int(topology_diag.loc[0, "max_sector_rank"]) == 3,
    "domain_operational_identities_complete": domain_nulls == 0,
}

print("=" * 108)
print("MODULE 5 — M05_03 CELL 04 — TOPOLOGY INVENTORY")
print("=" * 108)
print(topology_diag.to_string(index=False))
print("\nCU-domain operational identities")
print("-" * 108)
print(DOMAIN_MAP_DF.to_string(index=False))
print("-" * 108)
for name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL':<7} {name}")

failed = [name for name, passed in checks.items() if not passed]
if failed:
    raise RuntimeError(
        "M05_03 topology qualification failed: " + ", ".join(failed)
    )

M05_03_TOPOLOGY_PASS = True
print("\nM05_03_TOPOLOGY_PASS                :", M05_03_TOPOLOGY_PASS)
````

##### Cell 06 — Deterministic 32-Case Schedule and Split

Creates opaque scenario IDs, non-overlapping 4-hour windows, deterministic target scopes, severity profiles and DEVELOPMENT/EVAL assignments. The scenario ID carries **no semantic clue** about the fault.


````python
if not globals().get("M05_03_INCIDENT_DESIGN_PASS", False):
    raise RuntimeError("Cell 06 requires Cell 05.")

BASE_START = pd.Timestamp("2026-06-02T00:00:00Z")
WINDOW_SPACING_HOURS = 66
SEVERITY_MAX_BY_EPISODE = {1:0.65, 2:0.80, 3:0.95, 4:0.75}
INCIDENT_SHAPE = [0.40,0.70,1.00,1.00,1.00,1.00,0.70,0.40]
sites = sorted(CELL_MAP_DF["site_id"].unique().tolist())

def domain_record(cu_domain_id):
    return DOMAIN_MAP_DF.loc[
        DOMAIN_MAP_DF["cu_domain_id"] == cu_domain_id
    ].iloc[0].to_dict()

private_cases, timeline_rows, target_cell_rows = [], [], []
global_index = 0

for episode in range(1,5):
    for family_index, family in enumerate(CASE_FAMILIES):
        scenario_id = f"M05I{global_index+1:04d}"
        window_start = BASE_START + pd.Timedelta(hours=WINDOW_SPACING_HOURS*global_index)
        incident_start = window_start + pd.Timedelta(hours=1)
        incident_end = incident_start + pd.Timedelta(hours=2)
        window_end = window_start + pd.Timedelta(hours=4)
        split = "DEVELOPMENT" if episode <= 3 else "EVAL"
        max_severity = 0.0 if family == "HEALTHY_CONTROL" else SEVERITY_MAX_BY_EPISODE[episode]

        site_pos = (family_index*2 + episode*3) % len(sites)
        target_site = sites[site_pos]
        site_df = CELL_MAP_DF[CELL_MAP_DF["site_id"] == target_site].sort_values("sector_rank")
        sector = ((family_index + episode) % 3) + 1
        target_row = site_df[site_df["sector_rank"] == sector].iloc[0]
        target_domain = target_row["cu_domain_id"]
        domain = domain_record(target_domain)

        if family in {"RADIO_INTERFERENCE","QOS_CONFIGURATION","RRC_CONFIGURATION"}:
            target_rows = pd.DataFrame([target_row])
        elif family == "RADIO_RESOURCE_CONGESTION":
            target_rows = site_df
        elif family == "HEALTHY_CONTROL":
            target_rows = site_df.iloc[0:0]
        else:
            target_rows = CELL_MAP_DF[
                CELL_MAP_DF["cu_domain_id"] == target_domain
            ].sort_values(["site_id","sector_rank"])

        distractor_site = sites[(site_pos + 5) % len(sites)]
        distractor_df = CELL_MAP_DF[
            CELL_MAP_DF["site_id"] == distractor_site
        ].sort_values("sector_rank")
        distractor_domain = distractor_df.iloc[0]["cu_domain_id"]
        distractor_domain_rec = domain_record(distractor_domain)
        distractor_metric = [
            "processing.du_cpu_pct","radio.dl_bler_pct","transport.n2_util_pct"
        ][global_index % 3]

        private_cases.append({
            "scenario_id":scenario_id,"family_internal":family,
            "incident_present":INCIDENT_FAMILIES[family]["incident_present"],
            "episode":episode,"benchmark_split":split,
            "window_start_utc":window_start,"incident_start_utc":incident_start,
            "incident_end_utc":incident_end,"window_end_utc":window_end,
            "max_severity":max_severity,
            "target_site_id":None if family=="HEALTHY_CONTROL" else target_site,
            "target_sector_rank":None if family=="HEALTHY_CONTROL" else int(sector),
            "target_cu_domain_id":None if family=="HEALTHY_CONTROL" else target_domain,
            "target_cu_cp_id":None if family=="HEALTHY_CONTROL" else domain["cu_cp_id"],
            "target_n2_path_id":None if family=="HEALTHY_CONTROL" else domain["n2_path_id"],
            "target_amf_peer_id":None if family=="HEALTHY_CONTROL" else domain["amf_peer_id"],
            "target_5qi":INCIDENT_FAMILIES[family]["target_5qi"],
            "distractor_metric":None if family=="HEALTHY_CONTROL" else distractor_metric,
            "distractor_site_id":None if family=="HEALTHY_CONTROL" else distractor_site,
            "distractor_cu_domain_id":None if family=="HEALTHY_CONTROL" else distractor_domain,
            "distractor_n2_path_id":None if family=="HEALTHY_CONTROL" else distractor_domain_rec["n2_path_id"],
        })

        for tr in target_rows.itertuples(index=False):
            target_cell_rows.append({
                "scenario_id":scenario_id,
                "nr_cell_cu_id":tr.nr_cell_cu_id,
                "nr_cell_du_id":tr.nr_cell_du_id,
            })

        for i in range(16):
            ts = window_start + pd.Timedelta(minutes=15*i)
            if i < 4:
                phase, sev = "PRE", 0.0
            elif i < 12:
                phase, sev = "INCIDENT", INCIDENT_SHAPE[i-4]*max_severity
            else:
                phase, sev = "RECOVERY", 0.0
            timeline_rows.append({
                "scenario_id":scenario_id,"interval_start_utc":ts,
                "phase":phase,"severity":float(sev),
            })
        global_index += 1

PRIVATE_CASES_DF = pd.DataFrame(private_cases)
TIMELINE_DF = pd.DataFrame(timeline_rows)
TARGET_CELLS_DF = pd.DataFrame(target_cell_rows)

windows = PRIVATE_CASES_DF[
    ["scenario_id","window_start_utc","window_end_utc"]
].sort_values("window_start_utc").reset_index(drop=True)
overlaps = sum(
    windows.loc[i,"window_start_utc"] < windows.loc[i-1,"window_end_utc"]
    for i in range(1,len(windows))
)
split_counts = PRIVATE_CASES_DF.groupby(
    ["family_internal","benchmark_split"]
).size().unstack(fill_value=0)

telemetry_last = pd.Timestamp(runtime_summary.loc[0,"last_interval"])
checks = {
    "scenario_count_32":len(PRIVATE_CASES_DF)==32,
    "timeline_rows_512":len(TIMELINE_DF)==512,
    "opaque_ids_unique":PRIVATE_CASES_DF["scenario_id"].nunique()==32,
    "no_window_overlap":overlaps==0,
    "all_windows_inside_baseline":
        PRIVATE_CASES_DF["window_end_utc"].max() <= telemetry_last + pd.Timedelta(minutes=15),
    "three_development_per_family":bool((split_counts["DEVELOPMENT"]==3).all()),
    "one_eval_per_family":bool((split_counts["EVAL"]==1).all()),
    "healthy_zero_severity":bool((
        PRIVATE_CASES_DF.loc[
            PRIVATE_CASES_DF["family_internal"]=="HEALTHY_CONTROL","max_severity"
        ]==0
    ).all()),
}

print("=" * 108)
print("MODULE 5 — M05_03 CELL 06 — DETERMINISTIC EPISODE SCHEDULE")
print("=" * 108)
for name, passed in checks.items(): print(f"{'PASS' if passed else 'FAIL':<7} {name}")
print("-" * 108)
print("Benchmark cases                     :",len(PRIVATE_CASES_DF))
print("Timeline rows                       :",len(TIMELINE_DF))
print("DEVELOPMENT / EVAL                  :",
      int((PRIVATE_CASES_DF["benchmark_split"]=="DEVELOPMENT").sum()),"/",
      int((PRIVATE_CASES_DF["benchmark_split"]=="EVAL").sum()))
print("First case start                    :",PRIVATE_CASES_DF["window_start_utc"].min())
print("Last case end                       :",PRIVATE_CASES_DF["window_end_utc"].max())

failed=[n for n,p in checks.items() if not p]
if failed: raise RuntimeError("M05_03 episode schedule failed: "+", ".join(failed))

M05_03_EPISODE_SCHEDULE_PASS=True
print("\nM05_03_EPISODE_SCHEDULE_PASS        :",M05_03_EPISODE_SCHEDULE_PASS)
````

###### Observation

The deterministic episode schedule passed all checks, creating 32 unique, non-overlapping benchmark cases across 512 timeline intervals. The split is fixed at 24 DEVELOPMENT and 8 held-out EVAL cases, with one EVAL case per family, all fully contained within the healthy baseline period and with zero severity correctly preserved for healthy controls.

##### Cell 07 — Copy Healthy Case Windows into the M05_03 Runtime

# Section 4 — Controlled Incident Overlays


<a id="cell-08"></a>
##### Cell 08 — Apply Numeric Incident Overlays

Numeric overlays modify only copied M05_03 rows in the active incident phase. Standardized success counters are scaled in coordinated families so attempt/success and Added-DRB hierarchy remain valid.

**Formal KPI values are never directly modified.**


````python
if not globals().get("M05_03_BASE_SLICE_PASS",False):
    raise RuntimeError("Cell 08 requires Cell 07.")

CELL_CU_TARGET="""EXISTS(
 SELECT 1 FROM m05_03_target_cells tc
 WHERE tc.scenario_id=w.scenario_id AND tc.nr_cell_cu_id=w.entity_id
)"""
CELL_DU_TARGET="""EXISTS(
 SELECT 1 FROM m05_03_target_cells tc
 WHERE tc.scenario_id=w.scenario_id AND tc.nr_cell_du_id=w.entity_id
)"""
SITE_TARGET="""EXISTS(
 SELECT 1 FROM m05_03_private_cases pc
 WHERE pc.scenario_id=w.scenario_id AND pc.target_site_id=w.site_id
)"""
DOMAIN_TARGET="""EXISTS(
 SELECT 1 FROM m05_03_private_cases pc
 WHERE pc.scenario_id=w.scenario_id AND pc.target_cu_domain_id=w.cu_domain_id
)"""

def update_float(family,metric,scope,expr):
    con.execute(f"""
    UPDATE m05_03_working_telemetry AS w
    SET value_float={expr}
    WHERE phase='INCIDENT' AND family_internal='{family}'
      AND metric_name='{metric}' AND ({scope})
    """)

def update_int(family,metric,scope,expr,extra="TRUE"):
    con.execute(f"""
    UPDATE m05_03_working_telemetry AS w
    SET value_int=CAST(GREATEST(0,ROUND({expr})) AS BIGINT)
    WHERE phase='INCIDENT' AND family_internal='{family}'
      AND metric_name='{metric}' AND ({scope}) AND ({extra})
    """)

def scale_success(family,metrics,scope,reduction,extra="TRUE"):
    for metric in metrics:
        update_int(
            family,metric,scope,
            f"w.original_value_int*(1.0-{reduction}*w.severity)",
            extra,
        )

# RADIO INTERFERENCE — cell
update_float("RADIO_INTERFERENCE","radio.sinr_db",CELL_DU_TARGET,
             "w.original_value_float-9.0*w.severity")
update_float("RADIO_INTERFERENCE","radio.dl_bler_pct",CELL_DU_TARGET,
             "LEAST(100.0,w.original_value_float+12.0*w.severity)")
update_float("RADIO_INTERFERENCE","radio.ul_bler_pct",CELL_DU_TARGET,
             "LEAST(100.0,w.original_value_float+8.0*w.severity)")
scale_success("RADIO_INTERFERENCE",["RRC.ConnEstabSucc.Cause"],CELL_CU_TARGET,0.12)
scale_success("RADIO_INTERFERENCE",
              ["DRB.InitialEstabSucc.5QI","DRB.EstabSucc.5QI"],CELL_CU_TARGET,0.20)
scale_success("RADIO_INTERFERENCE",["DRB.ResumeSucc.5QI"],CELL_CU_TARGET,0.12)

# RADIO RESOURCE CONGESTION — site
for metric,inc in [
    ("resource.dl_prb_util_pct",35.0),("resource.ul_prb_util_pct",28.0),
    ("resource.scheduler_pressure_pct",40.0),
    ("resource.bearer_admission_pressure_pct",30.0),
]:
    update_float("RADIO_RESOURCE_CONGESTION",metric,SITE_TARGET,
                 f"LEAST(100.0,w.original_value_float+{inc}*w.severity)")
for metric,mult in [("load.connected_ues",0.55),("load.active_ues",0.65)]:
    update_int("RADIO_RESOURCE_CONGESTION",metric,SITE_TARGET,
               f"w.original_value_int*(1.0+{mult}*w.severity)")
scale_success("RADIO_RESOURCE_CONGESTION",["RRC.ConnEstabSucc.Cause"],SITE_TARGET,0.08)
scale_success("RADIO_RESOURCE_CONGESTION",
              ["DRB.InitialEstabSucc.5QI","DRB.EstabSucc.5QI"],SITE_TARGET,0.25)
scale_success("RADIO_RESOURCE_CONGESTION",["DRB.ResumeSucc.5QI"],SITE_TARGET,0.15)

# N2 PACKET LOSS / LATENCY — CU domain
update_float("N2_PACKET_LOSS_LATENCY","transport.n2_packet_loss_pct",DOMAIN_TARGET,
             "LEAST(100.0,w.original_value_float+6.0*w.severity)")
update_float("N2_PACKET_LOSS_LATENCY","transport.n2_latency_ms",DOMAIN_TARGET,
             "w.original_value_float+35.0*w.severity")
update_int("N2_PACKET_LOSS_LATENCY","sctp.retransmissions",DOMAIN_TARGET,
           "w.original_value_int+12.0*w.severity")
update_int("N2_PACKET_LOSS_LATENCY","sctp.heartbeat_failures",DOMAIN_TARGET,
           "w.original_value_int+3.0*w.severity")
scale_success("N2_PACKET_LOSS_LATENCY",["UECNTX.ConnEstabSucc.Cause"],DOMAIN_TARGET,0.25)

# CU-CP PROCESSING PRESSURE — CU domain
for metric,inc in [("processing.cu_cp_cpu_pct",35.0),("processing.cu_cp_memory_pct",25.0)]:
    update_float("CU_CP_PROCESSING_PRESSURE",metric,DOMAIN_TARGET,
                 f"LEAST(100.0,w.original_value_float+{inc}*w.severity)")
update_int("CU_CP_PROCESSING_PRESSURE","processing.cu_cp_control_queue_depth",DOMAIN_TARGET,
           "w.original_value_int*(1.0+3.0*w.severity)")
update_float("CU_CP_PROCESSING_PRESSURE","processing.cu_cp_control_delay_ms",DOMAIN_TARGET,
             "w.original_value_float*(1.0+3.0*w.severity)")
scale_success("CU_CP_PROCESSING_PRESSURE",["UECNTX.ConnEstabSucc.Cause"],DOMAIN_TARGET,0.22)
scale_success("CU_CP_PROCESSING_PRESSURE",["RRC.ConnEstabSucc.Cause"],DOMAIN_TARGET,0.04)

# QoS CONFIGURATION — cell / 5QI 1
qos_extra="CAST(w.dimension_5qi AS VARCHAR)='1'"
scale_success("QOS_CONFIGURATION",
              ["DRB.InitialEstabSucc.5QI","DRB.EstabSucc.5QI"],
              CELL_CU_TARGET,0.35,qos_extra)
scale_success("QOS_CONFIGURATION",["DRB.ResumeSucc.5QI"],
              CELL_CU_TARGET,0.20,qos_extra)

# RRC CONFIGURATION — cell
scale_success("RRC_CONFIGURATION",["RRC.ConnEstabSucc.Cause"],CELL_CU_TARGET,0.30)

# AMF-FACING BOUNDARY — CU domain
scale_success("AMF_FACING_BOUNDARY",["UECNTX.ConnEstabSucc.Cause"],DOMAIN_TARGET,0.30)

numeric_change_diag=con.execute("""
SELECT family_internal,phase,
 SUM(CASE WHEN COALESCE(
     ABS(COALESCE(value_float,CAST(value_int AS DOUBLE))-
         COALESCE(original_value_float,CAST(original_value_int AS DOUBLE))),0
 )>1e-12 THEN 1 ELSE 0 END) AS changed_numeric_rows
FROM m05_03_working_telemetry
GROUP BY family_internal,phase ORDER BY family_internal,phase
""").df()

outside=int(numeric_change_diag.loc[numeric_change_diag["phase"]!="INCIDENT",
                                     "changed_numeric_rows"].sum())
healthy=int(numeric_change_diag.loc[
    numeric_change_diag["family_internal"]=="HEALTHY_CONTROL","changed_numeric_rows"
].sum())
fault_active=numeric_change_diag[
    (numeric_change_diag["phase"]=="INCIDENT")&
    (numeric_change_diag["family_internal"]!="HEALTHY_CONTROL")
]
all_faults=(fault_active["family_internal"].nunique()==7 and
            bool((fault_active["changed_numeric_rows"]>0).all()))

print("="*108)
print("MODULE 5 — M05_03 CELL 08 — NUMERIC INCIDENT OVERLAYS")
print("="*108)
print(numeric_change_diag.to_string(index=False))
print("-"*108)
print("Numeric changes outside INCIDENT    :",outside)
print("Healthy-control numeric changes     :",healthy)
print("All 7 fault families changed        :",all_faults)

if outside or healthy or not all_faults:
    raise RuntimeError("Numeric overlay boundary validation failed.")

M05_03_NUMERIC_OVERLAY_PASS=True
print("\nM05_03_NUMERIC_OVERLAY_PASS         :",M05_03_NUMERIC_OVERLAY_PASS)
````

###### Observation

Numeric incident overlays were successfully applied to all 7 fault families only during their INCIDENT periods. PRE, RECOVERY and HEALTHY_CONTROL telemetry remained unchanged, confirming that the benchmark preserves clean healthy reference states while introducing controlled and family-specific fault signatures.

##### Cell 09 — Configuration / Boundary State, Events and Non-Causal Distractors

Adds QoS/RRC configuration states, AMF-facing state evidence, controlled event records and one mild **unrelated operational distractor** per non-control case. Distractors never modify standardized PM counters.


````python
if not globals().get("M05_03_NUMERIC_OVERLAY_PASS",False):
    raise RuntimeError("Cell 09 requires Cell 08.")

con.execute(f"""
UPDATE m05_03_working_telemetry w SET value_text='INCIDENT_PROFILE'
WHERE phase='INCIDENT' AND family_internal='QOS_CONFIGURATION'
  AND metric_name='config.qos_profile_state' AND ({CELL_CU_TARGET})
""")
con.execute(f"""
UPDATE m05_03_working_telemetry w SET value_text='INCIDENT_PROFILE'
WHERE phase='INCIDENT' AND family_internal='RRC_CONFIGURATION'
  AND metric_name='config.rrc_profile_state' AND ({CELL_CU_TARGET})
""")
con.execute(f"""
UPDATE m05_03_working_telemetry w
SET value_bool=CASE WHEN severity>=0.50 THEN FALSE ELSE original_value_bool END
WHERE phase='INCIDENT' AND family_internal='AMF_FACING_BOUNDARY'
  AND metric_name='boundary.amf_reachable' AND ({DOMAIN_TARGET})
""")
con.execute(f"""
UPDATE m05_03_working_telemetry w SET value_text='UNSTABLE'
WHERE phase='INCIDENT' AND family_internal='AMF_FACING_BOUNDARY'
  AND metric_name='boundary.amf_peer_state' AND ({DOMAIN_TARGET})
""")

event_rows=[]
for row in PRIVATE_CASES_DF.itertuples(index=False):
    if row.family_internal in {"QOS_CONFIGURATION","RRC_CONFIGURATION"}:
        target=TARGET_CELLS_DF.loc[
            TARGET_CELLS_DF["scenario_id"]==row.scenario_id,"nr_cell_cu_id"
        ].iloc[0]
        event_rows.append({
            "scenario_id":row.scenario_id,"event_time_utc":row.incident_start_utc,
            "event_metric_name":"event.config_change","entity_id":target,
            "site_id":row.target_site_id,"cu_domain_id":row.target_cu_domain_id,
            "state":"ACTIVE",
            "message":"Configuration change detected on the managed object.",
        })
    elif row.family_internal=="AMF_FACING_BOUNDARY":
        event_rows.append({
            "scenario_id":row.scenario_id,"event_time_utc":row.incident_start_utc,
            "event_metric_name":"event.interface_alarm_state","entity_id":row.target_amf_peer_id,
            "site_id":None,"cu_domain_id":row.target_cu_domain_id,"state":"ACTIVE",
            "message":"Intermittent peer/interface reachability alarm observed.",
        })
EVENTS_DF=pd.DataFrame(event_rows)

# Mild distractors outside causal target.
for case in PRIVATE_CASES_DF.itertuples(index=False):
    if case.family_internal=="HEALTHY_CONTROL": continue
    metric=case.distractor_metric
    if metric=="processing.du_cpu_pct":
        con.execute(f"""
        UPDATE m05_03_working_telemetry w
        SET value_float=LEAST(100.0,original_value_float+8.0*severity)
        WHERE scenario_id='{case.scenario_id}' AND phase='INCIDENT'
          AND metric_name='processing.du_cpu_pct' AND site_id='{case.distractor_site_id}'
        """)
    elif metric=="radio.dl_bler_pct":
        du=CELL_MAP_DF.loc[
            CELL_MAP_DF["site_id"]==case.distractor_site_id
        ].sort_values("sector_rank").iloc[0]["nr_cell_du_id"]
        con.execute(f"""
        UPDATE m05_03_working_telemetry w
        SET value_float=LEAST(100.0,original_value_float+1.5*severity)
        WHERE scenario_id='{case.scenario_id}' AND phase='INCIDENT'
          AND metric_name='radio.dl_bler_pct' AND entity_id='{du}'
        """)
    else:
        con.execute(f"""
        UPDATE m05_03_working_telemetry w
        SET value_float=LEAST(100.0,original_value_float+8.0*severity)
        WHERE scenario_id='{case.scenario_id}' AND phase='INCIDENT'
          AND metric_name='transport.n2_util_pct'
          AND entity_id='{case.distractor_n2_path_id}'
        """)

state_diag=con.execute("""
SELECT family_internal,phase,
 SUM(CASE WHEN value_text IS DISTINCT FROM original_value_text
               OR value_bool IS DISTINCT FROM original_value_bool
          THEN 1 ELSE 0 END) AS changed_state_rows
FROM m05_03_working_telemetry
GROUP BY family_internal,phase ORDER BY family_internal,phase
""").df()
outside=int(state_diag.loc[state_diag["phase"]!="INCIDENT","changed_state_rows"].sum())
expected=set(PRIVATE_CASES_DF.loc[
    PRIVATE_CASES_DF["family_internal"].isin(
        {"QOS_CONFIGURATION","RRC_CONFIGURATION","AMF_FACING_BOUNDARY"}
    ),"scenario_id"])
observed=set(EVENTS_DF["scenario_id"])

checks={
 "no_state_changes_outside_incident":outside==0,
 "events_only_for_expected_cases":observed==expected,
 "twelve_controlled_events":len(EVENTS_DF)==12,
 "healthy_control_has_no_events":
     not bool(set(PRIVATE_CASES_DF.loc[
         PRIVATE_CASES_DF["family_internal"]=="HEALTHY_CONTROL","scenario_id"
     ])&observed),
}

print("="*108)
print("MODULE 5 — M05_03 CELL 09 — CONTEXT / EVENTS / DISTRACTORS")
print("="*108)
for n,p in checks.items():print(f"{'PASS' if p else 'FAIL':<7} {n}")
print("-"*108)
print("Controlled event rows               :",len(EVENTS_DF))
print("Cases with controlled events        :",len(observed))
print("State changes outside incident      :",outside)
print("Distractor policy                   : 1 mild operational distractor per non-control case")

failed=[n for n,p in checks.items() if not p]
if failed:raise RuntimeError("M05_03 context overlay validation failed: "+", ".join(failed))

M05_03_CONTEXT_OVERLAY_PASS=True
print("\nM05_03_CONTEXT_OVERLAY_PASS         :",M05_03_CONTEXT_OVERLAY_PASS)
````

Controlled state changes remained confined to INCIDENT periods, exactly 12 expected cases generated one event each, healthy controls remained event-free, and every non-control case received one mild operational distractor to increase diagnostic realism without altering the true causal path.


<a id="section-5-sanitization-kpi"></a>
# Section 5 — Sanitization and Formal KPI Recalculation


<a id="cell-10"></a>
##### Cell 10 — Build Agent-Visible Evidence and Enforce Leakage Boundary

Creates sanitized telemetry/events. Internal family, severity, phase and original values are excluded, and text-bearing public fields are scanned for benchmark-answer vocabulary.


````python
if not globals().get("M05_03_CONTEXT_OVERLAY_PASS", False):
    raise RuntimeError("Cell 10 requires Cell 09.")

con.execute("""
CREATE OR REPLACE TABLE m05_03_agent_telemetry AS
SELECT
    scenario_id,
    observation_id,
    environment,
    observation_type,
    evidence_origin,
    metric_name,
    is_standardized_metric,
    operational_category,
    entity_id,
    entity_namespace,
    entity_type,
    operational_scope,
    parent_entity_id,
    site_id,
    cu_domain_id,
    interval_start_utc,
    interval_end_utc,
    pm_interval_minutes,
    value_kind,
    value_int,
    value_float,
    value_bool,
    value_text,
    unit,
    dimension_cause,
    dimension_5qi,
    dimension_snssai,
    dimensions_json,
    quality,
    graph_node_id,
    investigation_stage,
    source_system,
    metadata_json
FROM m05_03_working_telemetry
""")

con.register("events_input", EVENTS_DF)
con.execute("""
CREATE OR REPLACE TABLE m05_03_agent_events AS
SELECT * FROM events_input
""")

FORBIDDEN_PUBLIC_TERMS = {
    "root_cause",
    "ground_truth",
    "fault_label",
    "expected_diagnosis",
    "correct_action",
    "benchmark_answer",
    "family_internal",
    "severity",
    "phase",
}

agent_cols = set(
    con.execute("DESCRIBE m05_03_agent_telemetry")
    .df()["column_name"]
    .astype(str)
)

event_cols = set(
    con.execute("DESCRIBE m05_03_agent_events")
    .df()["column_name"]
    .astype(str)
)

schema_hits = sorted(
    FORBIDDEN_PUBLIC_TERMS & (agent_cols | event_cols)
)

text_hits = int(con.execute("""
SELECT COUNT(*)
FROM m05_03_agent_telemetry
WHERE LOWER(COALESCE(value_text,'')) SIMILAR TO
      '%(root_cause|ground_truth|fault_label|expected_diagnosis|correct_action|benchmark_answer)%'
   OR LOWER(COALESCE(metadata_json,'')) SIMILAR TO
      '%(root_cause|ground_truth|fault_label|expected_diagnosis|correct_action|benchmark_answer)%'
   OR LOWER(COALESCE(dimensions_json,'')) SIMILAR TO
      '%(root_cause|ground_truth|fault_label|expected_diagnosis|correct_action|benchmark_answer)%'
""").fetchone()[0])

text_hits += int(con.execute("""
SELECT COUNT(*)
FROM m05_03_agent_events
WHERE LOWER(COALESCE(message,'')) SIMILAR TO
      '%(root_cause|ground_truth|fault_label|expected_diagnosis|correct_action|benchmark_answer)%'
""").fetchone()[0])

agent_rows = int(
    con.execute(
        "SELECT COUNT(*) FROM m05_03_agent_telemetry"
    ).fetchone()[0]
)

checks = {
    "agent_schema_has_no_hidden_fields": not schema_hits,
    "public_text_has_no_truth_vocabulary": text_hits == 0,
    "agent_row_count_preserved": agent_rows == EXPECTED_CASE_ROWS,
}

print("=" * 108)
print("MODULE 5 — M05_03 CELL 10 — AGENT-VISIBLE SANITIZATION / LEAKAGE GATE")
print("=" * 108)

for name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL':<7} {name}")

print("-" * 108)
print("Agent-visible telemetry rows         :", f"{agent_rows:,}")
print("Agent-visible event rows             :", len(EVENTS_DF))
print("Forbidden schema hits                :", schema_hits)
print("Forbidden text hits                  :", text_hits)
print("Phase / severity visible to agent    : NO")
print("Root-cause labels visible to agent   : NO")

failed = [name for name, passed in checks.items() if not passed]

if failed:
    raise RuntimeError(
        "M05_03 truth-leakage gate failed: " + ", ".join(failed)
    )

M05_03_LEAKAGE_PASS = True

print(
    "\nM05_03_LEAKAGE_PASS                 :",
    M05_03_LEAKAGE_PASS
)
````

###### Observation

The agent-visible telemetry and event datasets passed the leakage gate. All 1,091,584 telemetry rows and 12 event records were preserved, while hidden benchmark fields such as phase, severity and root-cause labels were completely excluded from both schema and text content.


<a id="cell-11"></a>
##### Cell 11 — Recompute Formal Accessibility KPIs for Every Scenario

Reuses the **same deterministic M05_02 formula logic** on the sanitized scenario telemetry: control-plane aggregation, DRB hierarchy, Partial Accessibility, Total-RRC/RRC-Resume supporting ratios, Total Accessibility, and SubNetwork aggregation.

No KPI value is injected directly.


````python
if not globals().get("M05_03_LEAKAGE_PASS",False):
    raise RuntimeError("Cell 11 requires Cell 10.")

con.execute("""
CREATE OR REPLACE TABLE m05_03_control AS
SELECT scenario_id,entity_id,ANY_VALUE(site_id) AS site_id,
       ANY_VALUE(cu_domain_id) AS cu_domain_id,interval_start_utc,
 SUM(CASE WHEN metric_name='RRC.ConnEstabAtt.Cause' AND dimension_cause<>'mo-Signalling'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS rrc_conn_att_service,
 SUM(CASE WHEN metric_name='RRC.ConnEstabSucc.Cause' AND dimension_cause<>'mo-Signalling'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS rrc_conn_succ_service,
 SUM(CASE WHEN metric_name='UECNTX.ConnEstabAtt.Cause' AND dimension_cause<>'mo-Signalling'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS ng_att_service,
 SUM(CASE WHEN metric_name='UECNTX.ConnEstabSucc.Cause' AND dimension_cause<>'mo-Signalling'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS ng_succ_service,
 SUM(CASE WHEN metric_name='RRC.ResumeAtt.Cause'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS resume_att,
 SUM(CASE WHEN metric_name='RRC.ResumeSucc.Cause'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS resume_succ,
 SUM(CASE WHEN metric_name='RRC.ResumeFallbackToSetupAtt.cause' AND dimension_cause<>'mo-Signalling'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS resume_fallback_att_service,
 SUM(CASE WHEN metric_name='RRC.ResumeSuccByFallback.cause' AND dimension_cause<>'mo-Signalling'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS resume_fallback_succ_service,
 SUM(CASE WHEN metric_name='RRC.ResumeFallbackToSetupAtt.cause'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS resume_fallback_att_all,
 SUM(CASE WHEN metric_name='RRC.ResumeSuccByFallback.cause'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS resume_fallback_succ_all,
 SUM(CASE WHEN metric_name='RRC.ReEstabFallbackToSetupAtt'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS reestab_fallback_att,
 SUM(CASE WHEN metric_name='RRC.ReEstabSuccWithoutUeContext'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS reestab_succ_without_ctx
FROM m05_03_agent_telemetry
WHERE metric_name IN (
 'RRC.ConnEstabAtt.Cause','RRC.ConnEstabSucc.Cause',
 'UECNTX.ConnEstabAtt.Cause','UECNTX.ConnEstabSucc.Cause',
 'RRC.ResumeAtt.Cause','RRC.ResumeSucc.Cause',
 'RRC.ResumeFallbackToSetupAtt.cause','RRC.ResumeSuccByFallback.cause',
 'RRC.ReEstabFallbackToSetupAtt','RRC.ReEstabSuccWithoutUeContext'
)
GROUP BY scenario_id,entity_id,interval_start_utc
""")

con.execute("""
CREATE OR REPLACE TABLE m05_03_drb AS
SELECT scenario_id,entity_id,ANY_VALUE(site_id) AS site_id,
       ANY_VALUE(cu_domain_id) AS cu_domain_id,interval_start_utc,
       CAST(dimension_5qi AS VARCHAR) AS five_qi,
 SUM(CASE WHEN metric_name='DRB.InitialEstabAtt.5QI'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS initial_att,
 SUM(CASE WHEN metric_name='DRB.InitialEstabSucc.5QI'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS initial_succ,
 SUM(CASE WHEN metric_name='DRB.EstabAtt.5QI'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS estab_att,
 SUM(CASE WHEN metric_name='DRB.EstabSucc.5QI'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS estab_succ,
 SUM(CASE WHEN metric_name='DRB.ResumeAtt.5QI'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS drb_resume_att,
 SUM(CASE WHEN metric_name='DRB.ResumeSucc.5QI'
          THEN COALESCE(CAST(value_int AS DOUBLE),value_float) ELSE 0 END) AS drb_resume_succ
FROM m05_03_agent_telemetry
WHERE metric_name IN (
 'DRB.InitialEstabAtt.5QI','DRB.InitialEstabSucc.5QI',
 'DRB.EstabAtt.5QI','DRB.EstabSucc.5QI',
 'DRB.ResumeAtt.5QI','DRB.ResumeSucc.5QI'
)
GROUP BY scenario_id,entity_id,interval_start_utc,dimension_5qi
""")
con.execute("ALTER TABLE m05_03_drb ADD COLUMN added_att DOUBLE")
con.execute("UPDATE m05_03_drb SET added_att=estab_att-initial_att")
con.execute("ALTER TABLE m05_03_drb ADD COLUMN added_succ DOUBLE")
con.execute("UPDATE m05_03_drb SET added_succ=estab_succ-initial_succ")

con.execute("""
CREATE OR REPLACE TABLE m05_03_control_ratios AS
SELECT *,
 (rrc_conn_succ_service+resume_fallback_succ_service+reestab_succ_without_ctx)/
 NULLIF(rrc_conn_att_service+resume_fallback_att_service+reestab_fallback_att,0)
 AS rrc_total_setup_sr,
 resume_succ/NULLIF(resume_att-resume_fallback_att_all,0) AS rrc_resume_sr
FROM m05_03_control
""")

con.execute("""
CREATE OR REPLACE TABLE m05_03_partial AS
SELECT d.scenario_id,d.entity_id,d.site_id,d.cu_domain_id,d.interval_start_utc,d.five_qi,
 100.0*(c.rrc_conn_succ_service/NULLIF(c.rrc_conn_att_service,0))
      *(c.ng_succ_service/NULLIF(c.ng_att_service,0))
      *(d.estab_succ/NULLIF(d.estab_att,0))
 AS partial_drb_accessibility_pct
FROM m05_03_drb d
JOIN m05_03_control c USING(scenario_id,entity_id,interval_start_utc)
""")

con.execute("""
CREATE OR REPLACE TABLE m05_03_total AS
SELECT d.scenario_id,d.entity_id,d.site_id,d.cu_domain_id,d.interval_start_utc,d.five_qi,
 d.initial_att,d.initial_succ,d.added_att,d.added_succ,d.drb_resume_att,d.drb_resume_succ,
 c.rrc_total_setup_sr,
 c.ng_succ_service/NULLIF(c.ng_att_service,0) AS ng_connection_sr,
 c.rrc_resume_sr,
 d.initial_att/NULLIF(
   c.rrc_total_setup_sr*(c.ng_succ_service/NULLIF(c.ng_att_service,0)),0
 ) AS initial_service_intent,
 d.drb_resume_att/NULLIF(c.rrc_resume_sr,0) AS resume_service_intent,
 d.initial_succ+d.added_succ+d.drb_resume_succ AS total_success_numerator,
 (
   d.initial_att/NULLIF(
     c.rrc_total_setup_sr*(c.ng_succ_service/NULLIF(c.ng_att_service,0)),0
   )
   +d.added_att
   +d.drb_resume_att/NULLIF(c.rrc_resume_sr,0)
 ) AS total_service_intent_denominator,
 100.0*(d.initial_succ+d.added_succ+d.drb_resume_succ)/
 NULLIF(
   d.initial_att/NULLIF(
     c.rrc_total_setup_sr*(c.ng_succ_service/NULLIF(c.ng_att_service,0)),0
   )
   +d.added_att
   +d.drb_resume_att/NULLIF(c.rrc_resume_sr,0),
   0
 ) AS total_drb_accessibility_pct
FROM m05_03_drb d
JOIN m05_03_control_ratios c USING(scenario_id,entity_id,interval_start_utc)
""")

con.execute("""
CREATE OR REPLACE TABLE m05_03_cell_kpis AS
SELECT p.scenario_id,p.entity_id,p.site_id,p.cu_domain_id,p.interval_start_utc,p.five_qi,
       p.partial_drb_accessibility_pct,t.total_drb_accessibility_pct
FROM m05_03_partial p
JOIN m05_03_total t USING(
 scenario_id,entity_id,site_id,cu_domain_id,interval_start_utc,five_qi
)
ORDER BY scenario_id,interval_start_utc,entity_id,five_qi
""")

con.execute("""
CREATE OR REPLACE TABLE m05_03_subnetwork_kpis AS
SELECT scenario_id,interval_start_utc,five_qi,
 AVG(partial_drb_accessibility_pct) AS partial_drb_accessibility_pct,
 AVG(total_drb_accessibility_pct) AS total_drb_accessibility_pct,
 COUNT(*) AS contributing_nr_cell_cu
FROM m05_03_cell_kpis
GROUP BY scenario_id,interval_start_utc,five_qi
ORDER BY scenario_id,interval_start_utc,five_qi
""")

diag=con.execute("""
SELECT COUNT(*) AS cell_kpi_rows,COUNT(DISTINCT scenario_id) AS scenarios,
 COUNT(DISTINCT entity_id) AS cells,COUNT(DISTINCT five_qi) AS five_qi_values,
 MIN(partial_drb_accessibility_pct) AS partial_min_pct,
 AVG(partial_drb_accessibility_pct) AS partial_mean_pct,
 MAX(partial_drb_accessibility_pct) AS partial_max_pct,
 MIN(total_drb_accessibility_pct) AS total_min_pct,
 AVG(total_drb_accessibility_pct) AS total_mean_pct,
 MAX(total_drb_accessibility_pct) AS total_max_pct
FROM m05_03_cell_kpis
""").df()

print("="*108)
print("MODULE 5 — M05_03 CELL 11 — FORMAL SCENARIO KPI DERIVATION")
print("="*108)
print(diag.round(6).to_string(index=False))
print("-"*108)
print("Formal KPI values directly injected : NO")
print("Formula source                      : M05_02 deterministic logic")

M05_03_SCENARIO_KPI_PASS=True
print("\nM05_03_SCENARIO_KPI_PASS            :",M05_03_SCENARIO_KPI_PASS)
````

###### Observation

Formal Accessibility KPIs were successfully recomputed for all 46,080 scenario/cell/5QI combinations using the unchanged M05_02 deterministic formulas. No KPI values were directly injected; the observed degradation, including minima of 65.32% Partial and 70.66% Total Accessibility, therefore emerges solely from the controlled PM-counter changes introduced by the incident benchmark.

##### Cell 12 — Formal KPI Completeness, Mathematical Validity and Upstream Reproduction

Verifies exact `32 × 30 × 16 × 3 = 46,080` cell-KPI rows, positive denominators, valid 0–100% ranges, Added-DRB hierarchy, all 30 SubNetwork contributors, and exact reproduction of M05_02 in non-injected windows.


````python
if not globals().get("M05_03_SCENARIO_KPI_PASS", False):
    raise RuntimeError("Cell 12 requires Cell 11.")

m05_02_kpi_path_sql = str(NR_CELL_KPI_PATH).replace("\\", "/")

con.execute(f"""
CREATE OR REPLACE VIEW m05_02_cell_kpis_reference AS
SELECT * FROM read_parquet('{m05_02_kpi_path_sql}')
""")

EXPECTED_KPI_ROWS = 32 * 30 * 16 * 3

# ------------------------------------------------------------------
# 1. Formal KPI completeness and valid ranges
# ------------------------------------------------------------------

formal = con.execute("""
SELECT
    COUNT(*) AS rows,
    SUM(
        CASE
            WHEN partial_drb_accessibility_pct IS NULL
              OR NOT isfinite(partial_drb_accessibility_pct)
              OR partial_drb_accessibility_pct < 0
              OR partial_drb_accessibility_pct > 100
            THEN 1 ELSE 0
        END
    ) AS partial_failures,
    SUM(
        CASE
            WHEN total_drb_accessibility_pct IS NULL
              OR NOT isfinite(total_drb_accessibility_pct)
              OR total_drb_accessibility_pct < 0
              OR total_drb_accessibility_pct > 100
            THEN 1 ELSE 0
        END
    ) AS total_failures
FROM m05_03_cell_kpis
""").df()

# ------------------------------------------------------------------
# 2. Denominator and Added-DRB hierarchy validation
# ------------------------------------------------------------------

hier = con.execute("""
SELECT
    SUM(
        CASE
            WHEN total_service_intent_denominator <= 0
              OR total_service_intent_denominator IS NULL
              OR NOT isfinite(total_service_intent_denominator)
            THEN 1 ELSE 0
        END
    ) AS denominator_failures,

    SUM(
        CASE
            WHEN added_att < 0
              OR added_succ < 0
              OR added_succ > added_att
            THEN 1 ELSE 0
        END
    ) AS added_drb_failures

FROM m05_03_total
""").df()

# ------------------------------------------------------------------
# 3. SubNetwork contributor completeness
# ------------------------------------------------------------------

sub = con.execute("""
SELECT
    MIN(contributing_nr_cell_cu) AS min_cells,
    MAX(contributing_nr_cell_cu) AS max_cells
FROM m05_03_subnetwork_kpis
""").df()

# ------------------------------------------------------------------
# 4. Exact reproduction of M05_02 in untouched periods
# ------------------------------------------------------------------

repro = con.execute("""
WITH c AS (

    SELECT
        k.scenario_id,
        t.phase,
        pc.family_internal,

        ABS(
            k.partial_drb_accessibility_pct
            - r.partial_drb_accessibility_pct
        ) AS pd,

        ABS(
            k.total_drb_accessibility_pct
            - r.total_drb_accessibility_pct
        ) AS td

    FROM m05_03_cell_kpis k

    JOIN m05_03_timeline t
      ON k.scenario_id = t.scenario_id
     AND k.interval_start_utc = t.interval_start_utc

    JOIN m05_03_private_cases pc
      ON k.scenario_id = pc.scenario_id

    JOIN m05_02_cell_kpis_reference r
      ON k.entity_id = r.entity_id
     AND k.interval_start_utc = r.interval_start_utc
     AND k.five_qi = CAST(r.five_qi AS VARCHAR)
)

SELECT
    MAX(
        CASE
            WHEN phase IN ('PRE', 'RECOVERY')
            THEN pd ELSE 0
        END
    ) AS pre_recovery_partial_max_diff,

    MAX(
        CASE
            WHEN phase IN ('PRE', 'RECOVERY')
            THEN td ELSE 0
        END
    ) AS pre_recovery_total_max_diff,

    MAX(
        CASE
            WHEN family_internal = 'HEALTHY_CONTROL'
            THEN pd ELSE 0
        END
    ) AS healthy_partial_max_diff,

    MAX(
        CASE
            WHEN family_internal = 'HEALTHY_CONTROL'
            THEN td ELSE 0
        END
    ) AS healthy_total_max_diff

FROM c
""").df()

# ------------------------------------------------------------------
# 5. Validation gates
# ------------------------------------------------------------------

tol = 1e-10

checks = {
    "expected_cell_kpi_rows_46080":
        int(formal.loc[0, "rows"]) == EXPECTED_KPI_ROWS,

    "partial_range_valid":
        int(formal.loc[0, "partial_failures"]) == 0,

    "total_range_valid":
        int(formal.loc[0, "total_failures"]) == 0,

    "total_denominators_valid":
        int(hier.loc[0, "denominator_failures"]) == 0,

    "added_drb_hierarchy_valid":
        int(hier.loc[0, "added_drb_failures"]) == 0,

    "subnetwork_all_30_cells":
        int(sub.loc[0, "min_cells"]) == 30
        and int(sub.loc[0, "max_cells"]) == 30,

    "pre_recovery_reproduces_m05_02":
        float(repro.loc[0, "pre_recovery_partial_max_diff"]) <= tol
        and float(repro.loc[0, "pre_recovery_total_max_diff"]) <= tol,

    "healthy_control_reproduces_m05_02":
        float(repro.loc[0, "healthy_partial_max_diff"]) <= tol
        and float(repro.loc[0, "healthy_total_max_diff"]) <= tol,
}

# ------------------------------------------------------------------
# 6. Output
# ------------------------------------------------------------------

print("=" * 108)
print("MODULE 5 — M05_03 CELL 12 — FORMAL KPI VALIDATION")
print("=" * 108)

for name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL':<7} {name}")

print("-" * 108)

print(
    "Expected / observed KPI rows        :",
    f"{EXPECTED_KPI_ROWS:,} / {int(formal.loc[0, 'rows']):,}"
)

print("\nM05_02 reproduction diagnostics\n" + "-" * 108)
print(repro.to_string(index=False))

failed = [name for name, passed in checks.items() if not passed]

if failed:
    raise RuntimeError(
        "M05_03 formal KPI validation failed: "
        + ", ".join(failed)
    )

M05_03_KPI_VALIDATION_PASS = True

print(
    "\nM05_03_KPI_VALIDATION_PASS          :",
    M05_03_KPI_VALIDATION_PASS
)
````

Formal KPI validation passed all checks across the 46,080 expected cell/5QI records. KPI ranges, denominators, Added-DRB hierarchy and SubNetwork aggregation remain mathematically valid, while PRE, RECOVERY and HEALTHY_CONTROL periods reproduce the authoritative M05_02 KPIs exactly with zero numerical difference.

# Section 6 — Causal Diagnosability and Truth Separation


<a id="cell-13"></a>
##### Cell 13 — Causal Signature and Exclusion-Evidence Validation

For every non-control scenario, at least two declared direct evidence metrics must actually change in the intended causal scope while declared exclusion evidence remains unchanged there. Healthy controls must contain zero injected changes.


````python
if not globals().get("M05_03_KPI_VALIDATION_PASS",False):
    raise RuntimeError("Cell 13 requires Cell 12.")

def changed_expr(a="w"):
    return f"""(
      COALESCE(ABS(
        COALESCE({a}.value_float,CAST({a}.value_int AS DOUBLE))-
        COALESCE({a}.original_value_float,CAST({a}.original_value_int AS DOUBLE))
      ),0)>1e-12
      OR {a}.value_bool IS DISTINCT FROM {a}.original_value_bool
      OR {a}.value_text IS DISTINCT FROM {a}.original_value_text
    )"""

def scope_clause(family):
    sc=INCIDENT_FAMILIES[family]["scope_class"]
    if sc in {"CELL","CELL_5QI"}:
        return "(w.site_id=pc.target_site_id OR w.cu_domain_id=pc.target_cu_domain_id)"
    if sc=="SITE":
        return "w.site_id=pc.target_site_id"
    if sc in {"CU_DOMAIN","CU_DOMAIN_BOUNDARY"}:
        return "w.cu_domain_id=pc.target_cu_domain_id"
    return "TRUE"

audit=[]
for case in PRIVATE_CASES_DF.itertuples(index=False):
    family=case.family_internal
    s=INCIDENT_FAMILIES[family]

    if family=="HEALTHY_CONTROL":
        changed=int(con.execute(f"""
        SELECT COUNT(*) FROM m05_03_working_telemetry w
        WHERE scenario_id='{case.scenario_id}' AND {changed_expr()}
        """).fetchone()[0])
        audit.append({
            "scenario_id":case.scenario_id,"family_internal":family,
            "changed_positive_metrics":0,"declared_direct_metrics":0,
            "changed_exclusion_rows":0,"healthy_injected_changes":changed,
            "positive_evidence_pass":changed==0,"exclusion_evidence_pass":True,
        })
        continue

    sc=scope_clause(family)
    positive_changed=0
    for metric in s["direct_change_metrics"]:
        extra=""
        if s["target_5qi"] is not None and metric.startswith("DRB."):
            extra=f"AND CAST(w.dimension_5qi AS VARCHAR)='{s['target_5qi']}'"
        n=int(con.execute(f"""
        SELECT COUNT(*) FROM m05_03_working_telemetry w
        JOIN m05_03_private_cases pc USING(scenario_id)
        WHERE w.scenario_id='{case.scenario_id}' AND w.phase='INCIDENT'
          AND w.metric_name='{metric}' AND ({sc}) {extra}
          AND {changed_expr()}
        """).fetchone()[0])
        positive_changed += int(n>0)

    exclusion_changes=0
    for metric in s["exclusion_metrics"]:
        exclusion_changes += int(con.execute(f"""
        SELECT COUNT(*) FROM m05_03_working_telemetry w
        JOIN m05_03_private_cases pc USING(scenario_id)
        WHERE w.scenario_id='{case.scenario_id}' AND w.phase='INCIDENT'
          AND w.metric_name='{metric}' AND ({sc}) AND {changed_expr()}
        """).fetchone()[0])

    audit.append({
        "scenario_id":case.scenario_id,"family_internal":family,
        "changed_positive_metrics":positive_changed,
        "declared_direct_metrics":len(s["direct_change_metrics"]),
        "changed_exclusion_rows":exclusion_changes,"healthy_injected_changes":0,
        "positive_evidence_pass":positive_changed>=2,
        "exclusion_evidence_pass":exclusion_changes==0,
    })

INJECTION_AUDIT_DF=pd.DataFrame(audit)
family_audit=INJECTION_AUDIT_DF.groupby("family_internal").agg(
    scenarios=("scenario_id","count"),
    min_changed_positive_metrics=("changed_positive_metrics","min"),
    exclusion_change_rows=("changed_exclusion_rows","sum"),
    healthy_injected_changes=("healthy_injected_changes","sum"),
    positive_pass=("positive_evidence_pass","all"),
    exclusion_pass=("exclusion_evidence_pass","all"),
).reset_index()

checks={
 "all_fault_cases_have_2plus_positive_metrics":
     bool(INJECTION_AUDIT_DF.loc[
         INJECTION_AUDIT_DF["family_internal"]!="HEALTHY_CONTROL","positive_evidence_pass"
     ].all()),
 "all_target_exclusion_evidence_unchanged":
     bool(INJECTION_AUDIT_DF["exclusion_evidence_pass"].all()),
 "healthy_controls_unmodified":
     int(INJECTION_AUDIT_DF.loc[
         INJECTION_AUDIT_DF["family_internal"]=="HEALTHY_CONTROL","healthy_injected_changes"
     ].sum())==0,
}

print("="*108)
print("MODULE 5 — M05_03 CELL 13 — CAUSAL SIGNATURE VALIDATION")
print("="*108)
print(family_audit.to_string(index=False))
print("-"*108)
for n,p in checks.items():print(f"{'PASS' if p else 'FAIL':<7} {n}")

failed=[n for n,p in checks.items() if not p]
if failed:raise RuntimeError("M05_03 causal signature validation failed: "+", ".join(failed))

M05_03_CAUSAL_SIGNATURE_PASS=True
print("\nM05_03_CAUSAL_SIGNATURE_PASS        :",M05_03_CAUSAL_SIGNATURE_PASS)
````

###### Observation

Causal-signature validation passed across all 32 scenarios. Every fault family produced at least two intended direct-evidence changes while all declared exclusion evidence remained unchanged in the target scope, and healthy controls remained fully unmodified. This confirms that the benchmark preserves distinguishable causal patterns rather than generic KPI degradation.

##### Cell 14 — Diagnosability and Competing-Hypothesis Gate

A formal incident is admitted only if its evidence is diagnostically discriminating. This cell also confirms that the intended affected KPI population degrades during each non-control incident relative to its own pre-window.


````python
if not globals().get("M05_03_CAUSAL_SIGNATURE_PASS",False):
    raise RuntimeError("Cell 14 requires Cell 13.")

impact=[]
for case in PRIVATE_CASES_DF.itertuples(index=False):
    family=case.family_internal
    if family=="HEALTHY_CONTROL":
        impact.append({
            "scenario_id":case.scenario_id,"family_internal":family,
            "pre_total_mean_pct":np.nan,"incident_total_mean_pct":np.nan,
            "delta_pct_points":np.nan,"kpi_impact_pass":True,
        })
        continue

    s=INCIDENT_FAMILIES[family]
    target_cells=TARGET_CELLS_DF.loc[
        TARGET_CELLS_DF["scenario_id"]==case.scenario_id,"nr_cell_cu_id"
    ].tolist()
    if not target_cells:
        raise RuntimeError(f"No KPI target cells for {case.scenario_id}")
    qcells=", ".join(f"'{x}'" for x in target_cells)
    q5="" if s["target_5qi"] is None else f"AND k.five_qi='{s['target_5qi']}'"

    d=con.execute(f"""
    SELECT
      AVG(CASE WHEN t.phase='PRE' THEN k.total_drb_accessibility_pct END) AS pre_mean,
      AVG(CASE WHEN t.phase='INCIDENT' THEN k.total_drb_accessibility_pct END) AS incident_mean
    FROM m05_03_cell_kpis k
    JOIN m05_03_timeline t
      ON k.scenario_id=t.scenario_id AND k.interval_start_utc=t.interval_start_utc
    WHERE k.scenario_id='{case.scenario_id}' AND k.entity_id IN ({qcells}) {q5}
    """).df()
    pre=float(d.loc[0,"pre_mean"]); inc=float(d.loc[0,"incident_mean"])
    impact.append({
        "scenario_id":case.scenario_id,"family_internal":family,
        "pre_total_mean_pct":pre,"incident_total_mean_pct":inc,
        "delta_pct_points":inc-pre,"kpi_impact_pass":inc<pre,
    })

KPI_IMPACT_DF=pd.DataFrame(impact)
event_ids=set(EVENTS_DF["scenario_id"])
structural=[]
for family,s in non_control.items():
    if len(s["direct_change_metrics"])<2:structural.append(f"{family}: direct evidence")
    if len(s["exclusion_metrics"])<2:structural.append(f"{family}: exclusion evidence")
    if len(s["competing_hypotheses"])<2:structural.append(f"{family}: competitors")
    if s["event_expected"]:
        ids=set(PRIVATE_CASES_DF.loc[
            PRIVATE_CASES_DF["family_internal"]==family,"scenario_id"
        ])
        if not ids.issubset(event_ids):structural.append(f"{family}: missing expected event")

family_impact=KPI_IMPACT_DF[
    KPI_IMPACT_DF["family_internal"]!="HEALTHY_CONTROL"
].groupby("family_internal").agg(
    cases=("scenario_id","count"),
    mean_pre_pct=("pre_total_mean_pct","mean"),
    mean_incident_pct=("incident_total_mean_pct","mean"),
    mean_delta_pp=("delta_pct_points","mean"),
    all_cases_degrade=("kpi_impact_pass","all"),
).reset_index()

checks={
 "structural_diagnosability_complete":not structural,
 "every_non_control_case_degrades_target_kpi":
     bool(KPI_IMPACT_DF.loc[
         KPI_IMPACT_DF["family_internal"]!="HEALTHY_CONTROL","kpi_impact_pass"
     ].all()),
 "healthy_control_cases_4":
     int((PRIVATE_CASES_DF["family_internal"]=="HEALTHY_CONTROL").sum())==4,
}

print("="*108)
print("MODULE 5 — M05_03 CELL 14 — DIAGNOSABILITY GATE")
print("="*108)
print(family_impact.round(6).to_string(index=False))
print("-"*108)
for n,p in checks.items():print(f"{'PASS' if p else 'FAIL':<7} {n}")
print("Structural failures                 :",structural)

failed=[n for n,p in checks.items() if not p]
if failed:raise RuntimeError("M05_03 diagnosability gate failed: "+", ".join(failed))

M05_03_DIAGNOSABILITY_PASS=True
print("\nM05_03_DIAGNOSABILITY_PASS          :",M05_03_DIAGNOSABILITY_PASS)
````

###### Observation

The diagnosability gate passed for all benchmark families. Every non-control scenario produced a measurable target-KPI degradation, while structural evidence and competing-hypothesis definitions remained complete. The variation in degradation magnitude across fault families supports realistic diagnosis based on causal evidence rather than KPI severity alone.

##### Cell 17 — Immutable Benchmark Output Persistence

Creates new M05_03 artifacts only. If a versioned path already exists, an identical byte artifact is verified; a different artifact causes failure rather than silent overwrite. Parquet products are ordered before serialization.


````python
if not globals().get("M05_03_SPLIT_ISOLATION_PASS",False):
    raise RuntimeError("Cell 17 requires Cell 16.")

import shutil

M05_03_BENCHMARK_DIR.mkdir(parents=True,exist_ok=True)
M05_03_PRIVATE_DIR.mkdir(parents=True,exist_ok=True)
M05_03_MANIFEST_DIR.mkdir(parents=True,exist_ok=True)

AGENT_TELEMETRY_PATH=M05_03_BENCHMARK_DIR/"module5_m05_03_agent_visible_telemetry_v1.parquet"
CELL_KPI_OUTPUT_PATH=M05_03_BENCHMARK_DIR/"module5_m05_03_agent_visible_cell_kpis_v1.parquet"
SUBNETWORK_KPI_OUTPUT_PATH=M05_03_BENCHMARK_DIR/"module5_m05_03_agent_visible_subnetwork_kpis_v1.parquet"
EVENTS_OUTPUT_PATH=M05_03_BENCHMARK_DIR/"module5_m05_03_agent_visible_events_v1.parquet"
PUBLIC_CASES_PATH=M05_03_BENCHMARK_DIR/"module5_m05_03_benchmark_cases_v1.json"

PRIVATE_TRUTH_PATH=M05_03_PRIVATE_DIR/"module5_m05_03_ground_truth_v1.json"
DEVELOPMENT_LABELS_PATH=M05_03_PRIVATE_DIR/"module5_m05_03_development_labels_v1.parquet"
INJECTION_AUDIT_PATH=M05_03_PRIVATE_DIR/"module5_m05_03_injection_audit_summary_v1.parquet"

ROADMAP_AMENDMENT_PATH=M05_03_MANIFEST_DIR/"module5_m05_03_roadmap_amendment_v1.json"

TEMP_DIR=Path("/content/m05_03_persistence_temp")
TEMP_DIR.mkdir(parents=True,exist_ok=True)

def persist_temp_immutable(temp_path,destination):
    temp_path=Path(temp_path);destination=Path(destination)
    new_sha=sha256_file(temp_path)
    if destination.exists():
        if sha256_file(destination)!=new_sha:
            raise FileExistsError(f"Different artifact already exists at {destination}")
        operation="VERIFIED_EXISTING"
    else:
        shutil.copy2(temp_path,destination)
        operation="CREATED"
    return {
        "path":str(destination),"operation":operation,
        "bytes":destination.stat().st_size,
        "artifact_sha256":sha256_file(destination),
    }

def persist_json_semantic_immutable(payload,destination):
    destination=Path(destination)
    serialized=json.dumps(payload,indent=2,sort_keys=True,ensure_ascii=False)+"\n"
    if destination.exists():
        existing=load_json(destination)
        if existing.get("semantic_sha256")!=payload.get("semantic_sha256"):
            raise FileExistsError(f"Different semantic JSON artifact already exists at {destination}")
        operation="VERIFIED_EXISTING"
    else:
        destination.write_text(serialized,encoding="utf-8")
        operation="CREATED"
    return {
        "path":str(destination),"operation":operation,
        "bytes":destination.stat().st_size,
        "artifact_sha256":sha256_file(destination),
        "semantic_sha256":payload.get("semantic_sha256"),
    }

def copy_query(query,temp_name):
    p=TEMP_DIR/temp_name
    if p.exists():p.unlink()
    ps=str(p).replace("\\","/")
    con.execute(f"COPY ({query}) TO '{ps}' (FORMAT PARQUET, COMPRESSION ZSTD)")
    return p

agent_temp=copy_query("""
 SELECT * FROM m05_03_agent_telemetry
 ORDER BY scenario_id,interval_start_utc,metric_name,entity_id,
          COALESCE(dimension_cause,''),
          COALESCE(CAST(dimension_5qi AS VARCHAR),''),
          COALESCE(dimension_snssai,'')
""","agent_visible_telemetry.parquet")

cell_temp=copy_query("""
 SELECT * FROM m05_03_cell_kpis
 ORDER BY scenario_id,interval_start_utc,entity_id,five_qi
""","agent_visible_cell_kpis.parquet")

sub_temp=copy_query("""
 SELECT * FROM m05_03_subnetwork_kpis
 ORDER BY scenario_id,interval_start_utc,five_qi
""","agent_visible_subnetwork_kpis.parquet")

events_temp=copy_query("""
 SELECT * FROM m05_03_agent_events
 ORDER BY scenario_id,event_time_utc,event_metric_name,entity_id
""","agent_visible_events.parquet")

con.register("development_labels_input",DEVELOPMENT_LABELS_DF)
con.execute("CREATE OR REPLACE TABLE m05_03_development_labels AS SELECT * FROM development_labels_input")
dev_temp=copy_query("SELECT * FROM m05_03_development_labels ORDER BY scenario_id",
                    "development_labels.parquet")

con.register("injection_audit_input",INJECTION_AUDIT_DF)
con.execute("CREATE OR REPLACE TABLE m05_03_injection_audit AS SELECT * FROM injection_audit_input")
audit_temp=copy_query("SELECT * FROM m05_03_injection_audit ORDER BY scenario_id",
                      "injection_audit.parquet")

OUTPUT_PERSISTENCE={
 "agent_visible_telemetry":persist_temp_immutable(agent_temp,AGENT_TELEMETRY_PATH),
 "agent_visible_cell_kpis":persist_temp_immutable(cell_temp,CELL_KPI_OUTPUT_PATH),
 "agent_visible_subnetwork_kpis":persist_temp_immutable(sub_temp,SUBNETWORK_KPI_OUTPUT_PATH),
 "agent_visible_events":persist_temp_immutable(events_temp,EVENTS_OUTPUT_PATH),
 "benchmark_cases":persist_json_semantic_immutable(PUBLIC_CASES_PAYLOAD,PUBLIC_CASES_PATH),
 "private_ground_truth":persist_json_semantic_immutable(PRIVATE_TRUTH_PAYLOAD,PRIVATE_TRUTH_PATH),
 "development_labels":persist_temp_immutable(dev_temp,DEVELOPMENT_LABELS_PATH),
 "injection_audit":persist_temp_immutable(audit_temp,INJECTION_AUDIT_PATH),
 "roadmap_amendment":persist_json_semantic_immutable(
     ROADMAP_AMENDMENT_PAYLOAD,ROADMAP_AMENDMENT_PATH
 ),
}

print("="*108)
print("MODULE 5 — M05_03 CELL 17 — IMMUTABLE OUTPUT PERSISTENCE")
print("="*108)
for name,r in OUTPUT_PERSISTENCE.items():
    print(f"\n{name}\n"+"-"*108)
    print("Operation                           :",r["operation"])
    print("Bytes                               :",f"{r['bytes']:,}")
    print("SHA-256                             :",r["artifact_sha256"])
    print("Path                                :",r["path"])

M05_03_OUTPUT_PERSISTENCE_PASS=True
print("\nM05_03_OUTPUT_PERSISTENCE_PASS      :",M05_03_OUTPUT_PERSISTENCE_PASS)
````
