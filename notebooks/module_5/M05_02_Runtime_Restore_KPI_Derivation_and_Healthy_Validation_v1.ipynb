````python
from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import os
import sys

MODULE_ID = "M05_02"
NOTEBOOK_VERSION = "V1"

EXPECTED = {
    "m05_01_freeze_semantic_sha256":
        "4a9f6ecec30613c160b618dfe50e11e63b8f4c8c24d06eac351d5275c744a968",
    "foundation_bundle_index_semantic_sha256":
        "55c8ef51f905945fb825fad3bb35b932ccf7e140f5ffe9d85d3430071ff7143b",
    "foundation_bundle_index_artifact_sha256":
        "bdeee58930d9a13334640132e7a12fa9ace6222a19cfc2c8626e1fd430fd9b8f",
    "foundation_persistence_semantic_sha256":
        "936b605cb1ce3c2d2b2977fed537df65c5e61e19e26ff655f89b27b919d62879",
    "telemetry_qualification_semantic_sha256":
        "fc3c38f980209b1d5a9504b30f98efa6636ba7eedc153ef4f4d1e7bb49be156a",
    "canonical_manifest_artifact_sha256":
        "e856ffe3f7ed97e6ad4592e999e5fd3adf5993f80ac72cdada1f374ab6543411",
    "canonical_dataset_semantic_sha256":
        "c1fb3a49eacd63f45a98986322d536b74d21e585ff8bc49a4649ef11b3b34cd7",
}

SUPERSEDED = {
    "telemetry_v1_semantic_sha256":
        "d6b75444faf53d602edf952f4620fe02f2f3a4ea68431a4889e4829adf008f4d",
    "freeze_v1_semantic_sha256":
        "3a519602da5e211df684786441c41669fbfdaac1c0d13254d18187ad5ce874e8",
}

DRIVE_M05_01_ROOT = Path(
    "/content/drive/MyDrive/Telecom_AI_Engineering_Platform/Module_5/M05_01"
)
FOUNDATION_BUNDLE_DIR = (
    DRIVE_M05_01_ROOT / "foundation" / "foundation_bundle_v1_2"
)
TELEMETRY_DIR = (
    DRIVE_M05_01_ROOT / "telemetry" / "healthy_baseline_v1_1"
)

DRIVE_M05_02_ROOT = Path(
    "/content/drive/MyDrive/Telecom_AI_Engineering_Platform/Module_5/M05_02"
)
M05_02_KPI_DIR = DRIVE_M05_02_ROOT / "kpi"
M05_02_MANIFEST_DIR = DRIVE_M05_02_ROOT / "manifests"

def sha256_file(path, chunk_size=1024 * 1024):
    path = Path(path)
    hasher = hashlib.sha256()
    with path.open("rb") as handle:
        while True:
            chunk = handle.read(chunk_size)
            if not chunk:
                break
            hasher.update(chunk)
    return hasher.hexdigest()

def canonical_sha256(payload):
    raw = json.dumps(
        payload,
        sort_keys=True,
        separators=(",", ":"),
        ensure_ascii=False,
    ).encode("utf-8")
    return hashlib.sha256(raw).hexdigest()

def load_json(path):
    with Path(path).open("r", encoding="utf-8") as handle:
        return json.load(handle)

def embedded_semantic_sha(payload):
    return (
        payload.get("semantic_sha256")
        or payload.get("taxonomy_semantic_sha256")
    )

print("=" * 108)
print("MODULE 5 — M05_02 CELL 01 — AUTHORITATIVE RESTORE CONTRACT")
print("=" * 108)
print(f"Python version                      : {sys.version.split()[0]}")
print(f"Foundation bundle                  : {FOUNDATION_BUNDLE_DIR}")
print(f"Canonical telemetry                : {TELEMETRY_DIR}")
print(f"M05_02 output root                 : {DRIVE_M05_02_ROOT}")
print("Superseded telemetry V1 permitted  : NO")
print("Kaggle restore permitted           : NO")
print("GPU required                       : NO")
print("=" * 108)

M05_02_RESTORE_CONTRACT_DEFINED = True
````

##### Cell 04 — Verify All Frozen Foundation and Control Artifacts

Walks the explicit bundle inventory and independently verifies every indexed file against its frozen artifact SHA and, where available, embedded semantic SHA.


````python
if not globals().get("M05_02_BUNDLE_ROOT_PASS", False):
    raise RuntimeError("Cell 04 requires Cell 03.")

BUNDLE_ENTRIES = BUNDLE_INDEX["files"]

verification_rows = []
failures = []

for entry in BUNDLE_ENTRIES:
    path = FOUNDATION_BUNDLE_DIR / entry["bundle_relative_path"]
    exists = path.exists()

    observed_artifact_sha = sha256_file(path) if exists else None
    artifact_pass = (
        exists
        and observed_artifact_sha == entry["artifact_sha256"]
    )

    expected_semantic = entry.get("semantic_sha256")
    observed_semantic = None
    semantic_pass = True

    if exists and expected_semantic is not None:
        payload = load_json(path)
        observed_semantic = embedded_semantic_sha(payload)
        semantic_pass = observed_semantic == expected_semantic

    passed = exists and artifact_pass and semantic_pass

    verification_rows.append({
        "classification": entry["classification"],
        "role": entry["role"],
        "exists": exists,
        "artifact_pass": artifact_pass,
        "semantic_pass": semantic_pass,
        "passed": passed,
    })

    if not passed:
        failures.append(entry["role"])

foundation_count = sum(
    row["classification"] == "FOUNDATION"
    for row in verification_rows
)
control_count = sum(
    row["classification"] == "CONTROL"
    for row in verification_rows
)

print("=" * 108)
print("MODULE 5 — M05_02 CELL 04 — COMPLETE BUNDLE INVENTORY VERIFICATION")
print("=" * 108)
print(f"Foundation artifacts                : {foundation_count}")
print(f"Control artifacts                   : {control_count}")
print(f"Total indexed artifacts             : {len(verification_rows)}")
print(f"Verification failures               : {len(failures)}")
if failures:
    print(f"Failed roles                        : {failures}")

if foundation_count != 11 or control_count != 4:
    raise RuntimeError("Unexpected V1.2 bundle inventory size.")

if failures:
    raise RuntimeError(
        "Frozen bundle artifact verification failed: "
        + ", ".join(failures)
    )

M05_02_FOUNDATION_BUNDLE_FILES_PASS = True
print("M05_02_FOUNDATION_BUNDLE_FILES_PASS : True")
````

##### Cell 07 — Verify 92-Day Parquet Partition Inventory

Checks every canonical daily Parquet file against the manifest byte hash and row count, then verifies total rows, standardized/operational row accounting and the healthy event-channel artifact.


````python
if not globals().get("M05_02_CANONICAL_MANIFEST_PASS", False):
    raise RuntimeError("Cell 07 requires Cell 06.")

import pyarrow.parquet as pq

partition_failures = []
observed_total_rows = 0

for number, entry in enumerate(
    CANONICAL_MANIFEST["daily_partitions"],
    start=1,
):
    path = TELEMETRY_DIR / entry["relative_path"]

    if not path.exists():
        partition_failures.append(
            f"{entry['date']}: missing"
        )
        continue

    observed_sha = sha256_file(path)
    metadata_rows = pq.ParquetFile(path).metadata.num_rows
    observed_total_rows += int(metadata_rows)

    if observed_sha != entry["parquet_sha256"]:
        partition_failures.append(
            f"{entry['date']}: SHA mismatch"
        )

    if int(metadata_rows) != int(entry["rows"]):
        partition_failures.append(
            f"{entry['date']}: row mismatch"
        )

EVENT_CHANNEL_PATH = TELEMETRY_DIR / CANONICAL_MANIFEST[
    "event_channel"
]["relative_path"]

event_channel_pass = (
    EVENT_CHANNEL_PATH.exists()
    and sha256_file(EVENT_CHANNEL_PATH)
    == CANONICAL_MANIFEST["event_channel"]["sha256"]
)

expected_total_rows = int(
    CANONICAL_MANIFEST["rows"]["total_rows"]
)

checks = {
    "daily_partition_count_92":
        len(CANONICAL_MANIFEST["daily_partitions"]) == 92,
    "partition_byte_and_row_identity":
        len(partition_failures) == 0,
    "total_row_identity":
        observed_total_rows == expected_total_rows,
    "healthy_event_channel_identity":
        event_channel_pass,
}

failed = [name for name, passed in checks.items() if not passed]

print("=" * 108)
print("MODULE 5 — M05_02 CELL 07 — PARQUET INVENTORY VERIFICATION")
print("=" * 108)
for name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL':<7} {name}")
print("-" * 108)
print(f"Verified partitions                 : {len(CANONICAL_MANIFEST['daily_partitions'])}")
print(f"Observed total rows                 : {observed_total_rows:,}")
print(f"Expected total rows                 : {expected_total_rows:,}")
print(f"Partition failures                  : {len(partition_failures)}")

if partition_failures:
    print(partition_failures[:10])

if failed:
    raise RuntimeError(
        "Canonical partition verification failed: "
        + ", ".join(failed)
    )

M05_02_TELEMETRY_RESTORE_PASS = True
print("M05_02_TELEMETRY_RESTORE_PASS       : True")
````

<a id="cell-10"></a>
##### Cell 10 — Qualify Knowledge-to-Telemetry PM Bindings, Dimensions and Entity Scope

Binds every PM dependency from the restored TS 28.554 KPI knowledge contract to the canonical telemetry schema, including explicit reviewed aliases where foundation and telemetry naming differ. It then validates metric availability, 5QI and cause dimensions, NRCellCU scope, and the absence of RNA-update Resume causes in the healthy synthetic baseline.

````python
# ============================================================
# MODULE 5 — M05_02
# CELL 10 — KNOWLEDGE-TO-TELEMETRY PM BINDING QUALIFICATION
# ============================================================

if not globals().get(
    "M05_02_KPI_KNOWLEDGE_RESTORE_PASS",
    False,
):
    raise RuntimeError(
        "Cell 10 requires Cell 09."
    )


# ============================================================
# 1. DERIVE FORMAL PM DEPENDENCIES FROM THE RESTORED
#    STANDARDS-GROUNDED KPI KNOWLEDGE CONTRACT
# ============================================================

FORMAL_KPI_IDS = (
    "partial_drb_accessibility",
    "total_drb_accessibility",
)

knowledge_required_metrics = sorted({
    measurement
    for kpi_id in FORMAL_KPI_IDS
    for measurement
    in KPI_FORMULAS[kpi_id]["required_measurements"]
})


# ============================================================
# 2. EXPLICIT REVIEWED PM NAME ALIASES
# ============================================================
#
# The frozen KPI knowledge artifact preserves TS 28.554
# formula naming for the two RRC Resume measurements:
#
#   RRC.ResumeAtt.cause
#   RRC.ResumeSucc.cause
#
# The canonical synthetic telemetry uses normalized internal names:
#
#   RRC.ResumeAtt.Cause
#   RRC.ResumeSucc.Cause
#
# TS 28.552 itself uses the lowercase `.cause` form for these
# Resume measurements. The uppercase `Cause` variants are an
# explicit internal canonicalization in the M05_01 telemetry schema.
#
# No generic case normalisation is performed.
# Only explicitly reviewed aliases are permitted.
# ============================================================

EXPLICIT_PM_ALIASES = {
    "RRC.ResumeAtt.cause":
        "RRC.ResumeAtt.Cause",

    "RRC.ResumeSucc.cause":
        "RRC.ResumeSucc.Cause",
}


# ============================================================
# 3. BUILD KNOWLEDGE -> TELEMETRY BINDINGS
# ============================================================

KNOWLEDGE_TO_TELEMETRY_BINDINGS = {
    measurement:
        EXPLICIT_PM_ALIASES.get(
            measurement,
            measurement,
        )
    for measurement
    in knowledge_required_metrics
}

FORMAL_TELEMETRY_METRICS = set(
    KNOWLEDGE_TO_TELEMETRY_BINDINGS.values()
)


# ============================================================
# 4. INVENTORY STANDARDIZED METRICS ACTUALLY AVAILABLE
#    IN THE CANONICAL TELEMETRY
# ============================================================

observed_formal_metrics = set(
    con.execute(
        """
        SELECT DISTINCT metric_name
        FROM telemetry
        WHERE is_standardized_metric = TRUE
        """
    )
    .fetchnumpy()["metric_name"]
    .tolist()
)

missing_metrics = sorted(
    FORMAL_TELEMETRY_METRICS
    - observed_formal_metrics
)

binding_failures = sorted(
    knowledge_measurement
    for (
        knowledge_measurement,
        telemetry_metric,
    )
    in KNOWLEDGE_TO_TELEMETRY_BINDINGS.items()
    if telemetry_metric
    not in observed_formal_metrics
)


# ============================================================
# 5. BUILD CAUSE AND 5QI METRIC GROUPS DIRECTLY FROM
#    THE BOUND TELEMETRY DEPENDENCY SET
# ============================================================

cause_metrics = sorted(
    metric
    for metric in FORMAL_TELEMETRY_METRICS
    if metric.lower().endswith(".cause")
)

fiveqi_metrics = sorted(
    metric
    for metric in FORMAL_TELEMETRY_METRICS
    if metric.endswith(".5QI")
)


def sql_string_list(values):
    return ", ".join(
        "'"
        + value.replace("'", "''")
        + "'"
        for value in values
    )


cause_metrics_sql = sql_string_list(
    cause_metrics
)

fiveqi_metrics_sql = sql_string_list(
    fiveqi_metrics
)

formal_metrics_sql = sql_string_list(
    sorted(FORMAL_TELEMETRY_METRICS)
)


# ============================================================
# 6. INSPECT CAUSE DIMENSIONS
# ============================================================

cause_df = con.execute(
    f"""
    SELECT
        metric_name,

        LIST(
            DISTINCT dimension_cause
            ORDER BY dimension_cause
        )
        FILTER (
            WHERE dimension_cause IS NOT NULL
        ) AS causes

    FROM telemetry

    WHERE metric_name IN (
        {cause_metrics_sql}
    )

    GROUP BY metric_name

    ORDER BY metric_name
    """
).df()


# ============================================================
# 7. INSPECT 5QI DIMENSIONS
# ============================================================

fiveqi_df = con.execute(
    f"""
    SELECT
        metric_name,

        LIST(
            DISTINCT CAST(
                dimension_5qi AS VARCHAR
            )
            ORDER BY CAST(
                dimension_5qi AS VARCHAR
            )
        )
        FILTER (
            WHERE dimension_5qi IS NOT NULL
        ) AS five_qi

    FROM telemetry

    WHERE metric_name IN (
        {fiveqi_metrics_sql}
    )

    GROUP BY metric_name

    ORDER BY metric_name
    """
).df()


# ============================================================
# 8. VERIFY ENTITY SCOPE
# ============================================================

scope_df = con.execute(
    f"""
    SELECT
        metric_name,

        LIST(
            DISTINCT entity_type
            ORDER BY entity_type
        ) AS entity_types

    FROM telemetry

    WHERE metric_name IN (
        {formal_metrics_sql}
    )

    GROUP BY metric_name

    ORDER BY metric_name
    """
).df()


# ============================================================
# 9. EXPLICIT ENTITY-SCOPE NORMALISATION
# ============================================================
#
# Foundation / standards-facing knowledge representation:
#
#   NRCellCU
#
# Canonical telemetry EntityType enum representation:
#
#   NR_CELL_CU
#
# They represent the same logical managed-object scope.
# We preserve both representations and explicitly bind them.
# ============================================================

FOUNDATION_NR_CELL_CU_SCOPE = (
    "NRCellCU"
)

CANONICAL_NR_CELL_CU_ENTITY_TYPE = (
    "NR_CELL_CU"
)

ENTITY_SCOPE_BINDINGS = {
    FOUNDATION_NR_CELL_CU_SCOPE:
        CANONICAL_NR_CELL_CU_ENTITY_TYPE
}


# ============================================================
# 10. INSPECT RRC RESUME CAUSES
# ============================================================
#
# TS 28.554 requires exclusion of Resume causes associated
# with RNA update from the RRC Resume success-rate component.
#
# The current synthetic V1.1 baseline is expected to contain
# no such cause.
# ============================================================

observed_resume_causes = set(
    con.execute(
        """
        SELECT DISTINCT
            dimension_cause

        FROM telemetry

        WHERE
            metric_name =
                'RRC.ResumeAtt.Cause'

            AND dimension_cause
                IS NOT NULL
        """
    )
    .fetchnumpy()["dimension_cause"]
    .tolist()
)

rna_like_causes = sorted(
    cause
    for cause in observed_resume_causes
    if "rna" in str(cause).lower()
)


# ============================================================
# 11. CONSOLIDATE OBSERVED 5QI VALUES
# ============================================================

expected_5qi = {
    "1",
    "5",
    "9",
}

observed_5qi = set()

for values in fiveqi_df["five_qi"]:

    if values is not None:

        observed_5qi.update(
            str(value)
            for value in values
        )


# ============================================================
# 12. QUALIFICATION GATES
# ============================================================

checks = {

    "knowledge_dependency_count_16":
        len(
            knowledge_required_metrics
        ) == 16,

    "explicit_alias_count_2":
        len(
            EXPLICIT_PM_ALIASES
        ) == 2,

    "all_knowledge_dependencies_bound":
        len(
            binding_failures
        ) == 0,

    "all_formal_metrics_present":
        len(
            missing_metrics
        ) == 0,

    "expected_5qi_dimension":
        observed_5qi
        == expected_5qi,

    "no_rna_update_causes_in_synthetic_v1_1":
        len(
            rna_like_causes
        ) == 0,

    "complete_entity_scope_inventory":
        len(
            scope_df
        )
        == len(
            FORMAL_TELEMETRY_METRICS
        ),

    "all_formula_metrics_nr_cell_cu":
        (
            len(
                scope_df
            )
            == len(
                FORMAL_TELEMETRY_METRICS
            )

            and all(
                list(
                    entity_types
                )
                == [
                    CANONICAL_NR_CELL_CU_ENTITY_TYPE
                ]

                for entity_types
                in scope_df["entity_types"]
            )
        ),
}


failed = [
    name
    for name, passed
    in checks.items()
    if not passed
]


# ============================================================
# 13. REPORT KNOWLEDGE -> TELEMETRY BINDINGS
# ============================================================

print("=" * 108)

print(
    "MODULE 5 — M05_02 CELL 10 — "
    "KNOWLEDGE-TO-TELEMETRY PM "
    "BINDING QUALIFICATION"
)

print("=" * 108)


print(
    "\nKnowledge -> telemetry bindings"
)

print("-" * 108)


for knowledge_measurement in (
    knowledge_required_metrics
):

    telemetry_metric = (
        KNOWLEDGE_TO_TELEMETRY_BINDINGS[
            knowledge_measurement
        ]
    )

    binding_type = (
        "ALIAS"
        if knowledge_measurement
        in EXPLICIT_PM_ALIASES
        else "EXACT"
    )

    print(
        f"{binding_type:<6} "
        f"{knowledge_measurement:<40} "
        f"-> {telemetry_metric}"
    )


# ============================================================
# 14. REPORT ENTITY-SCOPE BINDING
# ============================================================

print(
    "\nEntity-scope binding"
)

print("-" * 108)

print(
    f"Foundation knowledge scope          : "
    f"{FOUNDATION_NR_CELL_CU_SCOPE}"
)

print(
    f"Canonical telemetry entity type     : "
    f"{CANONICAL_NR_CELL_CU_ENTITY_TYPE}"
)

print(
    "Scope interpretation                : "
    "EQUIVALENT"
)


# ============================================================
# 15. REPORT OBSERVED FORMULA-METRIC ENTITY SCOPES
# ============================================================

print(
    "\nObserved formula metric entity scopes"
)

print("-" * 108)

print(
    scope_df.to_string(
        index=False
    )
)


# ============================================================
# 16. REPORT QUALIFICATION CHECKS
# ============================================================

print(
    "\nQualification checks"
)

print("-" * 108)

for name, passed in checks.items():

    print(
        f"{'PASS' if passed else 'FAIL':<7} "
        f"{name}"
    )


print("-" * 108)

print(
    f"Knowledge dependencies              : "
    f"{len(knowledge_required_metrics)}"
)

print(
    f"Bound telemetry metrics             : "
    f"{len(FORMAL_TELEMETRY_METRICS)}"
)

print(
    f"Explicit reviewed PM aliases        : "
    f"{len(EXPLICIT_PM_ALIASES)}"
)

print(
    f"Explicit entity-scope bindings      : "
    f"{len(ENTITY_SCOPE_BINDINGS)}"
)

print(
    f"Missing telemetry metrics           : "
    f"{missing_metrics}"
)

print(
    f"Binding failures                    : "
    f"{binding_failures}"
)

print(
    f"Observed 5QI values                 : "
    f"{sorted(observed_5qi)}"
)

print(
    f"Observed Resume causes              : "
    f"{sorted(observed_resume_causes)}"
)

print(
    f"RNA-like Resume causes              : "
    f"{rna_like_causes}"
)


# ============================================================
# 17. FINAL GATE
# ============================================================

if failed:

    raise RuntimeError(
        "Knowledge-to-telemetry PM "
        "binding qualification failed: "
        + ", ".join(
            failed
        )
    )


M05_02_FORMULA_BINDING_PASS = True


print(
    "\nM05_02_FORMULA_BINDING_PASS         : True"
)
````

##### Cell 11 — Aggregate RRC and NG Control-Plane Counters

Scans the canonical Parquet telemetry and builds a compact per-NRCellCU/per-15-minute control-plane table for direct RRC setup, NG setup, Resume fallback and re-establishment counters.


````python
if not globals().get("M05_02_FORMULA_BINDING_PASS", False):
    raise RuntimeError("Cell 11 requires Cell 10.")

con.execute("""
CREATE OR REPLACE TABLE m05_02_control AS
SELECT
    entity_id,
    ANY_VALUE(site_id) AS site_id,
    ANY_VALUE(cu_domain_id) AS cu_domain_id,
    interval_start_utc,

    SUM(CASE
        WHEN metric_name = 'RRC.ConnEstabAtt.Cause'
         AND dimension_cause <> 'mo-Signalling'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS rrc_conn_att_service,

    SUM(CASE
        WHEN metric_name = 'RRC.ConnEstabSucc.Cause'
         AND dimension_cause <> 'mo-Signalling'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS rrc_conn_succ_service,

    SUM(CASE
        WHEN metric_name = 'UECNTX.ConnEstabAtt.Cause'
         AND dimension_cause <> 'mo-Signalling'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS ng_att_service,

    SUM(CASE
        WHEN metric_name = 'UECNTX.ConnEstabSucc.Cause'
         AND dimension_cause <> 'mo-Signalling'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS ng_succ_service,

    SUM(CASE
        WHEN metric_name = 'RRC.ResumeAtt.Cause'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS resume_att,

    SUM(CASE
        WHEN metric_name = 'RRC.ResumeSucc.Cause'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS resume_succ,

    SUM(CASE
        WHEN metric_name = 'RRC.ResumeFallbackToSetupAtt.cause'
         AND dimension_cause <> 'mo-Signalling'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS resume_fallback_att_service,

    SUM(CASE
        WHEN metric_name = 'RRC.ResumeSuccByFallback.cause'
         AND dimension_cause <> 'mo-Signalling'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS resume_fallback_succ_service,

    SUM(CASE
        WHEN metric_name = 'RRC.ResumeFallbackToSetupAtt.cause'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS resume_fallback_att_all,

    SUM(CASE
        WHEN metric_name = 'RRC.ResumeSuccByFallback.cause'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS resume_fallback_succ_all,

    SUM(CASE
        WHEN metric_name = 'RRC.ReEstabFallbackToSetupAtt'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS reestab_fallback_att,

    SUM(CASE
        WHEN metric_name = 'RRC.ReEstabSuccWithoutUeContext'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS reestab_succ_without_ctx

FROM telemetry
WHERE metric_name IN (
    'RRC.ConnEstabAtt.Cause',
    'RRC.ConnEstabSucc.Cause',
    'UECNTX.ConnEstabAtt.Cause',
    'UECNTX.ConnEstabSucc.Cause',
    'RRC.ResumeAtt.Cause',
    'RRC.ResumeSucc.Cause',
    'RRC.ResumeFallbackToSetupAtt.cause',
    'RRC.ResumeSuccByFallback.cause',
    'RRC.ReEstabFallbackToSetupAtt',
    'RRC.ReEstabSuccWithoutUeContext'
)
GROUP BY entity_id, interval_start_utc
""")

control_summary = con.execute("""
SELECT
    COUNT(*) AS rows,
    COUNT(DISTINCT entity_id) AS cells,
    COUNT(DISTINCT interval_start_utc) AS intervals,
    MIN(rrc_conn_att_service) AS min_rrc_attempts,
    MIN(ng_att_service) AS min_ng_attempts
FROM m05_02_control
""").df()

print("=" * 108)
print("MODULE 5 — M05_02 CELL 11 — CONTROL-PLANE AGGREGATION")
print("=" * 108)
print(control_summary.to_string(index=False))

if int(control_summary.loc[0, "min_rrc_attempts"]) <= 0:
    raise RuntimeError("Non-positive RRC service-attempt denominator found.")

if int(control_summary.loc[0, "min_ng_attempts"]) <= 0:
    raise RuntimeError("Non-positive NG service-attempt denominator found.")

M05_02_CONTROL_AGG_PASS = True
print("\nM05_02_CONTROL_AGG_PASS             : True")
````

The resulting table contains exactly **264,960 cell-interval records**, corresponding to **30 NRCellCU entities × 8,832 fifteen-minute intervals**. All RRC and NG service-attempt denominators remained positive, with minimum observed values of **231 RRC attempts** and **230 NG attempts**.

##### Cell 12 — Aggregate DRB Counters by 5QI

Builds the per-cell/per-interval/per-5QI Initial DRB, total DRB and DRB Resume counter table, then reconstructs Added DRB attempts/successes as Total minus Initial.


````python
if not globals().get("M05_02_CONTROL_AGG_PASS", False):
    raise RuntimeError("Cell 12 requires Cell 11.")

con.execute("""
CREATE OR REPLACE TABLE m05_02_drb AS
SELECT
    entity_id,
    ANY_VALUE(site_id) AS site_id,
    ANY_VALUE(cu_domain_id) AS cu_domain_id,
    interval_start_utc,
    CAST(dimension_5qi AS VARCHAR) AS five_qi,

    SUM(CASE
        WHEN metric_name = 'DRB.InitialEstabAtt.5QI'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS initial_att,

    SUM(CASE
        WHEN metric_name = 'DRB.InitialEstabSucc.5QI'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS initial_succ,

    SUM(CASE
        WHEN metric_name = 'DRB.EstabAtt.5QI'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS estab_att,

    SUM(CASE
        WHEN metric_name = 'DRB.EstabSucc.5QI'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS estab_succ,

    SUM(CASE
        WHEN metric_name = 'DRB.ResumeAtt.5QI'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS drb_resume_att,

    SUM(CASE
        WHEN metric_name = 'DRB.ResumeSucc.5QI'
        THEN COALESCE(CAST(value_int AS DOUBLE), value_float)
        ELSE 0 END) AS drb_resume_succ

FROM telemetry
WHERE metric_name IN (
    'DRB.InitialEstabAtt.5QI',
    'DRB.InitialEstabSucc.5QI',
    'DRB.EstabAtt.5QI',
    'DRB.EstabSucc.5QI',
    'DRB.ResumeAtt.5QI',
    'DRB.ResumeSucc.5QI'
)
GROUP BY entity_id, interval_start_utc, dimension_5qi
""")

con.execute("""
ALTER TABLE m05_02_drb
ADD COLUMN added_att DOUBLE
""")
con.execute("""
UPDATE m05_02_drb
SET added_att = estab_att - initial_att
""")

con.execute("""
ALTER TABLE m05_02_drb
ADD COLUMN added_succ DOUBLE
""")
con.execute("""
UPDATE m05_02_drb
SET added_succ = estab_succ - initial_succ
""")

drb_diag = con.execute("""
SELECT
    COUNT(*) AS rows,
    COUNT(DISTINCT entity_id) AS cells,
    COUNT(DISTINCT interval_start_utc) AS intervals,
    COUNT(DISTINCT five_qi) AS five_qi_values,
    SUM(CASE WHEN added_att < 0 THEN 1 ELSE 0 END) AS negative_added_att,
    SUM(CASE WHEN added_succ < 0 THEN 1 ELSE 0 END) AS negative_added_succ,
    SUM(CASE WHEN added_succ > added_att THEN 1 ELSE 0 END) AS added_succ_gt_att
FROM m05_02_drb
""").df()

print("=" * 108)
print("MODULE 5 — M05_02 CELL 12 — DRB AGGREGATION")
print("=" * 108)
print(drb_diag.to_string(index=False))

if any(
    int(drb_diag.loc[0, col]) != 0
    for col in [
        "negative_added_att",
        "negative_added_succ",
        "added_succ_gt_att",
    ]
):
    raise RuntimeError(
        "Corrected Added-DRB hierarchy failed."
    )

M05_02_DRB_AGG_PASS = True
print("\nM05_02_DRB_AGG_PASS                 : True")
````

##### Cell 13 — Derive Partial DRB Accessibility

Implements TS 28.554 Clause 6.2.4: RRC setup success ratio × UE-associated NG-connection success ratio × DRB establishment success ratio × 100, excluding `mo-Signalling` from RRC and NG cause sums.


````python
if not globals().get("M05_02_DRB_AGG_PASS", False):
    raise RuntimeError("Cell 13 requires Cell 12.")

con.execute("""
CREATE OR REPLACE TABLE m05_02_partial AS
SELECT
    d.entity_id,
    d.site_id,
    d.cu_domain_id,
    d.interval_start_utc,
    d.five_qi,

    c.rrc_conn_succ_service
        / NULLIF(c.rrc_conn_att_service, 0)
        AS partial_rrc_setup_sr,

    c.ng_succ_service
        / NULLIF(c.ng_att_service, 0)
        AS ng_connection_sr,

    d.estab_succ
        / NULLIF(d.estab_att, 0)
        AS drb_establishment_sr,

    100.0
        * (c.rrc_conn_succ_service
            / NULLIF(c.rrc_conn_att_service, 0))
        * (c.ng_succ_service
            / NULLIF(c.ng_att_service, 0))
        * (d.estab_succ
            / NULLIF(d.estab_att, 0))
        AS partial_drb_accessibility_pct

FROM m05_02_drb AS d
JOIN m05_02_control AS c
USING (entity_id, interval_start_utc)
""")

partial_diag = con.execute("""
SELECT
    COUNT(*) AS rows,
    SUM(CASE
        WHEN partial_drb_accessibility_pct IS NULL
        THEN 1 ELSE 0 END) AS null_kpi,
    SUM(CASE
        WHEN NOT isfinite(partial_drb_accessibility_pct)
        THEN 1 ELSE 0 END) AS nonfinite_kpi,
    SUM(CASE
        WHEN partial_drb_accessibility_pct < 0
          OR partial_drb_accessibility_pct > 100
        THEN 1 ELSE 0 END) AS range_failures,
    MIN(partial_drb_accessibility_pct) AS min_pct,
    AVG(partial_drb_accessibility_pct) AS mean_pct,
    MAX(partial_drb_accessibility_pct) AS max_pct
FROM m05_02_partial
""").df()

print("=" * 108)
print("MODULE 5 — M05_02 CELL 13 — PARTIAL DRB ACCESSIBILITY")
print("=" * 108)
print(partial_diag.round(6).to_string(index=False))

partial_failures = sum(
    int(partial_diag.loc[0, col])
    for col in ["null_kpi", "nonfinite_kpi", "range_failures"]
)

if partial_failures:
    raise RuntimeError(
        "Partial DRB Accessibility validation failed."
    )

M05_02_PARTIAL_KPI_PASS = True
print("\nM05_02_PARTIAL_KPI_PASS             : True")
````

##### Cell 14 — Derive Total-RRC and RRC-Resume Success Ratios

Implements the supporting TS 28.554 Clause 6.2.7 RRC ratios and independently repeats the corrected Resume/fallback/re-establishment hierarchy validation before Total Accessibility is allowed.


````python
if not globals().get("M05_02_PARTIAL_KPI_PASS", False):
    raise RuntimeError("Cell 14 requires Cell 13.")

con.execute("""
CREATE OR REPLACE TABLE m05_02_control_ratios AS
SELECT
    *,

    (
        rrc_conn_succ_service
        + resume_fallback_succ_service
        + reestab_succ_without_ctx
    )
    /
    NULLIF(
        rrc_conn_att_service
        + resume_fallback_att_service
        + reestab_fallback_att,
        0
    ) AS rrc_total_setup_sr,

    resume_succ
    /
    NULLIF(
        resume_att - resume_fallback_att_all,
        0
    ) AS rrc_resume_sr,

    resume_att - resume_succ AS resume_direct_failures,

    (
        resume_att
        - resume_succ
        - resume_fallback_succ_all
    ) AS resume_residual_failure

FROM m05_02_control
""")

hierarchy_diag = con.execute("""
SELECT
    SUM(CASE
        WHEN resume_fallback_att_all > resume_direct_failures
        THEN 1 ELSE 0 END) AS fallback_att_gt_direct_failures,

    SUM(CASE
        WHEN resume_fallback_succ_all > resume_fallback_att_all
        THEN 1 ELSE 0 END) AS fallback_succ_gt_att,

    SUM(CASE
        WHEN resume_succ + resume_fallback_succ_all > resume_att
        THEN 1 ELSE 0 END) AS resume_total_succ_gt_att,

    SUM(CASE
        WHEN resume_residual_failure < 0
        THEN 1 ELSE 0 END) AS negative_resume_residual,

    SUM(CASE
        WHEN reestab_fallback_att > resume_residual_failure
        THEN 1 ELSE 0 END) AS reestab_att_gt_residual,

    SUM(CASE
        WHEN reestab_succ_without_ctx > reestab_fallback_att
        THEN 1 ELSE 0 END) AS reestab_succ_gt_att,

    SUM(CASE
        WHEN rrc_total_setup_sr IS NULL
          OR NOT isfinite(rrc_total_setup_sr)
          OR rrc_total_setup_sr < 0
          OR rrc_total_setup_sr > 1
        THEN 1 ELSE 0 END) AS rrc_total_ratio_failures,

    SUM(CASE
        WHEN rrc_resume_sr IS NULL
          OR NOT isfinite(rrc_resume_sr)
          OR rrc_resume_sr < 0
          OR rrc_resume_sr > 1
        THEN 1 ELSE 0 END) AS rrc_resume_ratio_failures

FROM m05_02_control_ratios
""").df()

ratio_summary = con.execute("""
SELECT
    MIN(rrc_total_setup_sr) * 100 AS rrc_total_min_pct,
    AVG(rrc_total_setup_sr) * 100 AS rrc_total_mean_pct,
    MAX(rrc_total_setup_sr) * 100 AS rrc_total_max_pct,
    MIN(rrc_resume_sr) * 100 AS rrc_resume_min_pct,
    AVG(rrc_resume_sr) * 100 AS rrc_resume_mean_pct,
    MAX(rrc_resume_sr) * 100 AS rrc_resume_max_pct
FROM m05_02_control_ratios
""").df()

print("=" * 108)
print("MODULE 5 — M05_02 CELL 14 — RRC TOTAL / RESUME FORMAL RATIOS")
print("=" * 108)
print("\nHierarchy diagnostics")
print("-" * 108)
print(hierarchy_diag.to_string(index=False))
print("\nRatio summary")
print("-" * 108)
print(ratio_summary.round(6).to_string(index=False))

if int(hierarchy_diag.sum(axis=1).iloc[0]) != 0:
    raise RuntimeError(
        "RRC Resume/re-establishment hierarchy or ratio validation failed."
    )

M05_02_RRC_TOTAL_RESUME_PASS = True
print("\nM05_02_RRC_TOTAL_RESUME_PASS        : True")
````

##### Cell 15 — Derive Total DRB Accessibility

Implements TS 28.554 Clause 6.2.7 using Initial DRB service-intent adjustment, Added DRB attempts, DRB Resume service-intent adjustment, Total-RRC setup SR, NG SR and RRC Resume SR.


````python
if not globals().get("M05_02_RRC_TOTAL_RESUME_PASS", False):
    raise RuntimeError("Cell 15 requires Cell 14.")

con.execute("""
CREATE OR REPLACE TABLE m05_02_total AS
SELECT
    d.entity_id,
    d.site_id,
    d.cu_domain_id,
    d.interval_start_utc,
    d.five_qi,

    d.initial_att,
    d.initial_succ,
    d.added_att,
    d.added_succ,
    d.drb_resume_att,
    d.drb_resume_succ,

    c.rrc_total_setup_sr,
    c.ng_succ_service / NULLIF(c.ng_att_service, 0)
        AS ng_connection_sr,
    c.rrc_resume_sr,

    d.initial_att
        /
        NULLIF(
            c.rrc_total_setup_sr
            * (c.ng_succ_service / NULLIF(c.ng_att_service, 0)),
            0
        )
        AS initial_service_intent,

    d.drb_resume_att
        /
        NULLIF(c.rrc_resume_sr, 0)
        AS resume_service_intent,

    (
        d.initial_succ
        + d.added_succ
        + d.drb_resume_succ
    ) AS total_success_numerator,

    (
        d.initial_att
            /
            NULLIF(
                c.rrc_total_setup_sr
                * (c.ng_succ_service / NULLIF(c.ng_att_service, 0)),
                0
            )
        + d.added_att
        + d.drb_resume_att / NULLIF(c.rrc_resume_sr, 0)
    ) AS total_service_intent_denominator,

    100.0
        * (
            d.initial_succ
            + d.added_succ
            + d.drb_resume_succ
        )
        /
        NULLIF(
            (
                d.initial_att
                    /
                    NULLIF(
                        c.rrc_total_setup_sr
                        * (c.ng_succ_service / NULLIF(c.ng_att_service, 0)),
                        0
                    )
                + d.added_att
                + d.drb_resume_att / NULLIF(c.rrc_resume_sr, 0)
            ),
            0
        )
        AS total_drb_accessibility_pct

FROM m05_02_drb AS d
JOIN m05_02_control_ratios AS c
USING (entity_id, interval_start_utc)
""")

total_diag = con.execute("""
SELECT
    COUNT(*) AS rows,
    SUM(CASE
        WHEN total_service_intent_denominator <= 0
        THEN 1 ELSE 0 END) AS denominator_failures,
    SUM(CASE
        WHEN total_drb_accessibility_pct IS NULL
        THEN 1 ELSE 0 END) AS null_kpi,
    SUM(CASE
        WHEN NOT isfinite(total_drb_accessibility_pct)
        THEN 1 ELSE 0 END) AS nonfinite_kpi,
    SUM(CASE
        WHEN total_drb_accessibility_pct < 0
          OR total_drb_accessibility_pct > 100
        THEN 1 ELSE 0 END) AS range_failures,
    MIN(total_drb_accessibility_pct) AS min_pct,
    AVG(total_drb_accessibility_pct) AS mean_pct,
    MAX(total_drb_accessibility_pct) AS max_pct
FROM m05_02_total
""").df()

print("=" * 108)
print("MODULE 5 — M05_02 CELL 15 — TOTAL DRB ACCESSIBILITY")
print("=" * 108)
print(total_diag.round(6).to_string(index=False))

failure_cols = [
    "denominator_failures",
    "null_kpi",
    "nonfinite_kpi",
    "range_failures",
]

if sum(int(total_diag.loc[0, col]) for col in failure_cols) != 0:
    raise RuntimeError(
        "Total DRB Accessibility validation failed."
    )

M05_02_TOTAL_KPI_PASS = True
print("\nM05_02_TOTAL_KPI_PASS               : True")
````

##### Cell 17 — Establish Healthy Cell and SubNetwork Baselines

Produces descriptive healthy-baseline KPI statistics and applies the TS 28.554 SubNetwork rule by averaging NRCellCU KPI values. These summaries are observational; no arbitrary 3GPP pass threshold is introduced.


````python
if not globals().get("M05_02_FORMAL_KPI_VALIDATION_PASS", False):
    raise RuntimeError("Cell 17 requires Cell 16.")

healthy_summary_df = con.execute("""
SELECT
    five_qi,

    MIN(partial_drb_accessibility_pct) AS partial_min_pct,
    QUANTILE_CONT(partial_drb_accessibility_pct, 0.05) AS partial_p05_pct,
    AVG(partial_drb_accessibility_pct) AS partial_mean_pct,
    QUANTILE_CONT(partial_drb_accessibility_pct, 0.95) AS partial_p95_pct,
    MAX(partial_drb_accessibility_pct) AS partial_max_pct,

    MIN(total_drb_accessibility_pct) AS total_min_pct,
    QUANTILE_CONT(total_drb_accessibility_pct, 0.05) AS total_p05_pct,
    AVG(total_drb_accessibility_pct) AS total_mean_pct,
    QUANTILE_CONT(total_drb_accessibility_pct, 0.95) AS total_p95_pct,
    MAX(total_drb_accessibility_pct) AS total_max_pct

FROM m05_02_cell_kpis
GROUP BY five_qi
ORDER BY five_qi
""").df()

con.execute("""
CREATE OR REPLACE TABLE m05_02_subnetwork_kpis AS
SELECT
    interval_start_utc,
    five_qi,
    AVG(partial_drb_accessibility_pct)
        AS partial_drb_accessibility_pct,
    AVG(total_drb_accessibility_pct)
        AS total_drb_accessibility_pct,
    COUNT(*) AS contributing_nr_cell_cu
FROM m05_02_cell_kpis
GROUP BY interval_start_utc, five_qi
ORDER BY interval_start_utc, five_qi
""")

subnetwork_diag = con.execute("""
SELECT
    COUNT(*) AS rows,
    MIN(contributing_nr_cell_cu) AS min_contributing_cells,
    MAX(contributing_nr_cell_cu) AS max_contributing_cells,
    MIN(partial_drb_accessibility_pct) AS partial_min_pct,
    AVG(partial_drb_accessibility_pct) AS partial_mean_pct,
    MAX(partial_drb_accessibility_pct) AS partial_max_pct,
    MIN(total_drb_accessibility_pct) AS total_min_pct,
    AVG(total_drb_accessibility_pct) AS total_mean_pct,
    MAX(total_drb_accessibility_pct) AS total_max_pct
FROM m05_02_subnetwork_kpis
""").df()

expected_cells = int(
    CANONICAL_MANIFEST["topology"]["logical_cells"]
)

subnetwork_complete = (
    int(subnetwork_diag.loc[0, "min_contributing_cells"]) == expected_cells
    and
    int(subnetwork_diag.loc[0, "max_contributing_cells"]) == expected_cells
)

print("=" * 108)
print("MODULE 5 — M05_02 CELL 17 — HEALTHY BASELINE")
print("=" * 108)
print("\nCell-level descriptive baseline by 5QI")
print("-" * 108)
print(healthy_summary_df.round(6).to_string(index=False))
print("\nSubNetwork baseline")
print("-" * 108)
print(subnetwork_diag.round(6).to_string(index=False))
print("-" * 108)
print(f"All SubNetwork points use {expected_cells} cells : {subnetwork_complete}")

if not subnetwork_complete:
    raise RuntimeError(
        "SubNetwork averaging has incomplete NRCellCU coverage."
    )

M05_02_HEALTHY_VALIDATION_PASS = True
print("M05_02_HEALTHY_VALIDATION_PASS     : True")
````

````python
if not globals().get("M05_02_HEALTHY_VALIDATION_PASS", False):
    raise RuntimeError("Cell 18 requires Cell 17.")

import shutil

M05_02_KPI_DIR.mkdir(parents=True, exist_ok=True)
M05_02_MANIFEST_DIR.mkdir(parents=True, exist_ok=True)

LOCAL_STAGE = Path("/content/module5_m05_02_stage")
if LOCAL_STAGE.exists():
    shutil.rmtree(LOCAL_STAGE)
LOCAL_STAGE.mkdir(parents=True)

CELL_KPI_NAME = (
    "module5_m05_02_nr_cell_cu_accessibility_kpis_v1.parquet"
)
SUBNETWORK_KPI_NAME = (
    "module5_m05_02_subnetwork_accessibility_kpis_v1.parquet"
)

local_cell_kpi = LOCAL_STAGE / CELL_KPI_NAME
local_subnetwork_kpi = LOCAL_STAGE / SUBNETWORK_KPI_NAME

con.execute(f"""
COPY (
    SELECT *
    FROM m05_02_cell_kpis
    ORDER BY interval_start_utc, entity_id, five_qi
)
TO '{str(local_cell_kpi).replace("\\", "/")}'
(FORMAT PARQUET, COMPRESSION ZSTD)
""")

con.execute(f"""
COPY (
    SELECT *
    FROM m05_02_subnetwork_kpis
    ORDER BY interval_start_utc, five_qi
)
TO '{str(local_subnetwork_kpi).replace("\\", "/")}'
(FORMAT PARQUET, COMPRESSION ZSTD)
""")

def persist_immutable(source, destination):
    source = Path(source)
    destination = Path(destination)
    source_sha = sha256_file(source)

    if destination.exists():
        destination_sha = sha256_file(destination)
        if destination_sha != source_sha:
            raise FileExistsError(
                "Different artifact already exists: "
                + str(destination)
            )
        operation = "VERIFIED_EXISTING"
    else:
        shutil.copy2(source, destination)
        destination_sha = sha256_file(destination)
        if destination_sha != source_sha:
            raise RuntimeError(
                "Post-copy SHA verification failed: "
                + str(destination)
            )
        operation = "CREATED"

    return {
        "path": str(destination),
        "sha256": source_sha,
        "bytes": int(destination.stat().st_size),
        "operation": operation,
    }

CELL_KPI_PERSISTENCE = persist_immutable(
    local_cell_kpi,
    M05_02_KPI_DIR / CELL_KPI_NAME,
)

SUBNETWORK_KPI_PERSISTENCE = persist_immutable(
    local_subnetwork_kpi,
    M05_02_KPI_DIR / SUBNETWORK_KPI_NAME,
)

print("=" * 108)
print("MODULE 5 — M05_02 CELL 18 — KPI OUTPUT PERSISTENCE")
print("=" * 108)
for label, record in [
    ("NRCellCU KPI", CELL_KPI_PERSISTENCE),
    ("SubNetwork KPI", SUBNETWORK_KPI_PERSISTENCE),
]:
    print(f"\n{label}")
    print("-" * 108)
    print(f"Operation                           : {record['operation']}")
    print(f"Bytes                               : {record['bytes']:,}")
    print(f"SHA-256                             : {record['sha256']}")
    print(f"Path                                : {record['path']}")

M05_02_KPI_FILES_PERSISTED = True
print("\nM05_02_KPI_FILES_PERSISTED         : True")
````
