#### Module 5 — Agentic Telecom AI: M05_01 — Foundation + Healthy Telemetry

**Structured filename:** `M05_01_Foundation_and_Healthy_Telemetry_v1.ipynb`

This notebook builds, qualifies, persists and freezes the **5G RAN Accessibility foundation** for Module 5. It converts standards-aligned KPI knowledge into a causal investigation graph, telemetry contracts, deterministic synthetic topology, causal-observability configuration, a qualified 7-day pilot, a canonical 92-day healthy 15-minute telemetry baseline, verified private Kaggle persistence, provenance reconciliation, a frozen M05_01 root-of-trust manifest and a durable foundation-artifact bundle for runtime restoration.

> **Core principle:** The LLM orchestrates; deterministic telemetry supplies network truth. Standards knowledge provides grounding. ML supplies probabilistic evidence. Policy controls actions. Verification closes the loop.

For the full architecture, implementation roadmap, causal-observability rationale, fault-family design and V1 → V2 → live transition, use:

`M05_00_Implementation_and_Causal_Observability_Reference_v1.ipynb`

The uploaded M05_00 reference used for this review has byte SHA:

```text
9c4db1552791510d3237fd8c7ff93cd7ae66d614d42fed0da7a268bf77fa910e
```

This matches the authoritative M05_00 identity reconciled in Cell 16 and frozen in Cell 17.

##### Execution and rebuild protocol

The notebook records a **two-pass Cell 13 workflow**, so a clean rebuild is not a single uninterrupted top-to-bottom run:

1. Restart the runtime and execute Cells 01–12.
2. Run Cell 13 in `PILOT_7D` mode.
3. Run Cell 14 and require `PILOT_QUALIFICATION_PASS = True`.
4. Return to Cell 13 and rerun the unchanged generation engine in `CANONICAL_92D` mode after qualification authorization.
5. Continue with Cells 15–18 for canonical persistence, provenance reconciliation, foundation freeze and durable foundation-artifact persistence.

The saved notebook state reflects the completed canonical second pass. Cell 15 and Cell 18 are idempotent: when their private Kaggle datasets already exist, they verify the existing remote artifacts rather than silently uploading or creating new versions.

##### Historical internal identifiers

Several persisted Python variables retain identifiers established during earlier development numbering, such as `cell5_v1_2`, `cell7_v1_1` and `cell8_v1_1`. These names are intentionally preserved because renaming them would disturb frozen provenance without changing network truth. The current Cell 01–18 numbering and user-facing descriptions are authoritative.

<a id="cell-01-runtime-preflight"></a>
##### Cell 01 — Runtime Preflight

Inspect Python, platform, GPU/CUDA state, installed package versions, import health and `pip` dependency status **without modifying the runtime**. The output determines what must be installed or pinned in Cell 02.


````python
# ============================================================
# MODULE 5 — CELL 01
# FOUNDATION RUNTIME PREFLIGHT
# ============================================================
#
# Module:
#   Agentic Telecom AI
#
# Notebook:
#   M05_01 — Foundation and Healthy Telemetry
#
# Purpose:
#
#   Inspect the current CPU runtime before configuring the
#   Module 5 foundation environment.
#
#   This notebook requires only the packages needed to:
#
#       - define schemas and configuration
#       - construct KPI / telemetry artifacts
#       - generate deterministic synthetic telemetry
#       - persist Parquet / JSON artifacts
#       - calculate and verify hashes
#
# IMPORTANT:
#
#   - Diagnostic only.
#   - Does NOT install or modify packages.
#   - GPU / CUDA are NOT required or inspected.
#   - MCP, FAISS, LangGraph and LLM runtimes are out of scope.
#
# ============================================================


import sys
import os
import platform
import subprocess
import importlib
import importlib.metadata as metadata

from pathlib import Path


# ============================================================
# 1.1 — BASIC RUNTIME INFORMATION
# ============================================================

print("=" * 88)

print(
    "MODULE 5 — CELL 01 "
    "FOUNDATION RUNTIME PREFLIGHT"
)

print("=" * 88)


print(
    "\n[1] PYTHON / PLATFORM"
)

print("-" * 88)

print(
    f"Python version     : "
    f"{sys.version.split()[0]}"
)

print(
    f"Python executable  : "
    f"{sys.executable}"
)

print(
    f"Platform           : "
    f"{platform.platform()}"
)

print(
    f"Processor          : "
    f"{platform.processor() or 'Not reported'}"
)

print(
    f"Working directory  : "
    f"{Path.cwd()}"
)


# ============================================================
# 1.2 — NOTEBOOK ENVIRONMENT
# ============================================================

print(
    "\n[2] NOTEBOOK ENVIRONMENT"
)

print("-" * 88)


environment = "UNKNOWN"


if (
    "COLAB_RELEASE_TAG" in os.environ
    or "COLAB_GPU" in os.environ
):

    environment = "GOOGLE COLAB"


elif (
    "KAGGLE_KERNEL_RUN_TYPE"
    in os.environ
    or Path("/kaggle").exists()
):

    environment = "KAGGLE"


elif (
    "JPY_PARENT_PID"
    in os.environ
):

    environment = "JUPYTER / IPYTHON"


print(
    f"Detected environment : "
    f"{environment}"
)


# ============================================================
# 1.3 — FOUNDATION PACKAGE INVENTORY
# ============================================================

print(
    "\n[3] FOUNDATION PACKAGE VERSIONS"
)

print("-" * 88)


FOUNDATION_PACKAGES = [

    "numpy",

    "pandas",

    "pydantic",

    "pyarrow",

    "jedi",
]


installed_versions = {}


for package in FOUNDATION_PACKAGES:

    try:

        installed_versions[
            package
        ] = metadata.version(
            package
        )

    except metadata.PackageNotFoundError:

        installed_versions[
            package
        ] = "NOT INSTALLED"


width = max(

    len(package)

    for package
    in installed_versions
)


for (
    package,
    version
) in installed_versions.items():

    print(
        f"{package:<{width}} : "
        f"{version}"
    )


# ============================================================
# 1.4 — CRITICAL IMPORT TEST
# ============================================================

print(
    "\n[4] FOUNDATION IMPORT TEST"
)

print("-" * 88)


FOUNDATION_IMPORTS = {

    "numpy":
        "numpy",

    "pandas":
        "pandas",

    "pydantic":
        "pydantic",

    "pyarrow":
        "pyarrow",
}


import_results = {}


for (
    label,
    module_name
) in FOUNDATION_IMPORTS.items():

    try:

        importlib.import_module(
            module_name
        )

        import_results[
            label
        ] = True


        print(
            f"{label:<20} : PASS"
        )


    except Exception as exc:

        import_results[
            label
        ] = False


        print(

            f"{label:<20} : "
            f"FAIL "
            f"({type(exc).__name__}: "
            f"{str(exc)[:120]})"
        )


# ============================================================
# 1.5 — PARQUET ENGINE STATUS
# ============================================================

print(
    "\n[5] PARQUET SUPPORT"
)

print("-" * 88)


PYARROW_AVAILABLE = (
    import_results.get(
        "pyarrow",
        False,
    )
)


print(
    "PyArrow available      : "
    f"{PYARROW_AVAILABLE}"
)


print(
    "Parquet write expected : "
    f"{PYARROW_AVAILABLE}"
)


# ============================================================
# 1.6 — CURRENT DEPENDENCY HEALTH
# ============================================================

print(
    "\n[6] PIP DEPENDENCY HEALTH"
)

print("-" * 88)


try:

    result = subprocess.run(

        [
            sys.executable,
            "-m",
            "pip",
            "check",
        ],

        capture_output=True,

        text=True,

        timeout=60,
    )


    pip_check_output = (

        result.stdout
        + result.stderr

    ).strip()


    PIP_CHECK_PASS = (
        result.returncode
        == 0
    )


    if PIP_CHECK_PASS:

        print(
            "pip check : PASS"
        )

        print(
            pip_check_output
            or
            "No broken requirements found."
        )


    else:

        print(
            "pip check : ISSUES DETECTED"
        )

        print(
            pip_check_output
        )


except Exception as exc:

    PIP_CHECK_PASS = False


    print(
        f"pip check failed : "
        f"{exc}"
    )


# ============================================================
# 1.7 — PREFLIGHT READINESS
# ============================================================

FOUNDATION_IMPORTS_READY = all(

    import_results.values()
)


CELL02_ACTION_REQUIRED = (

    not FOUNDATION_IMPORTS_READY

    or

    not PIP_CHECK_PASS
)


print(
    "\n[7] FOUNDATION PREFLIGHT SUMMARY"
)

print("-" * 88)

print(
    "Foundation imports ready : "
    f"{FOUNDATION_IMPORTS_READY}"
)

print(
    "PyArrow ready            : "
    f"{PYARROW_AVAILABLE}"
)

print(
    "Dependency health        : "
    f"{PIP_CHECK_PASS}"
)

print(
    "Cell 02 action required  : "
    f"{CELL02_ACTION_REQUIRED}"
)


print(
    "\nOut-of-scope for M05_01"
)

print("-" * 88)

print(
    "GPU / CUDA               : "
    "NOT REQUIRED"
)

print(
    "Torch / Transformers     : "
    "NOT REQUIRED"
)

print(
    "FAISS                    : "
    "NOT REQUIRED"
)

print(
    "MCP                      : "
    "NOT REQUIRED"
)

print(
    "LangGraph                : "
    "NOT REQUIRED"
)

print(
    "LLM runtime / vLLM       : "
    "NOT REQUIRED"
)


print(
    "\n" + "=" * 88
)

print(
    "CELL 01 COMPLETE — "
    "FOUNDATION RUNTIME PREFLIGHT CAPTURED"
)

print("=" * 88)


print(
    "\nNo packages were installed or modified."
)
````

<a id="cell-02-runtime-initialization"></a>
##### Cell 02 — Runtime Initialization

Install and validate the qualified **CPU foundation stack**, establish deterministic seeds and artifact paths, and persist the runtime manifest. Full RAG restoration, Granite inference and GPU-intensive execution remain deferred to later Module 5 notebooks.


````python
# ============================================================
# MODULE 5 — CELL 02
# FOUNDATION CPU RUNTIME + GLOBAL CONFIGURATION
# ============================================================
#
# Module:
#   Agentic Telecom AI
#
# Notebook:
#   M05_01 — Foundation and Healthy Telemetry
#
# Purpose:
#   1. Resolve only missing foundation dependencies.
#   2. Configure deterministic CPU execution.
#   3. Create the Module 5 artifact structure.
#   4. Validate Parquet read/write capability.
#   5. Save the foundation runtime manifest.
#
# IMPORTANT:
#   - GPU is NOT required.
#   - MCP / FAISS / LangGraph are NOT required.
#   - LLM / vLLM / Transformers are NOT required.
# ============================================================


# ============================================================
# 2.1 — IMPORTS
# ============================================================

import sys
import os
import json
import random
import platform
import subprocess
import importlib
import importlib.metadata as metadata

from pathlib import Path
from datetime import datetime, timezone


# ============================================================
# 2.2 — FOUNDATION DEPENDENCY RESOLUTION
# ============================================================

FOUNDATION_REQUIREMENTS = {
    "numpy": "numpy",
    "pandas": "pandas",
    "pydantic": "pydantic",
    "pyarrow": "pyarrow",
    "jedi": "jedi",
}

installed_during_cell = []


for import_name, pip_name in FOUNDATION_REQUIREMENTS.items():

    try:
        importlib.import_module(import_name)

    except ImportError:

        subprocess.run(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                "-q",
                pip_name,
            ],
            check=True,
        )

        installed_during_cell.append(pip_name)


# ============================================================
# 2.3 — FOUNDATION PACKAGE IMPORTS
# ============================================================

import numpy as np
import pandas as pd
import pydantic
import pyarrow


# ============================================================
# 2.4 — EXECUTION MODE
# ============================================================

MODULE5_MODE = "FOUNDATION_CPU"

GPU_REQUIRED = False

assert MODULE5_MODE == "FOUNDATION_CPU"


# ============================================================
# 2.5 — REPRODUCIBILITY
# ============================================================

GLOBAL_SEED = 42

random.seed(GLOBAL_SEED)
np.random.seed(GLOBAL_SEED)

os.environ["PYTHONHASHSEED"] = str(GLOBAL_SEED)


# ============================================================
# 2.6 — TELEMETRY CONTROLS
# ============================================================

PM_INTERVAL_MINUTES = 15

NRT_DEFAULT_WINDOW_MINUTES = 5


# ============================================================
# 2.7 — ARTIFACT STRUCTURE
# ============================================================

MODULE5_ROOT = Path(
    "/content/module5_artifacts"
)


PATHS = {
    "root": MODULE5_ROOT,
    "config": MODULE5_ROOT / "config",
    "kpi": MODULE5_ROOT / "kpi",
    "telemetry": MODULE5_ROOT / "telemetry",
    "scenarios": MODULE5_ROOT / "scenarios",
    "ml": MODULE5_ROOT / "ml",
    "checkpoints": MODULE5_ROOT / "checkpoints",
    "formal_runs": MODULE5_ROOT / "formal_runs",
    "analysis": MODULE5_ROOT / "analysis",
    "manifests": MODULE5_ROOT / "manifests",
    "logs": MODULE5_ROOT / "logs",
}


for path in PATHS.values():

    path.mkdir(
        parents=True,
        exist_ok=True,
    )


# ============================================================
# 2.8 — PACKAGE VERSION SNAPSHOT
# ============================================================

PACKAGE_NAMES = [
    "numpy",
    "pandas",
    "pydantic",
    "pyarrow",
    "jedi",
]


package_versions = {}


for package in PACKAGE_NAMES:

    try:
        package_versions[package] = metadata.version(package)

    except metadata.PackageNotFoundError:
        package_versions[package] = "NOT INSTALLED"


# ============================================================
# 2.9 — PARQUET ROUND-TRIP VALIDATION
# ============================================================

parquet_test_path = (
    PATHS["analysis"]
    / "_module5_parquet_runtime_test.parquet"
)


parquet_test_df = pd.DataFrame(
    {
        "value": [1, 2, 3]
    }
)


parquet_test_df.to_parquet(
    parquet_test_path,
    index=False,
)


parquet_test_readback = pd.read_parquet(
    parquet_test_path
)


assert (
    parquet_test_readback["value"].tolist()
    == [1, 2, 3]
)


parquet_test_path.unlink()


PARQUET_READY = True


# ============================================================
# 2.10 — DEPENDENCY HEALTH
# ============================================================

pip_check = subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "check",
    ],
    capture_output=True,
    text=True,
)


if pip_check.returncode != 0:

    raise RuntimeError(
        "Foundation dependency health check failed:\n"
        + pip_check.stdout
        + pip_check.stderr
    )


DEPENDENCY_HEALTH_READY = True


# ============================================================
# 2.11 — FOUNDATION RUNTIME MANIFEST
# ============================================================

runtime_manifest = {

    "module":
        "Module 5 — Agentic Telecom AI",

    "notebook":
        "M05_01_Foundation_and_Healthy_Telemetry_v1",

    "execution_mode":
        MODULE5_MODE,

    "created_utc":
        datetime.now(timezone.utc).isoformat(),

    "python": {
        "version": sys.version.split()[0],
        "executable": sys.executable,
        "platform": platform.platform(),
    },

    "hardware": {
        "gpu_required": False,
        "gpu_used": False,
    },

    "reproducibility": {
        "global_seed": GLOBAL_SEED,
    },

    "telemetry": {
        "pm_interval_minutes": PM_INTERVAL_MINUTES,
        "nrt_default_window_minutes":
            NRT_DEFAULT_WINDOW_MINUTES,
    },

    "packages":
        package_versions,

    "capabilities": {
        "parquet_ready": PARQUET_READY,
        "dependency_health_ready":
            DEPENDENCY_HEALTH_READY,
        "gpu_required": False,
        "mcp_required": False,
        "faiss_required": False,
        "llm_runtime_required": False,
    },
}


# ============================================================
# 2.12 — SAVE RUNTIME MANIFEST
# ============================================================

runtime_manifest_path = (
    PATHS["manifests"]
    / "module5_foundation_cpu_runtime_manifest.json"
)


with open(
    runtime_manifest_path,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        runtime_manifest,
        f,
        indent=2,
        ensure_ascii=False,
    )


# ============================================================
# 2.13 — SUMMARY
# ============================================================

print("=" * 88)

print(
    "MODULE 5 — CELL 02 "
    "FOUNDATION CPU RUNTIME"
)

print("=" * 88)


print("\nExecution")
print("-" * 88)

print(
    f"Mode                         : "
    f"{MODULE5_MODE}"
)

print(
    f"Python                       : "
    f"{sys.version.split()[0]}"
)

print(
    f"GPU required                 : "
    f"{GPU_REQUIRED}"
)

print(
    f"Global seed                  : "
    f"{GLOBAL_SEED}"
)


print("\nFoundation capabilities")
print("-" * 88)

print(
    f"Parquet ready                : "
    f"{PARQUET_READY}"
)

print(
    f"Dependency health            : "
    f"{DEPENDENCY_HEALTH_READY}"
)

print(
    "MCP required                 : False"
)

print(
    "FAISS required               : False"
)

print(
    "LLM runtime required         : False"
)


print("\nFoundation packages")
print("-" * 88)

for package, version in package_versions.items():

    print(
        f"{package:<28} : "
        f"{version}"
    )


print("\nPackages installed by Cell 02")
print("-" * 88)

print(
    installed_during_cell
    if installed_during_cell
    else "None"
)


print("\nArtifact configuration")
print("-" * 88)

print(
    f"Artifact root                : "
    f"{MODULE5_ROOT}"
)

print(
    f"Runtime manifest             : "
    f"{runtime_manifest_path}"
)


print("\n" + "=" * 88)

print(
    "CELL 02 COMPLETE — "
    "FOUNDATION CPU ENVIRONMENT READY"
)

print("=" * 88)
````

<a id="section-2-canonical-standards-foundation"></a>
#### Section 2 — Canonical Standards Foundation

The foundation remains deliberately broader than the first RAN experiment. It defines reusable 5G KPI, measurement, entity and observation concepts so later Module 5 work can extend beyond Accessibility without redesigning the canonical model.


<a id="cell-03-canonical-schema"></a>
##### Cell 03 — 3GPP-Aligned KPI Taxonomy + Canonical Telemetry Schema

Define the broad 5G KPI taxonomy, standards references, entity types, PM interval and canonical telemetry records. This is the standards-facing foundation reused by all downstream Accessibility artifacts.


````python
# ============================================================
# MODULE 5 — CELL 03
# 3GPP-ALIGNED 5G E2E KPI TAXONOMY
# + CANONICAL TELEMETRY SCHEMA
# ============================================================
#
# Primary standards basis:
#
#   3GPP TS 28.554 v19.7.0 Release 19
#   Management and orchestration;
#   5G end to end Key Performance Indicators (KPI)
#
# Supporting standards:
#
#   3GPP TS 28.552 v19.7.0 Release 19
#   Management and orchestration;
#   5G performance measurements
#
#   3GPP TS 28.541 v19.7.0 Release 19
#   Management and orchestration;
#   5G Network Resource Model (NRM)
#
#   3GPP TS 28.545 v17.0.0 Release 17
#   Management and orchestration;
#   Fault Supervision (FS)
#
# Design principle:
#
#   Preserve standards terminology in the canonical truth layer.
#   Experimental/benchmark abstractions must NOT overwrite
#   standards-facing classifications.
#
# IMPORTANT:
#   - This cell does NOT generate telemetry.
#   - This cell does NOT build the KPI dependency graph.
#   - This cell does NOT define fault scenarios.
#   - This cell does NOT expose MCP tools.
#   - This cell does NOT load RAG or any LLM.
# ============================================================


# ============================================================
# 4.1 — IMPORTS
# ============================================================

from enum import Enum
from typing import Any, Dict, List, Optional, Literal
from datetime import datetime, timezone

from pydantic import (
    BaseModel,
    Field,
    ConfigDict,
    model_validator,
)

import json
import hashlib


# ============================================================
# 4.2 — STANDARDS REFERENCE MODEL
# ============================================================

class StandardsReference(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    organization: str

    specification: str

    title: str

    release: Optional[str] = None

    version: Optional[str] = None

    clause: Optional[str] = None

    reference_role: Literal[
        "PRIMARY_KPI_STANDARD",
        "PERFORMANCE_MEASUREMENT_STANDARD",
        "NETWORK_RESOURCE_MODEL",
        "FAULT_SUPERVISION_STANDARD",
        "TERMINOLOGY_REFERENCE",
        "INFORMATIVE_REFERENCE",
    ]


# ============================================================
# 4.3 — PINNED MODULE 5 STANDARDS BASELINES
# ============================================================
#
# The experiment uses explicit standards snapshots so that
# later changes to 3GPP specifications do not silently alter
# the meaning of a formal Module 5 run.
# ============================================================

TS_28_554_REL19 = StandardsReference(

    organization="3GPP",

    specification="TS 28.554",

    title=(
        "Management and orchestration; "
        "5G end to end Key Performance Indicators (KPI)"
    ),

    release="Release 19",

    version="19.7.0",

    clause="4-5",

    reference_role="PRIMARY_KPI_STANDARD",
)


TS_28_552_REL19 = StandardsReference(

    organization="3GPP",

    specification="TS 28.552",

    title=(
        "Management and orchestration; "
        "5G performance measurements"
    ),

    release="Release 19",

    version="19.7.0",

    clause=None,

    reference_role="PERFORMANCE_MEASUREMENT_STANDARD",
)


TS_28_541_REL19 = StandardsReference(

    organization="3GPP",

    specification="TS 28.541",

    title=(
        "Management and orchestration; "
        "5G Network Resource Model (NRM); "
        "Stage 2 and stage 3"
    ),

    release="Release 19",

    version="19.7.0",

    clause=None,

    reference_role="NETWORK_RESOURCE_MODEL",
)


TS_28_545_REL17 = StandardsReference(

    organization="3GPP",

    specification="TS 28.545",

    title=(
        "Management and orchestration; "
        "Fault Supervision (FS)"
    ),

    release="Release 17",

    version="17.0.0",

    clause=None,

    reference_role="FAULT_SUPERVISION_STANDARD",
)


STANDARD_BASELINES = {

    "kpi":
        TS_28_554_REL19,

    "performance_measurements":
        TS_28_552_REL19,

    "network_resource_model":
        TS_28_541_REL19,

    "fault_supervision":
        TS_28_545_REL17,
}


# ============================================================
# 4.4 — CANONICAL NETWORK ENTITY TYPES
# ============================================================
#
# The entity model is intentionally an operational abstraction
# rather than a complete implementation of the 3GPP NRM.
#
# Network-slice objects are included explicitly because
# TS 28.554 KPI object definitions make extensive use of
# SubNetwork, NetworkSlice and NetworkSliceSubnet objects.
# ============================================================

class EntityType(str, Enum):

    NETWORK = "NETWORK"

    SUBNETWORK = "SUBNETWORK"

    NETWORK_SLICE = "NETWORK_SLICE"

    NETWORK_SLICE_SUBNET = "NETWORK_SLICE_SUBNET"

    SITE = "SITE"

    CELL = "CELL"

    NR_CELL_DU = "NR_CELL_DU"

    NR_CELL_CU = "NR_CELL_CU"

    UE_GROUP = "UE_GROUP"

    AMF = "AMF"

    SMF = "SMF"

    UPF = "UPF"

    NETWORK_FUNCTION = "NETWORK_FUNCTION"

    KUBERNETES_CLUSTER = "KUBERNETES_CLUSTER"

    KUBERNETES_NAMESPACE = "KUBERNETES_NAMESPACE"

    KUBERNETES_WORKLOAD = "KUBERNETES_WORKLOAD"

    KUBERNETES_POD = "KUBERNETES_POD"

    TRANSPORT_LINK = "TRANSPORT_LINK"

    SERVICE_ENDPOINT = "SERVICE_ENDPOINT"


# ============================================================
# 4.5 — 3GPP TS 28.554 KPI CATEGORIES
# ============================================================
#
# 3GPP TS 28.554 v19.7.0 Release 19, Clause 4:
#
#   Accessibility
#   Integrity
#   Utilization
#   Retainability
#   Mobility
#   Energy Efficiency
#   Reliability
#   Availability
#
# IMPORTANT:
#   These names are preserved separately and are NOT collapsed
#   into an experimental Module 5 category.
# ============================================================

class StandardKPICategory(str, Enum):

    ACCESSIBILITY = "ACCESSIBILITY"

    INTEGRITY = "INTEGRITY"

    UTILIZATION = "UTILIZATION"

    RETAINABILITY = "RETAINABILITY"

    MOBILITY = "MOBILITY"

    ENERGY_EFFICIENCY = "ENERGY_EFFICIENCY"

    RELIABILITY = "RELIABILITY"

    AVAILABILITY = "AVAILABILITY"


# ============================================================
# 4.6 — OPERATIONAL CATEGORY DESCRIPTIONS
# ============================================================
#
# These are concise Module 5 operational paraphrases.
#
# They are NOT intended to replace normative 3GPP/ITU-T
# definitions. When exact standards wording is required,
# the agent should retrieve the cited specification.
# ============================================================

KPI_CATEGORY_DESCRIPTIONS = {

    StandardKPICategory.ACCESSIBILITY.value:
        "Ability to successfully obtain access to required "
        "network or service resources.",

    StandardKPICategory.INTEGRITY.value:
        "Performance quality experienced while the requested "
        "network or service function is being provided.",

    StandardKPICategory.UTILIZATION.value:
        "Extent to which network, service or infrastructure "
        "resources are being consumed.",

    StandardKPICategory.RETAINABILITY.value:
        "Ability to retain an established service or connection "
        "until normal completion.",

    StandardKPICategory.MOBILITY.value:
        "Performance associated with maintaining service and "
        "network continuity during mobility.",

    StandardKPICategory.ENERGY_EFFICIENCY.value:
        "Relationship between useful network/service output "
        "and energy consumption.",

    StandardKPICategory.RELIABILITY.value:
        "Ability of a network or service function to perform "
        "as required under stated conditions.",

    StandardKPICategory.AVAILABILITY.value:
        "Degree to which the relevant network or service resource "
        "is available for use when required.",
}


# ============================================================
# 4.7 — KPI MEASUREMENT METHOD
# ============================================================
#
# TS 28.554 v19.7.0 Clause 5 defines the KPI measurement
# method as one of:
#
#   MEAN
#   RATIO
#   CUM
#
# OTHER is retained only for experimental/non-standard
# Module 5 metrics that may coexist with standards KPIs.
# ============================================================

class KPIMeasurementMethod(str, Enum):

    MEAN = "MEAN"

    RATIO = "RATIO"

    CUM = "CUM"

    OTHER = "OTHER"


# ============================================================
# 4.8 — TELEMETRY PLANES
# ============================================================

class TelemetryPlane(str, Enum):

    PM_15MIN = "PM_15MIN"

    NEAR_REAL_TIME = "NEAR_REAL_TIME"


# ============================================================
# 4.9 — OBSERVATION TYPES
# ============================================================

class ObservationType(str, Enum):

    KPI = "KPI"

    COUNTER = "COUNTER"

    MEASUREMENT = "MEASUREMENT"

    ALARM = "ALARM"

    EVENT = "EVENT"

    LOG = "LOG"

    NF_HEALTH = "NF_HEALTH"

    RESOURCE = "RESOURCE"

    TOPOLOGY = "TOPOLOGY"

    ACTION = "ACTION"


# ============================================================
# 4.10 — TECHNICAL MEASUREMENT DOMAINS
# ============================================================
#
# This is a Module 5 operational construct.
#
# It is intentionally separate from the KPI category because:
#
#   KPI symptom domain != root-cause domain
#
# Example:
#
#   Integrity degradation
#       ↓
#   root cause may be RAN, transport, Core or cloud platform.
# ============================================================

class MeasurementDomain(str, Enum):

    RAN = "RAN"

    TRANSPORT = "TRANSPORT"

    CORE = "CORE"

    CLOUD_PLATFORM = "CLOUD_PLATFORM"

    SERVICE = "SERVICE"

    CROSS_DOMAIN = "CROSS_DOMAIN"


# ============================================================
# 4.11 — DATA SOURCE / CONFIDENCE / SEVERITY
# ============================================================

class DataSourceType(str, Enum):

    SYNTHETIC = "SYNTHETIC"

    DERIVED = "DERIVED"

    OPEN_SOURCE_NETWORK = "OPEN_SOURCE_NETWORK"

    EXTERNAL_DATASET = "EXTERNAL_DATASET"


class ConfidenceLevel(str, Enum):

    VALIDATED = "VALIDATED"

    PROBABLE = "PROBABLE"

    UNVERIFIED = "UNVERIFIED"


class Severity(str, Enum):

    INFO = "INFO"

    WARNING = "WARNING"

    MINOR = "MINOR"

    MAJOR = "MAJOR"

    CRITICAL = "CRITICAL"


# ============================================================
# 4.12 — COMMON PROVENANCE MODEL
# ============================================================

class Provenance(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    source_type: DataSourceType

    source_name: str

    source_version: Optional[str] = None

    generator_version: Optional[str] = None

    confidence: ConfidenceLevel = (
        ConfidenceLevel.VALIDATED
    )

    notes: Optional[str] = None


# ============================================================
# 4.13 — CANONICAL NETWORK ENTITY MODEL
# ============================================================

class NetworkEntity(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    entity_id: str

    entity_type: EntityType

    name: Optional[str] = None

    parent_id: Optional[str] = None

    site_id: Optional[str] = None

    cell_id: Optional[str] = None

    attributes: Dict[str, Any] = Field(
        default_factory=dict
    )


# ============================================================
# 4.14 — KPI DEFINITION MODEL
# ============================================================
#
# This structure mirrors the major information concepts in
# the TS 28.554 KPI definition template:
#
#   Name
#   Description
#   Measurement result
#   Measurement method
#   Formula
#   KPI object
#
# It also links KPI formulas to performance measurements,
# consistent with the TS 28.554 → TS 28.552 relationship.
# ============================================================

class KPIDefinition(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    kpi_id: str

    name: str

    standard_category: StandardKPICategory

    domain: MeasurementDomain

    description: str

    measurement_method: KPIMeasurementMethod

    unit: Optional[str] = None

    result_range_min: Optional[float] = None

    result_range_max: Optional[float] = None

    directionality: Literal[
        "HIGHER_IS_BETTER",
        "LOWER_IS_BETTER",
        "CONTEXT_DEPENDENT",
    ]

    aggregation_period_minutes: Optional[int] = None

    source_measurements: List[str] = Field(
        default_factory=list
    )

    numerator_counter: Optional[str] = None

    denominator_counter: Optional[str] = None

    formula: Optional[str] = None

    applicable_entity_types: List[EntityType]

    standards_references: List[
        StandardsReference
    ] = Field(
        default_factory=list
    )

    metadata: Dict[str, Any] = Field(
        default_factory=dict
    )


# ============================================================
# 4.15 — PM COUNTER RECORD
# ============================================================

class PMCounterRecord(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    timestamp: datetime

    interval_minutes: int = Field(
        default=PM_INTERVAL_MINUTES,
        gt=0,
    )

    entity_id: str

    entity_type: EntityType

    domain: MeasurementDomain

    counter_name: str

    value: float

    unit: str

    provenance: Provenance


# ============================================================
# 4.16 — PM KPI RECORD
# ============================================================

class PMKPIRecord(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    timestamp: datetime

    interval_minutes: int = Field(
        default=PM_INTERVAL_MINUTES,
        gt=0,
    )

    entity_id: str

    entity_type: EntityType

    standard_category: StandardKPICategory

    domain: MeasurementDomain

    kpi_id: str

    kpi_name: str

    measurement_method: KPIMeasurementMethod

    value: float

    unit: Optional[str] = None

    numerator: Optional[float] = None

    denominator: Optional[float] = None

    provenance: Provenance

    @model_validator(mode="after")
    def validate_ratio_fields(self):

        if (
            self.denominator is not None
            and self.denominator < 0
        ):
            raise ValueError(
                "KPI denominator cannot be negative."
            )

        if (
            self.measurement_method
            == KPIMeasurementMethod.RATIO
            and self.denominator == 0
        ):
            raise ValueError(
                "RATIO KPI denominator cannot be zero."
            )

        return self


# ============================================================
# 4.17 — NEAR-REAL-TIME MEASUREMENT RECORD
# ============================================================

class MeasurementRecord(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    timestamp: datetime

    entity_id: str

    entity_type: EntityType

    domain: MeasurementDomain

    measurement_group: str

    measurement_name: str

    value: float

    unit: str

    provenance: Provenance


# ============================================================
# 4.18 — ALARM RECORD
# ============================================================

class AlarmRecord(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    alarm_id: str

    timestamp: datetime

    entity_id: str

    entity_type: EntityType

    domain: MeasurementDomain

    alarm_type: str

    severity: Severity

    probable_cause: Optional[str] = None

    description: str

    active: bool = True

    cleared_at: Optional[datetime] = None

    provenance: Provenance


# ============================================================
# 4.19 — NETWORK EVENT RECORD
# ============================================================

class NetworkEventRecord(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    event_id: str

    timestamp: datetime

    entity_id: str

    entity_type: EntityType

    domain: MeasurementDomain

    event_type: str

    description: str

    attributes: Dict[str, Any] = Field(
        default_factory=dict
    )

    provenance: Provenance


# ============================================================
# 4.20 — NETWORK LOG RECORD
# ============================================================

class NetworkLogRecord(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    log_id: str

    timestamp: datetime

    entity_id: str

    entity_type: EntityType

    domain: MeasurementDomain

    component: str

    level: Literal[
        "DEBUG",
        "INFO",
        "WARNING",
        "ERROR",
        "CRITICAL",
    ]

    message: str

    structured_fields: Dict[str, Any] = Field(
        default_factory=dict
    )

    provenance: Provenance


# ============================================================
# 4.21 — NETWORK FUNCTION HEALTH RECORD
# ============================================================

class NFHealthRecord(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    timestamp: datetime

    entity_id: str

    entity_type: EntityType

    nf_type: str

    status: Literal[
        "HEALTHY",
        "DEGRADED",
        "UNAVAILABLE",
        "RESTARTING",
        "UNKNOWN",
    ]

    cpu_utilization_pct: Optional[float] = Field(
        default=None,
        ge=0,
        le=100,
    )

    memory_utilization_pct: Optional[float] = Field(
        default=None,
        ge=0,
        le=100,
    )

    restart_count: int = Field(
        default=0,
        ge=0,
    )

    active_sessions: Optional[int] = Field(
        default=None,
        ge=0,
    )

    provenance: Provenance


# ============================================================
# 4.22 — RESOURCE UTILIZATION RECORD
# ============================================================

class ResourceRecord(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    timestamp: datetime

    entity_id: str

    entity_type: EntityType

    domain: MeasurementDomain

    resource_name: str

    utilization_pct: Optional[float] = Field(
        default=None,
        ge=0,
        le=100,
    )

    absolute_value: Optional[float] = None

    unit: Optional[str] = None

    provenance: Provenance


# ============================================================
# 4.23 — TOPOLOGY RELATIONSHIP
# ============================================================

class TopologyRelation(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    relation_id: str

    source_entity_id: str

    target_entity_id: str

    relation_type: str

    attributes: Dict[str, Any] = Field(
        default_factory=dict
    )

    provenance: Provenance


# ============================================================
# 4.24 — CANONICAL TABLE REGISTRY
# ============================================================

CANONICAL_TABLES = {

    "network_entities":
        NetworkEntity,

    "pm_counters":
        PMCounterRecord,

    "pm_kpis":
        PMKPIRecord,

    "measurements":
        MeasurementRecord,

    "alarms":
        AlarmRecord,

    "network_events":
        NetworkEventRecord,

    "network_logs":
        NetworkLogRecord,

    "nf_health":
        NFHealthRecord,

    "resource_utilization":
        ResourceRecord,

    "network_topology":
        TopologyRelation,
}


# ============================================================
# 4.25 — STANDARDS-ALIGNED KPI TAXONOMY MANIFEST
# ============================================================

kpi_taxonomy = {

    "taxonomy_id":
        "module5_3gpp_aligned_5g_kpi_taxonomy_v1",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "primary_standard": {
        "organization":
            TS_28_554_REL19.organization,

        "specification":
            TS_28_554_REL19.specification,

        "release":
            TS_28_554_REL19.release,

        "version":
            TS_28_554_REL19.version,

        "clause":
            TS_28_554_REL19.clause,
    },

    "standards_baselines": {
        name:
            reference.model_dump(
                mode="json"
            )

        for name, reference
        in STANDARD_BASELINES.items()
    },

    "pm_interval_minutes":
        PM_INTERVAL_MINUTES,

    "standard_kpi_categories": {

        category.value:
            KPI_CATEGORY_DESCRIPTIONS[
                category.value
            ]

        for category
        in StandardKPICategory
    },

    "entity_types": [
        item.value
        for item in EntityType
    ],

    "measurement_domains": [
        item.value
        for item in MeasurementDomain
    ],

    "telemetry_planes": [
        item.value
        for item in TelemetryPlane
    ],

    "kpi_measurement_methods": [
        item.value
        for item in KPIMeasurementMethod
    ],

    "canonical_tables":
        list(
            CANONICAL_TABLES.keys()
        ),

    "design_notes": [

        (
            "Standard KPI category names are preserved "
            "from 3GPP TS 28.554."
        ),

        (
            "Availability and Energy Efficiency remain "
            "separate canonical categories."
        ),

        (
            "Reliability remains a separate canonical "
            "category."
        ),

        (
            "Module 5 operational groupings, if later used, "
            "must not overwrite standards classifications."
        ),

        (
            "Exact normative definitions remain retrievable "
            "through the knowledge layer rather than being "
            "reproduced in this schema."
        ),
    ],
}

# ============================================================
# DETERMINISTIC TAXONOMY CONTENT HASH
# ============================================================
#
# Runtime metadata such as created_utc must not affect the
# semantic identity of the taxonomy.
# ============================================================

def canonical_json_sha256(data):

    canonical = json.dumps(
        data,
        sort_keys=True,
        separators=(",", ":"),
        ensure_ascii=False,
    ).encode("utf-8")

    return hashlib.sha256(
        canonical
    ).hexdigest()


taxonomy_semantic_content = {
    key: value
    for key, value in kpi_taxonomy.items()
    if key != "created_utc"
}

taxonomy_semantic_sha = canonical_json_sha256(
    taxonomy_semantic_content
)

kpi_taxonomy[
    "taxonomy_semantic_sha256"
] = taxonomy_semantic_sha

# ============================================================
# 4.26 — GENERATE JSON SCHEMAS
# ============================================================

schema_bundle = {

    table_name:
        model.model_json_schema()

    for table_name, model
    in CANONICAL_TABLES.items()
}


# ============================================================
# 4.27 — SAVE TAXONOMY + SCHEMA ARTIFACTS
# ============================================================

taxonomy_path = (
    PATHS["kpi"]
    / "module5_3gpp_aligned_5g_kpi_taxonomy_v1.json"
)

schema_path = (
    PATHS["config"]
    / "module5_canonical_telemetry_schema_v1.json"
)


with open(
    taxonomy_path,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        kpi_taxonomy,
        f,
        indent=2,
        ensure_ascii=False,
    )


with open(
    schema_path,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        schema_bundle,
        f,
        indent=2,
        ensure_ascii=False,
    )


# ============================================================
# 4.28 — HASH ARTIFACTS
# ============================================================

def sha256_file(path):

    hasher = hashlib.sha256()

    with open(path, "rb") as f:

        for chunk in iter(
            lambda: f.read(
                1024 * 1024
            ),
            b"",
        ):

            hasher.update(chunk)

    return hasher.hexdigest()


taxonomy_sha = sha256_file(
    taxonomy_path
)

schema_sha = sha256_file(
    schema_path
)


# ============================================================
# 4.29 — SAMPLE VALIDATION OBJECTS
# ============================================================
#
# These objects test the schema only.
#
# They are NOT yet part of the synthetic network dataset or
# the formal Module 5 KPI catalogue.
# ============================================================

synthetic_provenance = Provenance(

    source_type=
        DataSourceType.SYNTHETIC,

    source_name=
        "module5_synthetic_engine",

    source_version=
        "v1",

    generator_version=
        "development",

    confidence=
        ConfidenceLevel.VALIDATED,

    notes=
        "Schema validation sample only.",
)


sample_entity = NetworkEntity(

    entity_id=
        "AMF_01",

    entity_type=
        EntityType.AMF,

    name=
        "Module 5 Development AMF",

    parent_id=
        "5GC_01",

    attributes={
        "network_function": "AMF",
        "environment": "synthetic",
    },
)


sample_kpi_definition = KPIDefinition(

    kpi_id=
        "development_accessibility_ratio",

    name=
        "Development Accessibility Ratio",

    standard_category=
        StandardKPICategory.ACCESSIBILITY,

    domain=
        MeasurementDomain.CORE,

    description=(
        "Development-only ratio KPI used to validate "
        "the Module 5 standards-shaped KPI schema."
    ),

    measurement_method=
        KPIMeasurementMethod.RATIO,

    unit=
        "percent",

    result_range_min=
        0,

    result_range_max=
        100,

    directionality=
        "HIGHER_IS_BETTER",

    aggregation_period_minutes=
        PM_INTERVAL_MINUTES,

    source_measurements=[
        "development_success_count",
        "development_attempt_count",
    ],

    numerator_counter=
        "development_success_count",

    denominator_counter=
        "development_attempt_count",

    formula=(
        "development_success_count / "
        "development_attempt_count * 100"
    ),

    applicable_entity_types=[
        EntityType.AMF
    ],

    standards_references=[
        TS_28_554_REL19,
        TS_28_552_REL19,
    ],

    metadata={
        "formal_kpi": False,
        "purpose": "schema_validation_only",
    },
)


sample_counter = PMCounterRecord(

    timestamp=datetime(
        2026,
        9,
        5,
        12,
        0,
        tzinfo=timezone.utc,
    ),

    entity_id=
        "AMF_01",

    entity_type=
        EntityType.AMF,

    domain=
        MeasurementDomain.CORE,

    counter_name=
        "development_attempt_count",

    value=
        1000,

    unit=
        "count",

    provenance=
        synthetic_provenance,
)


sample_kpi = PMKPIRecord(

    timestamp=datetime(
        2026,
        9,
        5,
        12,
        0,
        tzinfo=timezone.utc,
    ),

    entity_id=
        "AMF_01",

    entity_type=
        EntityType.AMF,

    standard_category=
        StandardKPICategory.ACCESSIBILITY,

    domain=
        MeasurementDomain.CORE,

    kpi_id=
        "development_accessibility_ratio",

    kpi_name=
        "Development Accessibility Ratio",

    measurement_method=
        KPIMeasurementMethod.RATIO,

    value=
        98.7,

    unit=
        "percent",

    numerator=
        987,

    denominator=
        1000,

    provenance=
        synthetic_provenance,
)


# ============================================================
# 4.30 — FINAL VALIDATION
# ============================================================

assert len(
    StandardKPICategory
) == 8


assert set(
    category.value
    for category
    in StandardKPICategory
) == {
    "ACCESSIBILITY",
    "INTEGRITY",
    "UTILIZATION",
    "RETAINABILITY",
    "MOBILITY",
    "ENERGY_EFFICIENCY",
    "RELIABILITY",
    "AVAILABILITY",
}


assert len(
    CANONICAL_TABLES
) == 10


assert (
    TS_28_554_REL19.version
    == "19.7.0"
)


assert (
    TS_28_552_REL19.version
    == "19.7.0"
)


assert (
    sample_kpi.standard_category
    == StandardKPICategory.ACCESSIBILITY
)


assert (
    sample_kpi.measurement_method
    == KPIMeasurementMethod.RATIO
)


assert (
    sample_counter.interval_minutes
    == PM_INTERVAL_MINUTES
)


# ============================================================
# 4.31 — SUMMARY
# ============================================================

print("=" * 88)

print(
    "MODULE 5 — CELL 03 "
    "3GPP-ALIGNED KPI + TELEMETRY ARCHITECTURE"
)

print("=" * 88)


print("\nPrimary standards baseline")
print("-" * 88)

print(
    f"{TS_28_554_REL19.specification} | "
    f"{TS_28_554_REL19.release} | "
    f"v{TS_28_554_REL19.version}"
)


print("\n3GPP KPI categories")
print("-" * 88)

for category in StandardKPICategory:

    print(
        f"{category.value:<24} : "
        f"{KPI_CATEGORY_DESCRIPTIONS[category.value]}"
    )


print("\nKPI measurement methods")
print("-" * 88)

for method in KPIMeasurementMethod:

    print(
        method.value
    )


print("\nCanonical telemetry tables")
print("-" * 88)

for table_name, model in CANONICAL_TABLES.items():

    print(
        f"{table_name:<28} : "
        f"{model.__name__}"
    )


print("\nCanonical entities")
print("-" * 88)

for entity_type in EntityType:

    print(
        entity_type.value
    )


print("\nSample schema validation")
print("-" * 88)

print(
    f"Entity                   : "
    f"{sample_entity.entity_id}"
)

print(
    f"KPI category             : "
    f"{sample_kpi.standard_category.value}"
)

print(
    f"KPI method               : "
    f"{sample_kpi.measurement_method.value}"
)

print(
    f"KPI value                : "
    f"{sample_kpi.value}%"
)


print("\nStandards references")
print("-" * 88)

for name, ref in STANDARD_BASELINES.items():

    print(
        f"{name:<28} : "
        f"{ref.specification} | "
        f"{ref.release} | "
        f"v{ref.version}"
    )


print("\nArtifacts")
print("-" * 88)

print(
    f"Taxonomy                 : "
    f"{taxonomy_path}"
)



print(
    f"Telemetry schema         : "
    f"{schema_path}"
)

print(
    f"Schema SHA-256           : "
    f"{schema_sha}"
)


print("\n" + "=" * 88)

print(
    "CELL 03 COMPLETE — "
    "3GPP-ALIGNED CANONICAL TELECOM SCHEMA READY"
)

print("=" * 88)
````

<a id="cell-04-accessibility-domain-v1"></a>
##### Cell 04 — 5G RAN Accessibility Reference Domain V1

Create the initial standards-grounded Accessibility dependency graph around **Partial DRB Accessibility** and **Total DRB Accessibility**, while representing RACH as a supporting pre-RRC operational diagnostic rather than a fabricated TS 28.554 Accessibility KPI.


````python
# ============================================================
# MODULE 5 — CELL 04
# 5G RAN ACCESSIBILITY REFERENCE DOMAIN
# + KPI DEPENDENCY GRAPH
# ============================================================
#
# Purpose:
#   1. Establish 5G RAN Accessibility as the first Module 5
#      standards-grounded troubleshooting reference domain.
#   2. Define selected 3GPP Accessibility KPIs that are directly
#      meaningful to NG-RAN / NRCellCU investigation.
#   3. Link high-level KPIs to RAN KPI components.
#   4. Link components to standardized PM measurements.
#   5. Link measurements/components to RAN procedures.
#   6. Add operational RAN diagnostics without falsely
#      representing them as standardized TS 28.554 KPIs.
#   7. Persist a deterministic machine-readable dependency graph.
#
# Primary standards baseline:
#
#   3GPP TS 28.554 v19.7.0 Release 19
#   3GPP TS 28.552 v19.7.0 Release 19
#
# Supporting procedure specifications:
#
#   3GPP TS 38.331 — NR RRC protocol
#   3GPP TS 38.413 — NG-RAN NGAP
#   3GPP TS 38.321 — NR MAC protocol / Random Access
#
# IMPORTANT:
#
#   - Module 5 V1 focuses on 5G RAN investigation.
#   - Cell 4 remains the broader 5G canonical standards/schema layer.
#   - This cell does NOT expose hidden fault ground truth.
#   - This cell does NOT generate telemetry.
#   - This cell does NOT define formal benchmark incidents.
#   - RACH Success behaviour is operational/derived evidence,
#     NOT labelled as a standardized TS 28.554 KPI.
#   - Core/E2E KPIs such as Registration, PDU Session and PFCP
#     are intentionally outside the Cell 5 formal reference set.
# ============================================================


# ============================================================
# 5.1 — IMPORTS + CELL 4 DEPENDENCY VALIDATION
# ============================================================

from enum import Enum
from typing import Any, Dict, List
from datetime import datetime, timezone

from pydantic import (
    BaseModel,
    Field,
    ConfigDict,
)

import json
import hashlib


REQUIRED_CELL4_OBJECTS = [

    "StandardsReference",
    "KPIDefinition",

    "StandardKPICategory",
    "KPIMeasurementMethod",

    "MeasurementDomain",
    "EntityType",

    "TS_28_554_REL19",
    "TS_28_552_REL19",

    "PM_INTERVAL_MINUTES",
    "PATHS",
    "sha256_file",
]


missing_cell4_objects = [

    name
    for name in REQUIRED_CELL4_OBJECTS

    if name not in globals()
]


if missing_cell4_objects:

    raise RuntimeError(
        "Cell 5 requires Cell 4 to be executed first. "
        f"Missing objects: {missing_cell4_objects}"
    )


# ============================================================
# 5.2 — SUPPORTING RAN PROCEDURE REFERENCES
# ============================================================
#
# TS 28.554 / TS 28.552 remain the pinned KPI/PM truth layer.
#
# Exact supporting-procedure version pinning can be added when
# the Cell 6 KPI / measurement knowledge supplement is frozen.
# ============================================================


TS_38_331_REL19 = StandardsReference(

    organization="3GPP",

    specification="TS 38.331",

    title=(
        "NR; Radio Resource Control (RRC); "
        "Protocol specification"
    ),

    release="Release 19",

    version=None,

    clause=None,

    reference_role="INFORMATIVE_REFERENCE",
)


TS_38_413_REL19 = StandardsReference(

    organization="3GPP",

    specification="TS 38.413",

    title=(
        "NG-RAN; NG Application Protocol (NGAP)"
    ),

    release="Release 19",

    version=None,

    clause=None,

    reference_role="INFORMATIVE_REFERENCE",
)


TS_38_321_REL19 = StandardsReference(

    organization="3GPP",

    specification="TS 38.321",

    title=(
        "NR; Medium Access Control (MAC) "
        "protocol specification"
    ),

    release="Release 19",

    version=None,

    clause=None,

    reference_role="INFORMATIVE_REFERENCE",
)


# ============================================================
# 5.3 — RAN ACCESSIBILITY GRAPH ENUMS
# ============================================================

class RANAccessibilityNodeLayer(str, Enum):

    KPI = "KPI"

    COMPONENT = "COMPONENT"

    MEASUREMENT = "MEASUREMENT"

    PROCEDURE = "PROCEDURE"

    OPERATIONAL_DIAGNOSTIC = (
        "OPERATIONAL_DIAGNOSTIC"
    )


class RANAlignmentType(str, Enum):

    STANDARD_KPI = "STANDARD_KPI"

    STANDARD_MEASUREMENT = (
        "STANDARD_MEASUREMENT"
    )

    STANDARD_DERIVED_COMPONENT = (
        "STANDARD_DERIVED_COMPONENT"
    )

    PROCEDURE_REFERENCE = (
        "PROCEDURE_REFERENCE"
    )

    OPERATIONAL_DIAGNOSTIC = (
        "OPERATIONAL_DIAGNOSTIC"
    )


class RANDependencyRelation(str, Enum):

    MEASURED_BY = "MEASURED_BY"

    IMPLEMENTED_BY = "IMPLEMENTED_BY"

    CONTRIBUTES_TO = "CONTRIBUTES_TO"

    PRECEDES = "PRECEDES"

    SUPPORTS_DIAGNOSIS_OF = (
        "SUPPORTS_DIAGNOSIS_OF"
    )

    BOUNDARY_DEPENDENCY = (
        "BOUNDARY_DEPENDENCY"
    )


# ============================================================
# 5.4 — GRAPH DATA MODELS
# ============================================================

class RANAccessibilityNode(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    node_id: str

    name: str

    layer: RANAccessibilityNodeLayer

    alignment_type: RANAlignmentType

    description: str

    domains: List[MeasurementDomain]

    standards_references: List[
        StandardsReference
    ] = Field(
        default_factory=list
    )

    metadata: Dict[str, Any] = Field(
        default_factory=dict
    )


class RANAccessibilityEdge(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    source_node_id: str

    target_node_id: str

    relation: RANDependencyRelation

    rationale: str


# ============================================================
# 5.5 — STANDARDIZED 5G RAN ACCESSIBILITY KPI DEFINITIONS
# ============================================================
#
# IMPORTANT:
#
# These are selected TS 28.554 Accessibility KPIs whose
# underlying measurements and managed-object applicability
# make them directly useful for NG-RAN investigation.
#
# We intentionally do NOT include:
#
#   - Registration Success Rate
#   - PDU Session Establishment Success Rate
#   - PFCP Session Establishment Success Rate
#
# Those remain valid 5G Accessibility KPIs but belong to
# future Core / E2E investigation domains.
# ============================================================


# ------------------------------------------------------------
# Partial DRB Accessibility
# TS 28.554 Clause 6.2.4
# ------------------------------------------------------------

partial_drb_accessibility = KPIDefinition(

    kpi_id="partial_drb_accessibility",

    name=(
        "Partial DRB Accessibility "
        "for UE Services"
    ),

    standard_category=(
        StandardKPICategory.ACCESSIBILITY
    ),

    domain=MeasurementDomain.RAN,

    description=(
        "RAN-focused service accessibility based on "
        "RRC connection establishment, UE-associated "
        "logical NG-connection establishment and DRB "
        "establishment success."
    ),

    measurement_method=(
        KPIMeasurementMethod.RATIO
    ),

    unit="percent",

    result_range_min=0,

    result_range_max=100,

    directionality="HIGHER_IS_BETTER",

    aggregation_period_minutes=(
        PM_INTERVAL_MINUTES
    ),

    source_measurements=[

        "RRC.ConnEstabSucc.Cause",
        "RRC.ConnEstabAtt.Cause",

        "UECNTXT.ConnEstabSucc.Cause",
        "UECNTXT.ConnEstabAtt.Cause",

        "DRB.EstabSucc.5QI",
        "DRB.EstabAtt.5QI",

        "DRB.EstabSucc.SNSSAI",
        "DRB.EstabAtt.SNSSAI",
    ],

    numerator_counter=None,

    denominator_counter=None,

    formula=(
        "RRC connection establishment success ratio "
        "* UE-associated logical NG-connection "
        "success ratio "
        "* DRB establishment success ratio "
        "* 100"
    ),

    applicable_entity_types=[

        EntityType.SUBNETWORK,
        EntityType.NR_CELL_CU,
    ],

    standards_references=[

        StandardsReference(

            organization="3GPP",

            specification="TS 28.554",

            title=TS_28_554_REL19.title,

            release="Release 19",

            version="19.7.0",

            clause="6.2.4",

            reference_role=(
                "PRIMARY_KPI_STANDARD"
            ),
        ),

        TS_28_552_REL19,
    ],
)


# ------------------------------------------------------------
# Total DRB Accessibility
# TS 28.554 Clause 6.2.7
# ------------------------------------------------------------

total_drb_accessibility = KPIDefinition(

    kpi_id="total_drb_accessibility",

    name=(
        "Total DRB Accessibility "
        "for UE Services"
    ),

    standard_category=(
        StandardKPICategory.ACCESSIBILITY
    ),

    domain=MeasurementDomain.RAN,

    description=(
        "Total RAN service accessibility considering "
        "initial DRB establishment, additional DRB "
        "establishment and Resume-based service "
        "establishment."
    ),

    measurement_method=(
        KPIMeasurementMethod.RATIO
    ),

    unit="percent",

    result_range_min=0,

    result_range_max=100,

    directionality="HIGHER_IS_BETTER",

    aggregation_period_minutes=(
        PM_INTERVAL_MINUTES
    ),

    source_measurements=[

        "DRB.InitialEstabSucc.5QI",
        "DRB.InitialEstabAtt.5QI",

        "DRB.EstabSucc.5QI",
        "DRB.EstabAtt.5QI",

        "DRB.ResumeSucc.5QI",
        "DRB.ResumeAtt.5QI",

        "RRC.ConnEstabSucc.Cause",
        "RRC.ConnEstabAtt.Cause",

        "RRC.ResumeSucc.Cause",
        "RRC.ResumeAtt.Cause",

        "UECNTXT.ConnEstabSucc.Cause",
        "UECNTXT.ConnEstabAtt.Cause",
    ],

    numerator_counter=None,

    denominator_counter=None,

    formula=(
        "Composite TS 28.554 Clause 6.2.7 "
        "accessibility calculation combining "
        "initial DRB establishment, additional DRB "
        "establishment, DRB Resume, RRC connection "
        "establishment, RRC Resume and UE-associated "
        "logical NG-connection success."
    ),

    applicable_entity_types=[

        EntityType.SUBNETWORK,
        EntityType.NR_CELL_CU,
    ],

    standards_references=[

        StandardsReference(

            organization="3GPP",

            specification="TS 28.554",

            title=TS_28_554_REL19.title,

            release="Release 19",

            version="19.7.0",

            clause="6.2.7",

            reference_role=(
                "PRIMARY_KPI_STANDARD"
            ),
        ),

        TS_28_552_REL19,
    ],
)


RAN_ACCESSIBILITY_REFERENCE_KPIS = {

    item.kpi_id: item

    for item in [

        partial_drb_accessibility,
        total_drb_accessibility,
    ]
}


# ============================================================
# 5.6 — GRAPH NODE HELPER
# ============================================================

def make_ran_node(
    node_id,
    name,
    layer,
    alignment_type,
    description,
    domains,
    standards_references=None,
    metadata=None,
):

    return RANAccessibilityNode(

        node_id=node_id,

        name=name,

        layer=layer,

        alignment_type=alignment_type,

        description=description,

        domains=domains,

        standards_references=(
            standards_references or []
        ),

        metadata=(
            metadata or {}
        ),
    )


# ============================================================
# 5.7 — GRAPH INITIALIZATION + STANDARD KPI NODES
# ============================================================

RAN_ACCESSIBILITY_NODES = {}


for kpi in RAN_ACCESSIBILITY_REFERENCE_KPIS.values():

    node_id = f"kpi::{kpi.kpi_id}"

    RAN_ACCESSIBILITY_NODES[
        node_id
    ] = make_ran_node(

        node_id=node_id,

        name=kpi.name,

        layer=(
            RANAccessibilityNodeLayer.KPI
        ),

        alignment_type=(
            RANAlignmentType.STANDARD_KPI
        ),

        description=kpi.description,

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=(
            kpi.standards_references
        ),

        metadata={

            "standard_category":
                kpi.standard_category.value,

            "measurement_method":
                kpi.measurement_method.value,

            "reference_domain":
                "5G_RAN_ACCESSIBILITY",
        },
    )


# ============================================================
# 5.8 — RAN KPI COMPONENT NODES
# ============================================================

component_nodes = [


    # --------------------------------------------------------
    # RRC CONNECTION ESTABLISHMENT
    # --------------------------------------------------------

    make_ran_node(

        node_id="component::rrc_setup_sr",

        name=(
            "RRC Connection Setup "
            "Success Rate"
        ),

        layer=(
            RANAccessibilityNodeLayer.COMPONENT
        ),

        alignment_type=(
            RANAlignmentType.STANDARD_DERIVED_COMPONENT
        ),

        description=(
            "Successful RRC connection establishments "
            "relative to attempted establishments."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[

            TS_28_554_REL19,
            TS_28_552_REL19,
            TS_38_331_REL19,
        ],
    ),


    # --------------------------------------------------------
    # UE-ASSOCIATED NG CONNECTION
    # --------------------------------------------------------

    make_ran_node(

        node_id="component::ng_signalling_sr",

        name=(
            "UE-Associated Logical "
            "NG-Connection Success Rate"
        ),

        layer=(
            RANAccessibilityNodeLayer.COMPONENT
        ),

        alignment_type=(
            RANAlignmentType.STANDARD_DERIVED_COMPONENT
        ),

        description=(
            "Successful establishment of the "
            "UE-associated logical NG signalling "
            "connection between NG-RAN and 5GC."
        ),

        domains=[

            MeasurementDomain.RAN,
            MeasurementDomain.TRANSPORT,
        ],

        standards_references=[

            TS_28_554_REL19,
            TS_28_552_REL19,
            TS_38_413_REL19,
        ],

        metadata={

            "investigation_boundary":
                "RAN_TO_5GC",

            "primary_focus":
                "RAN_SIDE_EVIDENCE",
        },
    ),


    # --------------------------------------------------------
    # DRB ESTABLISHMENT
    # --------------------------------------------------------

    make_ran_node(

        node_id="component::drb_setup_sr",

        name=(
            "DRB Establishment Success Rate"
        ),

        layer=(
            RANAccessibilityNodeLayer.COMPONENT
        ),

        alignment_type=(
            RANAlignmentType.STANDARD_DERIVED_COMPONENT
        ),

        description=(
            "Successful DRB establishments relative "
            "to attempted DRB establishments."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[

            TS_28_554_REL19,
            TS_28_552_REL19,
            TS_38_331_REL19,
        ],
    ),


    # --------------------------------------------------------
    # INITIAL DRB ESTABLISHMENT
    # --------------------------------------------------------

    make_ran_node(

        node_id=(
            "component::initial_drb_setup_sr"
        ),

        name=(
            "Initial DRB Establishment "
            "Success Rate"
        ),

        layer=(
            RANAccessibilityNodeLayer.COMPONENT
        ),

        alignment_type=(
            RANAlignmentType.STANDARD_DERIVED_COMPONENT
        ),

        description=(
            "Initial DRB establishment success "
            "component contributing to Total DRB "
            "Accessibility."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[

            TS_28_554_REL19,
            TS_28_552_REL19,
        ],
    ),


    # --------------------------------------------------------
    # DRB RESUME
    # --------------------------------------------------------

    make_ran_node(

        node_id="component::drb_resume_sr",

        name=(
            "DRB Resume Success Rate"
        ),

        layer=(
            RANAccessibilityNodeLayer.COMPONENT
        ),

        alignment_type=(
            RANAlignmentType.STANDARD_DERIVED_COMPONENT
        ),

        description=(
            "Successful DRB resume operations "
            "relative to attempted DRB resume "
            "operations."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[

            TS_28_554_REL19,
            TS_28_552_REL19,
        ],
    ),


    # --------------------------------------------------------
    # RRC RESUME
    # --------------------------------------------------------

    make_ran_node(

        node_id="component::rrc_resume_sr",

        name=(
            "RRC Resume Success Rate"
        ),

        layer=(
            RANAccessibilityNodeLayer.COMPONENT
        ),

        alignment_type=(
            RANAlignmentType.STANDARD_DERIVED_COMPONENT
        ),

        description=(
            "Successful RRC Resume procedures "
            "relative to attempted Resume procedures."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[

            TS_28_554_REL19,
            TS_28_552_REL19,
            TS_38_331_REL19,
        ],
    ),
]


for node in component_nodes:

    RAN_ACCESSIBILITY_NODES[
        node.node_id
    ] = node


# ============================================================
# 5.9 — STANDARD PM MEASUREMENT NODES
# ============================================================

measurement_specs = [


    # --------------------------------------------------------
    # RRC SETUP
    # --------------------------------------------------------

    (
        "measurement::rrc_conn_estab_att",

        "RRC Connection Establishment Attempts",

        "RRC.ConnEstabAtt.Cause",

        EntityType.NR_CELL_CU,
    ),

    (
        "measurement::rrc_conn_estab_succ",

        "RRC Connection Establishment Successes",

        "RRC.ConnEstabSucc.Cause",

        EntityType.NR_CELL_CU,
    ),


    # --------------------------------------------------------
    # UE CONTEXT / LOGICAL NG CONNECTION
    # --------------------------------------------------------

    (
        "measurement::uecntxt_conn_estab_att",

        "UE Context / Logical NG Connection Attempts",

        "UECNTXT.ConnEstabAtt.Cause",

        EntityType.NR_CELL_CU,
    ),

    (
        "measurement::uecntxt_conn_estab_succ",

        "UE Context / Logical NG Connection Successes",

        "UECNTXT.ConnEstabSucc.Cause",

        EntityType.NR_CELL_CU,
    ),


    # --------------------------------------------------------
    # DRB ESTABLISHMENT — 5QI
    # --------------------------------------------------------

    (
        "measurement::drb_estab_att_5qi",

        "DRB Establishment Attempts by 5QI",

        "DRB.EstabAtt.5QI",

        EntityType.NR_CELL_CU,
    ),

    (
        "measurement::drb_estab_succ_5qi",

        "DRB Establishment Successes by 5QI",

        "DRB.EstabSucc.5QI",

        EntityType.NR_CELL_CU,
    ),


    # --------------------------------------------------------
    # DRB ESTABLISHMENT — S-NSSAI
    # --------------------------------------------------------

    (
        "measurement::drb_estab_att_snssai",

        "DRB Establishment Attempts by S-NSSAI",

        "DRB.EstabAtt.SNSSAI",

        EntityType.NR_CELL_CU,
    ),

    (
        "measurement::drb_estab_succ_snssai",

        "DRB Establishment Successes by S-NSSAI",

        "DRB.EstabSucc.SNSSAI",

        EntityType.NR_CELL_CU,
    ),


    # --------------------------------------------------------
    # INITIAL DRB ESTABLISHMENT
    # --------------------------------------------------------

    (
        "measurement::drb_initial_att",

        "Initial DRB Establishment Attempts",

        "DRB.InitialEstabAtt.5QI",

        EntityType.NR_CELL_CU,
    ),

    (
        "measurement::drb_initial_succ",

        "Initial DRB Establishment Successes",

        "DRB.InitialEstabSucc.5QI",

        EntityType.NR_CELL_CU,
    ),


    # --------------------------------------------------------
    # DRB RESUME
    # --------------------------------------------------------

    (
        "measurement::drb_resume_att",

        "DRB Resume Attempts",

        "DRB.ResumeAtt.5QI",

        EntityType.NR_CELL_CU,
    ),

    (
        "measurement::drb_resume_succ",

        "DRB Resume Successes",

        "DRB.ResumeSucc.5QI",

        EntityType.NR_CELL_CU,
    ),


    # --------------------------------------------------------
    # RRC RESUME
    # --------------------------------------------------------

    (
        "measurement::rrc_resume_att",

        "RRC Resume Attempts",

        "RRC.ResumeAtt.Cause",

        EntityType.NR_CELL_CU,
    ),

    (
        "measurement::rrc_resume_succ",

        "RRC Resume Successes",

        "RRC.ResumeSucc.Cause",

        EntityType.NR_CELL_CU,
    ),
]


for (
    node_id,
    name,
    measurement_name,
    entity_type,
) in measurement_specs:

    RAN_ACCESSIBILITY_NODES[
        node_id
    ] = make_ran_node(

        node_id=node_id,

        name=name,

        layer=(
            RANAccessibilityNodeLayer.MEASUREMENT
        ),

        alignment_type=(
            RANAlignmentType.STANDARD_MEASUREMENT
        ),

        description=(
            "3GPP performance measurement used "
            "within the 5G RAN Accessibility "
            f"reference domain: {measurement_name}."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[
            TS_28_552_REL19
        ],

        metadata={

            "measurement_name":
                measurement_name,

            "entity_type":
                entity_type.value,
        },
    )


# ============================================================
# 5.10 — RAN PROCEDURE NODES
# ============================================================

procedure_nodes = [


    # --------------------------------------------------------
    # RANDOM ACCESS
    # --------------------------------------------------------

    make_ran_node(

        node_id="procedure::random_access",

        name=(
            "NR Random Access Procedure"
        ),

        layer=(
            RANAccessibilityNodeLayer.PROCEDURE
        ),

        alignment_type=(
            RANAlignmentType.PROCEDURE_REFERENCE
        ),

        description=(
            "NR MAC Random Access procedure that "
            "precedes successful RRC establishment."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[
            TS_38_321_REL19
        ],
    ),


    # --------------------------------------------------------
    # RRC ESTABLISHMENT
    # --------------------------------------------------------

    make_ran_node(

        node_id="procedure::rrc_establishment",

        name=(
            "RRC Connection Establishment"
        ),

        layer=(
            RANAccessibilityNodeLayer.PROCEDURE
        ),

        alignment_type=(
            RANAlignmentType.PROCEDURE_REFERENCE
        ),

        description=(
            "NR RRC procedure establishing "
            "the UE RRC connection."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[
            TS_38_331_REL19
        ],
    ),


    # --------------------------------------------------------
    # NG SIGNALLING
    # --------------------------------------------------------

    make_ran_node(

        node_id="procedure::ng_signalling",

        name=(
            "UE-Associated NG Signalling "
            "Connection Establishment"
        ),

        layer=(
            RANAccessibilityNodeLayer.PROCEDURE
        ),

        alignment_type=(
            RANAlignmentType.PROCEDURE_REFERENCE
        ),

        description=(
            "NGAP signalling context associated "
            "with the UE between NG-RAN and 5GC."
        ),

        domains=[

            MeasurementDomain.RAN,
            MeasurementDomain.TRANSPORT,
        ],

        standards_references=[
            TS_38_413_REL19
        ],
    ),


    # --------------------------------------------------------
    # DRB ESTABLISHMENT
    # --------------------------------------------------------

    make_ran_node(

        node_id="procedure::drb_establishment",

        name=(
            "DRB Establishment"
        ),

        layer=(
            RANAccessibilityNodeLayer.PROCEDURE
        ),

        alignment_type=(
            RANAlignmentType.PROCEDURE_REFERENCE
        ),

        description=(
            "RAN procedure establishing the "
            "Data Radio Bearer required for "
            "UE service delivery."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[
            TS_38_331_REL19
        ],
    ),


    # --------------------------------------------------------
    # RRC RESUME
    # --------------------------------------------------------

    make_ran_node(

        node_id="procedure::rrc_resume",

        name=(
            "RRC Resume"
        ),

        layer=(
            RANAccessibilityNodeLayer.PROCEDURE
        ),

        alignment_type=(
            RANAlignmentType.PROCEDURE_REFERENCE
        ),

        description=(
            "RRC Resume procedure used when a UE "
            "returns from RRC_INACTIVE to connected "
            "service."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[
            TS_38_331_REL19
        ],
    ),
]


for node in procedure_nodes:

    RAN_ACCESSIBILITY_NODES[
        node.node_id
    ] = node


# ============================================================
# 5.11 — OPERATIONAL RAN DIAGNOSTIC NODES
# ============================================================
#
# IMPORTANT:
#
# These nodes are intentionally NOT represented as TS 28.554
# standardized KPIs.
#
# They represent evidence an experienced RAN engineer would
# commonly inspect while determining the mechanism behind
# Accessibility degradation.
#
# Exact standardized measurement bindings and any derived
# formulae are deferred to Cell 6.
# ============================================================

operational_nodes = [


    # --------------------------------------------------------
    # RACH
    # --------------------------------------------------------

    make_ran_node(

        node_id=(
            "diagnostic::rach_access_success"
        ),

        name=(
            "Random Access / RACH "
            "Success Behaviour"
        ),

        layer=(
            RANAccessibilityNodeLayer.OPERATIONAL_DIAGNOSTIC
        ),

        alignment_type=(
            RANAlignmentType.OPERATIONAL_DIAGNOSTIC
        ),

        description=(
            "Derived/operational indication of "
            "Random Access performance used to "
            "determine whether Accessibility "
            "degradation starts before RRC "
            "connection establishment."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[

            TS_28_552_REL19,
            TS_38_321_REL19,
        ],

        metadata={

            "standard_kpi": False,

            "operational_metric_role":
                "DERIVED_RAN_KPI_CANDIDATE",

            "measurement_binding":
                "DEFERRED_TO_CELL_6",

            "formula_status":
                "NOT_FROZEN",

            "diagnostic_role":
                "PRE_RRC_ACCESS",
        },
    ),


    # --------------------------------------------------------
    # RADIO CONDITION
    # --------------------------------------------------------

    make_ran_node(

        node_id=(
            "diagnostic::radio_condition"
        ),

        name=(
            "Radio Coverage / Quality Condition"
        ),

        layer=(
            RANAccessibilityNodeLayer.OPERATIONAL_DIAGNOSTIC
        ),

        alignment_type=(
            RANAlignmentType.OPERATIONAL_DIAGNOSTIC
        ),

        description=(
            "Radio-condition evidence used to "
            "differentiate coverage or quality "
            "degradation from signalling, capacity "
            "or configuration faults."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[],

        metadata={

            "standard_kpi": False,

            "examples": [

                "RSRP",
                "RSRQ",
                "SINR",
            ],

            "measurement_binding":
                "DEFERRED_TO_CELL_6",
        },
    ),


    # --------------------------------------------------------
    # CELL LOAD / RESOURCES
    # --------------------------------------------------------

    make_ran_node(

        node_id=(
            "diagnostic::cell_resource_load"
        ),

        name=(
            "Cell Load / Radio Resource Pressure"
        ),

        layer=(
            RANAccessibilityNodeLayer.OPERATIONAL_DIAGNOSTIC
        ),

        alignment_type=(
            RANAlignmentType.OPERATIONAL_DIAGNOSTIC
        ),

        description=(
            "Cell-load evidence used to determine "
            "whether resource pressure contributes "
            "to Accessibility degradation."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[
            TS_28_552_REL19
        ],

        metadata={

            "standard_kpi": False,

            "examples": [

                "PRB utilisation",
                "connected users",
                "control-channel load",
            ],

            "measurement_binding":
                "DEFERRED_TO_CELL_6",
        },
    ),


    # --------------------------------------------------------
    # RAN ALARMS / EVENTS
    # --------------------------------------------------------

    make_ran_node(

        node_id=(
            "diagnostic::ran_alarms_events"
        ),

        name=(
            "RAN Alarms / Events"
        ),

        layer=(
            RANAccessibilityNodeLayer.OPERATIONAL_DIAGNOSTIC
        ),

        alignment_type=(
            RANAlignmentType.OPERATIONAL_DIAGNOSTIC
        ),

        description=(
            "RAN alarms and events that may provide "
            "causal or corroborating evidence during "
            "Accessibility investigation."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[],

        metadata={

            "standard_kpi": False,

            "telemetry_observation_type":
                "ALARM_EVENT",
        },
    ),


    # --------------------------------------------------------
    # CONFIGURATION
    # --------------------------------------------------------

    make_ran_node(

        node_id=(
            "diagnostic::ran_configuration"
        ),

        name=(
            "RAN Configuration State"
        ),

        layer=(
            RANAccessibilityNodeLayer.OPERATIONAL_DIAGNOSTIC
        ),

        alignment_type=(
            RANAlignmentType.OPERATIONAL_DIAGNOSTIC
        ),

        description=(
            "Configuration evidence used to detect "
            "parameter, feature, admission-control "
            "or software/configuration changes "
            "associated with Accessibility impact."
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[],

        metadata={

            "standard_kpi": False,

            "telemetry_observation_type":
                "CONFIGURATION",
        },
    ),


    # --------------------------------------------------------
    # TRANSPORT BOUNDARY
    # --------------------------------------------------------

    make_ran_node(

        node_id=(
            "diagnostic::transport_boundary"
        ),

        name=(
            "RAN-to-Core Transport Boundary"
        ),

        layer=(
            RANAccessibilityNodeLayer.OPERATIONAL_DIAGNOSTIC
        ),

        alignment_type=(
            RANAlignmentType.OPERATIONAL_DIAGNOSTIC
        ),

        description=(
            "Transport-facing evidence used when "
            "NG signalling degradation cannot be "
            "explained by local RAN conditions."
        ),

        domains=[

            MeasurementDomain.RAN,
            MeasurementDomain.TRANSPORT,
        ],

        standards_references=[],

        metadata={

            "standard_kpi": False,

            "examples": [

                "latency",
                "packet loss",
                "SCTP association condition",
                "interface reachability",
            ],

            "boundary_role":
                "RAN_TO_5GC",
        },
    ),
]


for node in operational_nodes:

    RAN_ACCESSIBILITY_NODES[
        node.node_id
    ] = node


# ============================================================
# 5.12 — DEPENDENCY GRAPH EDGES
# ============================================================

RAN_ACCESSIBILITY_EDGES = []


def add_ran_edge(
    source,
    target,
    relation,
    rationale,
):

    RAN_ACCESSIBILITY_EDGES.append(

        RANAccessibilityEdge(

            source_node_id=source,

            target_node_id=target,

            relation=relation,

            rationale=rationale,
        )
    )


# ============================================================
# MEASUREMENTS → KPI COMPONENTS
# ============================================================


# ------------------------------------------------------------
# RRC CONNECTION SETUP
# ------------------------------------------------------------

for measurement_id in [

    "measurement::rrc_conn_estab_att",
    "measurement::rrc_conn_estab_succ",
]:

    add_ran_edge(

        "component::rrc_setup_sr",

        measurement_id,

        RANDependencyRelation.MEASURED_BY,

        (
            "RRC setup success is derived from "
            "RRC establishment attempts and successes."
        ),
    )


# ------------------------------------------------------------
# UE-ASSOCIATED LOGICAL NG CONNECTION
# ------------------------------------------------------------

for measurement_id in [

    "measurement::uecntxt_conn_estab_att",
    "measurement::uecntxt_conn_estab_succ",
]:

    add_ran_edge(

        "component::ng_signalling_sr",

        measurement_id,

        RANDependencyRelation.MEASURED_BY,

        (
            "UE-associated NG signalling success "
            "is derived from corresponding attempts "
            "and successes."
        ),
    )


# ------------------------------------------------------------
# DRB ESTABLISHMENT
# ------------------------------------------------------------

for measurement_id in [

    "measurement::drb_estab_att_5qi",
    "measurement::drb_estab_succ_5qi",

    "measurement::drb_estab_att_snssai",
    "measurement::drb_estab_succ_snssai",
]:

    add_ran_edge(

        "component::drb_setup_sr",

        measurement_id,

        RANDependencyRelation.MEASURED_BY,

        (
            "DRB establishment success is derived "
            "from attempted and successful "
            "DRB establishments."
        ),
    )


# ------------------------------------------------------------
# INITIAL DRB
# ------------------------------------------------------------

for measurement_id in [

    "measurement::drb_initial_att",
    "measurement::drb_initial_succ",
]:

    add_ran_edge(

        "component::initial_drb_setup_sr",

        measurement_id,

        RANDependencyRelation.MEASURED_BY,

        (
            "Initial DRB establishment success is "
            "derived from initial DRB attempts "
            "and successes."
        ),
    )


# ------------------------------------------------------------
# DRB RESUME
# ------------------------------------------------------------

for measurement_id in [

    "measurement::drb_resume_att",
    "measurement::drb_resume_succ",
]:

    add_ran_edge(

        "component::drb_resume_sr",

        measurement_id,

        RANDependencyRelation.MEASURED_BY,

        (
            "DRB Resume success is derived from "
            "DRB Resume attempts and successes."
        ),
    )


# ------------------------------------------------------------
# RRC RESUME
# ------------------------------------------------------------

for measurement_id in [

    "measurement::rrc_resume_att",
    "measurement::rrc_resume_succ",
]:

    add_ran_edge(

        "component::rrc_resume_sr",

        measurement_id,

        RANDependencyRelation.MEASURED_BY,

        (
            "RRC Resume success is derived from "
            "Resume attempts and successes."
        ),
    )


# ============================================================
# COMPONENTS → STANDARDIZED ACCESSIBILITY KPIs
# ============================================================


# ------------------------------------------------------------
# PARTIAL DRB ACCESSIBILITY
# ------------------------------------------------------------

for component_id in [

    "component::rrc_setup_sr",
    "component::ng_signalling_sr",
    "component::drb_setup_sr",
]:

    add_ran_edge(

        component_id,

        "kpi::partial_drb_accessibility",

        RANDependencyRelation.CONTRIBUTES_TO,

        (
            "This RAN component contributes to "
            "Partial DRB Accessibility."
        ),
    )


# ------------------------------------------------------------
# TOTAL DRB ACCESSIBILITY
# ------------------------------------------------------------

for component_id in [

    "component::rrc_setup_sr",
    "component::ng_signalling_sr",

    "component::initial_drb_setup_sr",
    "component::drb_setup_sr",

    "component::rrc_resume_sr",
    "component::drb_resume_sr",
]:

    add_ran_edge(

        component_id,

        "kpi::total_drb_accessibility",

        RANDependencyRelation.CONTRIBUTES_TO,

        (
            "This RAN component contributes to "
            "Total DRB Accessibility."
        ),
    )


# ============================================================
# COMPONENTS → RAN PROCEDURES
# ============================================================

procedure_component_pairs = [

    (
        "component::rrc_setup_sr",
        "procedure::rrc_establishment",
    ),

    (
        "component::ng_signalling_sr",
        "procedure::ng_signalling",
    ),

    (
        "component::drb_setup_sr",
        "procedure::drb_establishment",
    ),

    (
        "component::initial_drb_setup_sr",
        "procedure::drb_establishment",
    ),

    (
        "component::rrc_resume_sr",
        "procedure::rrc_resume",
    ),

    (
        "component::drb_resume_sr",
        "procedure::rrc_resume",
    ),
]


for component_id, procedure_id in procedure_component_pairs:

    add_ran_edge(

        component_id,

        procedure_id,

        RANDependencyRelation.IMPLEMENTED_BY,

        (
            "The KPI component represents "
            "performance of this RAN procedure."
        ),
    )


# ============================================================
# PROCEDURAL ORDER
# ============================================================

add_ran_edge(

    "procedure::random_access",

    "procedure::rrc_establishment",

    RANDependencyRelation.PRECEDES,

    (
        "Successful Random Access normally precedes "
        "RRC connection establishment."
    ),
)


add_ran_edge(

    "procedure::rrc_establishment",

    "procedure::ng_signalling",

    RANDependencyRelation.PRECEDES,

    (
        "RRC establishment precedes the subsequent "
        "UE-associated NG signalling context."
    ),
)


add_ran_edge(

    "procedure::ng_signalling",

    "procedure::drb_establishment",

    RANDependencyRelation.PRECEDES,

    (
        "Successful access signalling supports "
        "subsequent DRB establishment."
    ),
)


# ============================================================
# OPERATIONAL DIAGNOSTICS → INVESTIGATION TARGETS
# ============================================================


# ------------------------------------------------------------
# RACH
# ------------------------------------------------------------

add_ran_edge(

    "diagnostic::rach_access_success",

    "component::rrc_setup_sr",

    RANDependencyRelation.SUPPORTS_DIAGNOSIS_OF,

    (
        "RACH behaviour helps determine whether "
        "Accessibility degradation starts before "
        "RRC establishment."
    ),
)


# ------------------------------------------------------------
# RADIO CONDITION
# ------------------------------------------------------------

for component_id in [

    "component::rrc_setup_sr",
    "component::drb_setup_sr",
    "component::initial_drb_setup_sr",
]:

    add_ran_edge(

        "diagnostic::radio_condition",

        component_id,

        RANDependencyRelation.SUPPORTS_DIAGNOSIS_OF,

        (
            "Radio coverage and quality may affect "
            "RRC or DRB Accessibility performance."
        ),
    )


# ------------------------------------------------------------
# LOAD / RESOURCE PRESSURE
# ------------------------------------------------------------

for component_id in [

    "component::rrc_setup_sr",
    "component::drb_setup_sr",
    "component::initial_drb_setup_sr",
]:

    add_ran_edge(

        "diagnostic::cell_resource_load",

        component_id,

        RANDependencyRelation.SUPPORTS_DIAGNOSIS_OF,

        (
            "Cell resource pressure may contribute "
            "to RAN Accessibility degradation."
        ),
    )


# ------------------------------------------------------------
# RAN ALARMS
# ------------------------------------------------------------

for kpi_id in [

    "kpi::partial_drb_accessibility",
    "kpi::total_drb_accessibility",
]:

    add_ran_edge(

        "diagnostic::ran_alarms_events",

        kpi_id,

        RANDependencyRelation.SUPPORTS_DIAGNOSIS_OF,

        (
            "RAN alarms and events provide "
            "corroborating fault evidence."
        ),
    )


# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

for kpi_id in [

    "kpi::partial_drb_accessibility",
    "kpi::total_drb_accessibility",
]:

    add_ran_edge(

        "diagnostic::ran_configuration",

        kpi_id,

        RANDependencyRelation.SUPPORTS_DIAGNOSIS_OF,

        (
            "Configuration state and recent changes "
            "may explain Accessibility degradation."
        ),
    )


# ------------------------------------------------------------
# TRANSPORT BOUNDARY
# ------------------------------------------------------------

add_ran_edge(

    "diagnostic::transport_boundary",

    "component::ng_signalling_sr",

    RANDependencyRelation.BOUNDARY_DEPENDENCY,

    (
        "NG signalling degradation may require "
        "transport-boundary investigation when "
        "local RAN evidence is insufficient."
    ),
)


# ============================================================
# 5.13 — GRAPH INTEGRITY VALIDATION
# ============================================================

node_ids = set(
    RAN_ACCESSIBILITY_NODES.keys()
)


for edge in RAN_ACCESSIBILITY_EDGES:

    assert (
        edge.source_node_id
        in node_ids
    ), (
        "Unknown source node: "
        f"{edge.source_node_id}"
    )

    assert (
        edge.target_node_id
        in node_ids
    ), (
        "Unknown target node: "
        f"{edge.target_node_id}"
    )


# Formal Cell 5 KPI set must contain only the selected
# RAN Accessibility reference KPIs.

assert (
    len(RAN_ACCESSIBILITY_REFERENCE_KPIS)
    == 2
)


assert (
    set(
        RAN_ACCESSIBILITY_REFERENCE_KPIS.keys()
    )
    == {
        "partial_drb_accessibility",
        "total_drb_accessibility",
    }
)


# Every formal KPI must remain Accessibility.

for kpi in (
    RAN_ACCESSIBILITY_REFERENCE_KPIS.values()
):

    assert (
        kpi.standard_category
        == StandardKPICategory.ACCESSIBILITY
    )


# No Core-only KPI is allowed into this formal RAN set.

FORBIDDEN_RAN_REFERENCE_KPIS = {

    "rsr_network_slice",
    "pdu_session_est_sr_snssai",
    "pfcp_session_est_sr",
}


assert (
    not (
        set(
            RAN_ACCESSIBILITY_REFERENCE_KPIS.keys()
        )
        & FORBIDDEN_RAN_REFERENCE_KPIS
    )
)


# Operational diagnostics must never masquerade as
# standardized KPIs.

for node in RAN_ACCESSIBILITY_NODES.values():

    if (
        node.layer
        == RANAccessibilityNodeLayer.OPERATIONAL_DIAGNOSTIC
    ):

        assert (
            node.alignment_type
            == RANAlignmentType.OPERATIONAL_DIAGNOSTIC
        )

        assert (
            node.metadata.get(
                "standard_kpi",
                False,
            )
            is False
        )


# RACH must exist but remain operational / derived.

assert (
    "diagnostic::rach_access_success"
    in RAN_ACCESSIBILITY_NODES
)


rach_node = RAN_ACCESSIBILITY_NODES[
    "diagnostic::rach_access_success"
]


assert (
    rach_node.alignment_type
    == RANAlignmentType.OPERATIONAL_DIAGNOSTIC
)


# ============================================================
# 5.14 — BUILD RAN ACCESSIBILITY DOMAIN ARTIFACT
# ============================================================

ran_accessibility_domain = {

    "domain_id":
        "module5_5g_ran_accessibility_reference_domain_v1",

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "module_scope":
        "MODULE_5_V1_5G_RAN",

    "reference_domain":
        "5G_RAN_ACCESSIBILITY",

    "standard_category":
        StandardKPICategory.ACCESSIBILITY.value,

    "primary_standard": {

        "specification":
            "TS 28.554",

        "release":
            "Release 19",

        "version":
            "19.7.0",

        "clauses": [
            "6.2.4",
            "6.2.7",
        ],
    },

    "measurement_standard": {

        "specification":
            "TS 28.552",

        "release":
            "Release 19",

        "version":
            "19.7.0",
    },

    "formal_reference_kpis": {

        kpi_id:
            kpi.model_dump(
                mode="json"
            )

        for kpi_id, kpi
        in RAN_ACCESSIBILITY_REFERENCE_KPIS.items()
    },

    "nodes": {

        node_id:
            node.model_dump(
                mode="json"
            )

        for node_id, node
        in RAN_ACCESSIBILITY_NODES.items()
    },

    "edges": [

        edge.model_dump(
            mode="json"
        )

        for edge
        in RAN_ACCESSIBILITY_EDGES
    ],

    "future_domain_extensions": [

        {
            "domain":
                "5G_CORE_ACCESSIBILITY",

            "examples": [

                "Registration Success Rate",
                "PDU Session Establishment Success Rate",
                "PFCP Session Establishment Success Rate",
            ],

            "status":
                "OUT_OF_SCOPE_MODULE5_V1_REFERENCE_DOMAIN",
        },

        {
            "domain":
                "5G_RAN_RETAINABILITY",

            "status":
                "FUTURE_EXTENSION",
        },

        {
            "domain":
                "5G_RAN_MOBILITY",

            "status":
                "FUTURE_EXTENSION",
        },

        {
            "domain":
                "5G_RAN_INTEGRITY",

            "status":
                "FUTURE_EXTENSION",
        },
    ],

    "design_notes": [

        (
            "5G RAN Accessibility is the first "
            "Module 5 operational investigation domain."
        ),

        (
            "Cell 4 remains the broader canonical "
            "5G standards and telemetry foundation."
        ),

        (
            "Partial DRB Accessibility and Total "
            "DRB Accessibility form the standardized "
            "formal KPI reference set."
        ),

        (
            "RACH access success behaviour is a "
            "high-value operational/derived diagnostic "
            "but is not labelled as a standardized "
            "TS 28.554 KPI."
        ),

        (
            "Registration, PDU Session and PFCP "
            "Accessibility KPIs are intentionally "
            "deferred to future Core/E2E extensions."
        ),

        (
            "The graph supports hierarchical "
            "investigation from KPI to component, "
            "measurement, procedure and supporting "
            "operational evidence."
        ),

        (
            "The dependency graph does not expose "
            "hidden incident ground truth."
        ),

        (
            "Exact RACH, radio-condition, resource "
            "and failure-cause measurement bindings "
            "are deferred to Cell 6."
        ),
    ],
}


# ============================================================
# 5.15 — DETERMINISTIC SEMANTIC HASH
# ============================================================

def canonical_json_sha256(data):

    canonical = json.dumps(

        data,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

    ).encode(
        "utf-8"
    )

    return hashlib.sha256(
        canonical
    ).hexdigest()


ran_accessibility_semantic_content = {

    key: value

    for key, value
    in ran_accessibility_domain.items()

    if key != "created_utc"
}


ran_accessibility_semantic_sha = (
    canonical_json_sha256(
        ran_accessibility_semantic_content
    )
)


ran_accessibility_domain[
    "semantic_sha256"
] = ran_accessibility_semantic_sha


# ============================================================
# 5.16 — SAVE ARTIFACT
# ============================================================

ran_accessibility_path = (

    PATHS["kpi"]

    / (
        "module5_5g_ran_accessibility_"
        "reference_domain_v1.json"
    )
)


with open(

    ran_accessibility_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        ran_accessibility_domain,

        f,

        indent=2,

        ensure_ascii=False,
    )


ran_accessibility_artifact_sha = (
    sha256_file(
        ran_accessibility_path
    )
)


# ============================================================
# 5.17 — SUMMARY STATISTICS
# ============================================================

layer_counts = {}


for layer in RANAccessibilityNodeLayer:

    layer_counts[
        layer.value
    ] = sum(

        1

        for node
        in RAN_ACCESSIBILITY_NODES.values()

        if node.layer == layer
    )


domain_counts = {}


for domain in MeasurementDomain:

    domain_counts[
        domain.value
    ] = sum(

        1

        for node
        in RAN_ACCESSIBILITY_NODES.values()

        if domain in node.domains
    )


# ============================================================
# 5.18 — FINAL VALIDATION
# ============================================================

assert (
    ran_accessibility_domain[
        "reference_domain"
    ]
    == "5G_RAN_ACCESSIBILITY"
)


assert (
    ran_accessibility_domain[
        "module_scope"
    ]
    == "MODULE_5_V1_5G_RAN"
)


assert (
    "kpi::partial_drb_accessibility"
    in RAN_ACCESSIBILITY_NODES
)


assert (
    "kpi::total_drb_accessibility"
    in RAN_ACCESSIBILITY_NODES
)


assert (
    "diagnostic::rach_access_success"
    in RAN_ACCESSIBILITY_NODES
)


assert (
    "component::rrc_setup_sr"
    in RAN_ACCESSIBILITY_NODES
)


assert (
    "component::drb_setup_sr"
    in RAN_ACCESSIBILITY_NODES
)


assert (
    "component::ng_signalling_sr"
    in RAN_ACCESSIBILITY_NODES
)


# ============================================================
# 5.19 — SUMMARY OUTPUT
# ============================================================

print("=" * 96)

print(
    "MODULE 5 — CELL 04 "
    "5G RAN ACCESSIBILITY REFERENCE DOMAIN"
)

print("=" * 96)


print(
    "\nModule scope"
)

print("-" * 96)

print(
    "Formal investigation scope : "
    "5G RAN"
)

print(
    "Reference domain           : "
    "5G RAN Accessibility"
)

print(
    "Future extension           : "
    "5G Core / E2E + other RAN KPI domains"
)


print(
    "\nStandards baseline"
)

print("-" * 96)

print(
    "KPI standard              : "
    "TS 28.554 Rel-19 v19.7.0"
)

print(
    "Measurement standard      : "
    "TS 28.552 Rel-19 v19.7.0"
)


print(
    "\nFormal standardized RAN Accessibility KPIs"
)

print("-" * 96)


for kpi in (
    RAN_ACCESSIBILITY_REFERENCE_KPIS.values()
):

    clause = (
        kpi.standards_references[0].clause
    )

    print(

        f"{kpi.kpi_id:<32} : "
        f"{kpi.name} [{clause}]"
    )


print(
    "\nKey RAN KPI components"
)

print("-" * 96)


for node in (
    RAN_ACCESSIBILITY_NODES.values()
):

    if (
        node.layer
        == RANAccessibilityNodeLayer.COMPONENT
    ):

        print(
            f"{node.node_id:<40} : "
            f"{node.name}"
        )


print(
    "\nOperational RAN diagnostics"
)

print("-" * 96)


for node in (
    RAN_ACCESSIBILITY_NODES.values()
):

    if (
        node.layer
        == RANAccessibilityNodeLayer.OPERATIONAL_DIAGNOSTIC
    ):

        print(

            f"{node.name:<44} : "
            "SUPPORTING EVIDENCE — NOT STANDARD KPI"
        )


print(
    "\nExplicitly deferred Core / E2E KPIs"
)

print("-" * 96)

print(
    "Registration Success Rate                : "
    "FUTURE CORE / E2E"
)

print(
    "PDU Session Establishment Success Rate   : "
    "FUTURE CORE / E2E"
)

print(
    "PFCP Session Establishment Success Rate  : "
    "FUTURE CORE / E2E"
)


print(
    "\nDependency graph"
)

print("-" * 96)

print(
    f"Nodes                     : "
    f"{len(RAN_ACCESSIBILITY_NODES)}"
)

print(
    f"Edges                     : "
    f"{len(RAN_ACCESSIBILITY_EDGES)}"
)


print(
    "\nNode layers"
)

print("-" * 96)


for layer, count in (
    layer_counts.items()
):

    print(
        f"{layer:<28} : "
        f"{count}"
    )


print(
    "\nTechnical-domain coverage"
)

print("-" * 96)


for domain, count in (
    domain_counts.items()
):

    if count > 0:

        print(
            f"{domain:<28} : "
            f"{count}"
        )


print(
    "\nInvestigation hierarchy"
)

print("-" * 96)

print(
    "Level 1 : Standardized RAN Accessibility KPI"
)

print(
    "Level 2 : RAN KPI component / procedure stage"
)

print(
    "Level 3 : Standardized PM measurement / counter"
)

print(
    "Context : RACH + radio condition + cell load "
    "+ alarms/events + configuration + transport boundary"
)


print(
    "\nArtifact"
)

print("-" * 96)

print(
    f"Path                      : "
    f"{ran_accessibility_path}"
)

print(
    f"Artifact SHA              : "
    f"{ran_accessibility_artifact_sha}"
)

print(
    f"Semantic SHA              : "
    f"{ran_accessibility_semantic_sha}"
)


print(
    "\n" + "=" * 96
)

print(
    "CELL 04 COMPLETE — "
    "5G RAN ACCESSIBILITY INVESTIGATION GRAPH READY"
)

print(
    "=" * 96
)
````

###### Cell 04 — Observation

Cell 04 establishes the first bounded Module 5 investigation domain: **5G RAN Accessibility**.

The graph contains **33 nodes and 44 edges**, connecting the two formal TS 28.554 Accessibility KPIs to their RRC, NG signalling, DRB and Resume components, standardized PM evidence, and supporting operational diagnostics such as RACH, radio condition, resource pressure, alarms, configuration and the transport boundary.

> **The benchmark is intentionally RAN-focused: supporting diagnostics are kept distinct from standardized KPIs, while Core/E2E Accessibility KPIs are deferred for later expansion.**


<a id="cell-05-kpi-knowledge-supplement"></a>
##### Cell 05 — Accessibility KPI Knowledge Supplement V1

Enrich the reference domain with exact TS 28.554 formula dependencies, TS 28.552 PM semantics, explicit aliases, investigation stages, supporting RACH measurements and graph-coverage validation.


````python
# ============================================================
# MODULE 5 — CELL 05
# 5G RAN ACCESSIBILITY KPI KNOWLEDGE SUPPLEMENT
# + EXACT FORMULA DEPENDENCY VALIDATION
# ============================================================
#
# Purpose:
#
#   1. Enrich the Cell 5 RAN Accessibility graph with
#      engineering interpretation.
#
#   2. Define a canonical catalogue of the 3GPP PM
#      measurements used by Partial / Total DRB Accessibility.
#
#   3. Preserve exact TS 28.554 KPI dependency rules,
#      including RRC fallback / re-establishment measurements.
#
#   4. Normalize source-specific naming differences without
#      silently changing either standard.
#
#   5. Define supporting operational evidence such as:
#         - RACH
#         - radio condition
#         - resource pressure
#         - alarms/events
#         - configuration
#         - RAN-to-Core transport boundary
#
#   6. Validate the completeness of the frozen Cell 5
#      dependency graph.
#
#   7. Persist a deterministic machine-readable knowledge
#      supplement for later Telemetry MCP / agent use.
#
#
# Standards baseline retained from Cells 4–5:
#
#   3GPP TS 28.554 v19.7.0 Release 19
#   3GPP TS 28.552 v19.7.0 Release 19
#
#
# IMPORTANT:
#
#   - Cell 6 DOES NOT mutate the Cell 5 graph.
#
#   - If an exact standardized KPI dependency is absent
#     from Cell 5, the gap is reported explicitly.
#
#   - 3GPP standardized measurements and operational /
#     derived diagnostics remain separate.
#
#   - RACH measurements are standardized PM evidence,
#     but "RACH Success Rate" is NOT represented here as
#     a standardized TS 28.554 Accessibility KPI.
#
# ============================================================


# ============================================================
# 6.1 — IMPORTS + DEPENDENCY VALIDATION
# ============================================================

from enum import Enum
from typing import Any, Dict, List, Optional
from datetime import datetime, timezone

from pydantic import (
    BaseModel,
    Field,
    ConfigDict,
)

import json
import hashlib


REQUIRED_CELL5_OBJECTS = [

    "RAN_ACCESSIBILITY_REFERENCE_KPIS",
    "RAN_ACCESSIBILITY_NODES",
    "RAN_ACCESSIBILITY_EDGES",

    "RANAccessibilityNodeLayer",
    "RANAlignmentType",

    "TS_28_554_REL19",
    "TS_28_552_REL19",

    "TS_38_331_REL19",
    "TS_38_413_REL19",
    "TS_38_321_REL19",

    "PATHS",
    "sha256_file",
]


missing_cell5_objects = [

    name
    for name in REQUIRED_CELL5_OBJECTS

    if name not in globals()
]


if missing_cell5_objects:

    raise RuntimeError(
        "Cell 6 requires Cells 4 and 5 first. "
        f"Missing objects: {missing_cell5_objects}"
    )


# ============================================================
# 6.2 — KNOWLEDGE ENUMS
# ============================================================

class RANInvestigationStage(str, Enum):

    RANDOM_ACCESS = "RANDOM_ACCESS"

    RRC_ESTABLISHMENT = (
        "RRC_ESTABLISHMENT"
    )

    NG_SIGNALLING = (
        "NG_SIGNALLING"
    )

    DRB_ESTABLISHMENT = (
        "DRB_ESTABLISHMENT"
    )

    RRC_REESTABLISHMENT = (
        "RRC_REESTABLISHMENT"
    )

    RRC_RESUME = (
        "RRC_RESUME"
    )

    DRB_RESUME = (
        "DRB_RESUME"
    )

    RADIO_CONDITION = (
        "RADIO_CONDITION"
    )

    RESOURCE_PRESSURE = (
        "RESOURCE_PRESSURE"
    )

    ALARM_EVENT = (
        "ALARM_EVENT"
    )

    CONFIGURATION = (
        "CONFIGURATION"
    )

    TRANSPORT_BOUNDARY = (
        "TRANSPORT_BOUNDARY"
    )


class PMMeasurementClass(str, Enum):

    CC = "CC"

    DER = "DER"

    SI = "SI"

    UNKNOWN = "UNKNOWN"


class EvidenceClassification(str, Enum):

    STANDARD_KPI_DEPENDENCY = (
        "STANDARD_KPI_DEPENDENCY"
    )

    STANDARD_SUPPORTING_PM = (
        "STANDARD_SUPPORTING_PM"
    )

    OPERATIONAL_DIAGNOSTIC = (
        "OPERATIONAL_DIAGNOSTIC"
    )


class GraphCoverageStatus(str, Enum):

    REPRESENTED = "REPRESENTED"

    ALIAS_REPRESENTED = (
        "ALIAS_REPRESENTED"
    )

    NOT_REPRESENTED = (
        "NOT_REPRESENTED"
    )


# ============================================================
# 6.3 — KNOWLEDGE MODELS
# ============================================================

class RANPMKnowledge(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    measurement_name: str

    aliases: List[str] = Field(
        default_factory=list
    )

    specification: str

    release: str

    version: str

    clause: str

    entity_scope: str

    measurement_class: PMMeasurementClass

    investigation_stage: RANInvestigationStage

    evidence_classification: EvidenceClassification

    description: str

    engineering_interpretation: str

    investigation_questions: List[str] = Field(
        default_factory=list
    )

    cell5_node_id: Optional[str] = None


class RANKPIFormulaKnowledge(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    kpi_id: str

    kpi_name: str

    specification: str

    clause: str

    formula_summary: str

    required_measurements: List[str]

    exclusion_rules: List[str]

    engineering_interpretation: str


class RANOperationalKnowledge(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    diagnostic_id: str

    name: str

    investigation_stage: RANInvestigationStage

    description: str

    evidence_examples: List[str]

    engineering_questions: List[str]

    standard_kpi: bool = False

    binding_status: str = (
        "OPERATIONAL_OR_DEFERRED"
    )


# ============================================================
# 6.4 — MEASUREMENT NAME NORMALIZATION
# ============================================================
#
# IMPORTANT SOURCE DIFFERENCE
#
# TS 28.552 v19.7.0 uses:
#
#   UECNTX.ConnEstabAtt.Cause
#   UECNTX.ConnEstabSucc.Cause
#
# TS 28.554 v19.7.0 KPI formula text uses:
#
#   UECNTXT.ConnEstabAtt.Cause
#   UECNTXT.ConnEstabSucc.Cause
#
# We retain TS 28.552 as the canonical PM catalogue name
# and preserve the TS 28.554 spelling as an alias.
# ============================================================


MEASUREMENT_ALIAS_MAP = {

    # --------------------------------------------------------
    # TS 28.554 -> TS 28.552 UE-context naming normalization
    # --------------------------------------------------------

    "UECNTXT.ConnEstabAtt.Cause":
        "UECNTX.ConnEstabAtt.Cause",

    "UECNTXT.ConnEstabSucc.Cause":
        "UECNTX.ConnEstabSucc.Cause",

    # --------------------------------------------------------
    # TS 28.554 -> TS 28.552 RRC Resume naming normalization
    # --------------------------------------------------------

    "RRC.ResumeAtt.cause":
        "RRC.ResumeAtt.Cause",

    "RRC.ResumeSucc.cause":
        "RRC.ResumeSucc.Cause",
}


def canonical_measurement_name(
    measurement_name: str,
) -> str:

    return MEASUREMENT_ALIAS_MAP.get(
        measurement_name,
        measurement_name,
    )


# ============================================================
# 6.5 — STANDARDIZED KPI-DEPENDENCY PM CATALOGUE
# ============================================================
#
# Exact measurements required by the selected Cell 5
# Accessibility KPIs.
#
# Most measurements are already represented in Cell 5.
#
# Four additional standardized PM dependencies required by
# the exact Total DRB Accessibility formula are deliberately
# included here even though Cell 5 currently lacks nodes for
# them.
# ============================================================


STANDARD_PM_KNOWLEDGE = {}


def add_pm_knowledge(
    measurement_name,
    clause,
    entity_scope,
    measurement_class,
    investigation_stage,
    description,
    engineering_interpretation,
    investigation_questions,
    cell5_node_id=None,
    aliases=None,
    evidence_classification=(
        EvidenceClassification.STANDARD_KPI_DEPENDENCY
    ),
):

    record = RANPMKnowledge(

        measurement_name=measurement_name,

        aliases=aliases or [],

        specification="TS 28.552",

        release="Release 19",

        version="19.7.0",

        clause=clause,

        entity_scope=entity_scope,

        measurement_class=measurement_class,

        investigation_stage=investigation_stage,

        evidence_classification=(
            evidence_classification
        ),

        description=description,

        engineering_interpretation=(
            engineering_interpretation
        ),

        investigation_questions=(
            investigation_questions
        ),

        cell5_node_id=cell5_node_id,
    )

    STANDARD_PM_KNOWLEDGE[
        measurement_name
    ] = record


# ============================================================
# RRC CONNECTION ESTABLISHMENT
# ============================================================

add_pm_knowledge(

    measurement_name=(
        "RRC.ConnEstabAtt.Cause"
    ),

    clause="5.1.1.15.1",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.RRC_ESTABLISHMENT
    ),

    description=(
        "Attempted RRC connection establishments "
        "split by establishment cause."
    ),

    engineering_interpretation=(
        "Represents demand entering the RRC "
        "connection-establishment stage. "
        "An attempt increase alone is not a failure; "
        "it must be interpreted together with "
        "successful establishments and cause mix."
    ),

    investigation_questions=[

        "Did RRC establishment attempts increase?",

        "Is the degradation broad or concentrated "
        "in specific establishment causes?",

        "Did attempts change without a corresponding "
        "increase in successes?",
    ],

    cell5_node_id=(
        "measurement::rrc_conn_estab_att"
    ),
)


add_pm_knowledge(

    measurement_name=(
        "RRC.ConnEstabSucc.Cause"
    ),

    clause="5.1.1.15.2",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.RRC_ESTABLISHMENT
    ),

    description=(
        "Successful RRC connection establishments "
        "split by establishment cause."
    ),

    engineering_interpretation=(
        "A reduction relative to RRC attempts "
        "indicates degradation within or before "
        "successful completion of RRC establishment."
    ),

    investigation_questions=[

        "Has the RRC success/attempt relationship "
        "degraded?",

        "Are failures localized to a subset of cells?",

        "Do RACH, radio quality, resource load or "
        "alarms explain the degradation?",
    ],

    cell5_node_id=(
        "measurement::rrc_conn_estab_succ"
    ),
)


# ============================================================
# UE-ASSOCIATED LOGICAL NG CONNECTION
# ============================================================

add_pm_knowledge(

    measurement_name=(
        "UECNTX.ConnEstabAtt.Cause"
    ),

    aliases=[
        "UECNTXT.ConnEstabAtt.Cause"
    ],

    clause="5.1.1.16.1",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.NG_SIGNALLING
    ),

    description=(
        "Attempted UE-associated logical "
        "NG-connection establishments from "
        "gNB toward AMF, split by establishment "
        "cause."
    ),

    engineering_interpretation=(
        "Represents signalling demand leaving the "
        "RAN toward the 5GC after RRC establishment."
    ),

    investigation_questions=[

        "Are RRC connections succeeding before "
        "NG signalling attempts occur?",

        "Did NG signalling attempts remain normal "
        "while successes declined?",

        "Is the symptom local to RAN or consistent "
        "with a transport / AMF-facing boundary issue?",
    ],

    cell5_node_id=(
        "measurement::uecntxt_conn_estab_att"
    ),
)


add_pm_knowledge(

    measurement_name=(
        "UECNTX.ConnEstabSucc.Cause"
    ),

    aliases=[
        "UECNTXT.ConnEstabSucc.Cause"
    ],

    clause="5.1.1.16.2",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.NG_SIGNALLING
    ),

    description=(
        "Successful UE-associated logical "
        "NG-connection establishments from "
        "gNB toward AMF."
    ),

    engineering_interpretation=(
        "If RRC establishment is healthy but this "
        "measurement degrades relative to attempts, "
        "the investigation should move toward NGAP, "
        "transport and RAN-to-Core boundary evidence."
    ),

    investigation_questions=[

        "Is RRC Setup healthy while NG connection "
        "success falls?",

        "Are SCTP / NG-interface conditions abnormal?",

        "Are there RAN alarms or configuration "
        "changes affecting N2 connectivity?",
    ],

    cell5_node_id=(
        "measurement::uecntxt_conn_estab_succ"
    ),
)


# ============================================================
# DRB ESTABLISHMENT
# ============================================================

add_pm_knowledge(

    measurement_name=(
        "DRB.EstabAtt.5QI"
    ),

    clause="5.1.1.10.1",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.DRB_ESTABLISHMENT
    ),

    description=(
        "DRBs attempted to be established, "
        "split by mapped 5QI."
    ),

    engineering_interpretation=(
        "Represents bearer-establishment demand "
        "after the required service signalling "
        "reaches the gNB."
    ),

    investigation_questions=[

        "Is DRB establishment demand changing?",

        "Is degradation limited to particular 5QIs?",

        "Are radio or admission/resource conditions "
        "restricting bearer establishment?",
    ],

    cell5_node_id=(
        "measurement::drb_estab_att_5qi"
    ),
)


add_pm_knowledge(

    measurement_name=(
        "DRB.EstabSucc.5QI"
    ),

    clause="5.1.1.10.2",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.DRB_ESTABLISHMENT
    ),

    description=(
        "DRBs successfully established, "
        "split by mapped 5QI."
    ),

    engineering_interpretation=(
        "A decline relative to attempts isolates "
        "degradation at the DRB-establishment stage "
        "when RRC and NG signalling remain healthy."
    ),

    investigation_questions=[

        "Is DRB SR degraded while RRC and NG "
        "signalling remain healthy?",

        "Is degradation 5QI-specific?",

        "Do resource pressure, radio condition or "
        "configuration changes align with the event?",
    ],

    cell5_node_id=(
        "measurement::drb_estab_succ_5qi"
    ),
)


add_pm_knowledge(

    measurement_name=(
        "DRB.EstabAtt.SNSSAI"
    ),

    clause="5.1.1.10.1",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.DRB_ESTABLISHMENT
    ),

    description=(
        "DRB establishment attempts "
        "split by S-NSSAI."
    ),

    engineering_interpretation=(
        "Provides slice-specific visibility into "
        "bearer-establishment demand."
    ),

    investigation_questions=[

        "Is the accessibility issue slice-specific?",

        "Does the affected S-NSSAI show abnormal "
        "attempt volume or success behaviour?",
    ],

    cell5_node_id=(
        "measurement::drb_estab_att_snssai"
    ),
)


add_pm_knowledge(

    measurement_name=(
        "DRB.EstabSucc.SNSSAI"
    ),

    clause="5.1.1.10.2",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.DRB_ESTABLISHMENT
    ),

    description=(
        "Successful DRB establishments "
        "split by S-NSSAI."
    ),

    engineering_interpretation=(
        "Used with DRB establishment attempts to "
        "identify slice-specific bearer "
        "accessibility degradation."
    ),

    investigation_questions=[

        "Is the failure limited to one S-NSSAI?",

        "Are other slices on the same cell healthy?",
    ],

    cell5_node_id=(
        "measurement::drb_estab_succ_snssai"
    ),
)


# ============================================================
# INITIAL DRB ESTABLISHMENT
# ============================================================

add_pm_knowledge(

    measurement_name=(
        "DRB.InitialEstabAtt.5QI"
    ),

    clause="5.1.1.10.5",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.DRB_ESTABLISHMENT
    ),

    description=(
        "Initial DRBs attempted to be established "
        "during Initial Context Setup."
    ),

    engineering_interpretation=(
        "Separates initial bearer establishment "
        "from later added DRB establishment."
    ),

    investigation_questions=[

        "Is degradation specific to initial "
        "service establishment?",

        "Are later added DRBs behaving differently?",
    ],

    cell5_node_id=(
        "measurement::drb_initial_att"
    ),
)


add_pm_knowledge(

    measurement_name=(
        "DRB.InitialEstabSucc.5QI"
    ),

    clause="5.1.1.10.6",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.DRB_ESTABLISHMENT
    ),

    description=(
        "Initial DRBs successfully established "
        "during Initial Context Setup."
    ),

    engineering_interpretation=(
        "Used to determine whether initial bearer "
        "creation is the stage degrading Total DRB "
        "Accessibility."
    ),

    investigation_questions=[

        "Did initial DRB success decline?",

        "Is the issue present before additional "
        "DRB or Resume procedures occur?",
    ],

    cell5_node_id=(
        "measurement::drb_initial_succ"
    ),
)


# ============================================================
# DRB RESUME
# ============================================================

add_pm_knowledge(

    measurement_name=(
        "DRB.ResumeAtt.5QI"
    ),

    clause="5.1.1.10.7",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.DRB_RESUME
    ),

    description=(
        "DRBs attempted to be resumed following "
        "an RRC Resume request."
    ),

    engineering_interpretation=(
        "Represents bearer demand associated with "
        "Resume-based service restoration."
    ),

    investigation_questions=[

        "Is the degradation specific to "
        "Resume-based access?",

        "Are initial DRB establishments healthy?",
    ],

    cell5_node_id=(
        "measurement::drb_resume_att"
    ),
)


add_pm_knowledge(

    measurement_name=(
        "DRB.ResumeSucc.5QI"
    ),

    clause="5.1.1.10.8",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.DRB_RESUME
    ),

    description=(
        "DRBs successfully resumed."
    ),

    engineering_interpretation=(
        "A decline relative to resume attempts may "
        "indicate a Resume-path accessibility issue "
        "rather than initial access degradation."
    ),

    investigation_questions=[

        "Are Resume attempts increasing?",

        "Are RRC Resume and DRB Resume degrading "
        "together?",
    ],

    cell5_node_id=(
        "measurement::drb_resume_succ"
    ),
)


# ============================================================
# RRC RESUME
# ============================================================

add_pm_knowledge(

    measurement_name=(
        "RRC.ResumeAtt.Cause"
    ),

    aliases=[
        "RRC.ResumeAtt.cause"
    ],

    clause="5.1.1.18.1",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.RRC_RESUME
    ),

    description=(
        "RRC connection Resume attempts "
        "split by Resume cause."
    ),

    engineering_interpretation=(
        "Represents demand entering the RRC Resume "
        "path from RRC_INACTIVE."
    ),

    investigation_questions=[

        "Has Resume demand changed?",

        "Which Resume causes dominate?",

        "Are RNA-update-related causes being "
        "excluded where required by the KPI formula?",
    ],

    cell5_node_id=(
        "measurement::rrc_resume_att"
    ),
)


add_pm_knowledge(

    measurement_name=(
        "RRC.ResumeSucc.Cause"
    ),

    aliases=[
        "RRC.ResumeSucc.cause"
    ],

    clause="5.1.1.18.2",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.RRC_RESUME
    ),

    description=(
        "Successful RRC Resume procedures "
        "split by Resume cause."
    ),

    engineering_interpretation=(
        "Used with Resume attempts and fallback "
        "attempts to derive the RRC Resume component "
        "required by Total DRB Accessibility."
    ),

    investigation_questions=[

        "Has RRC Resume success deteriorated?",

        "Did fallback-to-setup behaviour increase?",
    ],

    cell5_node_id=(
        "measurement::rrc_resume_succ"
    ),
)


# ============================================================
# EXACT TOTAL-DRB FORMULA DEPENDENCIES NOT YET IN CELL 5
# ============================================================
#
# These measurements are required by the exact TS 28.554
# Total DRB Accessibility formula.
#
# They are intentionally represented here even though the
# frozen Cell 5 graph does not yet include them.
# ============================================================


add_pm_knowledge(

    measurement_name=(
        "RRC.ResumeSuccByFallback.cause"
    ),

    clause="5.1.1.18.3",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.RRC_RESUME
    ),

    description=(
        "Successful RRC Resume procedures completed "
        "through fallback to RRC Setup."
    ),

    engineering_interpretation=(
        "Required by the exact RRC connection setup "
        "success component used in Total DRB "
        "Accessibility."
    ),

    investigation_questions=[

        "Is successful fallback masking deterioration "
        "in the normal Resume path?",

        "Did fallback activity increase during the "
        "incident?",
    ],

    cell5_node_id=None,
)


add_pm_knowledge(

    measurement_name=(
        "RRC.ResumeFallbackToSetupAtt.cause"
    ),

    clause="5.1.1.18.6",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.RRC_RESUME
    ),

    description=(
        "RRC Resume attempts followed by fallback "
        "to RRC Setup."
    ),

    engineering_interpretation=(
        "Required both for the exact RRC Resume "
        "success-rate denominator and the RRC setup "
        "component used in Total DRB Accessibility."
    ),

    investigation_questions=[

        "Has fallback-to-setup frequency increased?",

        "Is UE context retrieval or Resume handling "
        "showing abnormal behaviour?",
    ],

    cell5_node_id=None,
)


add_pm_knowledge(

    measurement_name=(
        "RRC.ReEstabSuccWithoutUeContext"
    ),

    clause="5.1.1.17.3",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.RRC_REESTABLISHMENT
    ),

    description=(
        "Successful RRC re-establishment where "
        "the prior UE context cannot be retrieved "
        "and setup completes through the fallback path."
    ),

    engineering_interpretation=(
        "Contributes to the exact RRC connection "
        "setup success component used in Total DRB "
        "Accessibility."
    ),

    investigation_questions=[

        "Is loss of UE context becoming more common?",

        "Does re-establishment behaviour correlate "
        "with accessibility degradation?",
    ],

    cell5_node_id=None,
)


add_pm_knowledge(

    measurement_name=(
        "RRC.ReEstabFallbackToSetupAtt"
    ),

    clause="5.1.1.17.4",

    entity_scope="NRCellCU",

    measurement_class=(
        PMMeasurementClass.CC
    ),

    investigation_stage=(
        RANInvestigationStage.RRC_REESTABLISHMENT
    ),

    description=(
        "RRC re-establishment attempts that fall "
        "back to RRC Setup because UE context "
        "cannot be retrieved."
    ),

    engineering_interpretation=(
        "Required in the denominator of the exact "
        "RRC connection setup component used in "
        "Total DRB Accessibility."
    ),

    investigation_questions=[

        "Did re-establishment fallback increase?",

        "Is UE-context recovery contributing to "
        "the accessibility symptom?",
    ],

    cell5_node_id=None,
)


# ============================================================
# 6.6 — STANDARDIZED RACH SUPPORTING PM
# ============================================================
#
# These are standardized TS 28.552 measurements.
#
# They support Random Access investigation, but they do NOT
# by themselves define a standardized TS 28.554
# "RACH Success Rate".
#
# TS 28.552 Clause 5.1.1.20.1 defines average received
# Random Access preambles per cell:
#
#   RACH.PreambleDedCell
#   RACH.PreambleACell
#   RACH.PreambleBCell
#
# Entity: NRCellDU
# Measurement class: DER
# ============================================================


for measurement_name, description in [

    (
        "RACH.PreambleDedCell",

        "Average received dedicated Random Access "
        "preambles per cell.",
    ),

    (
        "RACH.PreambleACell",

        "Average received randomly selected "
        "Group-A Random Access preambles per cell.",
    ),

    (
        "RACH.PreambleBCell",

        "Average received randomly selected "
        "Group-B Random Access preambles per cell.",
    ),
]:

    add_pm_knowledge(

        measurement_name=measurement_name,

        clause="5.1.1.20.1",

        entity_scope="NRCellDU",

        measurement_class=(
            PMMeasurementClass.DER
        ),

        investigation_stage=(
            RANInvestigationStage.RANDOM_ACCESS
        ),

        description=description,

        engineering_interpretation=(
            "Provides standardized RACH-load / "
            "access-attempt context. Increased "
            "preamble activity may indicate higher "
            "access demand or repeated access "
            "behaviour, but should not be interpreted "
            "as a standalone success-rate measurement."
        ),

        investigation_questions=[

            "Did RACH activity change before "
            "RRC degradation?",

            "Is the change cell-wide or beam-related?",

            "Does RACH behaviour align with radio "
            "quality or resource-pressure evidence?",
        ],

        cell5_node_id=None,

        evidence_classification=(
            EvidenceClassification.STANDARD_SUPPORTING_PM
        ),
    )


# ============================================================
# 6.7 — EXACT KPI FORMULA KNOWLEDGE
# ============================================================

KPI_FORMULA_KNOWLEDGE = {}


partial_drb_formula = RANKPIFormulaKnowledge(

    kpi_id=(
        "partial_drb_accessibility"
    ),

    kpi_name=(
        "Partial DRB Accessibility for UE Services"
    ),

    specification="TS 28.554",

    clause="6.2.4",

    formula_summary=(
        "RRC connection setup success ratio "
        "* UE-associated logical NG-connection "
        "success ratio "
        "* DRB establishment success ratio "
        "* 100"
    ),

    required_measurements=[

        "RRC.ConnEstabAtt.Cause",
        "RRC.ConnEstabSucc.Cause",

        "UECNTX.ConnEstabAtt.Cause",
        "UECNTX.ConnEstabSucc.Cause",

        "DRB.EstabAtt.5QI",
        "DRB.EstabSucc.5QI",
    ],

    exclusion_rules=[

        (
            "Exclude establishment cause "
            "mo-Signalling from the RRC connection "
            "setup success calculation."
        ),

        (
            "Exclude establishment cause "
            "mo-Signalling from the UE-associated "
            "logical NG-connection success calculation."
        ),
    ],

    engineering_interpretation=(
        "Partial DRB Accessibility can be decomposed "
        "into three sequential RAN stages: RRC access, "
        "NG signalling and DRB establishment."
    ),
)


KPI_FORMULA_KNOWLEDGE[
    partial_drb_formula.kpi_id
] = partial_drb_formula


total_drb_formula = RANKPIFormulaKnowledge(

    kpi_id=(
        "total_drb_accessibility"
    ),

    kpi_name=(
        "Total DRB Accessibility for UE Services"
    ),

    specification="TS 28.554",

    clause="6.2.7",

    formula_summary=(
        "Composite accessibility across initial DRB "
        "establishment, added DRB establishment and "
        "DRB Resume, adjusted by exact RRC setup, "
        "NG-connection and RRC Resume success ratios."
    ),

    required_measurements=[

        # RRC establishment
        "RRC.ConnEstabAtt.Cause",
        "RRC.ConnEstabSucc.Cause",

        # UE-associated NG signalling
        "UECNTX.ConnEstabAtt.Cause",
        "UECNTX.ConnEstabSucc.Cause",

        # Initial DRB
        "DRB.InitialEstabAtt.5QI",
        "DRB.InitialEstabSucc.5QI",

        # Added / total DRB
        "DRB.EstabAtt.5QI",
        "DRB.EstabSucc.5QI",

        # DRB Resume
        "DRB.ResumeAtt.5QI",
        "DRB.ResumeSucc.5QI",

        # RRC Resume
        "RRC.ResumeAtt.cause",
        "RRC.ResumeSucc.cause",

        # Exact RRC fallback dependencies
        "RRC.ResumeSuccByFallback.cause",
        "RRC.ResumeFallbackToSetupAtt.cause",

        # Exact RRC re-establishment dependencies
        "RRC.ReEstabSuccWithoutUeContext",
        "RRC.ReEstabFallbackToSetupAtt",
    ],

    exclusion_rules=[

        (
            "Exclude establishment cause "
            "mo-Signalling from the RRC connection "
            "setup success calculation."
        ),

        (
            "Exclude establishment cause "
            "mo-Signalling from the UE-associated "
            "logical NG-connection success calculation."
        ),

        (
            "Exclude RRC Resume causes related to "
            "RNA update from the RRC Resume "
            "success-rate calculation."
        ),
    ],

    engineering_interpretation=(
        "Total DRB Accessibility extends the "
        "initial-access view to include additional "
        "bearer setup and Resume-based service "
        "establishment. Correct diagnosis therefore "
        "requires distinguishing initial access, "
        "added bearer setup, Resume and fallback "
        "behaviour."
    ),
)


KPI_FORMULA_KNOWLEDGE[
    total_drb_formula.kpi_id
] = total_drb_formula


# ============================================================
# 6.8 — OPERATIONAL ENGINEERING KNOWLEDGE
# ============================================================

OPERATIONAL_RAN_KNOWLEDGE = {}


def add_operational_knowledge(
    diagnostic_id,
    name,
    stage,
    description,
    evidence_examples,
    engineering_questions,
    binding_status,
):

    record = RANOperationalKnowledge(

        diagnostic_id=diagnostic_id,

        name=name,

        investigation_stage=stage,

        description=description,

        evidence_examples=(
            evidence_examples
        ),

        engineering_questions=(
            engineering_questions
        ),

        standard_kpi=False,

        binding_status=binding_status,
    )

    OPERATIONAL_RAN_KNOWLEDGE[
        diagnostic_id
    ] = record


# ------------------------------------------------------------
# RACH
# ------------------------------------------------------------

add_operational_knowledge(

    diagnostic_id="rach_access",

    name=(
        "Random Access / RACH Behaviour"
    ),

    stage=(
        RANInvestigationStage.RANDOM_ACCESS
    ),

    description=(
        "Pre-RRC evidence used to determine whether "
        "Accessibility degradation begins during "
        "Random Access."
    ),

    evidence_examples=[

        "RACH.PreambleDedCell",
        "RACH.PreambleACell",
        "RACH.PreambleBCell",

        "RACH preamble distribution",

        "RACH access-delay distribution",
    ],

    engineering_questions=[

        "Did Random Access behaviour degrade before "
        "RRC Setup SR?",

        "Is excessive preamble activity consistent "
        "with repeated access attempts or contention?",

        "Is the problem limited to specific cells "
        "or beams?",

        "Does radio quality explain the RACH symptom?",
    ],

    binding_status=(
        "PARTIALLY_STANDARDIZED_TS_28_552"
    ),
)


# ------------------------------------------------------------
# RADIO CONDITION
# ------------------------------------------------------------

add_operational_knowledge(

    diagnostic_id="radio_condition",

    name=(
        "Radio Coverage / Quality Condition"
    ),

    stage=(
        RANInvestigationStage.RADIO_CONDITION
    ),

    description=(
        "Radio-condition evidence used to determine "
        "whether poor coverage or radio quality "
        "contributes to Accessibility degradation."
    ),

    evidence_examples=[

        "RSRP",
        "RSRQ",
        "SINR",

        "coverage distribution",
        "quality distribution",
    ],

    engineering_questions=[

        "Did coverage or quality deteriorate at the "
        "same time as Accessibility?",

        "Is the issue geographically localized?",

        "Does normal radio condition allow radio "
        "causes to be deprioritized?",
    ],

    binding_status=(
        "EXACT_PM_BINDING_DEFERRED"
    ),
)


# ------------------------------------------------------------
# RESOURCE PRESSURE
# ------------------------------------------------------------

add_operational_knowledge(

    diagnostic_id="resource_pressure",

    name=(
        "Cell Load / Radio Resource Pressure"
    ),

    stage=(
        RANInvestigationStage.RESOURCE_PRESSURE
    ),

    description=(
        "Resource evidence used to determine whether "
        "capacity or admission pressure contributes "
        "to access or bearer-establishment failure."
    ),

    evidence_examples=[

        "PRB utilisation",

        "connected UE population",

        "control-channel pressure",

        "admission / resource availability",
    ],

    engineering_questions=[

        "Did load increase before the accessibility "
        "degradation?",

        "Is degradation load-dependent?",

        "Are low-load cells healthy while high-load "
        "cells degrade?",
    ],

    binding_status=(
        "EXACT_PM_BINDING_DEFERRED"
    ),
)


# ------------------------------------------------------------
# ALARMS
# ------------------------------------------------------------

add_operational_knowledge(

    diagnostic_id="ran_alarms_events",

    name=(
        "RAN Alarms / Events"
    ),

    stage=(
        RANInvestigationStage.ALARM_EVENT
    ),

    description=(
        "Fault-management and event evidence used "
        "to corroborate or challenge KPI-based "
        "hypotheses."
    ),

    evidence_examples=[

        "cell alarms",

        "gNB function alarms",

        "interface events",

        "restart / availability events",
    ],

    engineering_questions=[

        "Did an alarm precede the KPI degradation?",

        "Is the alarm on the same affected network "
        "element or cell set?",

        "Does the alarm explain the affected "
        "procedure stage?",
    ],

    binding_status=(
        "OPERATIONAL_TELEMETRY"
    ),
)


# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

add_operational_knowledge(

    diagnostic_id="ran_configuration",

    name=(
        "RAN Configuration State"
    ),

    stage=(
        RANInvestigationStage.CONFIGURATION
    ),

    description=(
        "Configuration and change evidence used to "
        "determine whether a parameter, feature or "
        "software/configuration change coincides "
        "with Accessibility degradation."
    ),

    evidence_examples=[

        "parameter changes",

        "feature activation",

        "software upgrade",

        "cell administrative state",

        "admission-control configuration",
    ],

    engineering_questions=[

        "What changed before the incident?",

        "Is the change common to all affected cells?",

        "Do unaffected cells have a different "
        "configuration state?",
    ],

    binding_status=(
        "OPERATIONAL_CONFIGURATION"
    ),
)


# ------------------------------------------------------------
# TRANSPORT BOUNDARY
# ------------------------------------------------------------

add_operational_knowledge(

    diagnostic_id="transport_boundary",

    name=(
        "RAN-to-Core Transport Boundary"
    ),

    stage=(
        RANInvestigationStage.TRANSPORT_BOUNDARY
    ),

    description=(
        "Boundary evidence used when RRC access is "
        "healthy but UE-associated NG signalling "
        "toward the 5GC is impaired."
    ),

    evidence_examples=[

        "N2 reachability",

        "SCTP association state",

        "latency",

        "packet loss",

        "interface alarms",
    ],

    engineering_questions=[

        "Is RRC healthy but NG signalling degraded?",

        "Is the SCTP / N2 path healthy?",

        "Is the symptom common across cells sharing "
        "the same transport or AMF-facing path?",
    ],

    binding_status=(
        "TRANSPORT_BOUNDARY_EVIDENCE"
    ),
)


# ============================================================
# 6.9 — INVESTIGATION STAGE ORDER
# ============================================================
#
# This is a reasoning aid, not a mandatory LLM sequence.
#
# The future agent may skip stages when deterministic evidence
# already rules them out.
#
# This supports the Module 5 design principle:
#
#   Agentic AI selectively augments a mature troubleshooting
#   process rather than replacing every deterministic step.
# ============================================================


RAN_ACCESSIBILITY_INVESTIGATION_ORDER = [

    {
        "order": 1,

        "stage":
            RANInvestigationStage.RANDOM_ACCESS.value,

        "question":
            "Does the problem begin before RRC establishment?",
    },

    {
        "order": 2,

        "stage":
            RANInvestigationStage.RRC_ESTABLISHMENT.value,

        "question":
            "Is RRC connection establishment degraded?",
    },

    {
        "order": 3,

        "stage":
            RANInvestigationStage.NG_SIGNALLING.value,

        "question":
            "Is UE-associated NG signalling degraded after healthy RRC access?",
    },

    {
        "order": 4,

        "stage":
            RANInvestigationStage.DRB_ESTABLISHMENT.value,

        "question":
            "Is bearer establishment the failing stage?",
    },

    {
        "order": 5,

        "stage":
            RANInvestigationStage.RRC_RESUME.value,

        "question":
            "Is Resume / fallback behaviour contributing?",
    },

    {
        "order": 6,

        "stage":
            RANInvestigationStage.RADIO_CONDITION.value,

        "question":
            "Do radio conditions explain the symptom?",
    },

    {
        "order": 7,

        "stage":
            RANInvestigationStage.RESOURCE_PRESSURE.value,

        "question":
            "Does resource pressure explain the symptom?",
    },

    {
        "order": 8,

        "stage":
            RANInvestigationStage.ALARM_EVENT.value,

        "question":
            "Do alarms or events corroborate a fault mechanism?",
    },

    {
        "order": 9,

        "stage":
            RANInvestigationStage.CONFIGURATION.value,

        "question":
            "Did configuration or software state change?",
    },

    {
        "order": 10,

        "stage":
            RANInvestigationStage.TRANSPORT_BOUNDARY.value,

        "question":
            "If RAN-local evidence is insufficient, does the issue cross the RAN boundary?",
    },
]


# ============================================================
# 6.10 — CELL 5 GRAPH COVERAGE VALIDATION
# ============================================================


# ------------------------------------------------------------
# Extract Cell 5 standardized measurement strings.
# ------------------------------------------------------------

cell5_measurement_names = {}


for node_id, node in (
    RAN_ACCESSIBILITY_NODES.items()
):

    if (
        node.layer
        == RANAccessibilityNodeLayer.MEASUREMENT
    ):

        raw_name = node.metadata.get(
            "measurement_name"
        )

        if raw_name is not None:

            canonical_name = (
                canonical_measurement_name(
                    raw_name
                )
            )

            cell5_measurement_names[
                canonical_name
            ] = {

                "node_id":
                    node_id,

                "raw_name":
                    raw_name,

                "canonical_name":
                    canonical_name,
            }


# ------------------------------------------------------------
# Required standardized KPI measurements.
# ------------------------------------------------------------

required_formula_measurements = set()


for kpi_record in (
    KPI_FORMULA_KNOWLEDGE.values()
):

    required_formula_measurements.update(

        canonical_measurement_name(
            name
        )

        for name
        in kpi_record.required_measurements
    )


# ------------------------------------------------------------
# Determine Cell 5 coverage.
# ------------------------------------------------------------

GRAPH_COVERAGE_RESULTS = {}


for measurement_name in sorted(
    required_formula_measurements
):

    knowledge_record = (
        STANDARD_PM_KNOWLEDGE.get(
            measurement_name
        )
    )

    if knowledge_record is None:

        raise RuntimeError(
            "Required formula measurement has no "
            "Cell 6 PM knowledge record: "
            f"{measurement_name}"
        )


    if (
        measurement_name
        in cell5_measurement_names
    ):

        cell5_info = (
            cell5_measurement_names[
                measurement_name
            ]
        )

        raw_name = (
            cell5_info["raw_name"]
        )

        status = (

            GraphCoverageStatus.REPRESENTED

            if raw_name == measurement_name

            else (
                GraphCoverageStatus.ALIAS_REPRESENTED
            )
        )

        cell5_node_id = (
            cell5_info["node_id"]
        )

    else:

        status = (
            GraphCoverageStatus.NOT_REPRESENTED
        )

        cell5_node_id = None


    GRAPH_COVERAGE_RESULTS[
        measurement_name
    ] = {

        "measurement_name":
            measurement_name,

        "status":
            status.value,

        "cell5_node_id":
            cell5_node_id,

        "ts_28_552_clause":
            knowledge_record.clause,
    }


# ============================================================
# 6.11 — GAP ANALYSIS
# ============================================================

represented_measurements = [

    name

    for name, record
    in GRAPH_COVERAGE_RESULTS.items()

    if record["status"]
    in {

        GraphCoverageStatus.REPRESENTED.value,

        GraphCoverageStatus.ALIAS_REPRESENTED.value,
    }
]


missing_graph_measurements = [

    name

    for name, record
    in GRAPH_COVERAGE_RESULTS.items()

    if (
        record["status"]
        == GraphCoverageStatus.NOT_REPRESENTED.value
    )
]


alias_represented_measurements = [

    name

    for name, record
    in GRAPH_COVERAGE_RESULTS.items()

    if (
        record["status"]
        == GraphCoverageStatus.ALIAS_REPRESENTED.value
    )
]


CELL5_GRAPH_REVISION_REQUIRED = (
    len(missing_graph_measurements) > 0
)


# ============================================================
# 6.12 — VALIDATION
# ============================================================


# Formal RAN KPI set remains unchanged.

assert (
    set(
        RAN_ACCESSIBILITY_REFERENCE_KPIS.keys()
    )
    == {

        "partial_drb_accessibility",
        "total_drb_accessibility",
    }
)


# Every formula dependency must have a PM knowledge entry.

assert (
    required_formula_measurements
    <= set(
        STANDARD_PM_KNOWLEDGE.keys()
    )
)


# RACH supporting PM measurements must be present.

for rach_measurement in [

    "RACH.PreambleDedCell",
    "RACH.PreambleACell",
    "RACH.PreambleBCell",
]:

    assert (
        rach_measurement
        in STANDARD_PM_KNOWLEDGE
    )

    assert (
        STANDARD_PM_KNOWLEDGE[
            rach_measurement
        ].evidence_classification

        == EvidenceClassification.STANDARD_SUPPORTING_PM
    )


# RACH remains supporting PM evidence,
# NOT a standardized Accessibility KPI.

assert (
    "rach_access"
    in OPERATIONAL_RAN_KNOWLEDGE
)


assert (
    OPERATIONAL_RAN_KNOWLEDGE[
        "rach_access"
    ].standard_kpi
    is False
)


# Exact formula gap expected from current Cell 5.

EXPECTED_CURRENT_CELL5_GAPS = {

    "RRC.ResumeSuccByFallback.cause",

    "RRC.ResumeFallbackToSetupAtt.cause",

    "RRC.ReEstabSuccWithoutUeContext",

    "RRC.ReEstabFallbackToSetupAtt",
}


assert (
    set(
        missing_graph_measurements
    )
    == EXPECTED_CURRENT_CELL5_GAPS
), (
    "Unexpected Cell 5 formula-dependency gap. "
    f"Observed: {missing_graph_measurements}"
)


# ============================================================
# 6.13 — BUILD KNOWLEDGE SUPPLEMENT ARTIFACT
# ============================================================

ran_accessibility_knowledge_supplement = {

    "artifact_id":
        (
            "module5_5g_ran_accessibility_"
            "knowledge_supplement_v1"
        ),

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "module_scope":
        "MODULE_5_V1_5G_RAN",

    "reference_domain":
        "5G_RAN_ACCESSIBILITY",

    "standards_baseline": {

        "kpi_standard": {

            "specification":
                "TS 28.554",

            "release":
                "Release 19",

            "version":
                "19.7.0",
        },

        "measurement_standard": {

            "specification":
                "TS 28.552",

            "release":
                "Release 19",

            "version":
                "19.7.0",
        },
    },

    "measurement_aliases":
        MEASUREMENT_ALIAS_MAP,

    "standard_pm_knowledge": {

        measurement_name:
            record.model_dump(
                mode="json"
            )

        for measurement_name, record
        in STANDARD_PM_KNOWLEDGE.items()
    },

    "kpi_formula_knowledge": {

        kpi_id:
            record.model_dump(
                mode="json"
            )

        for kpi_id, record
        in KPI_FORMULA_KNOWLEDGE.items()
    },

    "operational_ran_knowledge": {

        diagnostic_id:
            record.model_dump(
                mode="json"
            )

        for diagnostic_id, record
        in OPERATIONAL_RAN_KNOWLEDGE.items()
    },

    "investigation_order":
        RAN_ACCESSIBILITY_INVESTIGATION_ORDER,

    "cell5_graph_coverage":
        GRAPH_COVERAGE_RESULTS,

    "cell5_graph_revision_required":
        CELL5_GRAPH_REVISION_REQUIRED,

    "cell5_missing_formula_dependencies":
        missing_graph_measurements,

    "design_notes": [

        (
            "Cell 6 adds engineering semantics to "
            "the structural Cell 5 dependency graph."
        ),

        (
            "TS 28.552 measurement names are used "
            "as canonical PM identifiers."
        ),

        (
            "Source naming differences between "
            "TS 28.552 and TS 28.554 are retained "
            "through an explicit alias map."
        ),

        (
            "RACH PM measurements are standardized "
            "supporting evidence but are not "
            "represented as a TS 28.554 "
            "Accessibility KPI."
        ),

        (
            "The exact Total DRB Accessibility "
            "formula exposes four standardized RRC "
            "dependencies not represented in the "
            "current Cell 5 graph."
        ),

        (
            "Cell 6 records the gap but does not "
            "silently mutate the frozen Cell 5 "
            "artifact."
        ),

        (
            "Operational evidence remains separate "
            "from standardized KPI and PM truth."
        ),

        (
            "The investigation order is guidance "
            "for evidence acquisition rather than "
            "a requirement that an LLM control "
            "every troubleshooting step."
        ),
    ],
}


# ============================================================
# 6.14 — DETERMINISTIC SEMANTIC HASH
# ============================================================

def canonical_json_sha256_cell6(
    data
):

    canonical = json.dumps(

        data,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

    ).encode(
        "utf-8"
    )

    return hashlib.sha256(
        canonical
    ).hexdigest()


cell6_semantic_content = {

    key: value

    for key, value
    in ran_accessibility_knowledge_supplement.items()

    if key != "created_utc"
}


cell6_semantic_sha = (
    canonical_json_sha256_cell6(
        cell6_semantic_content
    )
)


ran_accessibility_knowledge_supplement[
    "semantic_sha256"
] = cell6_semantic_sha


# ============================================================
# 6.15 — SAVE ARTIFACT
# ============================================================

cell6_path = (

    PATHS["kpi"]

    / (
        "module5_5g_ran_accessibility_"
        "knowledge_supplement_v1.json"
    )
)


with open(

    cell6_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        ran_accessibility_knowledge_supplement,

        f,

        indent=2,

        ensure_ascii=False,
    )


cell6_artifact_sha = (
    sha256_file(
        cell6_path
    )
)


# ============================================================
# 6.16 — SUMMARY STATISTICS
# ============================================================

kpi_dependency_pm_count = sum(

    1

    for record
    in STANDARD_PM_KNOWLEDGE.values()

    if (
        record.evidence_classification
        == (
            EvidenceClassification
            .STANDARD_KPI_DEPENDENCY
        )
    )
)


supporting_pm_count = sum(

    1

    for record
    in STANDARD_PM_KNOWLEDGE.values()

    if (
        record.evidence_classification
        == (
            EvidenceClassification
            .STANDARD_SUPPORTING_PM
        )
    )
)


# ============================================================
# 6.17 — FINAL SUMMARY
# ============================================================

print("=" * 104)

print(
    "MODULE 5 — CELL 05 "
    "5G RAN ACCESSIBILITY KPI KNOWLEDGE SUPPLEMENT"
)

print("=" * 104)


print(
    "\nStandards baseline"
)

print("-" * 104)

print(
    "KPI standard                     : "
    "TS 28.554 Rel-19 v19.7.0"
)

print(
    "Measurement standard             : "
    "TS 28.552 Rel-19 v19.7.0"
)


print(
    "\nKnowledge catalogue"
)

print("-" * 104)

print(
    f"Formal RAN Accessibility KPIs    : "
    f"{len(KPI_FORMULA_KNOWLEDGE)}"
)

print(
    f"Standard KPI-dependency PMs      : "
    f"{kpi_dependency_pm_count}"
)

print(
    f"Standard supporting PMs          : "
    f"{supporting_pm_count}"
)

print(
    f"Operational knowledge domains    : "
    f"{len(OPERATIONAL_RAN_KNOWLEDGE)}"
)


print(
    "\nCanonical naming"
)

print("-" * 104)

print(
    "TS 28.552 canonical             : "
    "UECNTX.ConnEstabAtt.Cause"
)

print(
    "TS 28.554 formula alias         : "
    "UECNTXT.ConnEstabAtt.Cause"
)

print(
    "Alias normalization             : ENABLED"
)


print(
    "\nRACH classification"
)

print("-" * 104)

print(
    "RACH standardized PM evidence   : YES"
)

print(
    "RACH TS 28.554 Accessibility KPI: NO"
)

print(
    "RACH success formula frozen     : NO"
)

print(
    "RACH role                       : "
    "PRE-RRC SUPPORTING DIAGNOSTIC"
)


print(
    "\nCell 5 graph coverage"
)

print("-" * 104)

print(
    f"Required exact-formula PMs       : "
    f"{len(required_formula_measurements)}"
)

print(
    f"Represented in Cell 5            : "
    f"{len(represented_measurements)}"
)

print(
    f"Alias-represented                : "
    f"{len(alias_represented_measurements)}"
)

print(
    f"Missing exact dependencies       : "
    f"{len(missing_graph_measurements)}"
)


if missing_graph_measurements:

    print(
        "\nMissing Cell 5 formula dependencies"
    )

    print("-" * 104)

    for measurement_name in (
        missing_graph_measurements
    ):

        record = (
            STANDARD_PM_KNOWLEDGE[
                measurement_name
            ]
        )

        print(

            f"{measurement_name:<42} : "
            f"TS 28.552 {record.clause}"
        )


print(
    "\nGraph revision assessment"
)

print("-" * 104)

print(
    "Cell 5 revision required         : "
    f"{CELL5_GRAPH_REVISION_REQUIRED}"
)

print(
    "Reason                           : "
    "Exact Total DRB Accessibility "
    "formula contains dependencies "
    "not yet represented in Cell 5."
)


print(
    "\nInvestigation path"
)

print("-" * 104)


for item in (
    RAN_ACCESSIBILITY_INVESTIGATION_ORDER
):

    print(

        f"{item['order']:>2}. "
        f"{item['stage']:<24} : "
        f"{item['question']}"
    )


print(
    "\nArtifact"
)

print("-" * 104)

print(
    f"Path                             : "
    f"{cell6_path}"
)

print(
    f"Artifact SHA                     : "
    f"{cell6_artifact_sha}"
)

print(
    f"Semantic SHA                     : "
    f"{cell6_semantic_sha}"
)


print(
    "\n" + "=" * 104
)

print(
    "CELL 05 COMPLETE — "
    "5G RAN ACCESSIBILITY KNOWLEDGE SUPPLEMENT READY"
)

print(
    "=" * 104
)
````

###### Cell 05 — Observation

Cell 05 confirms **18 standardized KPI-dependency PM families** plus **3 standardized supporting RACH PM families** and validates their relationship to the Accessibility investigation graph.

The standards-completeness check identifies four exact **Resume/re-establishment PM dependencies** required by the Total DRB Accessibility formula that are not yet represented in Cell 04.

> **This is a controlled graph-completeness gap rather than an architectural redesign; the next revision can patch the four missing standardized dependencies without changing the formal KPI scope.**

<a id="cell-06-standards-completeness-v1-1"></a>
##### Cell 06 — Accessibility Standards-Completeness Revision V1.1

Patch only the four exact standardized PM dependencies identified by Cell 05 and add their required graph edges. The original V1 artifact remains preserved as historical provenance.


````python
# ============================================================
# MODULE 5 — CELL 06
# CONTROLLED STANDARDS-COMPLETENESS REVISION
# 5G RAN ACCESSIBILITY DEPENDENCY GRAPH
# ============================================================
#
# Purpose:
#
#   1. Preserve the frozen Cell 5 V1 graph unchanged.
#
#   2. Apply only the four exact Total DRB Accessibility
#      dependencies identified by Cell 6 validation.
#
#   3. Update the V1.1 Total DRB KPI source-measurement list.
#
#   4. Revalidate exact formula coverage against Cell 6.
#
#   5. Persist a new immutable V1.1 artifact with new hashes.
#
#
# IMPORTANT:
#
#   - Run AFTER Cells 03, 04 and 05.
#
#   - RAN_ACCESSIBILITY_NODES and
#     RAN_ACCESSIBILITY_EDGES are NOT modified in place.
#
#   - The frozen Cell 5 V1 artifact is NOT overwritten.
#
# ============================================================


from copy import deepcopy
from datetime import datetime, timezone

import json
import hashlib


# ============================================================
# 5 V1.1.1 — DEPENDENCY VALIDATION
# ============================================================

REQUIRED_V1_1_OBJECTS = [

    # Cell 5 V1
    "RAN_ACCESSIBILITY_REFERENCE_KPIS",
    "RAN_ACCESSIBILITY_NODES",
    "RAN_ACCESSIBILITY_EDGES",

    "RANAccessibilityNodeLayer",
    "RANAlignmentType",
    "RANDependencyRelation",

    "RANAccessibilityEdge",
    "make_ran_node",

    "MeasurementDomain",
    "EntityType",

    "TS_28_552_REL19",

    "PATHS",
    "sha256_file",

    "ran_accessibility_domain",
    "ran_accessibility_path",
    "ran_accessibility_artifact_sha",
    "ran_accessibility_semantic_sha",

    # Cell 6
    "STANDARD_PM_KNOWLEDGE",
    "KPI_FORMULA_KNOWLEDGE",

    "canonical_measurement_name",

    "missing_graph_measurements",
    "CELL5_GRAPH_REVISION_REQUIRED",

    "cell6_semantic_sha",
]


missing_required_objects = [

    name

    for name
    in REQUIRED_V1_1_OBJECTS

    if name not in globals()
]


if missing_required_objects:

    raise RuntimeError(

        "Cell 06 requires Cells 03, 04 and 05 first. "
        f"Missing objects: {missing_required_objects}"
    )


# ============================================================
# 5 V1.1.2 — VERIFY FROZEN CELL 5 V1 BASELINE
# ============================================================

EXPECTED_CELL5_V1_SEMANTIC_SHA = (
    "11e52a727942b6d25e1b907e07539a82d37ad424d93039aa765c53489ee2896a"
)


assert (
    ran_accessibility_semantic_sha
    == EXPECTED_CELL5_V1_SEMANTIC_SHA
), (
    "Cell 5 V1 semantic baseline differs from the "
    "frozen reference. V1.1 patch aborted."
)


# Confirm that the V1 artifact currently on disk has not changed.

assert (
    sha256_file(
        ran_accessibility_path
    )
    == ran_accessibility_artifact_sha
), (
    "Cell 5 V1 artifact bytes changed after execution. "
    "V1.1 patch aborted."
)


# ============================================================
# 5 V1.1.3 — VERIFY CELL 6 GAP CONTRACT
# ============================================================

EXPECTED_CELL6_GAPS = {

    "RRC.ResumeSuccByFallback.cause",

    "RRC.ResumeFallbackToSetupAtt.cause",

    "RRC.ReEstabSuccWithoutUeContext",

    "RRC.ReEstabFallbackToSetupAtt",
}


assert (
    CELL5_GRAPH_REVISION_REQUIRED
    is True
)


assert (
    set(
        missing_graph_measurements
    )
    == EXPECTED_CELL6_GAPS
), (
    "Cell 6 gap set differs from the approved "
    "Cell 5 V1.1 patch scope. "
    f"Observed: {missing_graph_measurements}"
)


for measurement_name in EXPECTED_CELL6_GAPS:

    assert (
        measurement_name
        in STANDARD_PM_KNOWLEDGE
    ), (
        "Missing Cell 6 PM knowledge for approved "
        f"patch item: {measurement_name}"
    )


# ============================================================
# 5 V1.1.4 — CREATE NON-DESTRUCTIVE V1.1 WORKING COPIES
# ============================================================
#
# IMPORTANT:
#
# We copy the V1 structures instead of modifying them.
#
# Therefore:
#
#   RAN_ACCESSIBILITY_NODES
#   RAN_ACCESSIBILITY_EDGES
#   RAN_ACCESSIBILITY_REFERENCE_KPIS
#
# remain the frozen V1 baseline.
#
# ============================================================

RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_1 = dict(
    RAN_ACCESSIBILITY_REFERENCE_KPIS
)


RAN_ACCESSIBILITY_NODES_V1_1 = dict(
    RAN_ACCESSIBILITY_NODES
)


RAN_ACCESSIBILITY_EDGES_V1_1 = list(
    RAN_ACCESSIBILITY_EDGES
)


V1_NODE_COUNT = len(
    RAN_ACCESSIBILITY_NODES
)


V1_EDGE_COUNT = len(
    RAN_ACCESSIBILITY_EDGES
)


# ============================================================
# 5 V1.1.5 — REVISE TOTAL DRB KPI SOURCE DEPENDENCIES
# ============================================================
#
# Cell 5 V1 already contains the original Total DRB KPI.
#
# We do NOT modify that object.
#
# Instead we create a V1.1 copy and append only the four
# dependencies exposed by Cell 6.
#
# ============================================================

total_drb_v1 = (
    RAN_ACCESSIBILITY_REFERENCE_KPIS[
        "total_drb_accessibility"
    ]
)


total_drb_v1_1_source_measurements = list(
    total_drb_v1.source_measurements
)


for measurement_name in [

    "RRC.ResumeSuccByFallback.cause",

    "RRC.ResumeFallbackToSetupAtt.cause",

    "RRC.ReEstabSuccWithoutUeContext",

    "RRC.ReEstabFallbackToSetupAtt",

]:

    if (
        measurement_name
        not in total_drb_v1_1_source_measurements
    ):

        total_drb_v1_1_source_measurements.append(
            measurement_name
        )


total_drb_v1_1 = (
    total_drb_v1.model_copy(

        update={

            "source_measurements":
                total_drb_v1_1_source_measurements
        }
    )
)


RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_1[
    "total_drb_accessibility"
] = total_drb_v1_1


# ============================================================
# 5 V1.1.6 — DEFINE THE FOUR STANDARDIZED PM ADDITIONS
# ============================================================

V1_1_MEASUREMENT_PATCH_SPECS = [

    {
        "node_id":
            "measurement::rrc_resume_succ_by_fallback",

        "name":
            "RRC Resume Successes by Fallback",

        "measurement_name":
            "RRC.ResumeSuccByFallback.cause",

        "clause":
            "5.1.1.18.3",

        "entity_type":
            EntityType.NR_CELL_CU,
    },

    {
        "node_id":
            "measurement::rrc_resume_fallback_to_setup_att",

        "name":
            "RRC Resume Fallback-to-Setup Attempts",

        "measurement_name":
            "RRC.ResumeFallbackToSetupAtt.cause",

        "clause":
            "5.1.1.18.6",

        "entity_type":
            EntityType.NR_CELL_CU,
    },

    {
        "node_id":
            "measurement::rrc_reestab_succ_without_ue_context",

        "name":
            "RRC Re-establishment Successes Without UE Context",

        "measurement_name":
            "RRC.ReEstabSuccWithoutUeContext",

        "clause":
            "5.1.1.17.3",

        "entity_type":
            EntityType.NR_CELL_CU,
    },

    {
        "node_id":
            "measurement::rrc_reestab_fallback_to_setup_att",

        "name":
            "RRC Re-establishment Fallback-to-Setup Attempts",

        "measurement_name":
            "RRC.ReEstabFallbackToSetupAtt",

        "clause":
            "5.1.1.17.4",

        "entity_type":
            EntityType.NR_CELL_CU,
    },
]


# ============================================================
# 5 V1.1.7 — CREATE THE FOUR STANDARDIZED PM NODES
# ============================================================

for spec in V1_1_MEASUREMENT_PATCH_SPECS:

    node_id = (
        spec["node_id"]
    )

    measurement_name = (
        spec["measurement_name"]
    )


    # These nodes must genuinely be new relative to V1.

    assert (
        node_id
        not in RAN_ACCESSIBILITY_NODES
    ), (
        "Approved V1.1 node already exists in "
        f"frozen V1: {node_id}"
    )


    knowledge_record = (
        STANDARD_PM_KNOWLEDGE[
            measurement_name
        ]
    )


    # Ensure Cell 6 and this patch agree on the exact clause.

    assert (
        knowledge_record.clause
        == spec["clause"]
    ), (
        "Cell 6 PM clause differs from the V1.1 "
        f"patch contract for {measurement_name}. "
        f"Observed: {knowledge_record.clause}"
    )


    RAN_ACCESSIBILITY_NODES_V1_1[
        node_id
    ] = make_ran_node(

        node_id=node_id,

        name=spec["name"],

        layer=(
            RANAccessibilityNodeLayer.MEASUREMENT
        ),

        alignment_type=(
            RANAlignmentType.STANDARD_MEASUREMENT
        ),

        description=(

            "3GPP performance measurement added by the "
            "Cell 5 V1.1 standards-completeness revision: "
            f"{measurement_name}. "

            f"{knowledge_record.description}"
        ),

        domains=[
            MeasurementDomain.RAN
        ],

        standards_references=[
            TS_28_552_REL19
        ],

        metadata={

            "measurement_name":
                measurement_name,

            "entity_type":
                spec["entity_type"].value,

            "ts_28_552_clause":
                spec["clause"],

            "introduced_in_revision":
                "CELL_5_V1.1",

            "discovered_by":
                "CELL_6_V1_GRAPH_COVERAGE_VALIDATION",
        },
    )


# ============================================================
# 5 V1.1.8 — V1.1 EDGE HELPER
# ============================================================
#
# We intentionally do NOT use add_ran_edge().
#
# Cell 5's original helper appends to:
#
#   RAN_ACCESSIBILITY_EDGES
#
# which would mutate frozen V1 runtime state.
#
# This V1.1 helper appends only to the V1.1 copy.
#
# ============================================================

def add_ran_edge_v1_1(
    source,
    target,
    relation,
    rationale,
):

    edge_key = (

        source,

        target,

        relation.value,
    )


    existing_edge_keys = {

        (
            edge.source_node_id,

            edge.target_node_id,

            edge.relation.value,
        )

        for edge
        in RAN_ACCESSIBILITY_EDGES_V1_1
    }


    if edge_key in existing_edge_keys:

        raise AssertionError(

            "Duplicate V1.1 dependency edge: "
            f"{edge_key}"
        )


    RAN_ACCESSIBILITY_EDGES_V1_1.append(

        RANAccessibilityEdge(

            source_node_id=source,

            target_node_id=target,

            relation=relation,

            rationale=rationale,
        )
    )


# ============================================================
# 5 V1.1.9 — ADD EXACT FORMULA DEPENDENCY EDGES
# ============================================================
#
# Existing Cell 5 graph convention:
#
#   COMPONENT
#       │
#       │ MEASURED_BY
#       ▼
#   MEASUREMENT
#
#
# Three dependency concepts contribute only to the exact
# RRC-setup/fallback treatment:
#
#   RRC.ResumeSuccByFallback.cause
#   RRC.ReEstabSuccWithoutUeContext
#   RRC.ReEstabFallbackToSetupAtt
#
#
# RRC.ResumeFallbackToSetupAtt.cause participates in both:
#
#   - RRC setup / fallback treatment
#   - RRC Resume denominator
#
# Therefore:
#
#   4 new nodes
#   5 new edges
#
# ============================================================


# ------------------------------------------------------------
# RRC SETUP / FALLBACK CONTRIBUTIONS
# ------------------------------------------------------------

for measurement_id in [

    "measurement::rrc_resume_succ_by_fallback",

    "measurement::rrc_resume_fallback_to_setup_att",

    "measurement::rrc_reestab_succ_without_ue_context",

    "measurement::rrc_reestab_fallback_to_setup_att",

]:

    add_ran_edge_v1_1(

        "component::rrc_setup_sr",

        measurement_id,

        RANDependencyRelation.MEASURED_BY,

        (
            "Exact TS 28.554 Total DRB Accessibility "
            "dependency identified by Cell 6. "
            "This fallback / re-establishment measurement "
            "contributes to the RRC setup component used "
            "by the composite KPI."
        ),
    )


# ------------------------------------------------------------
# RRC RESUME DENOMINATOR CONTRIBUTION
# ------------------------------------------------------------

add_ran_edge_v1_1(

    "component::rrc_resume_sr",

    "measurement::rrc_resume_fallback_to_setup_att",

    RANDependencyRelation.MEASURED_BY,

    (
        "RRC.ResumeFallbackToSetupAtt.cause is required "
        "by the exact RRC Resume success-rate denominator "
        "used in Total DRB Accessibility."
    ),
)


# ============================================================
# 5 V1.1.10 — GRAPH INTEGRITY VALIDATION
# ============================================================

v1_1_node_ids = set(
    RAN_ACCESSIBILITY_NODES_V1_1.keys()
)


for edge in RAN_ACCESSIBILITY_EDGES_V1_1:

    assert (
        edge.source_node_id
        in v1_1_node_ids
    ), (
        "Unknown V1.1 source node: "
        f"{edge.source_node_id}"
    )


    assert (
        edge.target_node_id
        in v1_1_node_ids
    ), (
        "Unknown V1.1 target node: "
        f"{edge.target_node_id}"
    )


# ------------------------------------------------------------
# Confirm original V1 remains unchanged in memory
# ------------------------------------------------------------

assert (
    len(
        RAN_ACCESSIBILITY_NODES
    )
    == V1_NODE_COUNT
)


assert (
    len(
        RAN_ACCESSIBILITY_EDGES
    )
    == V1_EDGE_COUNT
)


# ------------------------------------------------------------
# Confirm exact expected V1.1 structural delta
# ------------------------------------------------------------

assert (
    len(
        RAN_ACCESSIBILITY_NODES_V1_1
    )
    == V1_NODE_COUNT + 4
)


assert (
    len(
        RAN_ACCESSIBILITY_EDGES_V1_1
    )
    == V1_EDGE_COUNT + 5
)


# Formal KPI scope remains unchanged.

assert (

    set(
        RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_1.keys()
    )

    == {

        "partial_drb_accessibility",

        "total_drb_accessibility",
    }
)


# ============================================================
# 5 V1.1.11 — EXACT FORMULA COVERAGE REVALIDATION
# ============================================================
#
# Reuse Cell 6's canonical naming and exact formula knowledge.
#
# We now ask:
#
#   "After the V1.1 patch, does every exact KPI formula
#    dependency have a measurement node?"
#
# ============================================================

v1_1_measurement_names = {}


for node_id, node in (
    RAN_ACCESSIBILITY_NODES_V1_1.items()
):

    if (
        node.layer
        == RANAccessibilityNodeLayer.MEASUREMENT
    ):

        raw_name = node.metadata.get(
            "measurement_name"
        )


        if raw_name is not None:

            canonical_name = (
                canonical_measurement_name(
                    raw_name
                )
            )


            v1_1_measurement_names[
                canonical_name
            ] = {

                "node_id":
                    node_id,

                "raw_name":
                    raw_name,

                "canonical_name":
                    canonical_name,
            }


required_formula_measurements_v1_1 = set()


for kpi_record in (
    KPI_FORMULA_KNOWLEDGE.values()
):

    required_formula_measurements_v1_1.update(

        canonical_measurement_name(
            name
        )

        for name
        in kpi_record.required_measurements
    )


missing_v1_1_formula_measurements = sorted(

    required_formula_measurements_v1_1

    - set(
        v1_1_measurement_names.keys()
    )
)


assert (
    missing_v1_1_formula_measurements
    == []
), (
    "Cell 5 V1.1 still has exact formula-dependency "
    f"gaps: {missing_v1_1_formula_measurements}"
)


# ------------------------------------------------------------
# Confirm all four approved gap measurements are represented
# ------------------------------------------------------------

for measurement_name in EXPECTED_CELL6_GAPS:

    canonical_name = (
        canonical_measurement_name(
            measurement_name
        )
    )


    assert (
        canonical_name
        in v1_1_measurement_names
    )


# ------------------------------------------------------------
# Confirm Total DRB KPI metadata also contains all four
# ------------------------------------------------------------

assert (

    EXPECTED_CELL6_GAPS

    <= set(

        RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_1[
            "total_drb_accessibility"
        ].source_measurements
    )
)


# ============================================================
# 5 V1.1.12 — BUILD VERSIONED V1.1 ARTIFACT
# ============================================================

ran_accessibility_domain_v1_1 = deepcopy(
    ran_accessibility_domain
)


# Remove inherited V1 semantic hash before recomputing.

ran_accessibility_domain_v1_1.pop(
    "semantic_sha256",
    None,
)


ran_accessibility_domain_v1_1[
    "domain_id"
] = (
    "module5_5g_ran_accessibility_"
    "reference_domain_v1_1"
)


ran_accessibility_domain_v1_1[
    "created_utc"
] = (
    datetime.now(
        timezone.utc
    ).isoformat()
)


# ------------------------------------------------------------
# Updated V1.1 formal KPI definitions
# ------------------------------------------------------------

ran_accessibility_domain_v1_1[
    "formal_reference_kpis"
] = {

    kpi_id:
        kpi.model_dump(
            mode="json"
        )

    for kpi_id, kpi
    in RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_1.items()
}


# ------------------------------------------------------------
# Updated V1.1 graph nodes
# ------------------------------------------------------------

ran_accessibility_domain_v1_1[
    "nodes"
] = {

    node_id:
        node.model_dump(
            mode="json"
        )

    for node_id, node
    in RAN_ACCESSIBILITY_NODES_V1_1.items()
}


# ------------------------------------------------------------
# Updated V1.1 graph edges
# ------------------------------------------------------------

ran_accessibility_domain_v1_1[
    "edges"
] = [

    edge.model_dump(
        mode="json"
    )

    for edge
    in RAN_ACCESSIBILITY_EDGES_V1_1
]


# ------------------------------------------------------------
# Explicit revision provenance
# ------------------------------------------------------------

ran_accessibility_domain_v1_1[
    "revision"
] = {

    "revision_id":
        "CELL_5_V1.1",

    "parent_domain_id":
        ran_accessibility_domain[
            "domain_id"
        ],

    "parent_semantic_sha256":
        ran_accessibility_semantic_sha,

    "validation_source":
        "CELL_6_V1",

    "validation_semantic_sha256":
        cell6_semantic_sha,

    "reason":
        (
            "Cell 6 exact-formula validation identified "
            "four standardized TS 28.552 fallback / "
            "re-establishment dependencies required by "
            "Total DRB Accessibility but absent from "
            "the frozen Cell 5 V1 graph."
        ),

    "added_measurements":
        sorted(
            EXPECTED_CELL6_GAPS
        ),

    "added_node_count":
        4,

    "added_edge_count":
        5,

    "parent_artifact_preserved":
        True,
}


# ------------------------------------------------------------
# Append V1.1 design notes
# ------------------------------------------------------------

ran_accessibility_domain_v1_1[
    "design_notes"
] = list(

    ran_accessibility_domain_v1_1.get(
        "design_notes",
        []
    )

) + [

    (
        "Cell 5 V1.1 is a controlled "
        "standards-completeness revision. "
        "The original Cell 5 V1 artifact remains "
        "unchanged and traceable."
    ),

    (
        "The revision adds only the four standardized "
        "fallback / re-establishment PM dependencies "
        "identified by Cell 6."
    ),

    (
        "RRC.ResumeFallbackToSetupAtt.cause is bound "
        "to both the RRC setup and RRC Resume "
        "components because the exact Total DRB "
        "formulation uses it in both dependency contexts."
    ),

    (
        "Cell 6 exact-formula coverage is complete "
        "after this revision."
    ),
]


# ============================================================
# 5 V1.1.13 — DETERMINISTIC SEMANTIC HASH
# ============================================================

def canonical_json_sha256_v1_1(
    data
):

    canonical = json.dumps(

        data,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

    ).encode(
        "utf-8"
    )


    return hashlib.sha256(
        canonical
    ).hexdigest()


ran_accessibility_v1_1_semantic_content = {

    key: value

    for key, value
    in ran_accessibility_domain_v1_1.items()

    if key != "created_utc"
}


ran_accessibility_v1_1_semantic_sha = (

    canonical_json_sha256_v1_1(

        ran_accessibility_v1_1_semantic_content
    )
)


ran_accessibility_domain_v1_1[
    "semantic_sha256"
] = ran_accessibility_v1_1_semantic_sha


# ============================================================
# 5 V1.1.14 — SAVE NEW ARTIFACT
# ============================================================

ran_accessibility_v1_1_path = (

    PATHS["kpi"]

    / (
        "module5_5g_ran_accessibility_"
        "reference_domain_v1_1.json"
    )
)


# Critical safety check:
# V1.1 must never point to the frozen V1 path.

assert (
    ran_accessibility_v1_1_path
    != ran_accessibility_path
)


with open(

    ran_accessibility_v1_1_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        ran_accessibility_domain_v1_1,

        f,

        indent=2,

        ensure_ascii=False,
    )


ran_accessibility_v1_1_artifact_sha = (

    sha256_file(
        ran_accessibility_v1_1_path
    )
)


# ============================================================
# 5 V1.1.15 — POST-SAVE V1 IMMUTABILITY CHECK
# ============================================================

assert (

    sha256_file(
        ran_accessibility_path
    )

    == ran_accessibility_artifact_sha

), (
    "Frozen Cell 5 V1 artifact changed during "
    "V1.1 creation."
)


assert (
    ran_accessibility_semantic_sha
    == EXPECTED_CELL5_V1_SEMANTIC_SHA
)


# ============================================================
# 5 V1.1.16 — SUMMARY STATISTICS
# ============================================================

layer_counts_v1_1 = {}


for layer in RANAccessibilityNodeLayer:

    layer_counts_v1_1[
        layer.value
    ] = sum(

        1

        for node
        in RAN_ACCESSIBILITY_NODES_V1_1.values()

        if node.layer == layer
    )


domain_counts_v1_1 = {}


for domain in MeasurementDomain:

    domain_counts_v1_1[
        domain.value
    ] = sum(

        1

        for node
        in RAN_ACCESSIBILITY_NODES_V1_1.values()

        if domain in node.domains
    )


# ============================================================
# 5 V1.1.17 — FINAL SUMMARY
# ============================================================

print("=" * 104)


print(
    "MODULE 5 — CELL 06 "
    "CONTROLLED STANDARDS-COMPLETENESS REVISION"
)


print("=" * 104)


# ------------------------------------------------------------
# Revision basis
# ------------------------------------------------------------

print(
    "\nRevision basis"
)

print("-" * 104)


print(
    "Parent graph                     : "
    "Cell 5 V1"
)


print(
    "Validation source                : "
    "Cell 6 V1"
)


print(
    "Parent V1 preserved              : "
    "YES"
)


print(
    "Parent semantic SHA              : "
    f"{ran_accessibility_semantic_sha}"
)


# ------------------------------------------------------------
# Patch contents
# ------------------------------------------------------------

print(
    "\nApproved structural patch"
)

print("-" * 104)


for measurement_name in sorted(
    EXPECTED_CELL6_GAPS
):

    print(
        measurement_name
    )


# ------------------------------------------------------------
# Graph delta
# ------------------------------------------------------------

print(
    "\nGraph delta"
)

print("-" * 104)


print(
    f"V1 nodes                         : "
    f"{V1_NODE_COUNT}"
)


print(
    f"V1.1 nodes                       : "
    f"{len(RAN_ACCESSIBILITY_NODES_V1_1)}"
)


print(
    f"Nodes added                      : "
    f"{len(RAN_ACCESSIBILITY_NODES_V1_1) - V1_NODE_COUNT}"
)


print(
    f"V1 edges                         : "
    f"{V1_EDGE_COUNT}"
)


print(
    f"V1.1 edges                       : "
    f"{len(RAN_ACCESSIBILITY_EDGES_V1_1)}"
)


print(
    f"Edges added                      : "
    f"{len(RAN_ACCESSIBILITY_EDGES_V1_1) - V1_EDGE_COUNT}"
)


# ------------------------------------------------------------
# Exact formula coverage
# ------------------------------------------------------------

print(
    "\nExact formula coverage after revision"
)

print("-" * 104)


print(
    f"Required exact-formula PMs       : "
    f"{len(required_formula_measurements_v1_1)}"
)


print(
    f"Missing exact dependencies       : "
    f"{len(missing_v1_1_formula_measurements)}"
)


print(
    "Exact Total DRB dependency gap   : "
    "CLOSED"
)


# ------------------------------------------------------------
# Node layers
# ------------------------------------------------------------

print(
    "\nNode layers"
)

print("-" * 104)


for layer, count in (
    layer_counts_v1_1.items()
):

    print(
        f"{layer:<32} : "
        f"{count}"
    )


# ------------------------------------------------------------
# Domain coverage
# ------------------------------------------------------------

print(
    "\nTechnical-domain coverage"
)

print("-" * 104)


for domain, count in (
    domain_counts_v1_1.items()
):

    if count > 0:

        print(
            f"{domain:<32} : "
            f"{count}"
        )


# ------------------------------------------------------------
# Artifacts
# ------------------------------------------------------------

print(
    "\nArtifact"
)

print("-" * 104)


print(
    f"V1 path                          : "
    f"{ran_accessibility_path}"
)


print(
    f"V1 artifact SHA                  : "
    f"{ran_accessibility_artifact_sha}"
)


print(
    f"V1.1 path                        : "
    f"{ran_accessibility_v1_1_path}"
)


print(
    f"V1.1 artifact SHA                : "
    f"{ran_accessibility_v1_1_artifact_sha}"
)


print(
    f"V1.1 semantic SHA                : "
    f"{ran_accessibility_v1_1_semantic_sha}"
)


print(
    "\n" + "=" * 104
)


print(
    "CELL 06 COMPLETE — "
    "EXACT TOTAL DRB ACCESSIBILITY DEPENDENCY GAP CLOSED"
)


print(
    "=" * 104
)
````

###### Cell 06 — Observation

Cell 06 applies a controlled standards-completeness revision to the Accessibility graph, adding the four missing Resume/re-establishment PM dependencies identified in Cell 05.

```text
Graph nodes                 : 33 → 37
Graph edges                 : 44 → 49
Missing formula dependencies: 4 → 0
```

The formal KPI set and broader architecture remain unchanged.

> **The exact Total DRB Accessibility dependency gap is now closed while preserving the original graph as a frozen parent revision.**


<a id="section-4-observation-and-topology-foundation"></a>
#### Section 4 — Observation and Topology Foundation

The knowledge model is now translated into a runtime observation contract and a deterministic synthetic RAN environment. Telemetry values and fault behaviour remain deliberately absent at this stage.


<a id="cell-07-observation-contract-v1"></a>
##### Cell 07 — 5G RAN Telemetry Observation Contract V1

Define the canonical observation envelope, entity references, metric bindings, data-quality states and the three telemetry environments: `SYNTHETIC_V1`, `EXECUTABLE_V2` and `EXTERNAL_LIVE`.


````python
# ============================================================
# MODULE 5 — CELL 07
# 5G RAN TELEMETRY OBSERVATION CONTRACT
# ============================================================
#
# Purpose:
#
#   1. Define the canonical runtime representation of
#      network observations used by Module 5.
#
#   2. Reuse the broad Cell 03 entity/domain foundation.
#
#   3. Bind Cell 06 V1.1 measurement nodes to the richer
#      Cell 05 PM knowledge catalogue.
#
#   4. Preserve standardized PM evidence separately from
#      operational telemetry and future ML inference.
#
#   5. Establish one logical observation contract that can
#      later serve both:
#
#         V1 synthetic telemetry
#         V2 executable / live-like telemetry
#
#   6. Persist a deterministic machine-readable telemetry
#      contract for the future Telemetry MCP.
#
#
# IMPORTANT:
#
#   - This cell DEFINES telemetry structure.
#
#   - This cell DOES NOT generate telemetry values.
#
#   - This cell DOES NOT inject incidents or faults.
#
#   - This cell DOES NOT expose hidden ground truth.
#
#   - This cell DOES NOT perform anomaly detection.
#
#   - This cell DOES NOT create ML predictions.
#
#   - Observation truth and ML inference remain separate.
#
# ============================================================


from enum import Enum
from typing import Any, Dict, List, Optional, Union
from datetime import datetime, timezone

from pydantic import (
    BaseModel,
    Field,
    ConfigDict,
    model_validator,
)

import json
import hashlib


# ============================================================
# 7.1 — DEPENDENCY VALIDATION
# ============================================================

REQUIRED_CELL7_OBJECTS = [

    # Cell 4
    "MeasurementDomain",
    "EntityType",
    "PM_INTERVAL_MINUTES",
    "PATHS",
    "sha256_file",

    # Cell 5 V1.1
    "RAN_ACCESSIBILITY_NODES_V1_1",
    "RANAccessibilityNodeLayer",
    "ran_accessibility_v1_1_semantic_sha",

    # Cell 6
    "STANDARD_PM_KNOWLEDGE",
    "EvidenceClassification",
    "RANInvestigationStage",
    "canonical_measurement_name",
    "KPI_FORMULA_KNOWLEDGE",
    "cell6_semantic_sha",
]


missing_cell7_objects = [

    name

    for name
    in REQUIRED_CELL7_OBJECTS

    if name not in globals()
]


if missing_cell7_objects:

    raise RuntimeError(

        "Cell 07 requires Cells 03, 05 and 06 first. "
        f"Missing objects: {missing_cell7_objects}"
    )


# ============================================================
# 7.2 — TELEMETRY ENVIRONMENT
# ============================================================
#
# The logical telemetry interface must not depend on whether
# the observations came from synthetic V1 or future V2.
#
# ============================================================

class TelemetryEnvironment(str, Enum):

    SYNTHETIC_V1 = "SYNTHETIC_V1"

    EXECUTABLE_V2 = "EXECUTABLE_V2"

    EXTERNAL_LIVE = "EXTERNAL_LIVE"


# ============================================================
# 7.3 — OBSERVATION TYPE
# ============================================================
#
# This defines WHAT KIND of network truth is being observed.
#
# It deliberately does not contain ML inference types.
#
# ============================================================

class TelemetryObservationType(str, Enum):

    STANDARD_PM = "STANDARD_PM"

    DERIVED_KPI = "DERIVED_KPI"

    RADIO_MEASUREMENT = "RADIO_MEASUREMENT"

    RESOURCE_MEASUREMENT = "RESOURCE_MEASUREMENT"

    ALARM_EVENT = "ALARM_EVENT"

    CONFIGURATION_STATE = "CONFIGURATION_STATE"

    TRANSPORT_MEASUREMENT = "TRANSPORT_MEASUREMENT"

    NETWORK_EVENT = "NETWORK_EVENT"


# ============================================================
# 7.4 — OBSERVATION QUALITY
# ============================================================
#
# Data quality is part of network truth.
#
# Missing / late / suspect data must not silently become
# valid evidence.
#
# ============================================================

class TelemetryDataQuality(str, Enum):

    VALID = "VALID"

    MISSING = "MISSING"

    LATE = "LATE"

    SUSPECT = "SUSPECT"

    UNAVAILABLE = "UNAVAILABLE"


# ============================================================
# 7.5 — EVIDENCE ORIGIN
# ============================================================
#
# Separates:
#
#   standardized PM truth
#   standardized supporting PM
#   operational telemetry
#
# ML evidence is intentionally absent.
#
# ============================================================

class TelemetryEvidenceOrigin(str, Enum):

    STANDARD_KPI_DEPENDENCY = (
        "STANDARD_KPI_DEPENDENCY"
    )

    STANDARD_SUPPORTING_PM = (
        "STANDARD_SUPPORTING_PM"
    )

    OPERATIONAL_TELEMETRY = (
        "OPERATIONAL_TELEMETRY"
    )


# ============================================================
# 7.6 — TELEMETRY VALUE TYPE
# ============================================================

TelemetryValue = Union[
    int,
    float,
    str,
    bool,
]


# ============================================================
# 7.7 — ENTITY REFERENCE MODEL
# ============================================================
#
# An observation must identify the network object that
# produced the evidence.
#
# Example:
#
#   entity_id   = "NRCellCU_001"
#   entity_type = NR_CELL_CU
#
# ============================================================

class RANTelemetryEntityRef(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    entity_id: str

    entity_type: EntityType

    parent_entity_id: Optional[str] = None

    site_id: Optional[str] = None

    metadata: Dict[str, Any] = Field(
        default_factory=dict
    )


# ============================================================
# 7.8 — METRIC BINDING MODEL
# ============================================================
#
# This is the deterministic bridge:
#
# Cell 5 graph
#      +
# Cell 6 knowledge
#      ↓
# Telemetry metric contract
#
# It describes WHAT a metric means before any values exist.
#
# ============================================================

class RANTelemetryMetricBinding(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    metric_name: str

    aliases: List[str] = Field(
        default_factory=list
    )

    graph_node_id: Optional[str] = None

    entity_type: EntityType

    domain: MeasurementDomain

    observation_type: TelemetryObservationType

    evidence_origin: TelemetryEvidenceOrigin

    investigation_stage: RANInvestigationStage

    pm_interval_minutes: int

    specification: Optional[str] = None

    clause: Optional[str] = None

    measurement_class: Optional[str] = None

    description: str

    engineering_interpretation: str


# ============================================================
# 7.9 — OBSERVATION MODEL
# ============================================================
#
# This is the primary network-truth envelope.
#
# Future synthetic generation and Telemetry MCP responses
# should produce records in this form.
#
# ============================================================

class RANTelemetryObservation(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    observation_id: str

    environment: TelemetryEnvironment

    observation_type: TelemetryObservationType

    evidence_origin: TelemetryEvidenceOrigin

    metric_name: str

    entity: RANTelemetryEntityRef

    interval_start: datetime

    interval_end: datetime

    pm_interval_minutes: int

    value: Optional[TelemetryValue] = None

    unit: Optional[str] = None

    dimensions: Dict[
        str,
        TelemetryValue
    ] = Field(
        default_factory=dict
    )

    quality: TelemetryDataQuality = (
        TelemetryDataQuality.VALID
    )

    graph_node_id: Optional[str] = None

    investigation_stage: Optional[
        RANInvestigationStage
    ] = None

    source_system: Optional[str] = None

    metadata: Dict[str, Any] = Field(
        default_factory=dict
    )


    @model_validator(
        mode="after"
    )
    def validate_interval(
        self
    ):

        if (
            self.interval_end
            <= self.interval_start
        ):

            raise ValueError(
                "interval_end must be later "
                "than interval_start."
            )


        duration_minutes = (

            self.interval_end
            - self.interval_start

        ).total_seconds() / 60


        if (
            self.observation_type
            == TelemetryObservationType.STANDARD_PM
        ):

            if (
                duration_minutes
                != self.pm_interval_minutes
            ):

                raise ValueError(

                    "STANDARD_PM observation interval "
                    "must match pm_interval_minutes."
                )


        if (
            self.quality
            in {
                TelemetryDataQuality.MISSING,
                TelemetryDataQuality.UNAVAILABLE,
            }
            and self.value is not None
        ):

            raise ValueError(

                "Missing or unavailable telemetry "
                "must not contain a value."
            )


        return self


# ============================================================
# 7.10 — OBSERVATION BATCH
# ============================================================
#
# Telemetry MCP will normally return collections of evidence,
# not isolated scalar values.
#
# ============================================================

class RANTelemetryBatch(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    batch_id: str

    environment: TelemetryEnvironment

    generated_at: datetime

    observations: List[
        RANTelemetryObservation
    ]

    query_context: Dict[
        str,
        Any
    ] = Field(
        default_factory=dict
    )


# ============================================================
# 7.11 — ENTITY-SCOPE NORMALIZATION
# ============================================================
#
# Cell 6 retains standards-style entity naming:
#
#   NRCellCU
#   NRCellDU
#
# Cell 4 provides canonical EntityType values.
#
# ============================================================

CELL6_ENTITY_SCOPE_TO_ENTITY_TYPE = {

    "NRCellCU":
        EntityType.NR_CELL_CU,

    "NRCellDU":
        EntityType.NR_CELL_DU,
}


# ============================================================
# 7.12 — CELL 5 V1.1 STANDARD MEASUREMENT LOOKUP
# ============================================================

cell5_v1_1_measurement_lookup = {}


for node_id, node in (
    RAN_ACCESSIBILITY_NODES_V1_1.items()
):

    if (
        node.layer
        == RANAccessibilityNodeLayer.MEASUREMENT
    ):

        raw_name = (
            node.metadata.get(
                "measurement_name"
            )
        )


        if raw_name is not None:

            canonical_name = (
                canonical_measurement_name(
                    raw_name
                )
            )


            cell5_v1_1_measurement_lookup[
                canonical_name
            ] = {

                "node_id":
                    node_id,

                "raw_name":
                    raw_name,

                "canonical_name":
                    canonical_name,
            }


# ============================================================
# 7.13 — BUILD STANDARDIZED TELEMETRY BINDING REGISTRY
# ============================================================
#
# Scope for this first telemetry contract:
#
#   A. All exact Accessibility KPI-dependency PMs
#
#   B. Standardized RACH supporting PMs
#
#
# Exact radio/load/alarm/config/transport metric binding
# remains intentionally deferred.
#
# ============================================================

RAN_TELEMETRY_METRIC_BINDINGS = {}


for (
    measurement_name,
    knowledge_record
) in STANDARD_PM_KNOWLEDGE.items():

    canonical_name = (
        canonical_measurement_name(
            measurement_name
        )
    )


    if (
        knowledge_record.evidence_classification
        not in {

            EvidenceClassification
            .STANDARD_KPI_DEPENDENCY,

            EvidenceClassification
            .STANDARD_SUPPORTING_PM,
        }
    ):

        continue


    entity_scope = (
        knowledge_record.entity_scope
    )


    if (
        entity_scope
        not in CELL6_ENTITY_SCOPE_TO_ENTITY_TYPE
    ):

        raise RuntimeError(

            "Cell 7 has no canonical EntityType "
            "mapping for Cell 6 entity scope: "
            f"{entity_scope}"
        )


    entity_type = (
        CELL6_ENTITY_SCOPE_TO_ENTITY_TYPE[
            entity_scope
        ]
    )


    graph_info = (
        cell5_v1_1_measurement_lookup.get(
            canonical_name
        )
    )


    graph_node_id = (

        graph_info["node_id"]

        if graph_info is not None

        else None
    )


    if (
        knowledge_record.evidence_classification
        == (
            EvidenceClassification
            .STANDARD_KPI_DEPENDENCY
        )
    ):

        evidence_origin = (
            TelemetryEvidenceOrigin
            .STANDARD_KPI_DEPENDENCY
        )


    else:

        evidence_origin = (
            TelemetryEvidenceOrigin
            .STANDARD_SUPPORTING_PM
        )


    RAN_TELEMETRY_METRIC_BINDINGS[
        canonical_name
    ] = RANTelemetryMetricBinding(

        metric_name=canonical_name,

        aliases=list(
            knowledge_record.aliases
        ),

        graph_node_id=graph_node_id,

        entity_type=entity_type,

        domain=MeasurementDomain.RAN,

        observation_type=(
            TelemetryObservationType.STANDARD_PM
        ),

        evidence_origin=evidence_origin,

        investigation_stage=(
            knowledge_record.investigation_stage
        ),

        pm_interval_minutes=(
            PM_INTERVAL_MINUTES
        ),

        specification=(
            knowledge_record.specification
        ),

        clause=(
            knowledge_record.clause
        ),

        measurement_class=(
            knowledge_record
            .measurement_class
            .value
        ),

        description=(
            knowledge_record.description
        ),

        engineering_interpretation=(
            knowledge_record
            .engineering_interpretation
        ),
    )


# ============================================================
# 7.14 — EXACT KPI FORMULA COVERAGE VALIDATION
# ============================================================

required_formula_measurements_cell7 = set()


for kpi_record in (
    KPI_FORMULA_KNOWLEDGE.values()
):

    required_formula_measurements_cell7.update(

        canonical_measurement_name(
            measurement_name
        )

        for measurement_name
        in kpi_record.required_measurements
    )


missing_binding_measurements = sorted(

    required_formula_measurements_cell7

    - set(
        RAN_TELEMETRY_METRIC_BINDINGS.keys()
    )
)


assert (
    missing_binding_measurements
    == []
), (
    "Telemetry contract does not bind every exact "
    "Accessibility KPI dependency: "
    f"{missing_binding_measurements}"
)


# ============================================================
# 7.15 — EXACT GRAPH BINDING VALIDATION
# ============================================================
#
# Every standardized KPI dependency must resolve to a
# Cell 5 V1.1 measurement node.
#
# Supporting RACH PMs are allowed to remain graph-unbound
# because Cell 5 represents RACH at the diagnostic level.
#
# ============================================================

unbound_kpi_dependency_metrics = sorted([

    metric_name

    for (
        metric_name,
        binding
    ) in (
        RAN_TELEMETRY_METRIC_BINDINGS.items()
    )

    if (

        binding.evidence_origin
        == (
            TelemetryEvidenceOrigin
            .STANDARD_KPI_DEPENDENCY
        )

        and

        binding.graph_node_id is None
    )
])


assert (
    unbound_kpi_dependency_metrics
    == []
), (
    "Exact KPI dependency lacks Cell 5 V1.1 "
    "graph binding: "
    f"{unbound_kpi_dependency_metrics}"
)


# ============================================================
# 7.16 — RACH CONTRACT VALIDATION
# ============================================================

EXPECTED_RACH_SUPPORTING_PM = {

    "RACH.PreambleDedCell",
    "RACH.PreambleACell",
    "RACH.PreambleBCell",
}


for rach_metric in (
    EXPECTED_RACH_SUPPORTING_PM
):

    assert (
        rach_metric
        in RAN_TELEMETRY_METRIC_BINDINGS
    )


    binding = (
        RAN_TELEMETRY_METRIC_BINDINGS[
            rach_metric
        ]
    )


    assert (

        binding.evidence_origin

        == (
            TelemetryEvidenceOrigin
            .STANDARD_SUPPORTING_PM
        )
    )


    assert (
        binding.entity_type
        == EntityType.NR_CELL_DU
    )


# ============================================================
# 7.17 — COUNT EVIDENCE CLASSES
# ============================================================

kpi_dependency_binding_count = sum(

    1

    for binding
    in RAN_TELEMETRY_METRIC_BINDINGS.values()

    if (

        binding.evidence_origin
        == (
            TelemetryEvidenceOrigin
            .STANDARD_KPI_DEPENDENCY
        )
    )
)


supporting_pm_binding_count = sum(

    1

    for binding
    in RAN_TELEMETRY_METRIC_BINDINGS.values()

    if (

        binding.evidence_origin
        == (
            TelemetryEvidenceOrigin
            .STANDARD_SUPPORTING_PM
        )
    )
)


# ============================================================
# 7.18 — VALIDATE 15-MINUTE PM CONTRACT
# ============================================================

assert (
    PM_INTERVAL_MINUTES
    == 15
), (
    "Module 5 V1 telemetry contract expects "
    "the frozen 15-minute PM interval."
)


for binding in (
    RAN_TELEMETRY_METRIC_BINDINGS.values()
):

    assert (
        binding.pm_interval_minutes
        == PM_INTERVAL_MINUTES
    )


# ============================================================
# 7.19 — PROHIBITED HIDDEN-TRUTH FIELDS
# ============================================================
#
# Network-observation records must never contain benchmark
# ground truth.
#
# ============================================================

PROHIBITED_OBSERVATION_FIELDS = {

    "root_cause",

    "ground_truth",

    "fault_label",

    "expected_diagnosis",

    "correct_action",

    "benchmark_answer",
}


observation_fields = set(
    RANTelemetryObservation.model_fields.keys()
)


assert (
    not (
        observation_fields
        & PROHIBITED_OBSERVATION_FIELDS
    )
)


# ============================================================
# 7.20 — BUILD TELEMETRY CONTRACT ARTIFACT
# ============================================================

ran_telemetry_observation_contract = {

    "artifact_id":
        (
            "module5_5g_ran_telemetry_"
            "observation_contract_v1"
        ),

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "module_scope":
        "MODULE_5_V1_5G_RAN",

    "reference_domain":
        "5G_RAN_ACCESSIBILITY",

    "contract_role":
        "NETWORK_OBSERVATION_TRUTH",

    "pm_interval_minutes":
        PM_INTERVAL_MINUTES,

    "supported_environments": [

        item.value

        for item
        in TelemetryEnvironment
    ],

    "observation_types": [

        item.value

        for item
        in TelemetryObservationType
    ],

    "data_quality_states": [

        item.value

        for item
        in TelemetryDataQuality
    ],

    "evidence_origins": [

        item.value

        for item
        in TelemetryEvidenceOrigin
    ],

    "metric_bindings": {

        metric_name:
            binding.model_dump(
                mode="json"
            )

        for metric_name, binding
        in RAN_TELEMETRY_METRIC_BINDINGS.items()
    },

    "schema": {

        "entity_reference":
            RANTelemetryEntityRef
            .model_json_schema(),

        "metric_binding":
            RANTelemetryMetricBinding
            .model_json_schema(),

        "observation":
            RANTelemetryObservation
            .model_json_schema(),

        "batch":
            RANTelemetryBatch
            .model_json_schema(),
    },

    "provenance": {

        "cell5_v1_1_semantic_sha256":
            (
                ran_accessibility_v1_1_semantic_sha
            ),

        "cell6_semantic_sha256":
            cell6_semantic_sha,
    },

    "design_notes": [

        (
            "The observation contract is independent "
            "of telemetry source implementation so the "
            "same logical interface can support V1 "
            "synthetic and future V2 telemetry."
        ),

        (
            "The contract represents observed network "
            "evidence only. ML inference remains a "
            "separate evidence type and is not stored "
            "as network truth."
        ),

        (
            "All exact Total / Partial DRB "
            "Accessibility formula dependencies are "
            "bound to Cell 5 V1.1 measurement nodes."
        ),

        (
            "Standardized RACH PM measurements are "
            "included as supporting evidence but remain "
            "distinct from a standardized TS 28.554 "
            "Accessibility KPI."
        ),

        (
            "Operational radio, resource, alarm, "
            "configuration and transport evidence "
            "share the observation envelope but exact "
            "metric bindings are intentionally deferred "
            "until the synthetic telemetry design."
        ),

        (
            "No hidden benchmark root-cause, fault or "
            "expected-answer fields are permitted in "
            "telemetry observations."
        ),
    ],
}


# ============================================================
# 7.21 — DETERMINISTIC SEMANTIC HASH
# ============================================================

def canonical_json_sha256_cell7(
    data
):

    canonical = json.dumps(

        data,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

    ).encode(
        "utf-8"
    )


    return hashlib.sha256(
        canonical
    ).hexdigest()


cell7_semantic_content = {

    key: value

    for key, value
    in ran_telemetry_observation_contract.items()

    if key != "created_utc"
}


cell7_semantic_sha = (

    canonical_json_sha256_cell7(
        cell7_semantic_content
    )
)


ran_telemetry_observation_contract[
    "semantic_sha256"
] = cell7_semantic_sha


# ============================================================
# 7.22 — SAVE ARTIFACT
# ============================================================

cell7_path = (

    PATHS["kpi"]

    / (
        "module5_5g_ran_telemetry_"
        "observation_contract_v1.json"
    )
)


with open(

    cell7_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        ran_telemetry_observation_contract,

        f,

        indent=2,

        ensure_ascii=False,
    )


cell7_artifact_sha = (

    sha256_file(
        cell7_path
    )
)


# ============================================================
# 7.23 — FINAL VALIDATION
# ============================================================

assert (
    len(
        required_formula_measurements_cell7
    )
    == 16
)


assert (
    kpi_dependency_binding_count
    == 18
)


assert (
    supporting_pm_binding_count
    == 3
)


assert (
    len(
        RAN_TELEMETRY_METRIC_BINDINGS
    )
    == 21
)


assert (
    len(
        missing_binding_measurements
    )
    == 0
)


assert (
    len(
        unbound_kpi_dependency_metrics
    )
    == 0
)


# ============================================================
# 7.24 — SUMMARY OUTPUT
# ============================================================

print("=" * 104)

print(
    "MODULE 5 — CELL 07 "
    "5G RAN TELEMETRY OBSERVATION CONTRACT"
)

print("=" * 104)


print(
    "\nContract scope"
)

print("-" * 104)

print(
    "Reference domain                  : "
    "5G RAN Accessibility"
)

print(
    "Contract role                     : "
    "NETWORK OBSERVATION TRUTH"
)

print(
    f"PM interval                       : "
    f"{PM_INTERVAL_MINUTES} minutes"
)

print(
    "Telemetry values generated        : "
    "NO"
)

print(
    "Hidden fault truth exposed        : "
    "NO"
)

print(
    "ML inference included             : "
    "NO"
)


print(
    "\nStandardized metric bindings"
)

print("-" * 104)

print(
    f"KPI-dependency PM bindings        : "
    f"{kpi_dependency_binding_count}"
)

print(
    f"Supporting PM bindings             : "
    f"{supporting_pm_binding_count}"
)

print(
    f"Total standardized PM bindings    : "
    f"{len(RAN_TELEMETRY_METRIC_BINDINGS)}"
)


print(
    "\nExact formula coverage"
)

print("-" * 104)

print(
    f"Required exact-formula PMs         : "
    f"{len(required_formula_measurements_cell7)}"
)

print(
    f"Missing telemetry bindings         : "
    f"{len(missing_binding_measurements)}"
)

print(
    f"Unbound KPI-dependency graph PMs   : "
    f"{len(unbound_kpi_dependency_metrics)}"
)


print(
    "\nRACH treatment"
)

print("-" * 104)

print(
    "Standardized supporting PMs       : "
    "3"
)

print(
    "Managed-object scope              : "
    "NRCellDU"
)

print(
    "Standardized Accessibility KPI    : "
    "NO"
)


print(
    "\nEnvironment abstraction"
)

print("-" * 104)

for environment in (
    TelemetryEnvironment
):

    print(
        f"{environment.value}"
    )


print(
    "\nObservation model"
)

print("-" * 104)

print(
    "Entity identity                    : REQUIRED"
)

print(
    "Interval timestamps                : REQUIRED"
)

print(
    "Data-quality state                 : REQUIRED"
)

print(
    "Metric dimensions                  : SUPPORTED"
)

print(
    "Graph-node binding                 : SUPPORTED"
)

print(
    "Investigation-stage binding        : SUPPORTED"
)


print(
    "\nProvenance"
)

print("-" * 104)

print(
    "Cell 5 V1.1 semantic SHA           : "
    f"{ran_accessibility_v1_1_semantic_sha}"
)

print(
    "Cell 6 semantic SHA                : "
    f"{cell6_semantic_sha}"
)


print(
    "\nArtifact"
)

print("-" * 104)

print(
    f"Path                               : "
    f"{cell7_path}"
)

print(
    f"Artifact SHA                       : "
    f"{cell7_artifact_sha}"
)

print(
    f"Semantic SHA                       : "
    f"{cell7_semantic_sha}"
)


print(
    "\n" + "=" * 104
)

print(
    "CELL 07 COMPLETE — "
    "RAN TELEMETRY OBSERVATION CONTRACT READY"
)

print(
    "=" * 104
)
````

###### Cell 07 — Observation

Cell 07 establishes the canonical **5G RAN telemetry observation contract** for the Accessibility domain.

All **21 standardized PM bindings** resolve successfully:

```text
KPI-dependency PMs   : 18
Supporting RACH PMs  : 3
Missing bindings     : 0
```

The contract supports synthetic, executable and future live telemetry while preserving explicit entity identity, interval timing, data quality, metric dimensions and investigation-stage linkage.

> **Network observation truth is now formally separated from ML evidence and hidden benchmark ground truth, providing a common telemetry envelope for all later Module 5 environments.**


<a id="cell-08-synthetic-topology-v1"></a>
##### Cell 08 — Synthetic 5G RAN Topology + Normal Operating Profile V1

Freeze the initial deterministic network structure before generating telemetry: **10 sites × 3 sectors**, paired `NRCellCU`/`NRCellDU` identities, two shared CU domains, and three synthetic workload archetypes.


````python
# ============================================================
# MODULE 5 — CELL 08
# SYNTHETIC 5G RAN TOPOLOGY
# + NORMAL OPERATING PROFILE FOUNDATION
# ============================================================
#
# Purpose:
#
#   1. Define the deterministic synthetic 5G RAN environment
#      used by Module 5 V1.
#
#   2. Represent the CU / DU / cell hierarchy required to
#      support realistic hierarchical investigation.
#
#   3. Create paired NRCellCU and NRCellDU identities for each
#      logical NR cell.
#
#   4. Define synthetic traffic/load archetypes that will later
#      control healthy 15-minute telemetry generation.
#
#   5. Freeze network structure BEFORE telemetry generation
#      and BEFORE incident/fault injection.
#
#
# IMPORTANT:
#
#   - NO telemetry values are generated in this cell.
#
#   - NO incidents are generated in this cell.
#
#   - NO faults are injected in this cell.
#
#   - NO root-cause labels are created in this cell.
#
#   - NO ML inference is performed in this cell.
#
#   - Traffic/load archetypes are SYNTHETIC ENGINEERING
#     assumptions, NOT 3GPP standardized thresholds.
#
# ============================================================


from enum import Enum
from typing import Dict, List
from datetime import datetime, timezone

from pydantic import (
    BaseModel,
    Field,
    ConfigDict,
)

import json
import hashlib


# ============================================================
# 8.1 — DEPENDENCY VALIDATION
# ============================================================

REQUIRED_CELL8_OBJECTS = [

    # Cell 4
    "EntityType",
    "MeasurementDomain",
    "PM_INTERVAL_MINUTES",
    "PATHS",
    "sha256_file",

    # Cell 7
    "RANTelemetryEntityRef",
    "TelemetryEnvironment",
    "cell7_semantic_sha",
]


missing_cell8_objects = [

    name

    for name
    in REQUIRED_CELL8_OBJECTS

    if name not in globals()
]


if missing_cell8_objects:

    raise RuntimeError(

        "Cell 08 requires Cell 07 and its upstream "
        "dependencies to be executed first. "
        f"Missing objects: {missing_cell8_objects}"
    )


# ============================================================
# 8.2 — VERIFY FROZEN CELL 7 BASELINE
# ============================================================

EXPECTED_CELL7_SEMANTIC_SHA = (
    "cc735d9a189f3346bfe7591f1f8e74bac0abd35f2c6e7235cb91f4bdb488a53e"
)


assert (
    cell7_semantic_sha
    == EXPECTED_CELL7_SEMANTIC_SHA
), (
    "Cell 7 semantic baseline differs from the "
    "frozen telemetry-observation contract."
)


assert (
    PM_INTERVAL_MINUTES
    == 15
)


# ============================================================
# 8.3 — SYNTHETIC ENVIRONMENT CONTROLS
# ============================================================
#
# These controls are frozen BEFORE telemetry generation.
#
# ============================================================

MODULE5_SYNTHETIC_SEED = 20260909

SYNTHETIC_SITE_COUNT = 10

SECTORS_PER_SITE = 3

SYNTHETIC_LOGICAL_CELL_COUNT = (
    SYNTHETIC_SITE_COUNT
    * SECTORS_PER_SITE
)

SYNTHETIC_CU_DOMAIN_COUNT = 2


# ============================================================
# 8.4 — SYNTHETIC SITE ARCHETYPE
# ============================================================
#
# These are engineering simulation categories only.
#
# They are NOT standardized 3GPP classifications.
#
# ============================================================

class SyntheticRANSiteArchetype(str, Enum):

    URBAN_HIGH = "URBAN_HIGH"

    URBAN_MEDIUM = "URBAN_MEDIUM"

    SUBURBAN = "SUBURBAN"


# ============================================================
# 8.5 — NORMAL OPERATING PROFILE
# ============================================================
#
# These parameters control SCALE and DAILY VARIATION only.
#
# Success/failure probabilities are deliberately deferred to
# the healthy telemetry generator.
#
# Therefore Cell 8 does not yet determine KPI outcomes.
#
# ============================================================

class SyntheticRANOperatingProfile(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    archetype: SyntheticRANSiteArchetype

    mean_connected_ues: int

    base_rrc_attempts_per_15m: int

    base_resume_attempts_per_15m: int

    relative_traffic_scale: float

    daily_load_amplitude: float

    description: str

    standardized_threshold: bool = False


# ============================================================
# 8.6 — FROZEN NORMAL-LOAD ARCHETYPES
# ============================================================
#
# IMPORTANT:
#
# Values below are synthetic workload assumptions.
#
# They are NOT:
#
#   - 3GPP thresholds
#   - alarm thresholds
#   - KPI acceptance criteria
#   - vendor recommendations
#
# ============================================================

SYNTHETIC_OPERATING_PROFILES = {

    SyntheticRANSiteArchetype.URBAN_HIGH:

        SyntheticRANOperatingProfile(

            archetype=(
                SyntheticRANSiteArchetype
                .URBAN_HIGH
            ),

            mean_connected_ues=420,

            base_rrc_attempts_per_15m=1800,

            base_resume_attempts_per_15m=800,

            relative_traffic_scale=1.00,

            daily_load_amplitude=0.45,

            description=(
                "High-traffic synthetic urban "
                "macro-cell operating profile."
            ),
        ),


    SyntheticRANSiteArchetype.URBAN_MEDIUM:

        SyntheticRANOperatingProfile(

            archetype=(
                SyntheticRANSiteArchetype
                .URBAN_MEDIUM
            ),

            mean_connected_ues=260,

            base_rrc_attempts_per_15m=1200,

            base_resume_attempts_per_15m=500,

            relative_traffic_scale=0.70,

            daily_load_amplitude=0.35,

            description=(
                "Moderate-traffic synthetic urban "
                "macro-cell operating profile."
            ),
        ),


    SyntheticRANSiteArchetype.SUBURBAN:

        SyntheticRANOperatingProfile(

            archetype=(
                SyntheticRANSiteArchetype
                .SUBURBAN
            ),

            mean_connected_ues=140,

            base_rrc_attempts_per_15m=700,

            base_resume_attempts_per_15m=250,

            relative_traffic_scale=0.45,

            daily_load_amplitude=0.25,

            description=(
                "Lower-traffic synthetic suburban "
                "macro-cell operating profile."
            ),
        ),
}


# ============================================================
# 8.7 — SYNTHETIC RAN CELL MODEL
# ============================================================
#
# One logical NR cell is represented at two management layers:
#
#                 LOGICAL NR CELL
#                       │
#             ┌─────────┴─────────┐
#             ▼                   ▼
#         NRCellCU            NRCellDU
#
#
# This allows:
#
#   CU-scoped Accessibility PM
#   +
#   DU-scoped RACH PM
#
# to refer to the same logical radio sector.
#
# ============================================================

class SyntheticRANCell(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    logical_cell_id: str

    site_id: str

    gnb_id: str

    cu_domain_id: str

    du_id: str

    sector_id: int

    sector_name: str

    azimuth_deg: int

    pci: int

    nr_band: str

    bandwidth_mhz: int

    archetype: SyntheticRANSiteArchetype

    nr_cell_cu: RANTelemetryEntityRef

    nr_cell_du: RANTelemetryEntityRef

    metadata: Dict[
        str,
        str
    ] = Field(
        default_factory=dict
    )


# ============================================================
# 8.8 — SYNTHETIC RAN SITE MODEL
# ============================================================

class SyntheticRANSite(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    site_id: str

    gnb_id: str

    cu_domain_id: str

    du_id: str

    archetype: SyntheticRANSiteArchetype

    cells: List[
        SyntheticRANCell
    ]


# ============================================================
# 8.9 — SYNTHETIC RAN TOPOLOGY MODEL
# ============================================================

class SyntheticRANTopology(BaseModel):

    model_config = ConfigDict(
        extra="forbid"
    )

    topology_id: str

    environment: TelemetryEnvironment

    pm_interval_minutes: int

    cu_domains: List[str]

    sites: List[
        SyntheticRANSite
    ]

    seed: int

    design_notes: List[str] = Field(
        default_factory=list
    )


# ============================================================
# 8.10 — SITE ARCHETYPE ASSIGNMENT
# ============================================================
#
# Fixed assignment prevents workload class from changing
# between experiment runs.
#
# ============================================================

SITE_ARCHETYPE_MAP = {

    "SITE_001":
        SyntheticRANSiteArchetype.URBAN_HIGH,

    "SITE_002":
        SyntheticRANSiteArchetype.URBAN_HIGH,

    "SITE_003":
        SyntheticRANSiteArchetype.URBAN_HIGH,

    "SITE_004":
        SyntheticRANSiteArchetype.URBAN_MEDIUM,

    "SITE_005":
        SyntheticRANSiteArchetype.URBAN_MEDIUM,

    "SITE_006":
        SyntheticRANSiteArchetype.URBAN_MEDIUM,

    "SITE_007":
        SyntheticRANSiteArchetype.URBAN_MEDIUM,

    "SITE_008":
        SyntheticRANSiteArchetype.SUBURBAN,

    "SITE_009":
        SyntheticRANSiteArchetype.SUBURBAN,

    "SITE_010":
        SyntheticRANSiteArchetype.SUBURBAN,
}


# ============================================================
# 8.11 — CU DOMAIN ASSIGNMENT
# ============================================================
#
# Two shared CU domains allow future investigation of:
#
#   - single-cell faults
#   - site-wide faults
#   - multi-site / shared-CU symptoms
#
#
# This grouping is a synthetic architecture choice,
# not a claim about a particular vendor implementation.
#
# ============================================================

def synthetic_cu_domain_for_site(
    site_number: int
) -> str:

    if site_number <= 5:

        return "CU_DOMAIN_01"

    return "CU_DOMAIN_02"


# ============================================================
# 8.12 — BUILD DETERMINISTIC SYNTHETIC TOPOLOGY
# ============================================================

SYNTHETIC_RAN_SITES = []


sector_azimuths = {

    1: 0,

    2: 120,

    3: 240,
}


for site_number in range(
    1,
    SYNTHETIC_SITE_COUNT + 1,
):

    site_id = (
        f"SITE_{site_number:03d}"
    )

    gnb_id = (
        f"GNB_{site_number:03d}"
    )

    du_id = (
        f"DU_{site_number:03d}"
    )

    cu_domain_id = (
        synthetic_cu_domain_for_site(
            site_number
        )
    )

    archetype = (
        SITE_ARCHETYPE_MAP[
            site_id
        ]
    )

    site_cells = []


    for sector_id in range(
        1,
        SECTORS_PER_SITE + 1,
    ):

        logical_cell_id = (

            f"{site_id}_CELL_{sector_id}"
        )


        nr_cell_cu_id = (

            f"NRCELL_CU_"
            f"{site_number:03d}_"
            f"{sector_id}"
        )


        nr_cell_du_id = (

            f"NRCELL_DU_"
            f"{site_number:03d}_"
            f"{sector_id}"
        )


        # Deterministic PCI assignment.
        #
        # Unique within this synthetic reference topology.

        pci = (

            ((site_number - 1) * 3)

            + sector_id

            + 99
        )


        nr_cell_cu_ref = (
            RANTelemetryEntityRef(

                entity_id=nr_cell_cu_id,

                entity_type=(
                    EntityType.NR_CELL_CU
                ),

                parent_entity_id=(
                    cu_domain_id
                ),

                site_id=site_id,

                metadata={

                    "logical_cell_id":
                        logical_cell_id,

                    "gnb_id":
                        gnb_id,

                    "du_id":
                        du_id,

                    "sector_id":
                        sector_id,
                },
            )
        )


        nr_cell_du_ref = (
            RANTelemetryEntityRef(

                entity_id=nr_cell_du_id,

                entity_type=(
                    EntityType.NR_CELL_DU
                ),

                parent_entity_id=du_id,

                site_id=site_id,

                metadata={

                    "logical_cell_id":
                        logical_cell_id,

                    "gnb_id":
                        gnb_id,

                    "cu_domain_id":
                        cu_domain_id,

                    "sector_id":
                        sector_id,
                },
            )
        )


        site_cells.append(

            SyntheticRANCell(

                logical_cell_id=(
                    logical_cell_id
                ),

                site_id=site_id,

                gnb_id=gnb_id,

                cu_domain_id=(
                    cu_domain_id
                ),

                du_id=du_id,

                sector_id=sector_id,

                sector_name=(
                    f"SECTOR_{sector_id}"
                ),

                azimuth_deg=(
                    sector_azimuths[
                        sector_id
                    ]
                ),

                pci=pci,

                nr_band="n78",

                bandwidth_mhz=100,

                archetype=archetype,

                nr_cell_cu=(
                    nr_cell_cu_ref
                ),

                nr_cell_du=(
                    nr_cell_du_ref
                ),

                metadata={

                    "topology_role":
                        "SYNTHETIC_REFERENCE_CELL",

                    "standards_threshold":
                        "FALSE",
                },
            )
        )


    SYNTHETIC_RAN_SITES.append(

        SyntheticRANSite(

            site_id=site_id,

            gnb_id=gnb_id,

            cu_domain_id=(
                cu_domain_id
            ),

            du_id=du_id,

            archetype=archetype,

            cells=site_cells,
        )
    )


# ============================================================
# 8.13 — BUILD TOPOLOGY OBJECT
# ============================================================

SYNTHETIC_RAN_TOPOLOGY = (

    SyntheticRANTopology(

        topology_id=(
            "module5_synthetic_"
            "5g_ran_topology_v1"
        ),

        environment=(
            TelemetryEnvironment
            .SYNTHETIC_V1
        ),

        pm_interval_minutes=(
            PM_INTERVAL_MINUTES
        ),

        cu_domains=[

            "CU_DOMAIN_01",

            "CU_DOMAIN_02",
        ],

        sites=(
            SYNTHETIC_RAN_SITES
        ),

        seed=(
            MODULE5_SYNTHETIC_SEED
        ),

        design_notes=[

            (
                "The topology contains 10 synthetic "
                "three-sector 5G NR sites."
            ),

            (
                "Each logical NR cell has paired "
                "NRCellCU and NRCellDU identities so "
                "CU-scoped Accessibility PM and "
                "DU-scoped RACH PM can be correlated."
            ),

            (
                "Two synthetic shared CU domains permit "
                "future single-cell, site-wide and "
                "multi-site correlation experiments."
            ),

            (
                "Site operating archetypes describe "
                "synthetic workload scale only and are "
                "not standardized KPI thresholds."
            ),

            (
                "No telemetry values, incidents, fault "
                "labels or hidden ground truth are "
                "created by this topology artifact."
            ),
        ],
    )
)


# ============================================================
# 8.14 — FLAT CELL LOOKUP
# ============================================================
#
# Later generators and Telemetry MCP should not repeatedly
# walk the nested site structure.
#
# ============================================================

SYNTHETIC_RAN_CELL_LOOKUP = {

    cell.logical_cell_id:
        cell

    for site
    in SYNTHETIC_RAN_TOPOLOGY.sites

    for cell
    in site.cells
}


# ============================================================
# 8.15 — ENTITY LOOKUPS
# ============================================================

SYNTHETIC_NR_CELL_CU_LOOKUP = {

    cell.nr_cell_cu.entity_id:
        cell.nr_cell_cu

    for cell
    in SYNTHETIC_RAN_CELL_LOOKUP.values()
}


SYNTHETIC_NR_CELL_DU_LOOKUP = {

    cell.nr_cell_du.entity_id:
        cell.nr_cell_du

    for cell
    in SYNTHETIC_RAN_CELL_LOOKUP.values()
}


# ============================================================
# 8.16 — TOPOLOGY INTEGRITY VALIDATION
# ============================================================

assert (
    len(
        SYNTHETIC_RAN_TOPOLOGY.sites
    )
    == SYNTHETIC_SITE_COUNT
)


assert (
    len(
        SYNTHETIC_RAN_CELL_LOOKUP
    )
    == SYNTHETIC_LOGICAL_CELL_COUNT
)


assert (
    len(
        SYNTHETIC_NR_CELL_CU_LOOKUP
    )
    == SYNTHETIC_LOGICAL_CELL_COUNT
)


assert (
    len(
        SYNTHETIC_NR_CELL_DU_LOOKUP
    )
    == SYNTHETIC_LOGICAL_CELL_COUNT
)


# ------------------------------------------------------------
# Every site must contain exactly three sectors.
# ------------------------------------------------------------

for site in (
    SYNTHETIC_RAN_TOPOLOGY.sites
):

    assert (
        len(
            site.cells
        )
        == SECTORS_PER_SITE
    )


# ------------------------------------------------------------
# Every logical cell must map to one CU and one DU identity.
# ------------------------------------------------------------

for cell in (
    SYNTHETIC_RAN_CELL_LOOKUP.values()
):

    assert (
        cell.nr_cell_cu.entity_type
        == EntityType.NR_CELL_CU
    )

    assert (
        cell.nr_cell_du.entity_type
        == EntityType.NR_CELL_DU
    )

    assert (
        cell.nr_cell_cu.metadata[
            "logical_cell_id"
        ]
        == cell.logical_cell_id
    )

    assert (
        cell.nr_cell_du.metadata[
            "logical_cell_id"
        ]
        == cell.logical_cell_id
    )


# ------------------------------------------------------------
# PCI uniqueness in this reference topology.
# ------------------------------------------------------------

all_pcis = [

    cell.pci

    for cell
    in SYNTHETIC_RAN_CELL_LOOKUP.values()
]


assert (
    len(
        all_pcis
    )
    == len(
        set(
            all_pcis
        )
    )
)


# ------------------------------------------------------------
# CU-domain distribution.
# ------------------------------------------------------------

cu_domain_site_counts = {}


for site in (
    SYNTHETIC_RAN_TOPOLOGY.sites
):

    cu_domain_site_counts[
        site.cu_domain_id
    ] = (

        cu_domain_site_counts.get(
            site.cu_domain_id,
            0,
        )

        + 1
    )


assert (
    cu_domain_site_counts
    == {

        "CU_DOMAIN_01": 5,

        "CU_DOMAIN_02": 5,
    }
)


# ============================================================
# 8.17 — ARCHETYPE DISTRIBUTION VALIDATION
# ============================================================

site_archetype_counts = {

    archetype.value: 0

    for archetype
    in SyntheticRANSiteArchetype
}


for site in (
    SYNTHETIC_RAN_TOPOLOGY.sites
):

    site_archetype_counts[
        site.archetype.value
    ] += 1


assert (
    site_archetype_counts
    == {

        "URBAN_HIGH": 3,

        "URBAN_MEDIUM": 4,

        "SUBURBAN": 3,
    }
)


# ============================================================
# 8.18 — HIDDEN-TRUTH SAFETY VALIDATION
# ============================================================

PROHIBITED_TOPOLOGY_FIELDS = {

    "root_cause",

    "ground_truth",

    "fault_label",

    "expected_diagnosis",

    "correct_action",

    "benchmark_answer",
}


cell_schema_fields = set(
    SyntheticRANCell.model_fields.keys()
)


site_schema_fields = set(
    SyntheticRANSite.model_fields.keys()
)


topology_schema_fields = set(
    SyntheticRANTopology.model_fields.keys()
)


assert (
    not (
        cell_schema_fields
        & PROHIBITED_TOPOLOGY_FIELDS
    )
)


assert (
    not (
        site_schema_fields
        & PROHIBITED_TOPOLOGY_FIELDS
    )
)


assert (
    not (
        topology_schema_fields
        & PROHIBITED_TOPOLOGY_FIELDS
    )
)


# ============================================================
# 8.19 — BUILD SYNTHETIC ENVIRONMENT ARTIFACT
# ============================================================

synthetic_ran_environment_artifact = {

    "artifact_id":
        (
            "module5_synthetic_"
            "5g_ran_environment_v1"
        ),

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "module_scope":
        "MODULE_5_V1_5G_RAN",

    "reference_domain":
        "5G_RAN_ACCESSIBILITY",

    "environment":
        (
            TelemetryEnvironment
            .SYNTHETIC_V1
            .value
        ),

    "role":
        "SYNTHETIC_NETWORK_ENVIRONMENT",

    "seed":
        MODULE5_SYNTHETIC_SEED,

    "pm_interval_minutes":
        PM_INTERVAL_MINUTES,

    "topology":
        SYNTHETIC_RAN_TOPOLOGY.model_dump(
            mode="json"
        ),

    "operating_profiles": {

        archetype.value:
            profile.model_dump(
                mode="json"
            )

        for archetype, profile
        in SYNTHETIC_OPERATING_PROFILES.items()
    },

    "provenance": {

        "cell7_semantic_sha256":
            cell7_semantic_sha,
    },

    "design_constraints": {

        "telemetry_generated":
            False,

        "incident_generated":
            False,

        "fault_injected":
            False,

        "hidden_ground_truth_exposed":
            False,

        "ml_inference_included":
            False,

        "standardized_thresholds_defined":
            False,
    },
}


# ============================================================
# 8.20 — DETERMINISTIC SEMANTIC HASH
# ============================================================

def canonical_json_sha256_cell8(
    data
):

    canonical = json.dumps(

        data,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

    ).encode(
        "utf-8"
    )


    return hashlib.sha256(
        canonical
    ).hexdigest()


cell8_semantic_content = {

    key: value

    for key, value
    in synthetic_ran_environment_artifact.items()

    if key != "created_utc"
}


cell8_semantic_sha = (

    canonical_json_sha256_cell8(
        cell8_semantic_content
    )
)


synthetic_ran_environment_artifact[
    "semantic_sha256"
] = cell8_semantic_sha


# ============================================================
# 8.21 — ARTIFACT DIRECTORY
# ============================================================

telemetry_artifact_dir = (

    PATHS["kpi"].parent
    / "telemetry"
)


telemetry_artifact_dir.mkdir(

    parents=True,

    exist_ok=True,
)


# ============================================================
# 8.22 — SAVE ARTIFACT
# ============================================================

cell8_path = (

    telemetry_artifact_dir

    / (
        "module5_synthetic_"
        "5g_ran_environment_v1.json"
    )
)


with open(

    cell8_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        synthetic_ran_environment_artifact,

        f,

        indent=2,

        ensure_ascii=False,
    )


cell8_artifact_sha = (

    sha256_file(
        cell8_path
    )
)


# ============================================================
# 8.23 — FINAL SUMMARY
# ============================================================

print("=" * 104)

print(
    "MODULE 5 — CELL 08 "
    "SYNTHETIC 5G RAN TOPOLOGY "
    "+ NORMAL OPERATING PROFILE FOUNDATION"
)

print("=" * 104)


print(
    "\nEnvironment"
)

print("-" * 104)

print(
    "Environment                       : "
    "SYNTHETIC_V1"
)

print(
    "Reference domain                  : "
    "5G RAN Accessibility"
)

print(
    f"Deterministic seed                : "
    f"{MODULE5_SYNTHETIC_SEED}"
)

print(
    f"PM interval                       : "
    f"{PM_INTERVAL_MINUTES} minutes"
)


print(
    "\nTopology"
)

print("-" * 104)

print(
    f"Shared CU domains                 : "
    f"{len(SYNTHETIC_RAN_TOPOLOGY.cu_domains)}"
)

print(
    f"Sites / DUs                       : "
    f"{len(SYNTHETIC_RAN_TOPOLOGY.sites)}"
)

print(
    f"Sectors per site                  : "
    f"{SECTORS_PER_SITE}"
)

print(
    f"Logical NR cells                  : "
    f"{len(SYNTHETIC_RAN_CELL_LOOKUP)}"
)

print(
    f"NRCellCU entities                 : "
    f"{len(SYNTHETIC_NR_CELL_CU_LOOKUP)}"
)

print(
    f"NRCellDU entities                 : "
    f"{len(SYNTHETIC_NR_CELL_DU_LOOKUP)}"
)

print(
    "Synthetic carrier                : "
    "n78 / 100 MHz"
)


print(
    "\nCU-domain distribution"
)

print("-" * 104)

for (
    cu_domain,
    site_count
) in sorted(
    cu_domain_site_counts.items()
):

    print(
        f"{cu_domain:<32} : "
        f"{site_count} sites"
    )


print(
    "\nSite operating archetypes"
)

print("-" * 104)

for (
    archetype,
    count
) in (
    site_archetype_counts.items()
):

    print(
        f"{archetype:<32} : "
        f"{count} sites"
    )


print(
    "\nGeneration boundaries"
)

print("-" * 104)

print(
    "Telemetry values generated        : "
    "NO"
)

print(
    "Incidents generated               : "
    "NO"
)

print(
    "Faults injected                   : "
    "NO"
)

print(
    "Hidden ground truth exposed       : "
    "NO"
)

print(
    "ML inference included             : "
    "NO"
)

print(
    "Standard KPI thresholds defined   : "
    "NO"
)


print(
    "\nProvenance"
)

print("-" * 104)

print(
    "Cell 7 semantic SHA               : "
    f"{cell7_semantic_sha}"
)


print(
    "\nArtifact"
)

print("-" * 104)

print(
    f"Path                               : "
    f"{cell8_path}"
)

print(
    f"Artifact SHA                       : "
    f"{cell8_artifact_sha}"
)

print(
    f"Semantic SHA                       : "
    f"{cell8_semantic_sha}"
)


print(
    "\n" + "=" * 104
)

print(
    "CELL 08 COMPLETE — "
    "SYNTHETIC 5G RAN ENVIRONMENT READY"
)

print(
    "=" * 104
)
````

###### Cell 08 — Observation

Cell 08 establishes the deterministic **synthetic 5G RAN topology and healthy operating-profile foundation** for the Accessibility experiment.

```text
CU domains        : 2
Sites / DUs       : 10
Logical NR cells  : 30
NRCellCU entities : 30
NRCellDU entities : 30
```

The three operating archetypes provide controlled workload diversity across the synthetic network.

> **These archetypes are synthetic engineering assumptions only; they are not 3GPP thresholds, vendor limits or SLA criteria, and no telemetry or fault behaviour is generated at this stage.**


<a id="cell-09-causal-observability"></a>
##### Cell 09 — Accessibility Causal Observability Configuration + Topology Readiness Gate

Convert the causal-observability reference into machine-readable runtime configuration. Define operational telemetry families, map fault families to required evidence, and test whether the V1 topology can support every declared diagnostic scope.


````python
# ============================================================
# MODULE 5 — CELL 09
# 5G RAN ACCESSIBILITY CAUSAL OBSERVABILITY CONFIGURATION
# + TOPOLOGY READINESS GATE
# ============================================================
#
# Purpose:
#
#   1. Convert the separate Accessibility Causal Observability
#      reference notebook into deterministic runtime
#      configuration.
#
#   2. Define the operational telemetry concepts required to
#      explain the selected Accessibility KPI components.
#
#   3. Map plausible synthetic fault families to observable
#      evidence rather than directly to hidden root-cause labels.
#
#   4. Verify that every declared fault family has sufficient
#      observable evidence.
#
#   5. Verify whether the Cell 08 topology contains the
#      entity/path identities required to generate those metrics.
#
#
# IMPORTANT:
#
#   - NO telemetry values are generated.
#
#   - NO incidents are generated.
#
#   - NO faults are injected.
#
#   - NO hidden root-cause truth is exposed.
#
#   - Operational canonical metrics below are NOT automatically
#     claimed to be standardized 3GPP PM measurements.
#
#   - Standardized PM truth remains the Cell 05 / Cell 07 layer.
#
# ============================================================


from enum import Enum

from typing import (
    Dict,
    List,
)

from datetime import (
    datetime,
    timezone,
)

from pydantic import (
    BaseModel,
    Field,
    ConfigDict,
)

import json
import hashlib


# ============================================================
# 9.1 — DEPENDENCY VALIDATION
# ============================================================

REQUIRED_CELL9_OBJECTS = [

    # Cell 4
    "EntityType",
    "PM_INTERVAL_MINUTES",
    "PATHS",
    "sha256_file",

    # Cell 6
    "RANInvestigationStage",

    # Cell 7
    "TelemetryObservationType",
    "TelemetryEvidenceOrigin",
    "RAN_TELEMETRY_METRIC_BINDINGS",
    "cell7_semantic_sha",

    # Cell 8
    "SyntheticRANCell",
    "SyntheticRANSite",
    "SYNTHETIC_RAN_TOPOLOGY",
    "SYNTHETIC_RAN_CELL_LOOKUP",
    "cell8_semantic_sha",
]


missing_cell9_objects = [

    name

    for name
    in REQUIRED_CELL9_OBJECTS

    if name not in globals()
]


if missing_cell9_objects:

    raise RuntimeError(

        "Cell 09 requires Cells 05, 07 and 08 "
        "to be executed first. "
        f"Missing objects: {missing_cell9_objects}"
    )


# ============================================================
# 9.2 — VERIFY FROZEN UPSTREAM BASELINES
# ============================================================

EXPECTED_CELL7_SEMANTIC_SHA = (
    "cc735d9a189f3346bfe7591f1f8e74bac0abd35f2c6e7235cb91f4bdb488a53e"
)


EXPECTED_CELL8_SEMANTIC_SHA = (
    "9e23832e8191e0dca58bde2c9d15631cd20539e2033e458fb39b54629f8e86d3"
)


assert (
    cell7_semantic_sha
    == EXPECTED_CELL7_SEMANTIC_SHA
)


assert (
    cell8_semantic_sha
    == EXPECTED_CELL8_SEMANTIC_SHA
)


assert (
    PM_INTERVAL_MINUTES
    == 15
)


# ============================================================
# 9.3 — REFERENCE NOTEBOOK PROVENANCE
# ============================================================

CAUSAL_REFERENCE_NOTEBOOK = (
    "Module5_Accessibility_"
    "Causal_Observability_Reference_v1.ipynb"
)


CAUSAL_REFERENCE_NOTEBOOK_SHA256 = (
    "6a58fb847fa47224c7289ccffb3126ffe"
    "76701d4d9c6b7fc5fd1e2a213858747"
)


# ============================================================
# 9.4 — ACCESSIBILITY INVESTIGATION COMPONENTS
# ============================================================

class AccessibilityInvestigationComponent(
    str,
    Enum,
):

    RACH_ACCESS = (
        "RACH_ACCESS"
    )

    RRC_ESTABLISHMENT = (
        "RRC_ESTABLISHMENT"
    )

    NG_SIGNALLING = (
        "NG_SIGNALLING"
    )

    INITIAL_DRB_ESTABLISHMENT = (
        "INITIAL_DRB_ESTABLISHMENT"
    )

    DRB_ESTABLISHMENT = (
        "DRB_ESTABLISHMENT"
    )

    RRC_RESUME = (
        "RRC_RESUME"
    )

    DRB_RESUME = (
        "DRB_RESUME"
    )


# ============================================================
# 9.5 — COMPONENT CLASSIFICATION
# ============================================================

class AccessibilityComponentClassification(
    str,
    Enum,
):

    KPI_COMPONENT = (
        "KPI_COMPONENT"
    )

    SUPPORTING_PRE_RRC_DIAGNOSTIC = (
        "SUPPORTING_PRE_RRC_DIAGNOSTIC"
    )


# ============================================================
# 9.6 — OPERATIONAL ENTITY SCOPES
# ============================================================
#
# These are CANONICAL OBSERVABILITY scopes.
#
# They are not claims that each item is itself a 3GPP
# managed-object class.
#
# ============================================================

class OperationalEntityScope(
    str,
    Enum,
):

    LOGICAL_CELL = (
        "LOGICAL_CELL"
    )

    NR_CELL_CU = (
        "NR_CELL_CU"
    )

    NR_CELL_DU = (
        "NR_CELL_DU"
    )

    SITE = (
        "SITE"
    )

    DU = (
        "DU"
    )

    CU_DOMAIN = (
        "CU_DOMAIN"
    )

    CU_CP = (
        "CU_CP"
    )

    CU_UP = (
        "CU_UP"
    )

    N2_PATH = (
        "N2_PATH"
    )

    AMF_PEER = (
        "AMF_PEER"
    )

    F1_PATH = (
        "F1_PATH"
    )

    E1_PATH = (
        "E1_PATH"
    )


# ============================================================
# 9.7 — OPERATIONAL METRIC CATEGORY
# ============================================================

class OperationalMetricCategory(
    str,
    Enum,
):

    RADIO = "RADIO"

    RESOURCE = "RESOURCE"

    RACH = "RACH"

    PAGING = "PAGING"

    PROCESSING = "PROCESSING"

    N2_TRANSPORT = (
        "N2_TRANSPORT"
    )

    SCTP = "SCTP"

    AMF_BOUNDARY = (
        "AMF_BOUNDARY"
    )

    INTERNAL_TRANSPORT = (
        "INTERNAL_TRANSPORT"
    )

    CONTEXT = "CONTEXT"

    CONFIGURATION = (
        "CONFIGURATION"
    )

    EVENT = "EVENT"


# ============================================================
# 9.8 — METRIC CADENCE
# ============================================================

class OperationalMetricCadence(
    str,
    Enum,
):

    PM_15_MIN = (
        "PM_15_MIN"
    )

    STATE_SNAPSHOT_15_MIN = (
        "STATE_SNAPSHOT_15_MIN"
    )

    EVENT_DRIVEN = (
        "EVENT_DRIVEN"
    )


# ============================================================
# 9.9 — METRIC VALUE KIND
# ============================================================

class OperationalMetricValueKind(
    str,
    Enum,
):

    FLOAT = "FLOAT"

    INTEGER = "INTEGER"

    BOOLEAN = "BOOLEAN"

    STATE = "STATE"


# ============================================================
# 9.10 — OPERATIONAL METRIC BINDING MODEL
# ============================================================

class RANOperationalTelemetryBinding(
    BaseModel
):

    model_config = ConfigDict(
        extra="forbid"
    )

    metric_name: str

    category: OperationalMetricCategory

    observation_type: (
        TelemetryObservationType
    )

    evidence_origin: (
        TelemetryEvidenceOrigin
    ) = (
        TelemetryEvidenceOrigin
        .OPERATIONAL_TELEMETRY
    )

    preferred_scope: (
        OperationalEntityScope
    )

    allowed_scopes: List[
        OperationalEntityScope
    ]

    cadence: OperationalMetricCadence

    value_kind: OperationalMetricValueKind

    unit: str

    description: str

    standardized_3gpp_measurement: bool = False

    live_mapping_expected: bool = True


# ============================================================
# 9.11 — COMPONENT CONFIG MODEL
# ============================================================

class AccessibilityCausalComponentConfig(
    BaseModel
):

    model_config = ConfigDict(
        extra="forbid"
    )

    component: (
        AccessibilityInvestigationComponent
    )

    display_name: str

    classification: (
        AccessibilityComponentClassification
    )

    investigation_stage: (
        RANInvestigationStage
    )

    standardized_formula_metrics: List[
        str
    ]

    fault_evidence_map: Dict[
        str,
        List[str],
    ]

    required_operational_metrics: List[
        str
    ]

    required_topology_scopes: List[
        OperationalEntityScope
    ]


# ============================================================
# 9.12 — OPERATIONAL METRIC REGISTRY HELPER
# ============================================================

RAN_OPERATIONAL_METRIC_BINDINGS = {}


def add_operational_metric(
    *,
    metric_name,
    category,
    observation_type,
    preferred_scope,
    cadence,
    value_kind,
    unit,
    description,
    allowed_scopes=None,
):

    if (
        metric_name
        in RAN_OPERATIONAL_METRIC_BINDINGS
    ):

        raise ValueError(

            "Duplicate operational metric: "
            f"{metric_name}"
        )


    if allowed_scopes is None:

        allowed_scopes = [
            preferred_scope
        ]


    RAN_OPERATIONAL_METRIC_BINDINGS[
        metric_name
    ] = (
        RANOperationalTelemetryBinding(

            metric_name=metric_name,

            category=category,

            observation_type=(
                observation_type
            ),

            preferred_scope=(
                preferred_scope
            ),

            allowed_scopes=(
                allowed_scopes
            ),

            cadence=cadence,

            value_kind=value_kind,

            unit=unit,

            description=description,
        )
    )


# ============================================================
# 9.13 — RADIO CONDITION METRICS
# ============================================================

for (
    metric_name,
    unit,
    description
) in [

    (
        "radio.rsrp_dbm",
        "dBm",
        (
            "Canonical cell radio-coverage "
            "evidence."
        ),
    ),

    (
        "radio.sinr_db",
        "dB",
        (
            "Canonical radio-quality and "
            "interference evidence."
        ),
    ),

    (
        "radio.dl_bler_pct",
        "percent",
        (
            "Canonical downlink block-error "
            "evidence."
        ),
    ),

    (
        "radio.ul_bler_pct",
        "percent",
        (
            "Canonical uplink block-error "
            "evidence."
        ),
    ),
]:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .RADIO
        ),

        observation_type=(
            TelemetryObservationType
            .RADIO_MEASUREMENT
        ),

        preferred_scope=(
            OperationalEntityScope
            .NR_CELL_DU
        ),

        cadence=(
            OperationalMetricCadence
            .PM_15_MIN
        ),

        value_kind=(
            OperationalMetricValueKind
            .FLOAT
        ),

        unit=unit,

        description=description,
    )


# ============================================================
# 9.14 — RADIO RESOURCE / LOAD METRICS
# ============================================================

RESOURCE_METRIC_DEFINITIONS = [

    (
        "resource.dl_prb_util_pct",
        OperationalEntityScope.NR_CELL_DU,
        OperationalMetricValueKind.FLOAT,
        "percent",
        "Downlink PRB utilisation evidence.",
    ),

    (
        "resource.ul_prb_util_pct",
        OperationalEntityScope.NR_CELL_DU,
        OperationalMetricValueKind.FLOAT,
        "percent",
        "Uplink PRB utilisation evidence.",
    ),

    (
        "load.connected_ues",
        OperationalEntityScope.NR_CELL_CU,
        OperationalMetricValueKind.INTEGER,
        "count",
        "Connected UE load.",
    ),

    (
        "load.active_ues",
        OperationalEntityScope.NR_CELL_CU,
        OperationalMetricValueKind.INTEGER,
        "count",
        "Active UE load.",
    ),

    (
        "resource.scheduler_pressure_pct",
        OperationalEntityScope.NR_CELL_DU,
        OperationalMetricValueKind.FLOAT,
        "percent",
        (
            "Canonical scheduler-resource "
            "pressure indicator."
        ),
    ),

    (
        "resource.bearer_admission_pressure_pct",
        OperationalEntityScope.NR_CELL_CU,
        OperationalMetricValueKind.FLOAT,
        "percent",
        (
            "Canonical bearer-admission "
            "pressure indicator."
        ),
    ),
]


for (
    metric_name,
    scope,
    value_kind,
    unit,
    description
) in RESOURCE_METRIC_DEFINITIONS:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .RESOURCE
        ),

        observation_type=(
            TelemetryObservationType
            .RESOURCE_MEASUREMENT
        ),

        preferred_scope=scope,

        cadence=(
            OperationalMetricCadence
            .PM_15_MIN
        ),

        value_kind=value_kind,

        unit=unit,

        description=description,
    )


# ============================================================
# 9.15 — OPERATIONAL RACH DIAGNOSTICS
# ============================================================
#
# These supplement the standardized RACH.Preamble* PMs.
#
# They are NOT represented as standardized TS 28.554 KPIs.
#
# ============================================================

for (
    metric_name,
    unit,
    description
) in [

    (
        "rach.contention_index",
        "index",
        (
            "Synthetic operational indicator "
            "of Random Access contention."
        ),
    ),

    (
        "rach.access_delay_ms",
        "milliseconds",
        (
            "Synthetic operational Random "
            "Access delay indicator."
        ),
    ),
]:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .RACH
        ),

        observation_type=(
            TelemetryObservationType
            .RADIO_MEASUREMENT
        ),

        preferred_scope=(
            OperationalEntityScope
            .NR_CELL_DU
        ),

        cadence=(
            OperationalMetricCadence
            .PM_15_MIN
        ),

        value_kind=(
            OperationalMetricValueKind
            .FLOAT
        ),

        unit=unit,

        description=description,
    )


# ============================================================
# 9.16 — PAGING
# ============================================================

PAGING_METRIC_DEFINITIONS = [

    (
        "paging.attempts",
        OperationalMetricValueKind.INTEGER,
        "count",
        "Paging attempts during the observation interval.",
    ),

    (
        "paging.responses",
        OperationalMetricValueKind.INTEGER,
        "count",
        "Observed paging responses.",
    ),

    (
        "paging.load_index",
        OperationalMetricValueKind.FLOAT,
        "index",
        "Synthetic paging load indicator.",
    ),
]


for (
    metric_name,
    value_kind,
    unit,
    description
) in PAGING_METRIC_DEFINITIONS:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .PAGING
        ),

        observation_type=(
            TelemetryObservationType
            .RESOURCE_MEASUREMENT
        ),

        preferred_scope=(
            OperationalEntityScope
            .NR_CELL_CU
        ),

        cadence=(
            OperationalMetricCadence
            .PM_15_MIN
        ),

        value_kind=value_kind,

        unit=unit,

        description=description,
    )


# ============================================================
# 9.17 — RAN PROCESSING
# ============================================================

PROCESSING_METRIC_DEFINITIONS = [

    (
        "processing.du_cpu_pct",
        OperationalEntityScope.DU,
        "percent",
        "DU processing utilisation.",
    ),

    (
        "processing.du_queue_depth",
        OperationalEntityScope.DU,
        "count",
        "DU processing queue depth.",
    ),

    (
        "processing.cu_cp_cpu_pct",
        OperationalEntityScope.CU_CP,
        "percent",
        "CU-CP CPU utilisation.",
    ),

    (
        "processing.cu_cp_memory_pct",
        OperationalEntityScope.CU_CP,
        "percent",
        "CU-CP memory utilisation.",
    ),

    (
        "processing.cu_cp_control_queue_depth",
        OperationalEntityScope.CU_CP,
        "count",
        "CU-CP control-plane queue depth.",
    ),

    (
        "processing.cu_cp_control_delay_ms",
        OperationalEntityScope.CU_CP,
        "milliseconds",
        "CU-CP control processing delay.",
    ),

    (
        "processing.cu_up_cpu_pct",
        OperationalEntityScope.CU_UP,
        "percent",
        "CU-UP CPU utilisation.",
    ),

    (
        "processing.cu_up_memory_pct",
        OperationalEntityScope.CU_UP,
        "percent",
        "CU-UP memory utilisation.",
    ),
]


for (
    metric_name,
    scope,
    unit,
    description
) in PROCESSING_METRIC_DEFINITIONS:

    value_kind = (

        OperationalMetricValueKind.INTEGER

        if unit == "count"

        else OperationalMetricValueKind.FLOAT
    )


    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .PROCESSING
        ),

        observation_type=(
            TelemetryObservationType
            .RESOURCE_MEASUREMENT
        ),

        preferred_scope=scope,

        cadence=(
            OperationalMetricCadence
            .PM_15_MIN
        ),

        value_kind=value_kind,

        unit=unit,

        description=description,
    )


# ============================================================
# 9.18 — N2 TRANSPORT
# ============================================================

for (
    metric_name,
    unit,
    description
) in [

    (
        "transport.n2_util_pct",
        "percent",
        "N2 transport utilisation.",
    ),

    (
        "transport.n2_latency_ms",
        "milliseconds",
        "N2 transport latency.",
    ),

    (
        "transport.n2_packet_loss_pct",
        "percent",
        "N2 packet-loss evidence.",
    ),
]:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .N2_TRANSPORT
        ),

        observation_type=(
            TelemetryObservationType
            .TRANSPORT_MEASUREMENT
        ),

        preferred_scope=(
            OperationalEntityScope
            .N2_PATH
        ),

        cadence=(
            OperationalMetricCadence
            .PM_15_MIN
        ),

        value_kind=(
            OperationalMetricValueKind
            .FLOAT
        ),

        unit=unit,

        description=description,
    )


# ============================================================
# 9.19 — SCTP
# ============================================================

SCTP_METRIC_DEFINITIONS = [

    (
        "sctp.association_state",
        OperationalMetricValueKind.STATE,
        "state",
        "SCTP association state.",
    ),

    (
        "sctp.retransmissions",
        OperationalMetricValueKind.INTEGER,
        "count",
        "SCTP retransmission count.",
    ),

    (
        "sctp.heartbeat_failures",
        OperationalMetricValueKind.INTEGER,
        "count",
        "SCTP heartbeat/path failure count.",
    ),
]


for (
    metric_name,
    value_kind,
    unit,
    description
) in SCTP_METRIC_DEFINITIONS:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .SCTP
        ),

        observation_type=(
            TelemetryObservationType
            .TRANSPORT_MEASUREMENT
        ),

        preferred_scope=(
            OperationalEntityScope
            .N2_PATH
        ),

        cadence=(
            OperationalMetricCadence
            .STATE_SNAPSHOT_15_MIN
        ),

        value_kind=value_kind,

        unit=unit,

        description=description,
    )


# ============================================================
# 9.20 — AMF-FACING BOUNDARY
# ============================================================

AMF_BOUNDARY_METRICS = [

    (
        "boundary.amf_reachable",
        OperationalMetricValueKind.BOOLEAN,
        "boolean",
        (
            "AMF reachability as observed "
            "from the RAN boundary."
        ),
    ),

    (
        "boundary.amf_peer_state",
        OperationalMetricValueKind.STATE,
        "state",
        (
            "RAN-observed AMF peer/path state."
        ),
    ),

    (
        "boundary.amf_overload_indication",
        OperationalMetricValueKind.BOOLEAN,
        "boolean",
        (
            "Observed AMF-facing overload indication."
        ),
    ),
]


for (
    metric_name,
    value_kind,
    unit,
    description
) in AMF_BOUNDARY_METRICS:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .AMF_BOUNDARY
        ),

        observation_type=(
            TelemetryObservationType
            .TRANSPORT_MEASUREMENT
        ),

        preferred_scope=(
            OperationalEntityScope
            .AMF_PEER
        ),

        cadence=(
            OperationalMetricCadence
            .STATE_SNAPSHOT_15_MIN
        ),

        value_kind=value_kind,

        unit=unit,

        description=description,
    )


# ============================================================
# 9.21 — INTERNAL RAN INTERFACE EVIDENCE
# ============================================================

INTERNAL_TRANSPORT_METRICS = [

    (
        "transport.f1_latency_ms",
        OperationalEntityScope.F1_PATH,
        OperationalMetricValueKind.FLOAT,
        "milliseconds",
        "F1 path latency.",
    ),

    (
        "transport.f1_packet_loss_pct",
        OperationalEntityScope.F1_PATH,
        OperationalMetricValueKind.FLOAT,
        "percent",
        "F1 path packet loss.",
    ),

    (
        "transport.f1_state",
        OperationalEntityScope.F1_PATH,
        OperationalMetricValueKind.STATE,
        "state",
        "F1 interface/path state.",
    ),

    (
        "transport.e1_latency_ms",
        OperationalEntityScope.E1_PATH,
        OperationalMetricValueKind.FLOAT,
        "milliseconds",
        "E1 path latency.",
    ),

    (
        "transport.e1_state",
        OperationalEntityScope.E1_PATH,
        OperationalMetricValueKind.STATE,
        "state",
        "E1 interface/path state.",
    ),
]


for (
    metric_name,
    scope,
    value_kind,
    unit,
    description
) in INTERNAL_TRANSPORT_METRICS:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .INTERNAL_TRANSPORT
        ),

        observation_type=(
            TelemetryObservationType
            .TRANSPORT_MEASUREMENT
        ),

        preferred_scope=scope,

        cadence=(
            OperationalMetricCadence
            .STATE_SNAPSHOT_15_MIN

            if value_kind
            == OperationalMetricValueKind.STATE

            else
            OperationalMetricCadence.PM_15_MIN
        ),

        value_kind=value_kind,

        unit=unit,

        description=description,
    )


# ============================================================
# 9.22 — CONTEXT AVAILABILITY
# ============================================================

CONTEXT_METRICS = [

    (
        "context.ue_context_availability_pct",
        OperationalEntityScope.CU_CP,
        (
            "Synthetic indicator representing "
            "availability of retained UE context."
        ),
    ),

    (
        "context.bearer_context_availability_pct",
        OperationalEntityScope.CU_UP,
        (
            "Synthetic indicator representing "
            "availability of bearer context."
        ),
    ),
]


for (
    metric_name,
    scope,
    description
) in CONTEXT_METRICS:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .CONTEXT
        ),

        observation_type=(
            TelemetryObservationType
            .RESOURCE_MEASUREMENT
        ),

        preferred_scope=scope,

        cadence=(
            OperationalMetricCadence
            .STATE_SNAPSHOT_15_MIN
        ),

        value_kind=(
            OperationalMetricValueKind
            .FLOAT
        ),

        unit="percent",

        description=description,
    )


# ============================================================
# 9.23 — CONFIGURATION STATE
# ============================================================

CONFIGURATION_METRICS = [

    (
        "config.prach_profile_state",
        OperationalEntityScope.NR_CELL_DU,
        "PRACH configuration-state evidence.",
    ),

    (
        "config.rrc_profile_state",
        OperationalEntityScope.NR_CELL_CU,
        "RRC configuration-state evidence.",
    ),

    (
        "config.n2_peer_state",
        OperationalEntityScope.AMF_PEER,
        "N2 / AMF peer configuration state.",
    ),

    (
        "config.qos_profile_state",
        OperationalEntityScope.NR_CELL_CU,
        "QoS / 5QI configuration state.",
    ),

    (
        "config.slice_profile_state",
        OperationalEntityScope.NR_CELL_CU,
        "S-NSSAI / slice configuration state.",
    ),

    (
        "config.resume_profile_state",
        OperationalEntityScope.CU_CP,
        "RRC Resume-related configuration state.",
    ),
]


for (
    metric_name,
    scope,
    description
) in CONFIGURATION_METRICS:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .CONFIGURATION
        ),

        observation_type=(
            TelemetryObservationType
            .CONFIGURATION_STATE
        ),

        preferred_scope=scope,

        cadence=(
            OperationalMetricCadence
            .STATE_SNAPSHOT_15_MIN
        ),

        value_kind=(
            OperationalMetricValueKind
            .STATE
        ),

        unit="state",

        description=description,
    )


# ============================================================
# 9.24 — ALARM / CHANGE / SOFTWARE EVENTS
# ============================================================

EVENT_DEFINITIONS = [

    (
        "event.alarm_state",
        TelemetryObservationType.ALARM_EVENT,
        OperationalEntityScope.CU_DOMAIN,
        (
            "Generic RAN operational alarm/event "
            "with affected-entity dimensions."
        ),
    ),

    (
        "event.interface_alarm_state",
        TelemetryObservationType.ALARM_EVENT,
        OperationalEntityScope.N2_PATH,
        (
            "Interface/path alarm state for "
            "N2/F1/E1 investigations."
        ),
    ),

    (
        "event.config_change",
        TelemetryObservationType.NETWORK_EVENT,
        OperationalEntityScope.CU_DOMAIN,
        (
            "Configuration-change event with "
            "affected-domain dimensions."
        ),
    ),

    (
        "event.software_restart",
        TelemetryObservationType.NETWORK_EVENT,
        OperationalEntityScope.CU_DOMAIN,
        (
            "Software restart/change event."
        ),
    ),
]


EVENT_ALLOWED_SCOPES = [

    OperationalEntityScope.NR_CELL_CU,

    OperationalEntityScope.NR_CELL_DU,

    OperationalEntityScope.DU,

    OperationalEntityScope.CU_DOMAIN,

    OperationalEntityScope.CU_CP,

    OperationalEntityScope.CU_UP,

    OperationalEntityScope.N2_PATH,

    OperationalEntityScope.AMF_PEER,

    OperationalEntityScope.F1_PATH,

    OperationalEntityScope.E1_PATH,
]


for (
    metric_name,
    observation_type,
    preferred_scope,
    description
) in EVENT_DEFINITIONS:

    add_operational_metric(

        metric_name=metric_name,

        category=(
            OperationalMetricCategory
            .EVENT
        ),

        observation_type=(
            observation_type
        ),

        preferred_scope=(
            preferred_scope
        ),

        allowed_scopes=(
            EVENT_ALLOWED_SCOPES
        ),

        cadence=(
            OperationalMetricCadence
            .EVENT_DRIVEN
        ),

        value_kind=(
            OperationalMetricValueKind
            .STATE
        ),

        unit="state",

        description=description,
    )


# ============================================================
# 9.25 — REGISTRY COUNT VALIDATION
# ============================================================

assert (
    len(
        RAN_OPERATIONAL_METRIC_BINDINGS
    )
    == 49
)


# ============================================================
# 9.26 — STANDARDIZED FORMULA / SUPPORTING EVIDENCE
# ============================================================
#
# Seven investigation components.
#
# RACH is supporting pre-RRC evidence.
#
# The other six are Accessibility investigation components.
#
# ============================================================

FORMULA_MEASUREMENTS_BY_COMPONENT = {

    AccessibilityInvestigationComponent
    .RACH_ACCESS: [

        "RACH.PreambleDedCell",

        "RACH.PreambleACell",

        "RACH.PreambleBCell",
    ],


    AccessibilityInvestigationComponent
    .RRC_ESTABLISHMENT: [

        "RRC.ConnEstabAtt.Cause",

        "RRC.ConnEstabSucc.Cause",
    ],


    AccessibilityInvestigationComponent
    .NG_SIGNALLING: [

        "UECNTX.ConnEstabAtt.Cause",

        "UECNTX.ConnEstabSucc.Cause",
    ],


    AccessibilityInvestigationComponent
    .INITIAL_DRB_ESTABLISHMENT: [

        "DRB.InitialEstabAtt.5QI",

        "DRB.InitialEstabSucc.5QI",
    ],


    AccessibilityInvestigationComponent
    .DRB_ESTABLISHMENT: [

        "DRB.EstabAtt.5QI",

        "DRB.EstabSucc.5QI",

        "DRB.EstabAtt.SNSSAI",

        "DRB.EstabSucc.SNSSAI",
    ],


    AccessibilityInvestigationComponent
    .RRC_RESUME: [

        "RRC.ResumeAtt.Cause",

        "RRC.ResumeSucc.Cause",

        "RRC.ResumeSuccByFallback.cause",

        "RRC.ResumeFallbackToSetupAtt.cause",

        "RRC.ReEstabSuccWithoutUeContext",

        "RRC.ReEstabFallbackToSetupAtt",
    ],


    AccessibilityInvestigationComponent
    .DRB_RESUME: [

        "DRB.ResumeAtt.5QI",

        "DRB.ResumeSucc.5QI",
    ],
}


# ============================================================
# 9.27 — EXACT STANDARDIZED COVERAGE
# ============================================================

configured_standardized_metrics = set()


for metric_list in (
    FORMULA_MEASUREMENTS_BY_COMPONENT.values()
):

    configured_standardized_metrics.update(
        metric_list
    )


assert (
    configured_standardized_metrics
    == set(
        RAN_TELEMETRY_METRIC_BINDINGS.keys()
    )
), (
    "The causal-observability configuration "
    "does not exactly cover the frozen Cell 7 "
    "standardized metric registry."
)


assert (
    len(
        configured_standardized_metrics
    )
    == 21
)


# ============================================================
# 9.28 — FAULT → OBSERVABLE EVIDENCE MAP
# ============================================================
#
# IMPORTANT:
#
# These are not hidden labels placed into telemetry.
#
# This is DESIGN-TIME configuration describing which evidence
# must exist if that fault family is later injected by the
# separate hidden experiment harness.
#
# ============================================================

FAULT_EVIDENCE_BY_COMPONENT = {

    # --------------------------------------------------------
    # RACH
    # --------------------------------------------------------

    AccessibilityInvestigationComponent
    .RACH_ACCESS: {

        "POOR_COVERAGE": [

            "RACH.PreambleACell",

            "radio.rsrp_dbm",

            "radio.sinr_db",
        ],

        "UL_INTERFERENCE": [

            "RACH.PreambleACell",

            "radio.sinr_db",

            "radio.ul_bler_pct",
        ],

        "RACH_CONTENTION": [

            "RACH.PreambleACell",

            "RACH.PreambleBCell",

            "rach.contention_index",

            "rach.access_delay_ms",

            "load.active_ues",
        ],

        "DU_PROCESSING_PRESSURE": [

            "processing.du_cpu_pct",

            "processing.du_queue_depth",

            "rach.access_delay_ms",
        ],

        "PRACH_CONFIGURATION": [

            "config.prach_profile_state",

            "event.config_change",

            "rach.contention_index",
        ],

        "RADIO_DU_ALARM": [

            "event.alarm_state",

            "processing.du_cpu_pct",

            "RACH.PreambleACell",
        ],
    },


    # --------------------------------------------------------
    # RRC ESTABLISHMENT
    # --------------------------------------------------------

    AccessibilityInvestigationComponent
    .RRC_ESTABLISHMENT: {

        "UPSTREAM_RACH": [

            "RACH.PreambleACell",

            "rach.contention_index",

            "rach.access_delay_ms",

            "RRC.ConnEstabSucc.Cause",
        ],

        "POOR_COVERAGE": [

            "radio.rsrp_dbm",

            "radio.sinr_db",

            "RRC.ConnEstabSucc.Cause",
        ],

        "RADIO_INTERFERENCE": [

            "radio.sinr_db",

            "radio.dl_bler_pct",

            "radio.ul_bler_pct",

            "RRC.ConnEstabSucc.Cause",
        ],

        "RADIO_RESOURCE_CONGESTION": [

            "resource.dl_prb_util_pct",

            "resource.ul_prb_util_pct",

            "load.connected_ues",

            "load.active_ues",

            "resource.scheduler_pressure_pct",
        ],

        "DU_PROCESSING_PRESSURE": [

            "processing.du_cpu_pct",

            "processing.du_queue_depth",

            "RRC.ConnEstabSucc.Cause",
        ],

        "CU_CP_PROCESSING_PRESSURE": [

            "processing.cu_cp_cpu_pct",

            "processing.cu_cp_memory_pct",

            "processing.cu_cp_control_queue_depth",

            "processing.cu_cp_control_delay_ms",
        ],

        "RRC_CONFIGURATION": [

            "config.rrc_profile_state",

            "event.config_change",

            "RRC.ConnEstabSucc.Cause",
        ],

        "SOFTWARE_EVENT": [

            "event.software_restart",

            "event.alarm_state",

            "RRC.ConnEstabSucc.Cause",
        ],

        "PAGING_MT_ACCESS": [

            "paging.attempts",

            "paging.responses",

            "paging.load_index",

            "RRC.ConnEstabSucc.Cause",
        ],
    },


    # --------------------------------------------------------
    # NG SIGNALLING
    # --------------------------------------------------------

    AccessibilityInvestigationComponent
    .NG_SIGNALLING: {

        "UPSTREAM_RRC": [

            "RRC.ConnEstabAtt.Cause",

            "RRC.ConnEstabSucc.Cause",

            "UECNTX.ConnEstabAtt.Cause",
        ],

        "CU_CP_PROCESSING_PRESSURE": [

            "processing.cu_cp_cpu_pct",

            "processing.cu_cp_memory_pct",

            "processing.cu_cp_control_queue_depth",

            "processing.cu_cp_control_delay_ms",
        ],

        "N2_CONGESTION": [

            "transport.n2_util_pct",

            "transport.n2_latency_ms",

            "sctp.retransmissions",
        ],

        "N2_PACKET_LOSS_LATENCY": [

            "transport.n2_packet_loss_pct",

            "transport.n2_latency_ms",

            "sctp.retransmissions",
        ],

        "SCTP_INSTABILITY": [

            "sctp.association_state",

            "sctp.retransmissions",

            "sctp.heartbeat_failures",

            "event.interface_alarm_state",
        ],

        "N2_PEER_CONFIGURATION": [

            "config.n2_peer_state",

            "event.config_change",

            "sctp.association_state",
        ],

        "SOFTWARE_EVENT": [

            "event.software_restart",

            "event.alarm_state",

            "UECNTX.ConnEstabSucc.Cause",
        ],

        "AMF_FACING_BOUNDARY": [

            "boundary.amf_reachable",

            "boundary.amf_peer_state",

            "boundary.amf_overload_indication",

            "transport.n2_packet_loss_pct",
        ],
    },


    # --------------------------------------------------------
    # INITIAL DRB
    # --------------------------------------------------------

    AccessibilityInvestigationComponent
    .INITIAL_DRB_ESTABLISHMENT: {

        "UPSTREAM_RRC_NG": [

            "RRC.ConnEstabSucc.Cause",

            "UECNTX.ConnEstabSucc.Cause",

            "DRB.InitialEstabAtt.5QI",
        ],

        "POOR_RADIO_QUALITY": [

            "radio.rsrp_dbm",

            "radio.sinr_db",

            "radio.dl_bler_pct",

            "radio.ul_bler_pct",
        ],

        "RADIO_RESOURCE_CONGESTION": [

            "resource.dl_prb_util_pct",

            "resource.ul_prb_util_pct",

            "load.connected_ues",

            "load.active_ues",

            "resource.scheduler_pressure_pct",
        ],

        "ADMISSION_RESOURCE_PRESSURE": [

            "resource.bearer_admission_pressure_pct",

            "resource.scheduler_pressure_pct",

            "DRB.InitialEstabSucc.5QI",
        ],

        "DU_PROCESSING_PRESSURE": [

            "processing.du_cpu_pct",

            "processing.du_queue_depth",

            "DRB.InitialEstabSucc.5QI",
        ],

        "CU_UP_RESOURCE_PRESSURE": [

            "processing.cu_up_cpu_pct",

            "processing.cu_up_memory_pct",

            "DRB.InitialEstabSucc.5QI",
        ],

        "QOS_CONFIGURATION": [

            "config.qos_profile_state",

            "event.config_change",

            "DRB.InitialEstabSucc.5QI",
        ],

        "INTERNAL_INTERFACE": [

            "transport.f1_latency_ms",

            "transport.f1_packet_loss_pct",

            "transport.f1_state",

            "transport.e1_latency_ms",

            "transport.e1_state",

            "event.interface_alarm_state",
        ],

        "SOFTWARE_CONFIGURATION_EVENT": [

            "event.software_restart",

            "event.config_change",

            "event.alarm_state",

            "DRB.InitialEstabSucc.5QI",
        ],
    },


    # --------------------------------------------------------
    # DRB ESTABLISHMENT
    # --------------------------------------------------------

    AccessibilityInvestigationComponent
    .DRB_ESTABLISHMENT: {

        "RADIO_RESOURCE_CONGESTION": [

            "resource.dl_prb_util_pct",

            "resource.ul_prb_util_pct",

            "load.connected_ues",

            "load.active_ues",

            "resource.scheduler_pressure_pct",
        ],

        "POOR_RADIO_QUALITY": [

            "radio.rsrp_dbm",

            "radio.sinr_db",

            "radio.dl_bler_pct",

            "radio.ul_bler_pct",
        ],

        "FIVE_QI_SPECIFIC": [

            "DRB.EstabAtt.5QI",

            "DRB.EstabSucc.5QI",

            "config.qos_profile_state",

            "resource.bearer_admission_pressure_pct",
        ],

        "SNSSAI_SPECIFIC": [

            "DRB.EstabAtt.SNSSAI",

            "DRB.EstabSucc.SNSSAI",

            "config.slice_profile_state",

            "resource.bearer_admission_pressure_pct",
        ],

        "DU_PROCESSING_PRESSURE": [

            "processing.du_cpu_pct",

            "processing.du_queue_depth",

            "DRB.EstabSucc.5QI",
        ],

        "CU_UP_RESOURCE_PRESSURE": [

            "processing.cu_up_cpu_pct",

            "processing.cu_up_memory_pct",

            "DRB.EstabSucc.5QI",
        ],

        "INTERNAL_INTERFACE": [

            "transport.f1_latency_ms",

            "transport.f1_packet_loss_pct",

            "transport.f1_state",

            "transport.e1_latency_ms",

            "transport.e1_state",

            "event.interface_alarm_state",
        ],

        "SOFTWARE_CONFIGURATION_EVENT": [

            "event.software_restart",

            "event.config_change",

            "event.alarm_state",

            "DRB.EstabSucc.5QI",
        ],
    },


    # --------------------------------------------------------
    # RRC RESUME
    # --------------------------------------------------------

    AccessibilityInvestigationComponent
    .RRC_RESUME: {

        "RADIO_RESUME_DEGRADATION": [

            "radio.rsrp_dbm",

            "radio.sinr_db",

            "RRC.ResumeSucc.Cause",
        ],

        "RACH_RESUME_ACCESS": [

            "RACH.PreambleACell",

            "rach.contention_index",

            "rach.access_delay_ms",

            "RRC.ResumeSucc.Cause",
        ],

        "CU_CP_PROCESSING_PRESSURE": [

            "processing.cu_cp_cpu_pct",

            "processing.cu_cp_memory_pct",

            "processing.cu_cp_control_queue_depth",

            "processing.cu_cp_control_delay_ms",
        ],

        "UE_CONTEXT_UNAVAILABLE": [

            "context.ue_context_availability_pct",

            "RRC.ResumeFallbackToSetupAtt.cause",

            "RRC.ReEstabSuccWithoutUeContext",
        ],

        "FALLBACK_ABNORMALITY": [

            "RRC.ResumeFallbackToSetupAtt.cause",

            "RRC.ResumeSuccByFallback.cause",

            "context.ue_context_availability_pct",
        ],

        "REESTABLISHMENT_ABNORMALITY": [

            "RRC.ReEstabSuccWithoutUeContext",

            "RRC.ReEstabFallbackToSetupAtt",

            "context.ue_context_availability_pct",
        ],

        "RESUME_CONFIGURATION": [

            "config.resume_profile_state",

            "event.config_change",

            "RRC.ResumeSucc.Cause",
        ],

        "CU_RESTART_SOFTWARE_EVENT": [

            "event.software_restart",

            "event.alarm_state",

            "context.ue_context_availability_pct",
        ],

        "PAGING_MT_ACCESS": [

            "paging.attempts",

            "paging.responses",

            "paging.load_index",

            "RRC.ResumeAtt.Cause",
        ],
    },


    # --------------------------------------------------------
    # DRB RESUME
    # --------------------------------------------------------

    AccessibilityInvestigationComponent
    .DRB_RESUME: {

        "UPSTREAM_RRC_RESUME": [

            "RRC.ResumeSucc.Cause",

            "RRC.ResumeSuccByFallback.cause",

            "DRB.ResumeAtt.5QI",
        ],

        "POOR_RADIO_QUALITY": [

            "radio.rsrp_dbm",

            "radio.sinr_db",

            "radio.dl_bler_pct",

            "radio.ul_bler_pct",
        ],

        "RADIO_RESOURCE_CONGESTION": [

            "resource.dl_prb_util_pct",

            "resource.ul_prb_util_pct",

            "load.connected_ues",

            "load.active_ues",

            "resource.scheduler_pressure_pct",
        ],

        "BEARER_CONTEXT_UNAVAILABLE": [

            "context.bearer_context_availability_pct",

            "DRB.ResumeSucc.5QI",

            "event.software_restart",
        ],

        "DU_PROCESSING_PRESSURE": [

            "processing.du_cpu_pct",

            "processing.du_queue_depth",

            "DRB.ResumeSucc.5QI",
        ],

        "CU_UP_RESOURCE_PRESSURE": [

            "processing.cu_up_cpu_pct",

            "processing.cu_up_memory_pct",

            "DRB.ResumeSucc.5QI",
        ],

        "FIVE_QI_SPECIFIC": [

            "DRB.ResumeAtt.5QI",

            "DRB.ResumeSucc.5QI",

            "config.qos_profile_state",
        ],

        "INTERNAL_INTERFACE": [

            "transport.f1_latency_ms",

            "transport.f1_packet_loss_pct",

            "transport.f1_state",

            "transport.e1_latency_ms",

            "transport.e1_state",

            "event.interface_alarm_state",
        ],

        "SOFTWARE_CONFIGURATION_EVENT": [

            "event.software_restart",

            "event.config_change",

            "event.alarm_state",

            "DRB.ResumeSucc.5QI",
        ],
    },
}


# ============================================================
# 9.29 — COMPONENT DISPLAY / STAGE METADATA
# ============================================================

COMPONENT_DISPLAY_NAMES = {

    AccessibilityInvestigationComponent
    .RACH_ACCESS:
        "Random Access / RACH",

    AccessibilityInvestigationComponent
    .RRC_ESTABLISHMENT:
        "RRC Connection Establishment",

    AccessibilityInvestigationComponent
    .NG_SIGNALLING:
        (
            "UE-Associated Logical "
            "NG-Connection Establishment"
        ),

    AccessibilityInvestigationComponent
    .INITIAL_DRB_ESTABLISHMENT:
        "Initial DRB Establishment",

    AccessibilityInvestigationComponent
    .DRB_ESTABLISHMENT:
        "DRB Establishment",

    AccessibilityInvestigationComponent
    .RRC_RESUME:
        (
            "RRC Resume / Fallback / "
            "Re-establishment"
        ),

    AccessibilityInvestigationComponent
    .DRB_RESUME:
        "DRB Resume",
}


COMPONENT_STAGES = {

    AccessibilityInvestigationComponent
    .RACH_ACCESS:
        RANInvestigationStage
        .RANDOM_ACCESS,

    AccessibilityInvestigationComponent
    .RRC_ESTABLISHMENT:
        RANInvestigationStage
        .RRC_ESTABLISHMENT,

    AccessibilityInvestigationComponent
    .NG_SIGNALLING:
        RANInvestigationStage
        .NG_SIGNALLING,

    AccessibilityInvestigationComponent
    .INITIAL_DRB_ESTABLISHMENT:
        RANInvestigationStage
        .DRB_ESTABLISHMENT,

    AccessibilityInvestigationComponent
    .DRB_ESTABLISHMENT:
        RANInvestigationStage
        .DRB_ESTABLISHMENT,

    AccessibilityInvestigationComponent
    .RRC_RESUME:
        RANInvestigationStage
        .RRC_RESUME,

    AccessibilityInvestigationComponent
    .DRB_RESUME:
        RANInvestigationStage
        .DRB_RESUME,
}


# ============================================================
# 9.30 — BUILD COMPONENT CONFIGURATION
# ============================================================

ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS = {}


combined_metric_registry = (

    set(
        RAN_TELEMETRY_METRIC_BINDINGS.keys()
    )

    | set(
        RAN_OPERATIONAL_METRIC_BINDINGS.keys()
    )
)


for component in (
    AccessibilityInvestigationComponent
):

    formula_metrics = (

        FORMULA_MEASUREMENTS_BY_COMPONENT[
            component
        ]
    )


    fault_map = (

        FAULT_EVIDENCE_BY_COMPONENT[
            component
        ]
    )


    # --------------------------------------------------------
    # Every fault-family evidence item must resolve to either:
    #
    #   standardized Cell 7 metric
    #       OR
    #   operational Cell 9 metric
    # --------------------------------------------------------

    referenced_metrics = set()


    for (
        fault_family,
        evidence_metrics
    ) in fault_map.items():

        assert (
            len(
                evidence_metrics
            )
            >= 2
        ), (
            "Each declared fault family must have "
            "at least two observable evidence items. "
            f"{component.value} / {fault_family}"
        )


        unresolved = (

            set(
                evidence_metrics
            )

            - combined_metric_registry
        )


        assert (
            len(
                unresolved
            )
            == 0
        ), (
            "Unresolved evidence metrics for "
            f"{component.value} / "
            f"{fault_family}: "
            f"{sorted(unresolved)}"
        )


        referenced_metrics.update(
            evidence_metrics
        )


    required_operational_metrics = sorted(

        referenced_metrics

        & set(
            RAN_OPERATIONAL_METRIC_BINDINGS.keys()
        )
    )


    required_topology_scopes = {


        RAN_OPERATIONAL_METRIC_BINDINGS[
            metric_name
        ].preferred_scope

        for metric_name
        in required_operational_metrics
    }


    classification = (

        AccessibilityComponentClassification
        .SUPPORTING_PRE_RRC_DIAGNOSTIC

        if (
            component
            == AccessibilityInvestigationComponent
            .RACH_ACCESS
        )

        else
        AccessibilityComponentClassification
        .KPI_COMPONENT
    )


    ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS[
        component.value
    ] = (

        AccessibilityCausalComponentConfig(

            component=component,

            display_name=(
                COMPONENT_DISPLAY_NAMES[
                    component
                ]
            ),

            classification=(
                classification
            ),

            investigation_stage=(
                COMPONENT_STAGES[
                    component
                ]
            ),

            standardized_formula_metrics=(
                formula_metrics
            ),

            fault_evidence_map=(
                fault_map
            ),

            required_operational_metrics=(
                required_operational_metrics
            ),

            required_topology_scopes=sorted(

                required_topology_scopes,

                key=lambda scope:
                    scope.value,
            ),
        )
    )


# ============================================================
# 9.31 — COMPLETE OPERATIONAL METRIC UTILIZATION CHECK
# ============================================================
#
# Every operational metric we define must be justified by at
# least one declared fault-family evidence map.
#
# ============================================================

referenced_operational_metrics = set()


for component_config in (
    ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS.values()
):

    referenced_operational_metrics.update(

        component_config
        .required_operational_metrics
    )


unused_operational_metrics = (

    set(
        RAN_OPERATIONAL_METRIC_BINDINGS.keys()
    )

    - referenced_operational_metrics
)


assert (
    len(
        unused_operational_metrics
    )
    == 0
), (
    "Operational metrics exist without a "
    "declared diagnostic purpose: "
    f"{sorted(unused_operational_metrics)}"
)


# ============================================================
# 9.32 — CAUSAL FAMILY COVERAGE SUMMARY
# ============================================================

unique_fault_families = set()


component_fault_family_count = 0


for component_config in (
    ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS.values()
):

    unique_fault_families.update(

        component_config
        .fault_evidence_map
        .keys()
    )


    component_fault_family_count += (

        len(
            component_config
            .fault_evidence_map
        )
    )


assert (
    len(
        ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS
    )
    == 7
)


assert (
    len(
        unique_fault_families
    )
    == 37
)


assert (
    component_fault_family_count
    == 58
)


# ============================================================
# 9.33 — REQUIRED TOPOLOGY SCOPES
# ============================================================
#
# Add cross-cutting correlation scopes even when they are not
# the preferred scope of a scalar metric.
#
# ============================================================

required_topology_scopes_global = {

    OperationalEntityScope
    .LOGICAL_CELL,

    OperationalEntityScope
    .NR_CELL_CU,

    OperationalEntityScope
    .NR_CELL_DU,

    OperationalEntityScope
    .SITE,

    OperationalEntityScope
    .DU,

    OperationalEntityScope
    .CU_DOMAIN,
}


for component_config in (
    ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS.values()
):

    required_topology_scopes_global.update(

        component_config
        .required_topology_scopes
    )


# ============================================================
# 9.34 — CELL 8 TOPOLOGY READINESS ASSESSMENT
# ============================================================

cell8_cell_fields = set(
    SyntheticRANCell.model_fields.keys()
)


cell8_site_fields = set(
    SyntheticRANSite.model_fields.keys()
)


CELL8_TOPOLOGY_SCOPE_SUPPORT = {

    OperationalEntityScope
    .LOGICAL_CELL:

        (
            "logical_cell_id"
            in cell8_cell_fields
        ),


    OperationalEntityScope
    .NR_CELL_CU:

        (
            "nr_cell_cu"
            in cell8_cell_fields
        ),


    OperationalEntityScope
    .NR_CELL_DU:

        (
            "nr_cell_du"
            in cell8_cell_fields
        ),


    OperationalEntityScope
    .SITE:

        (
            "site_id"
            in cell8_site_fields
        ),


    OperationalEntityScope
    .DU:

        (
            "du_id"
            in cell8_site_fields
        ),


    OperationalEntityScope
    .CU_DOMAIN:

        (
            "cu_domain_id"
            in cell8_site_fields
        ),


    OperationalEntityScope
    .CU_CP:

        (
            "cu_cp_id"
            in cell8_site_fields
        ),


    OperationalEntityScope
    .CU_UP:

        (
            "cu_up_id"
            in cell8_site_fields
        ),


    OperationalEntityScope
    .N2_PATH:

        (
            "n2_path_id"
            in cell8_site_fields
        ),


    OperationalEntityScope
    .AMF_PEER:

        (
            "amf_peer_id"
            in cell8_site_fields
        ),


    OperationalEntityScope
    .F1_PATH:

        (
            "f1_path_id"
            in cell8_site_fields
        ),


    OperationalEntityScope
    .E1_PATH:

        (
            "e1_path_id"
            in cell8_site_fields
        ),
}


missing_cell8_topology_scopes = sorted(

    [

        scope

        for scope
        in required_topology_scopes_global

        if not (
            CELL8_TOPOLOGY_SCOPE_SUPPORT[
                scope
            ]
        )
    ],

    key=lambda scope:
        scope.value,
)


EXPECTED_CELL8_TOPOLOGY_GAPS = {

    OperationalEntityScope
    .CU_CP,

    OperationalEntityScope
    .CU_UP,

    OperationalEntityScope
    .N2_PATH,

    OperationalEntityScope
    .AMF_PEER,

    OperationalEntityScope
    .F1_PATH,

    OperationalEntityScope
    .E1_PATH,
}


assert (
    set(
        missing_cell8_topology_scopes
    )
    == EXPECTED_CELL8_TOPOLOGY_GAPS
), (
    "Unexpected Cell 8 topology-readiness result. "
    "Observed missing scopes: "
    f"{[
        scope.value
        for scope
        in missing_cell8_topology_scopes
    ]}"
)


CELL8_CAUSAL_TOPOLOGY_REVISION_REQUIRED = (

    len(
        missing_cell8_topology_scopes
    )
    > 0
)


TELEMETRY_GENERATION_READY = (

    not (
        CELL8_CAUSAL_TOPOLOGY_REVISION_REQUIRED
    )
)


# ============================================================
# 9.35 — HIDDEN-TRUTH SAFETY VALIDATION
# ============================================================

PROHIBITED_CAUSAL_CONFIG_FIELDS = {

    "root_cause",

    "ground_truth",

    "expected_diagnosis",

    "correct_action",

    "benchmark_answer",

    "fault_label",
}


assert (
    not (
        set(
            RANOperationalTelemetryBinding
            .model_fields.keys()
        )

        & PROHIBITED_CAUSAL_CONFIG_FIELDS
    )
)


assert (
    not (
        set(
            AccessibilityCausalComponentConfig
            .model_fields.keys()
        )

        & PROHIBITED_CAUSAL_CONFIG_FIELDS
    )
)


# ============================================================
# 9.36 — BUILD ARTIFACT
# ============================================================

cell9_artifact = {

    "artifact_id":
        (
            "module5_5g_ran_accessibility_"
            "causal_observability_config_v1"
        ),

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "module_scope":
        "MODULE_5_V1_5G_RAN",

    "reference_domain":
        "5G_RAN_ACCESSIBILITY",

    "role":
        "CAUSAL_OBSERVABILITY_CONFIGURATION",

    "formal_standardized_kpi_count":
        2,

    "investigation_component_count":
        7,

    "standardized_metric_family_count":
        len(
            configured_standardized_metrics
        ),

    "operational_metric_family_count":
        len(
            RAN_OPERATIONAL_METRIC_BINDINGS
        ),

    "total_observable_metric_family_count":
        (
            len(
                configured_standardized_metrics
            )

            + len(
                RAN_OPERATIONAL_METRIC_BINDINGS
            )
        ),

    "unique_fault_family_count":
        len(
            unique_fault_families
        ),

    "component_fault_family_count":
        component_fault_family_count,

    "operational_metric_bindings": {

        metric_name:
            binding.model_dump(
                mode="json"
            )

        for metric_name, binding
        in sorted(
            RAN_OPERATIONAL_METRIC_BINDINGS.items()
        )
    },

    "component_configs": {

        component_id:
            config.model_dump(
                mode="json"
            )

        for component_id, config
        in sorted(
            ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS.items()
        )
    },

    "topology_readiness": {

        "required_scopes": [

            scope.value

            for scope
            in sorted(

                required_topology_scopes_global,

                key=lambda scope:
                    scope.value,
            )
        ],

        "cell8_supported_scopes": [

            scope.value

            for scope, supported
            in sorted(

                CELL8_TOPOLOGY_SCOPE_SUPPORT.items(),

                key=lambda item:
                    item[0].value,
            )

            if supported
        ],

        "cell8_missing_scopes": [

            scope.value

            for scope
            in missing_cell8_topology_scopes
        ],

        "cell8_revision_required":
            (
                CELL8_CAUSAL_TOPOLOGY_REVISION_REQUIRED
            ),

        "telemetry_generation_ready":
            TELEMETRY_GENERATION_READY,
    },

    "provenance": {

        "cell7_semantic_sha256":
            cell7_semantic_sha,

        "cell8_semantic_sha256":
            cell8_semantic_sha,

        "reference_notebook":
            CAUSAL_REFERENCE_NOTEBOOK,

        "reference_notebook_sha256":
            CAUSAL_REFERENCE_NOTEBOOK_SHA256,
    },

    "design_constraints": {

        "telemetry_generated":
            False,

        "incident_generated":
            False,

        "fault_injected":
            False,

        "hidden_ground_truth_exposed":
            False,

        "ml_inference_included":
            False,

        "operational_metrics_claimed_as_3gpp_standard":
            False,

        "fault_families_require_observable_evidence":
            True,
    },

    "design_notes": [

        (
            "The standardized Cell 7 PM registry "
            "remains the formula-evidence layer."
        ),

        (
            "Cell 9 adds operational evidence required "
            "to distinguish plausible causes of "
            "Accessibility degradation."
        ),

        (
            "Fault-family definitions belong to the "
            "design/evaluation configuration and are "
            "not exposed as telemetry fields."
        ),

        (
            "A synthetic incident must not enter the "
            "formal benchmark unless its cause can be "
            "distinguished from competing hypotheses "
            "using observable evidence."
        ),

        (
            "Cell 8 currently lacks several explicit "
            "shared processing and interface identities; "
            "these must be added before healthy telemetry "
            "generation begins."
        ),
    ],
}


# ============================================================
# 9.37 — SEMANTIC HASH
# ============================================================

def canonical_json_sha256_cell9(
    data
):

    canonical = json.dumps(

        data,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

    ).encode(
        "utf-8"
    )


    return hashlib.sha256(
        canonical
    ).hexdigest()


cell9_semantic_content = {

    key: value

    for key, value
    in cell9_artifact.items()

    if key != "created_utc"
}


cell9_semantic_sha = (

    canonical_json_sha256_cell9(
        cell9_semantic_content
    )
)


cell9_artifact[
    "semantic_sha256"
] = cell9_semantic_sha


# ============================================================
# 9.38 — SAVE ARTIFACT
# ============================================================

telemetry_artifact_dir = (

    PATHS["kpi"].parent

    / "telemetry"
)


telemetry_artifact_dir.mkdir(

    parents=True,

    exist_ok=True,
)


cell9_path = (

    telemetry_artifact_dir

    / (
        "module5_5g_ran_accessibility_"
        "causal_observability_config_v1.json"
    )
)


if cell9_path.exists():

    raise FileExistsError(

        "Cell 9 V1 causal-observability artifact "
        "already exists. Create an explicit revision "
        "rather than silently overwriting it."
    )


with open(

    cell9_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        cell9_artifact,

        f,

        indent=2,

        ensure_ascii=False,
    )


cell9_artifact_sha = (

    sha256_file(
        cell9_path
    )
)


# ============================================================
# 9.39 — FINAL VALIDATIONS
# ============================================================

assert (
    len(
        ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS
    )
    == 7
)


assert (
    len(
        configured_standardized_metrics
    )
    == 21
)


assert (
    len(
        RAN_OPERATIONAL_METRIC_BINDINGS
    )
    == 49
)


assert (
    (
        len(
            configured_standardized_metrics
        )

        + len(
            RAN_OPERATIONAL_METRIC_BINDINGS
        )
    )
    == 70
)


assert (
    len(
        unique_fault_families
    )
    == 37
)


assert (
    component_fault_family_count
    == 58
)


assert (
    CELL8_CAUSAL_TOPOLOGY_REVISION_REQUIRED
    is True
)


assert (
    TELEMETRY_GENERATION_READY
    is False
)


# ============================================================
# 9.40 — SUMMARY OUTPUT
# ============================================================

print(
    "=" * 108
)

print(
    "MODULE 5 — CELL 09 "
    "5G RAN ACCESSIBILITY CAUSAL OBSERVABILITY CONFIGURATION"
)

print(
    "=" * 108
)


print(
    "\nReference scope"
)

print("-" * 108)

print(
    "Formal TS 28.554 RAN KPIs        : "
    "2"
)

print(
    "Investigation components          : "
    f"{len(ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS)}"
)

print(
    "RACH classification               : "
    "SUPPORTING PRE-RRC DIAGNOSTIC"
)


print(
    "\nObservable metric catalogue"
)

print("-" * 108)

print(
    "Standardized metric families      : "
    f"{len(configured_standardized_metrics)}"
)

print(
    "Operational metric families       : "
    f"{len(RAN_OPERATIONAL_METRIC_BINDINGS)}"
)

print(
    "Total observable metric families  : "
    f"{len(configured_standardized_metrics) + len(RAN_OPERATIONAL_METRIC_BINDINGS)}"
)


print(
    "\nCausal coverage"
)

print("-" * 108)

print(
    "Unique fault-family concepts      : "
    f"{len(unique_fault_families)}"
)

print(
    "Component/fault-family mappings   : "
    f"{component_fault_family_count}"
)

print(
    "Unresolved evidence metrics       : "
    "0"
)

print(
    "Unused operational metrics        : "
    f"{len(unused_operational_metrics)}"
)


print(
    "\nComponent observability"
)

print("-" * 108)


for (
    component_id,
    config
) in (
    ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS.items()
):

    print(

        f"{component_id:<32} : "

        f"{len(config.standardized_formula_metrics):>2} standard | "

        f"{len(config.required_operational_metrics):>2} operational | "

        f"{len(config.fault_evidence_map):>2} fault families"
    )


print(
    "\nCell 8 topology readiness"
)

print("-" * 108)

print(
    "Existing topology supports"
)

for scope in sorted(

    [

        scope

        for scope, supported
        in CELL8_TOPOLOGY_SCOPE_SUPPORT.items()

        if supported
    ],

    key=lambda scope:
        scope.value,
):

    print(
        f"  OK      {scope.value}"
    )


print(
    "\nRequired topology additions"
)

for scope in (
    missing_cell8_topology_scopes
):

    print(
        f"  MISSING {scope.value}"
    )


print(
    "\nReadiness decision"
)

print("-" * 108)

print(
    "Cell 8 topology revision required : "
    f"{CELL8_CAUSAL_TOPOLOGY_REVISION_REQUIRED}"
)

print(
    "Healthy telemetry generation ready: "
    f"{TELEMETRY_GENERATION_READY}"
)


print(
    "\nGeneration boundaries"
)

print("-" * 108)

print(
    "Telemetry generated               : "
    "NO"
)

print(
    "Incidents generated               : "
    "NO"
)

print(
    "Faults injected                   : "
    "NO"
)

print(
    "Hidden root-cause truth exposed   : "
    "NO"
)

print(
    "ML inference included             : "
    "NO"
)


print(
    "\nProvenance"
)

print("-" * 108)

print(
    "Cell 7 semantic SHA               : "
    f"{cell7_semantic_sha}"
)

print(
    "Cell 8 semantic SHA               : "
    f"{cell8_semantic_sha}"
)

print(
    "Reference notebook SHA            : "
    f"{CAUSAL_REFERENCE_NOTEBOOK_SHA256}"
)


print(
    "\nArtifact"
)

print("-" * 108)

print(
    f"Path                               : "
    f"{cell9_path}"
)

print(
    f"Artifact SHA                       : "
    f"{cell9_artifact_sha}"
)

print(
    f"Semantic SHA                       : "
    f"{cell9_semantic_sha}"
)


print(
    "\n" + "=" * 108
)

print(
    "CELL 09 COMPLETE — "
    "CAUSAL OBSERVABILITY CONFIGURED; "
    "TOPOLOGY EXTENSION REQUIRED BEFORE GENERATION"
)

print(
    "=" * 108
)
````

#### Section 5 — Causal-Observability Closure

Cells 10–12 close the gaps intentionally surfaced by Cell 09. They enrich the reasoning graph, topology and observation namespace **without altering the standardized KPI/PM definitions**.

##### Cell 10 — Accessibility Causal Diagnostic Graph Extension V1.2

Extend the high-level Accessibility graph with the missing causal diagnostic domains: RAN processing/compute, paging, UE/bearer context and internal RAN interface state. Exact operational metric definitions remain in Cell 09.


````python
# ============================================================
# MODULE 5 — CELL 10
# 5G RAN ACCESSIBILITY CAUSAL DIAGNOSTIC GRAPH EXTENSION
# ============================================================
#
# Purpose:
#
#   1. Extend the Cell 06 V1.1 investigation graph using
#      the causal-observability findings frozen in Cell 09.
#
#   2. Add ONLY missing high-level diagnostic concepts.
#
#   3. Keep exact operational telemetry metrics in Cell 09.
#
#   4. Preserve the Cell 06 V1.1 artifact without mutation.
#
#
# V1.2 adds:
#
#   - RAN Processing / Compute State
#   - Paging Behaviour
#   - UE / Bearer Context State
#   - Internal RAN Interface State
#
#
# IMPORTANT:
#
#   - NO standardized KPI is added.
#   - NO standardized PM is added.
#   - NO telemetry is generated.
#   - NO fault truth is added.
#   - NO Cell 06 V1.1 object is intentionally mutated.
#
# ============================================================


from datetime import (
    datetime,
    timezone,
)

from pathlib import Path

import copy
import hashlib
import json


# ============================================================
# 5 V1.2.1 — DEPENDENCY VALIDATION
# ============================================================

REQUIRED_CELL5_V1_2_OBJECTS = [

    # Cell 4 / Cell 5
    "MeasurementDomain",
    "RANAccessibilityNodeLayer",
    "RANAlignmentType",
    "RANDependencyRelation",
    "RANAccessibilityEdge",
    "make_ran_node",

    # Cell 5 V1.1
    "RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_1",
    "RAN_ACCESSIBILITY_NODES_V1_1",
    "RAN_ACCESSIBILITY_EDGES_V1_1",
    "ran_accessibility_v1_1_semantic_sha",

    # Cell 9
    "ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS",
    "RAN_OPERATIONAL_METRIC_BINDINGS",
    "cell9_semantic_sha",

    # shared
    "sha256_file",
]


missing_cell5_v1_2_objects = [

    name

    for name
    in REQUIRED_CELL5_V1_2_OBJECTS

    if name not in globals()
]


if missing_cell5_v1_2_objects:

    raise RuntimeError(

        "Cell 10 requires the V1.1 Accessibility graph "
        "and Cell 09 to be executed first. "
        f"Missing objects: "
        f"{missing_cell5_v1_2_objects}"
    )


# ============================================================
# 5 V1.2.2 — VERIFY FROZEN PARENT BASELINES
# ============================================================

EXPECTED_CELL5_V1_1_SEMANTIC_SHA = (
    "28af15d5d97c8ccfc8f636e3175650b"
    "67d55de8b48dfb021365d3206110aee42"
)


EXPECTED_CELL9_SEMANTIC_SHA = (
    "8ccce01ee1375bbef52611a4b008ea01"
    "38bae88874993be67f7577aa0838f42a"
)


assert (
    ran_accessibility_v1_1_semantic_sha
    == EXPECTED_CELL5_V1_1_SEMANTIC_SHA
)


assert (
    cell9_semantic_sha
    == EXPECTED_CELL9_SEMANTIC_SHA
)


# ============================================================
# 5 V1.2.3 — COPY FROZEN V1.1
# ============================================================

RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_2 = (
    copy.deepcopy(
        RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_1
    )
)


RAN_ACCESSIBILITY_NODES_V1_2 = (
    copy.deepcopy(
        RAN_ACCESSIBILITY_NODES_V1_1
    )
)


RAN_ACCESSIBILITY_EDGES_V1_2 = (
    copy.deepcopy(
        RAN_ACCESSIBILITY_EDGES_V1_1
    )
)


# ============================================================
# 5 V1.2.4 — VERIFY PARENT STRUCTURE
# ============================================================

assert (
    len(
        RAN_ACCESSIBILITY_NODES_V1_1
    )
    == 37
)


assert (
    len(
        RAN_ACCESSIBILITY_EDGES_V1_1
    )
    == 49
)


assert (
    len(
        RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_1
    )
    == 2
)


# ============================================================
# 5 V1.2.5 — VERIFY EXISTING DIAGNOSTIC FOUNDATION
# ============================================================

EXPECTED_EXISTING_DIAGNOSTICS = {

    "diagnostic::rach_access_success",

    "diagnostic::radio_condition",

    "diagnostic::cell_resource_load",

    "diagnostic::ran_alarms_events",

    "diagnostic::ran_configuration",

    "diagnostic::transport_boundary",
}


assert (

    EXPECTED_EXISTING_DIAGNOSTICS

    <= set(
        RAN_ACCESSIBILITY_NODES_V1_2.keys()
    )
)


# ============================================================
# 5 V1.2.6 — NEW DIAGNOSTIC NODE DEFINITIONS
# ============================================================

NEW_CAUSAL_DIAGNOSTIC_NODES = {


    # --------------------------------------------------------
    # PROCESSING / COMPUTE
    # --------------------------------------------------------

    "diagnostic::ran_processing_state":

        make_ran_node(

            node_id=(
                "diagnostic::ran_processing_state"
            ),

            name=(
                "RAN Processing / Compute State"
            ),

            layer=(
                RANAccessibilityNodeLayer
                .OPERATIONAL_DIAGNOSTIC
            ),

            alignment_type=(
                RANAlignmentType
                .OPERATIONAL_DIAGNOSTIC
            ),

            description=(
                "Processing and compute evidence used "
                "to determine whether DU, CU-CP or "
                "CU-UP resource pressure contributes "
                "to Accessibility degradation."
            ),

            domains=[
                MeasurementDomain.RAN
            ],

            metadata={

                "introduced_revision":
                    "CELL_5_V1_2",

                "discovered_by":
                    "CELL_9_CAUSAL_OBSERVABILITY",

                "cell9_evidence_categories": [

                    "DU processing",

                    "CU-CP processing",

                    "CU-UP processing",
                ],

                "metric_examples": [

                    "processing.du_cpu_pct",

                    "processing.du_queue_depth",

                    "processing.cu_cp_cpu_pct",

                    (
                        "processing."
                        "cu_cp_control_delay_ms"
                    ),

                    "processing.cu_up_cpu_pct",
                ],

                "standardized_kpi":
                    False,
            },
        ),


    # --------------------------------------------------------
    # PAGING
    # --------------------------------------------------------

    "diagnostic::paging_behaviour":

        make_ran_node(

            node_id=(
                "diagnostic::paging_behaviour"
            ),

            name=(
                "Paging Behaviour"
            ),

            layer=(
                RANAccessibilityNodeLayer
                .OPERATIONAL_DIAGNOSTIC
            ),

            alignment_type=(
                RANAlignmentType
                .OPERATIONAL_DIAGNOSTIC
            ),

            description=(
                "Paging evidence used where "
                "Accessibility degradation is "
                "associated with mobile-terminated "
                "access or inactive-to-connected "
                "activation behaviour."
            ),

            domains=[
                MeasurementDomain.RAN
            ],

            metadata={

                "introduced_revision":
                    "CELL_5_V1_2",

                "discovered_by":
                    "CELL_9_CAUSAL_OBSERVABILITY",

                "metric_examples": [

                    "paging.attempts",

                    "paging.responses",

                    "paging.load_index",
                ],

                "applicability":
                    (
                        "CAUSE_OR_SCENARIO_DEPENDENT"
                    ),

                "standardized_kpi":
                    False,
            },
        ),


    # --------------------------------------------------------
    # UE / BEARER CONTEXT
    # --------------------------------------------------------

    "diagnostic::ue_bearer_context_state":

        make_ran_node(

            node_id=(
                "diagnostic::"
                "ue_bearer_context_state"
            ),

            name=(
                "UE / Bearer Context State"
            ),

            layer=(
                RANAccessibilityNodeLayer
                .OPERATIONAL_DIAGNOSTIC
            ),

            alignment_type=(
                RANAlignmentType
                .OPERATIONAL_DIAGNOSTIC
            ),

            description=(
                "Context-state evidence used to "
                "distinguish Resume and bearer "
                "Resume degradation caused by "
                "lost, unavailable or inconsistent "
                "UE or bearer context."
            ),

            domains=[
                MeasurementDomain.RAN
            ],

            metadata={

                "introduced_revision":
                    "CELL_5_V1_2",

                "discovered_by":
                    "CELL_9_CAUSAL_OBSERVABILITY",

                "metric_examples": [

                    (
                        "context."
                        "ue_context_availability_pct"
                    ),

                    (
                        "context."
                        "bearer_context_availability_pct"
                    ),
                ],

                "standardized_kpi":
                    False,
            },
        ),


    # --------------------------------------------------------
    # INTERNAL RAN INTERFACES
    # --------------------------------------------------------

    "diagnostic::internal_ran_interface_state":

        make_ran_node(

            node_id=(
                "diagnostic::"
                "internal_ran_interface_state"
            ),

            name=(
                "Internal RAN Interface State"
            ),

            layer=(
                RANAccessibilityNodeLayer
                .OPERATIONAL_DIAGNOSTIC
            ),

            alignment_type=(
                RANAlignmentType
                .OPERATIONAL_DIAGNOSTIC
            ),

            description=(
                "Internal RAN interface evidence "
                "used to determine whether F1 or E1 "
                "state, latency or packet loss "
                "contributes to Accessibility "
                "degradation."
            ),

            domains=[

                MeasurementDomain.RAN,

                MeasurementDomain.TRANSPORT,
            ],

            metadata={

                "introduced_revision":
                    "CELL_5_V1_2",

                "discovered_by":
                    "CELL_9_CAUSAL_OBSERVABILITY",

                "interface_scope": [

                    "F1",

                    "E1",
                ],

                "metric_examples": [

                    "transport.f1_latency_ms",

                    (
                        "transport."
                        "f1_packet_loss_pct"
                    ),

                    "transport.f1_state",

                    "transport.e1_latency_ms",

                    "transport.e1_state",
                ],

                "standardized_kpi":
                    False,
            },
        ),
}


# ============================================================
# 5 V1.2.7 — ADD NEW NODES NON-DESTRUCTIVELY
# ============================================================

for (
    node_id,
    node
) in (
    NEW_CAUSAL_DIAGNOSTIC_NODES.items()
):

    assert (
        node_id
        not in RAN_ACCESSIBILITY_NODES_V1_1
    ), (
        "Cell 5 V1.1 unexpectedly already "
        f"contains {node_id}"
    )


    RAN_ACCESSIBILITY_NODES_V1_2[
        node_id
    ] = node


# ============================================================
# 5 V1.2.8 — EXISTING DIAGNOSTIC ↔ CELL 9 METADATA
# ============================================================
#
# Cell 5 remains conceptual.
#
# We add only metric-family hints rather than 49 measurement
# nodes.
#
# ============================================================

EXISTING_DIAGNOSTIC_CELL9_MAP = {

    "diagnostic::rach_access_success": [

        "RACH.Preamble*",

        "rach.*",
    ],

    "diagnostic::radio_condition": [

        "radio.*",
    ],

    "diagnostic::cell_resource_load": [

        "resource.*",

        "load.*",
    ],

    "diagnostic::ran_alarms_events": [

        "event.*",
    ],

    "diagnostic::ran_configuration": [

        "config.*",
    ],

    "diagnostic::transport_boundary": [

        "transport.n2_*",

        "sctp.*",

        "boundary.amf_*",
    ],
}


for (
    node_id,
    metric_families
) in (
    EXISTING_DIAGNOSTIC_CELL9_MAP.items()
):

    node_copy = (
        RAN_ACCESSIBILITY_NODES_V1_2[
            node_id
        ]
    )


    node_copy.metadata[
        "cell9_causal_observability_metric_families"
    ] = metric_families


    node_copy.metadata[
        "cell9_semantic_sha256"
    ] = cell9_semantic_sha


# ============================================================
# 5 V1.2.9 — V1.2 EDGE HELPER
# ============================================================

def add_ran_edge_v1_2(
    source_node_id,
    target_node_id,
    relation,
    rationale,
):

    assert (
        source_node_id
        in RAN_ACCESSIBILITY_NODES_V1_2
    )


    assert (
        target_node_id
        in RAN_ACCESSIBILITY_NODES_V1_2
    )


    candidate = (
        RANAccessibilityEdge(

            source_node_id=source_node_id,

            target_node_id=target_node_id,

            relation=relation,

            rationale=rationale,
        )
    )


    duplicate = any(

        (
            edge.source_node_id
            == candidate.source_node_id
        )

        and

        (
            edge.target_node_id
            == candidate.target_node_id
        )

        and

        (
            edge.relation
            == candidate.relation
        )

        for edge
        in RAN_ACCESSIBILITY_EDGES_V1_2
    )


    if not duplicate:

        RAN_ACCESSIBILITY_EDGES_V1_2.append(
            candidate
        )


# ============================================================
# 5 V1.2.10 — COMPONENT IDS
# ============================================================

CELL5_COMPONENT_IDS = {

    "RRC":
        "component::rrc_setup_sr",

    "NG":
        "component::ng_signalling_sr",

    "INITIAL_DRB":
        "component::initial_drb_setup_sr",

    "DRB":
        "component::drb_setup_sr",

    "RRC_RESUME":
        "component::rrc_resume_sr",

    "DRB_RESUME":
        "component::drb_resume_sr",
}


for component_id in (
    CELL5_COMPONENT_IDS.values()
):

    assert (
        component_id
        in RAN_ACCESSIBILITY_NODES_V1_2
    )


# ============================================================
# 5 V1.2.11 — PROCESSING STATE EDGES
# ============================================================
#
# Processing pressure may affect control-plane or bearer
# procedures at multiple stages.
#
# ============================================================

for component_id in [

    CELL5_COMPONENT_IDS["RRC"],

    CELL5_COMPONENT_IDS["NG"],

    CELL5_COMPONENT_IDS[
        "INITIAL_DRB"
    ],

    CELL5_COMPONENT_IDS["DRB"],

    CELL5_COMPONENT_IDS[
        "RRC_RESUME"
    ],

    CELL5_COMPONENT_IDS[
        "DRB_RESUME"
    ],
]:

    add_ran_edge_v1_2(

        "diagnostic::ran_processing_state",

        component_id,

        RANDependencyRelation
        .SUPPORTS_DIAGNOSIS_OF,

        (
            "DU, CU-CP or CU-UP processing "
            "pressure may explain degradation "
            "at this Accessibility stage."
        ),
    )


# ============================================================
# 5 V1.2.12 — PAGING EDGES
# ============================================================

for component_id in [

    CELL5_COMPONENT_IDS["RRC"],

    CELL5_COMPONENT_IDS[
        "RRC_RESUME"
    ],
]:

    add_ran_edge_v1_2(

        "diagnostic::paging_behaviour",

        component_id,

        RANDependencyRelation
        .SUPPORTS_DIAGNOSIS_OF,

        (
            "Paging behaviour may contribute "
            "to mobile-terminated Accessibility "
            "or inactive-to-connected activation "
            "symptoms."
        ),
    )


# ============================================================
# 5 V1.2.13 — CONTEXT STATE EDGES
# ============================================================

for component_id in [

    CELL5_COMPONENT_IDS[
        "RRC_RESUME"
    ],

    CELL5_COMPONENT_IDS[
        "DRB_RESUME"
    ],
]:

    add_ran_edge_v1_2(

        "diagnostic::"
        "ue_bearer_context_state",

        component_id,

        RANDependencyRelation
        .SUPPORTS_DIAGNOSIS_OF,

        (
            "UE or bearer context availability "
            "is directly relevant to Resume and "
            "fallback behaviour."
        ),
    )


# ============================================================
# 5 V1.2.14 — INTERNAL INTERFACE EDGES
# ============================================================
#
# F1 and E1 relevance differs by procedure, but the high-level
# graph intentionally represents the INTERNAL RAN INTERFACE
# diagnostic domain rather than encoding individual transport
# counters here.
#
# ============================================================

for component_id in [

    CELL5_COMPONENT_IDS["RRC"],

    CELL5_COMPONENT_IDS["NG"],

    CELL5_COMPONENT_IDS[
        "INITIAL_DRB"
    ],

    CELL5_COMPONENT_IDS["DRB"],

    CELL5_COMPONENT_IDS[
        "RRC_RESUME"
    ],

    CELL5_COMPONENT_IDS[
        "DRB_RESUME"
    ],
]:

    add_ran_edge_v1_2(

        "diagnostic::"
        "internal_ran_interface_state",

        component_id,

        RANDependencyRelation
        .SUPPORTS_DIAGNOSIS_OF,

        (
            "F1/E1 interface state may provide "
            "supporting evidence for degradation "
            "at this Accessibility stage."
        ),
    )


# ============================================================
# 5 V1.2.15 — NON-DESTRUCTIVE VALIDATION
# ============================================================

assert (
    len(
        RAN_ACCESSIBILITY_NODES_V1_1
    )
    == 37
)


assert (
    len(
        RAN_ACCESSIBILITY_EDGES_V1_1
    )
    == 49
)


assert (
    len(
        RAN_ACCESSIBILITY_NODES_V1_2
    )
    == 41
)


assert (
    len(
        RAN_ACCESSIBILITY_EDGES_V1_2
    )
    == 65
)


# ============================================================
# 5 V1.2.16 — FORMAL KPI / MEASUREMENT PRESERVATION
# ============================================================

assert (
    set(
        RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_2
        .keys()
    )
    == set(
        RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_1
        .keys()
    )
)


assert (
    len(
        RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_2
    )
    == 2
)


v1_1_measurement_nodes = {

    node_id

    for node_id, node
    in RAN_ACCESSIBILITY_NODES_V1_1.items()

    if (
        node.layer
        == RANAccessibilityNodeLayer
        .MEASUREMENT
    )
}


v1_2_measurement_nodes = {

    node_id

    for node_id, node
    in RAN_ACCESSIBILITY_NODES_V1_2.items()

    if (
        node.layer
        == RANAccessibilityNodeLayer
        .MEASUREMENT
    )
}


assert (
    v1_2_measurement_nodes
    == v1_1_measurement_nodes
)


assert (
    len(
        v1_2_measurement_nodes
    )
    == 18
)


# ============================================================
# 5 V1.2.17 — DIAGNOSTIC COVERAGE VALIDATION
# ============================================================

v1_2_diagnostic_nodes = {

    node_id

    for node_id, node
    in RAN_ACCESSIBILITY_NODES_V1_2.items()

    if (
        node.layer
        == RANAccessibilityNodeLayer
        .OPERATIONAL_DIAGNOSTIC
    )
}


EXPECTED_V1_2_DIAGNOSTICS = (

    EXPECTED_EXISTING_DIAGNOSTICS

    | {

        "diagnostic::ran_processing_state",

        "diagnostic::paging_behaviour",

        (
            "diagnostic::"
            "ue_bearer_context_state"
        ),

        (
            "diagnostic::"
            "internal_ran_interface_state"
        ),
    }
)


assert (
    v1_2_diagnostic_nodes
    == EXPECTED_V1_2_DIAGNOSTICS
)


assert (
    len(
        v1_2_diagnostic_nodes
    )
    == 10
)


# ============================================================
# 5 V1.2.18 — CELL 9 CONFIGURATION LINK VALIDATION
# ============================================================

assert (
    len(
        ACCESSIBILITY_CAUSAL_COMPONENT_CONFIGS
    )
    == 7
)


assert (
    len(
        RAN_OPERATIONAL_METRIC_BINDINGS
    )
    == 49
)


# Cell 5 remains conceptual:
#
# none of the Cell 9 operational metric names should have been
# inserted as Cell 5 measurement nodes.

cell5_v1_2_measurement_names = {

    node.metadata.get(
        "measurement_name"
    )

    for node
    in RAN_ACCESSIBILITY_NODES_V1_2.values()

    if (
        node.layer
        == RANAccessibilityNodeLayer
        .MEASUREMENT
    )
}


cell5_v1_2_measurement_names.discard(
    None
)


unexpected_operational_metric_nodes = (

    cell5_v1_2_measurement_names

    & set(
        RAN_OPERATIONAL_METRIC_BINDINGS.keys()
    )
)


assert (
    len(
        unexpected_operational_metric_nodes
    )
    == 0
)


# ============================================================
# 5 V1.2.19 — LAYER + DOMAIN COUNTS
# ============================================================

layer_counts_v1_2 = {}


for node in (
    RAN_ACCESSIBILITY_NODES_V1_2.values()
):

    layer_counts_v1_2[
        node.layer.value
    ] = (

        layer_counts_v1_2.get(
            node.layer.value,
            0,
        )

        + 1
    )


domain_counts_v1_2 = {}


for node in (
    RAN_ACCESSIBILITY_NODES_V1_2.values()
):

    for domain in node.domains:

        domain_counts_v1_2[
            domain.value
        ] = (

            domain_counts_v1_2.get(
                domain.value,
                0,
            )

            + 1
        )


assert (
    layer_counts_v1_2
    == {

        "KPI": 2,

        "COMPONENT": 6,

        "MEASUREMENT": 18,

        "PROCEDURE": 5,

        "OPERATIONAL_DIAGNOSTIC": 10,
    }
)


# ============================================================
# 5 V1.2.20 — BUILD VERSIONED ARTIFACT
# ============================================================

cell5_v1_2_artifact = {

    "artifact_id":
        (
            "module5_5g_ran_accessibility_"
            "reference_domain_v1_2"
        ),

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "revision":
        "V1.2",

    "parent_revision":
        "V1.1",

    "revision_role":
        (
            "CAUSAL_DIAGNOSTIC_GRAPH_EXTENSION"
        ),

    "reference_domain":
        "5G RAN Accessibility",

    "reference_kpis": {

        kpi_id:
            kpi.model_dump(
                mode="json"
            )

        for kpi_id, kpi
        in (
            RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_2
            .items()
        )
    },

    "nodes": {

        node_id:
            node.model_dump(
                mode="json"
            )

        for node_id, node
        in (
            RAN_ACCESSIBILITY_NODES_V1_2
            .items()
        )
    },

    "edges": [

        edge.model_dump(
            mode="json"
        )

        for edge
        in RAN_ACCESSIBILITY_EDGES_V1_2
    ],

    "revision_delta": {

        "nodes_added":
            4,

        "edges_added":
            16,

        "standardized_kpis_added":
            0,

        "standardized_measurements_added":
            0,

        "operational_diagnostics_added": [

            (
                "RAN Processing / "
                "Compute State"
            ),

            "Paging Behaviour",

            "UE / Bearer Context State",

            "Internal RAN Interface State",
        ],
    },

    "provenance": {

        "parent_cell5_v1_1_semantic_sha256":
            (
                ran_accessibility_v1_1_semantic_sha
            ),

        "cell9_causal_observability_semantic_sha256":
            cell9_semantic_sha,
    },

    "design_constraints": {

        "parent_v1_1_mutated":
            False,

        "formal_kpi_set_changed":
            False,

        "standardized_measurement_set_changed":
            False,

        "cell9_operational_metrics_embedded_as_measurement_nodes":
            False,

        "hidden_ground_truth_exposed":
            False,
    },
}


# ============================================================
# 5 V1.2.21 — SEMANTIC HASH
# ============================================================

def canonical_json_sha256_cell5_v1_2(
    data
):

    canonical = json.dumps(

        data,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

    ).encode(
        "utf-8"
    )


    return hashlib.sha256(
        canonical
    ).hexdigest()


cell5_v1_2_semantic_content = {

    key: value

    for key, value
    in cell5_v1_2_artifact.items()

    if key != "created_utc"
}


ran_accessibility_v1_2_semantic_sha = (

    canonical_json_sha256_cell5_v1_2(

        cell5_v1_2_semantic_content
    )
)


cell5_v1_2_artifact[
    "semantic_sha256"
] = (
    ran_accessibility_v1_2_semantic_sha
)


# ============================================================
# 5 V1.2.22 — SAVE ARTIFACT
# ============================================================

cell5_v1_2_path = Path(

    "/content/module5_artifacts/kpi/"
    "module5_5g_ran_accessibility_"
    "reference_domain_v1_2.json"
)


if cell5_v1_2_path.exists():

    raise FileExistsError(

        "Cell 5 V1.2 artifact already exists. "
        "Create an explicit later revision rather "
        "than silently overwriting it."
    )


with open(

    cell5_v1_2_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        cell5_v1_2_artifact,

        f,

        indent=2,

        ensure_ascii=False,
    )


cell5_v1_2_artifact_sha = (

    sha256_file(
        cell5_v1_2_path
    )
)


# ============================================================
# 5 V1.2.23 — FINAL SUMMARY
# ============================================================

print(
    "=" * 108
)

print(
    "MODULE 5 — CELL 10 "
    "CAUSAL DIAGNOSTIC GRAPH EXTENSION"
)

print(
    "=" * 108
)


print(
    "\nRevision provenance"
)

print("-" * 108)

print(
    "Parent graph                     : "
    "Cell 5 V1.1"
)

print(
    "Validation source                : "
    "Cell 9 V1"
)

print(
    "Parent V1.1 preserved             : "
    "YES"
)

print(
    "Parent semantic SHA              : "
    f"{ran_accessibility_v1_1_semantic_sha}"
)

print(
    "Cell 9 semantic SHA              : "
    f"{cell9_semantic_sha}"
)


print(
    "\nStructural delta"
)

print("-" * 108)

print(
    "V1.1 nodes                       : "
    f"{len(RAN_ACCESSIBILITY_NODES_V1_1)}"
)

print(
    "V1.2 nodes                       : "
    f"{len(RAN_ACCESSIBILITY_NODES_V1_2)}"
)

print(
    "Nodes added                      : "
    f"{len(RAN_ACCESSIBILITY_NODES_V1_2) - len(RAN_ACCESSIBILITY_NODES_V1_1)}"
)

print(
    "V1.1 edges                       : "
    f"{len(RAN_ACCESSIBILITY_EDGES_V1_1)}"
)

print(
    "V1.2 edges                       : "
    f"{len(RAN_ACCESSIBILITY_EDGES_V1_2)}"
)

print(
    "Edges added                      : "
    f"{len(RAN_ACCESSIBILITY_EDGES_V1_2) - len(RAN_ACCESSIBILITY_EDGES_V1_1)}"
)


print(
    "\nNew causal diagnostic domains"
)

print("-" * 108)

for diagnostic_name in [

    "RAN Processing / Compute State",

    "Paging Behaviour",

    "UE / Bearer Context State",

    "Internal RAN Interface State",
]:

    print(
        diagnostic_name
    )


print(
    "\nLayer counts"
)

print("-" * 108)

for (
    layer_name,
    count
) in (
    layer_counts_v1_2.items()
):

    print(
        f"{layer_name:<32} : "
        f"{count}"
    )


print(
    "\nStandards preservation"
)

print("-" * 108)

print(
    "Formal RAN Accessibility KPIs    : "
    f"{len(RAN_ACCESSIBILITY_REFERENCE_KPIS_V1_2)}"
)

print(
    "Standardized PM nodes            : "
    f"{len(v1_2_measurement_nodes)}"
)

print(
    "Operational diagnostic nodes     : "
    f"{len(v1_2_diagnostic_nodes)}"
)

print(
    "Formal KPI set changed            : "
    "NO"
)

print(
    "Standardized PM set changed       : "
    "NO"
)

print(
    "Cell 9 metrics embedded as PMs    : "
    "NO"
)


print(
    "\nGeneration boundaries"
)

print("-" * 108)

print(
    "Telemetry generated               : "
    "NO"
)

print(
    "Faults injected                   : "
    "NO"
)

print(
    "Hidden ground truth exposed       : "
    "NO"
)


print(
    "\nArtifact"
)

print("-" * 108)

print(
    f"Path                               : "
    f"{cell5_v1_2_path}"
)

print(
    f"Artifact SHA                       : "
    f"{cell5_v1_2_artifact_sha}"
)

print(
    f"Semantic SHA                       : "
    f"{ran_accessibility_v1_2_semantic_sha}"
)


print(
    "\n" + "=" * 108
)

print(
    "CELL 10 COMPLETE — "
    "CAUSAL DIAGNOSTIC GRAPH READY"
)

print(
    "=" * 108
)
````

###### Cell 10 — Observation

Cell 10 extends the Accessibility reasoning graph with four additional causal diagnostic domains identified by the observability design.

```text
Graph nodes              : 37 → 41
Graph edges              : 49 → 65
Formal Accessibility KPIs: 2
Standardized PM nodes    : 18
Operational diagnostics  : 10
```

The revision preserves the standardized KPI and PM definitions while adding the causal concepts required for processing, paging, context and internal-interface investigation.

> **The reasoning graph now aligns with the causal-observability model without misclassifying operational telemetry as standardized PM evidence.**

##### Cell 11 — Synthetic 5G RAN Causal Topology Extension V1.1

Add the shared and local causal entities required for realistic correlation: CU-CP, CU-UP, N2 path, AMF-facing boundary peer, F1 path and E1 path.


````python
# ============================================================
# MODULE 5 — CELL 11
# SYNTHETIC 5G RAN CAUSAL TOPOLOGY EXTENSION
# ============================================================
#
# Purpose:
#
#   1. Extend the Cell 08 V1 synthetic topology with
#      the shared processing and interface identities required
#      by the Cell 09 causal-observability configuration.
#
#   2. Add explicit:
#
#         CU-CP
#         CU-UP
#         N2 path
#         AMF-facing peer
#         F1 path
#         E1 path
#
#   3. Preserve the original 10-site × 3-sector topology.
#
#   4. Preserve the Cell 08 V1 artifact without mutation.
#
#
# IMPORTANT:
#
#   - NO telemetry values generated.
#
#   - NO incident generated.
#
#   - NO fault injected.
#
#   - NO hidden root-cause truth.
#
#   - AMF_PEER represents only a RAN-observed boundary peer.
#     It does NOT model internal 5GC behaviour.
#
#   - N2 / F1 / E1 path identities are synthetic logical
#     observability scopes, not vendor-specific deployment
#     topology claims.
#
# ============================================================


from datetime import (
    datetime,
    timezone,
)

from typing import (
    Dict,
    List,
    Optional,
)

from pydantic import (
    BaseModel,
    ConfigDict,
)

import copy
import hashlib
import json


# ============================================================
# 8 V1.1.1 — DEPENDENCY VALIDATION
# ============================================================

REQUIRED_CELL8_V1_1_OBJECTS = [

    # Cell 7
    "TelemetryEnvironment",

    # Cell 8 V1
    "SyntheticRANCell",
    "SyntheticRANSite",
    "SYNTHETIC_RAN_TOPOLOGY",
    "SYNTHETIC_RAN_CELL_LOOKUP",
    "MODULE5_SYNTHETIC_SEED",
    "cell8_semantic_sha",

    # Cell 9
    "OperationalEntityScope",
    "required_topology_scopes_global",
    "missing_cell8_topology_scopes",
    "CELL8_CAUSAL_TOPOLOGY_REVISION_REQUIRED",
    "cell9_semantic_sha",

    # Cell 5 V1.2
    "ran_accessibility_v1_2_semantic_sha",

    # shared
    "PM_INTERVAL_MINUTES",
    "PATHS",
    "sha256_file",
]


missing_cell8_v1_1_objects = [

    name

    for name
    in REQUIRED_CELL8_V1_1_OBJECTS

    if name not in globals()
]


if missing_cell8_v1_1_objects:

    raise RuntimeError(

        "Cell 11 requires the V1 topology, "
        "Cell 09 and Cell 10 first. "
        f"Missing objects: "
        f"{missing_cell8_v1_1_objects}"
    )


# ============================================================
# 8 V1.1.2 — VERIFY FROZEN PARENT BASELINES
# ============================================================

EXPECTED_CELL8_V1_SEMANTIC_SHA = (
    "9e23832e8191e0dca58bde2c9d15631"
    "cd20539e2033e458fb39b54629f8e86d3"
)


EXPECTED_CELL9_SEMANTIC_SHA = (
    "8ccce01ee1375bbef52611a4b008ea01"
    "38bae88874993be67f7577aa0838f42a"
)


EXPECTED_CELL5_V1_2_SEMANTIC_SHA = (
    "aaeb1acf5739a9d80d7745469faa71fe"
    "142181c7e96df6bc2c9e5f8b8b031b68"
)


assert (
    cell8_semantic_sha
    == EXPECTED_CELL8_V1_SEMANTIC_SHA
)


assert (
    cell9_semantic_sha
    == EXPECTED_CELL9_SEMANTIC_SHA
)


assert (
    ran_accessibility_v1_2_semantic_sha
    == EXPECTED_CELL5_V1_2_SEMANTIC_SHA
)


assert (
    CELL8_CAUSAL_TOPOLOGY_REVISION_REQUIRED
    is True
)


assert (
    PM_INTERVAL_MINUTES
    == 15
)


# ============================================================
# 8 V1.1.3 — VERIFY EXPECTED CELL 9 TOPOLOGY GAPS
# ============================================================

EXPECTED_CAUSAL_TOPOLOGY_GAPS = {

    OperationalEntityScope.AMF_PEER,

    OperationalEntityScope.CU_CP,

    OperationalEntityScope.CU_UP,

    OperationalEntityScope.E1_PATH,

    OperationalEntityScope.F1_PATH,

    OperationalEntityScope.N2_PATH,
}


assert (
    set(
        missing_cell8_topology_scopes
    )
    == EXPECTED_CAUSAL_TOPOLOGY_GAPS
)


# ============================================================
# 8 V1.1.4 — EXTENDED CELL MODEL
# ============================================================

class SyntheticRANCellV1_1(
    SyntheticRANCell
):

    cu_cp_id: str

    cu_up_id: str

    n2_path_id: str

    amf_peer_id: str

    f1_path_id: str

    e1_path_id: str


# ============================================================
# 8 V1.1.5 — EXTENDED SITE MODEL
# ============================================================

class SyntheticRANSiteV1_1(
    SyntheticRANSite
):

    cu_cp_id: str

    cu_up_id: str

    n2_path_id: str

    amf_peer_id: str

    f1_path_id: str

    e1_path_id: str

    cells: List[
        SyntheticRANCellV1_1
    ]


# ============================================================
# 8 V1.1.6 — PROCESSING ENTITY MODEL
# ============================================================

class SyntheticRANProcessingEntity(
    BaseModel
):

    model_config = ConfigDict(
        extra="forbid"
    )

    entity_id: str

    scope: OperationalEntityScope

    cu_domain_id: str

    site_ids: List[str]

    du_ids: List[str]

    metadata: Dict[
        str,
        object,
    ] = {}


# ============================================================
# 8 V1.1.7 — AMF-FACING PEER MODEL
# ============================================================
#
# This is deliberately a RAN-facing BOUNDARY object.
#
# It does not represent internal AMF telemetry or Core truth.
#
# ============================================================

class SyntheticAMFPeerBoundary(
    BaseModel
):

    model_config = ConfigDict(
        extra="forbid"
    )

    amf_peer_id: str

    scope: OperationalEntityScope

    cu_domain_id: str

    n2_path_id: str

    metadata: Dict[
        str,
        object,
    ] = {}


# ============================================================
# 8 V1.1.8 — SYNTHETIC PATH MODEL
# ============================================================

class SyntheticRANPath(
    BaseModel
):

    model_config = ConfigDict(
        extra="forbid"
    )

    path_id: str

    scope: OperationalEntityScope

    source_entity_id: str

    target_entity_id: str

    cu_domain_id: str

    site_id: Optional[str] = None

    du_id: Optional[str] = None

    metadata: Dict[
        str,
        object,
    ] = {}


# ============================================================
# 8 V1.1.9 — EXTENDED TOPOLOGY MODEL
# ============================================================

class SyntheticRANTopologyV1_1(
    BaseModel
):

    model_config = ConfigDict(
        extra="forbid"
    )

    topology_id: str

    environment: TelemetryEnvironment

    pm_interval_minutes: int

    cu_domains: List[str]

    cu_cp_entities: Dict[
        str,
        SyntheticRANProcessingEntity,
    ]

    cu_up_entities: Dict[
        str,
        SyntheticRANProcessingEntity,
    ]

    amf_peers: Dict[
        str,
        SyntheticAMFPeerBoundary,
    ]

    n2_paths: Dict[
        str,
        SyntheticRANPath,
    ]

    f1_paths: Dict[
        str,
        SyntheticRANPath,
    ]

    e1_paths: Dict[
        str,
        SyntheticRANPath,
    ]

    sites: List[
        SyntheticRANSiteV1_1
    ]

    seed: int

    design_notes: List[str]


# ============================================================
# 8 V1.1.10 — DOMAIN-LEVEL IDENTITY PLAN
# ============================================================
#
# Five sites share each synthetic CU domain.
#
# Each CU domain receives:
#
#   1 CU-CP
#   1 CU-UP
#   1 N2 logical path
#   1 AMF-facing peer
#   1 E1 logical path
#
# Each site / DU receives:
#
#   1 F1 logical path
#
# ============================================================

CU_DOMAIN_CAUSAL_IDENTITIES = {

    "CU_DOMAIN_01": {

        "cu_cp_id":
            "CU_CP_01",

        "cu_up_id":
            "CU_UP_01",

        "n2_path_id":
            "N2_PATH_01",

        "amf_peer_id":
            "AMF_PEER_01",

        "e1_path_id":
            "E1_PATH_01",
    },


    "CU_DOMAIN_02": {

        "cu_cp_id":
            "CU_CP_02",

        "cu_up_id":
            "CU_UP_02",

        "n2_path_id":
            "N2_PATH_02",

        "amf_peer_id":
            "AMF_PEER_02",

        "e1_path_id":
            "E1_PATH_02",
    },
}


assert (
    set(
        CU_DOMAIN_CAUSAL_IDENTITIES.keys()
    )
    == set(
        SYNTHETIC_RAN_TOPOLOGY
        .cu_domains
    )
)


# ============================================================
# 8 V1.1.11 — BUILD DOMAIN MEMBERSHIP MAP
# ============================================================

sites_by_cu_domain = {

    cu_domain_id: []

    for cu_domain_id
    in SYNTHETIC_RAN_TOPOLOGY.cu_domains
}


for site in (
    SYNTHETIC_RAN_TOPOLOGY.sites
):

    sites_by_cu_domain[
        site.cu_domain_id
    ].append(
        site
    )


assert (
    {
        cu_domain_id:
            len(sites)

        for cu_domain_id, sites
        in sites_by_cu_domain.items()
    }
    == {

        "CU_DOMAIN_01": 5,

        "CU_DOMAIN_02": 5,
    }
)


# ============================================================
# 8 V1.1.12 — BUILD CU-CP / CU-UP ENTITIES
# ============================================================

SYNTHETIC_CU_CP_LOOKUP_V1_1 = {}


SYNTHETIC_CU_UP_LOOKUP_V1_1 = {}


for (
    cu_domain_id,
    domain_sites
) in (
    sites_by_cu_domain.items()
):

    identity = (
        CU_DOMAIN_CAUSAL_IDENTITIES[
            cu_domain_id
        ]
    )


    site_ids = [

        site.site_id

        for site
        in domain_sites
    ]


    du_ids = [

        site.du_id

        for site
        in domain_sites
    ]


    cu_cp = (
        SyntheticRANProcessingEntity(

            entity_id=(
                identity[
                    "cu_cp_id"
                ]
            ),

            scope=(
                OperationalEntityScope
                .CU_CP
            ),

            cu_domain_id=(
                cu_domain_id
            ),

            site_ids=(
                site_ids
            ),

            du_ids=(
                du_ids
            ),

            metadata={

                "synthetic_role":
                    "CONTROL_PLANE_PROCESSING",

                "vendor_specific":
                    False,
            },
        )
    )


    cu_up = (
        SyntheticRANProcessingEntity(

            entity_id=(
                identity[
                    "cu_up_id"
                ]
            ),

            scope=(
                OperationalEntityScope
                .CU_UP
            ),

            cu_domain_id=(
                cu_domain_id
            ),

            site_ids=(
                site_ids
            ),

            du_ids=(
                du_ids
            ),

            metadata={

                "synthetic_role":
                    "USER_PLANE_PROCESSING",

                "vendor_specific":
                    False,
            },
        )
    )


    SYNTHETIC_CU_CP_LOOKUP_V1_1[
        cu_cp.entity_id
    ] = cu_cp


    SYNTHETIC_CU_UP_LOOKUP_V1_1[
        cu_up.entity_id
    ] = cu_up


# ============================================================
# 8 V1.1.13 — BUILD N2 + AMF-BOUNDARY OBJECTS
# ============================================================

SYNTHETIC_N2_PATH_LOOKUP_V1_1 = {}


SYNTHETIC_AMF_PEER_LOOKUP_V1_1 = {}


for (
    cu_domain_id,
    identity
) in (
    CU_DOMAIN_CAUSAL_IDENTITIES.items()
):

    n2_path = SyntheticRANPath(

        path_id=(
            identity[
                "n2_path_id"
            ]
        ),

        scope=(
            OperationalEntityScope
            .N2_PATH
        ),

        source_entity_id=(
            identity[
                "cu_cp_id"
            ]
        ),

        target_entity_id=(
            identity[
                "amf_peer_id"
            ]
        ),

        cu_domain_id=(
            cu_domain_id
        ),

        metadata={

            "interface":
                "N2",

            "logical_observability_path":
                True,

            "vendor_specific":
                False,
        },
    )


    amf_peer = SyntheticAMFPeerBoundary(

        amf_peer_id=(
            identity[
                "amf_peer_id"
            ]
        ),

        scope=(
            OperationalEntityScope
            .AMF_PEER
        ),

        cu_domain_id=(
            cu_domain_id
        ),

        n2_path_id=(
            identity[
                "n2_path_id"
            ]
        ),

        metadata={

            "boundary_only":
                True,

            "internal_5gc_modelled":
                False,

            "description":
                (
                    "Synthetic RAN-observed "
                    "AMF-facing peer identity."
                ),
        },
    )


    SYNTHETIC_N2_PATH_LOOKUP_V1_1[
        n2_path.path_id
    ] = n2_path


    SYNTHETIC_AMF_PEER_LOOKUP_V1_1[
        amf_peer.amf_peer_id
    ] = amf_peer


# ============================================================
# 8 V1.1.14 — BUILD E1 PATHS
# ============================================================

SYNTHETIC_E1_PATH_LOOKUP_V1_1 = {}


for (
    cu_domain_id,
    identity
) in (
    CU_DOMAIN_CAUSAL_IDENTITIES.items()
):

    e1_path = SyntheticRANPath(

        path_id=(
            identity[
                "e1_path_id"
            ]
        ),

        scope=(
            OperationalEntityScope
            .E1_PATH
        ),

        source_entity_id=(
            identity[
                "cu_cp_id"
            ]
        ),

        target_entity_id=(
            identity[
                "cu_up_id"
            ]
        ),

        cu_domain_id=(
            cu_domain_id
        ),

        metadata={

            "interface":
                "E1",

            "logical_observability_path":
                True,

            "vendor_specific":
                False,
        },
    )


    SYNTHETIC_E1_PATH_LOOKUP_V1_1[
        e1_path.path_id
    ] = e1_path


# ============================================================
# 8 V1.1.15 — BUILD EXTENDED SITES + F1 PATHS
# ============================================================

SYNTHETIC_F1_PATH_LOOKUP_V1_1 = {}


SYNTHETIC_RAN_SITES_V1_1 = []


SYNTHETIC_RAN_CELL_LOOKUP_V1_1 = {}


SYNTHETIC_NR_CELL_CU_LOOKUP_V1_1 = {}


SYNTHETIC_NR_CELL_DU_LOOKUP_V1_1 = {}


for site in sorted(

    SYNTHETIC_RAN_TOPOLOGY.sites,

    key=lambda x:
        x.site_id,
):

    identity = (

        CU_DOMAIN_CAUSAL_IDENTITIES[
            site.cu_domain_id
        ]
    )


    # --------------------------------------------------------
    # One F1 path per DU / site
    # --------------------------------------------------------

    site_number = int(

        site.site_id.split(
            "_"
        )[-1]
    )


    f1_path_id = (
        f"F1_PATH_{site_number:03d}"
    )


    f1_path = SyntheticRANPath(

        path_id=f1_path_id,

        scope=(
            OperationalEntityScope
            .F1_PATH
        ),

        source_entity_id=(
            site.du_id
        ),

        target_entity_id=(
            identity[
                "cu_cp_id"
            ]
        ),

        cu_domain_id=(
            site.cu_domain_id
        ),

        site_id=(
            site.site_id
        ),

        du_id=(
            site.du_id
        ),

        metadata={

            "interface":
                "F1",

            "logical_observability_path":
                True,

            "vendor_specific":
                False,
        },
    )


    SYNTHETIC_F1_PATH_LOOKUP_V1_1[
        f1_path_id
    ] = f1_path


    # --------------------------------------------------------
    # Extend logical cells non-destructively
    # --------------------------------------------------------

    extended_cells = []


    for cell in site.cells:

        cell_payload = (
            cell.model_dump(
                mode="python"
            )
        )


        extended_cell = (
            SyntheticRANCellV1_1(

                **cell_payload,

                cu_cp_id=(
                    identity[
                        "cu_cp_id"
                    ]
                ),

                cu_up_id=(
                    identity[
                        "cu_up_id"
                    ]
                ),

                n2_path_id=(
                    identity[
                        "n2_path_id"
                    ]
                ),

                amf_peer_id=(
                    identity[
                        "amf_peer_id"
                    ]
                ),

                f1_path_id=(
                    f1_path_id
                ),

                e1_path_id=(
                    identity[
                        "e1_path_id"
                    ]
                ),
            )
        )


        extended_cells.append(
            extended_cell
        )


        SYNTHETIC_RAN_CELL_LOOKUP_V1_1[
            extended_cell.logical_cell_id
        ] = extended_cell


        SYNTHETIC_NR_CELL_CU_LOOKUP_V1_1[
            extended_cell.nr_cell_cu.entity_id
        ] = (
            extended_cell.nr_cell_cu
        )


        SYNTHETIC_NR_CELL_DU_LOOKUP_V1_1[
            extended_cell.nr_cell_du.entity_id
        ] = (
            extended_cell.nr_cell_du
        )


    # --------------------------------------------------------
    # Extend site non-destructively
    # --------------------------------------------------------

    site_payload = (
        site.model_dump(
            mode="python"
        )
    )


    site_payload[
        "cells"
    ] = extended_cells


    extended_site = (
        SyntheticRANSiteV1_1(

            **site_payload,

            cu_cp_id=(
                identity[
                    "cu_cp_id"
                ]
            ),

            cu_up_id=(
                identity[
                    "cu_up_id"
                ]
            ),

            n2_path_id=(
                identity[
                    "n2_path_id"
                ]
            ),

            amf_peer_id=(
                identity[
                    "amf_peer_id"
                ]
            ),

            f1_path_id=(
                f1_path_id
            ),

            e1_path_id=(
                identity[
                    "e1_path_id"
                ]
            ),
        )
    )


    SYNTHETIC_RAN_SITES_V1_1.append(
        extended_site
    )


# ============================================================
# 8 V1.1.16 — BUILD EXTENDED TOPOLOGY
# ============================================================

SYNTHETIC_RAN_TOPOLOGY_V1_1 = (

    SyntheticRANTopologyV1_1(

        topology_id=(
            "module5_synthetic_"
            "5g_ran_topology_v1_1"
        ),

        environment=(
            TelemetryEnvironment
            .SYNTHETIC_V1
        ),

        pm_interval_minutes=(
            PM_INTERVAL_MINUTES
        ),

        cu_domains=list(
            SYNTHETIC_RAN_TOPOLOGY
            .cu_domains
        ),

        cu_cp_entities=(
            SYNTHETIC_CU_CP_LOOKUP_V1_1
        ),

        cu_up_entities=(
            SYNTHETIC_CU_UP_LOOKUP_V1_1
        ),

        amf_peers=(
            SYNTHETIC_AMF_PEER_LOOKUP_V1_1
        ),

        n2_paths=(
            SYNTHETIC_N2_PATH_LOOKUP_V1_1
        ),

        f1_paths=(
            SYNTHETIC_F1_PATH_LOOKUP_V1_1
        ),

        e1_paths=(
            SYNTHETIC_E1_PATH_LOOKUP_V1_1
        ),

        sites=(
            SYNTHETIC_RAN_SITES_V1_1
        ),

        seed=(
            MODULE5_SYNTHETIC_SEED
        ),

        design_notes=[

            (
                "Cell 8 V1 topology is preserved "
                "and extended non-destructively."
            ),

            (
                "Each five-site CU domain shares "
                "one synthetic CU-CP, CU-UP, N2 "
                "path, AMF-facing peer and E1 path."
            ),

            (
                "Each DU/site has one independent "
                "synthetic F1 path."
            ),

            (
                "AMF peer identity represents only "
                "the observable RAN-to-Core boundary "
                "and does not model internal 5GC."
            ),

            (
                "N2, F1 and E1 path identities are "
                "logical diagnostic scopes, not "
                "vendor deployment claims."
            ),
        ],
    )
)


# ============================================================
# 8 V1.1.17 — PARENT V1 PRESERVATION
# ============================================================

assert (
    len(
        SYNTHETIC_RAN_TOPOLOGY.sites
    )
    == 10
)


assert (
    len(
        SYNTHETIC_RAN_CELL_LOOKUP
    )
    == 30
)


assert (
    all(

        not hasattr(
            site,
            "cu_cp_id",
        )

        for site
        in SYNTHETIC_RAN_TOPOLOGY.sites
    )
)


# ============================================================
# 8 V1.1.18 — EXTENDED TOPOLOGY COUNTS
# ============================================================

assert (
    len(
        SYNTHETIC_RAN_TOPOLOGY_V1_1
        .cu_domains
    )
    == 2
)


assert (
    len(
        SYNTHETIC_CU_CP_LOOKUP_V1_1
    )
    == 2
)


assert (
    len(
        SYNTHETIC_CU_UP_LOOKUP_V1_1
    )
    == 2
)


assert (
    len(
        SYNTHETIC_AMF_PEER_LOOKUP_V1_1
    )
    == 2
)


assert (
    len(
        SYNTHETIC_N2_PATH_LOOKUP_V1_1
    )
    == 2
)


assert (
    len(
        SYNTHETIC_E1_PATH_LOOKUP_V1_1
    )
    == 2
)


assert (
    len(
        SYNTHETIC_F1_PATH_LOOKUP_V1_1
    )
    == 10
)


assert (
    len(
        SYNTHETIC_RAN_SITES_V1_1
    )
    == 10
)


assert (
    len(
        SYNTHETIC_RAN_CELL_LOOKUP_V1_1
    )
    == 30
)


assert (
    len(
        SYNTHETIC_NR_CELL_CU_LOOKUP_V1_1
    )
    == 30
)


assert (
    len(
        SYNTHETIC_NR_CELL_DU_LOOKUP_V1_1
    )
    == 30
)


# ============================================================
# 8 V1.1.19 — DOMAIN CORRELATION VALIDATION
# ============================================================

for (
    cu_domain_id,
    domain_sites
) in (
    sites_by_cu_domain.items()
):

    identity = (

        CU_DOMAIN_CAUSAL_IDENTITIES[
            cu_domain_id
        ]
    )


    extended_domain_sites = [

        site

        for site
        in SYNTHETIC_RAN_SITES_V1_1

        if (
            site.cu_domain_id
            == cu_domain_id
        )
    ]


    assert (
        len(
            extended_domain_sites
        )
        == 5
    )


    assert (
        {
            site.cu_cp_id
            for site
            in extended_domain_sites
        }
        == {
            identity[
                "cu_cp_id"
            ]
        }
    )


    assert (
        {
            site.cu_up_id
            for site
            in extended_domain_sites
        }
        == {
            identity[
                "cu_up_id"
            ]
        }
    )


    assert (
        {
            site.n2_path_id
            for site
            in extended_domain_sites
        }
        == {
            identity[
                "n2_path_id"
            ]
        }
    )


    assert (
        {
            site.amf_peer_id
            for site
            in extended_domain_sites
        }
        == {
            identity[
                "amf_peer_id"
            ]
        }
    )


    assert (
        {
            site.e1_path_id
            for site
            in extended_domain_sites
        }
        == {
            identity[
                "e1_path_id"
            ]
        }
    )


    # Each site retains an independent F1 path.

    assert (
        len(
            {
                site.f1_path_id
                for site
                in extended_domain_sites
            }
        )
        == 5
    )


# ============================================================
# 8 V1.1.20 — PATH ENDPOINT VALIDATION
# ============================================================

for (
    path_id,
    path
) in (
    SYNTHETIC_N2_PATH_LOOKUP_V1_1.items()
):

    assert (
        path.source_entity_id
        in SYNTHETIC_CU_CP_LOOKUP_V1_1
    )


    assert (
        path.target_entity_id
        in SYNTHETIC_AMF_PEER_LOOKUP_V1_1
    )


for (
    path_id,
    path
) in (
    SYNTHETIC_E1_PATH_LOOKUP_V1_1.items()
):

    assert (
        path.source_entity_id
        in SYNTHETIC_CU_CP_LOOKUP_V1_1
    )


    assert (
        path.target_entity_id
        in SYNTHETIC_CU_UP_LOOKUP_V1_1
    )


for (
    path_id,
    path
) in (
    SYNTHETIC_F1_PATH_LOOKUP_V1_1.items()
):

    assert (
        path.target_entity_id
        in SYNTHETIC_CU_CP_LOOKUP_V1_1
    )


    assert (
        path.du_id
        is not None
    )


    assert (
        path.site_id
        is not None
    )


# ============================================================
# 8 V1.1.21 — CELL 9 TOPOLOGY READINESS RECHECK
# ============================================================

CELL8_V1_1_TOPOLOGY_SCOPE_SUPPORT = {

    OperationalEntityScope.LOGICAL_CELL:
        (
            len(
                SYNTHETIC_RAN_CELL_LOOKUP_V1_1
            )
            > 0
        ),

    OperationalEntityScope.NR_CELL_CU:
        (
            len(
                SYNTHETIC_NR_CELL_CU_LOOKUP_V1_1
            )
            > 0
        ),

    OperationalEntityScope.NR_CELL_DU:
        (
            len(
                SYNTHETIC_NR_CELL_DU_LOOKUP_V1_1
            )
            > 0
        ),

    OperationalEntityScope.SITE:
        (
            len(
                SYNTHETIC_RAN_SITES_V1_1
            )
            > 0
        ),

    OperationalEntityScope.DU:
        (
            len(
                {
                    site.du_id
                    for site
                    in SYNTHETIC_RAN_SITES_V1_1
                }
            )
            == 10
        ),

    OperationalEntityScope.CU_DOMAIN:
        (
            len(
                SYNTHETIC_RAN_TOPOLOGY_V1_1
                .cu_domains
            )
            == 2
        ),

    OperationalEntityScope.CU_CP:
        (
            len(
                SYNTHETIC_CU_CP_LOOKUP_V1_1
            )
            == 2
        ),

    OperationalEntityScope.CU_UP:
        (
            len(
                SYNTHETIC_CU_UP_LOOKUP_V1_1
            )
            == 2
        ),

    OperationalEntityScope.N2_PATH:
        (
            len(
                SYNTHETIC_N2_PATH_LOOKUP_V1_1
            )
            == 2
        ),

    OperationalEntityScope.AMF_PEER:
        (
            len(
                SYNTHETIC_AMF_PEER_LOOKUP_V1_1
            )
            == 2
        ),

    OperationalEntityScope.F1_PATH:
        (
            len(
                SYNTHETIC_F1_PATH_LOOKUP_V1_1
            )
            == 10
        ),

    OperationalEntityScope.E1_PATH:
        (
            len(
                SYNTHETIC_E1_PATH_LOOKUP_V1_1
            )
            == 2
        ),
}


missing_cell8_v1_1_topology_scopes = [

    scope

    for scope
    in required_topology_scopes_global

    if not (
        CELL8_V1_1_TOPOLOGY_SCOPE_SUPPORT[
            scope
        ]
    )
]


assert (
    len(
        missing_cell8_v1_1_topology_scopes
    )
    == 0
)


CELL8_V1_1_CAUSAL_TOPOLOGY_READY = True


TELEMETRY_GENERATION_READY_V1_1 = True


# ============================================================
# 8 V1.1.22 — HIDDEN-TRUTH SAFETY VALIDATION
# ============================================================

PROHIBITED_TOPOLOGY_FIELDS = {

    "root_cause",

    "ground_truth",

    "fault_label",

    "expected_diagnosis",

    "correct_action",

    "benchmark_answer",
}


for topology_model in [

    SyntheticRANCellV1_1,

    SyntheticRANSiteV1_1,

    SyntheticRANProcessingEntity,

    SyntheticAMFPeerBoundary,

    SyntheticRANPath,

    SyntheticRANTopologyV1_1,
]:

    assert (
        len(
            set(
                topology_model.model_fields.keys()
            )
            & PROHIBITED_TOPOLOGY_FIELDS
        )
        == 0
    )


# ============================================================
# 8 V1.1.23 — BUILD VERSIONED ARTIFACT
# ============================================================

cell8_v1_1_artifact = {

    "artifact_id":
        (
            "module5_synthetic_"
            "5g_ran_environment_v1_1"
        ),

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "revision":
        "V1.1",

    "parent_revision":
        "CELL_8_V1",

    "revision_role":
        "CAUSAL_TOPOLOGY_EXTENSION",

    "environment":
        "SYNTHETIC_V1",

    "reference_domain":
        "5G RAN Accessibility",

    "topology":
        (
            SYNTHETIC_RAN_TOPOLOGY_V1_1
            .model_dump(
                mode="json"
            )
        ),

    "revision_delta": {

        "original_sites_preserved":
            10,

        "original_logical_cells_preserved":
            30,

        "cu_cp_entities_added":
            2,

        "cu_up_entities_added":
            2,

        "n2_paths_added":
            2,

        "amf_boundary_peers_added":
            2,

        "f1_paths_added":
            10,

        "e1_paths_added":
            2,
    },

    "topology_readiness": {

        "required_scopes": [

            scope.value

            for scope
            in sorted(

                required_topology_scopes_global,

                key=lambda x:
                    x.value,
            )
        ],

        "missing_scopes":
            [],

        "causal_topology_ready":
            True,

        "telemetry_generation_ready":
            True,
    },

    "provenance": {

        "parent_cell8_v1_semantic_sha256":
            cell8_semantic_sha,

        "cell9_causal_observability_semantic_sha256":
            cell9_semantic_sha,

        "cell5_v1_2_semantic_sha256":
            (
                ran_accessibility_v1_2_semantic_sha
            ),
    },

    "design_constraints": {

        "parent_cell8_v1_mutated":
            False,

        "telemetry_generated":
            False,

        "incident_generated":
            False,

        "fault_injected":
            False,

        "hidden_ground_truth_exposed":
            False,

        "ml_inference_included":
            False,

        "internal_5gc_modelled":
            False,

        "vendor_specific_topology_claim":
            False,
    },
}


# ============================================================
# 8 V1.1.24 — SEMANTIC HASH
# ============================================================

def canonical_json_sha256_cell8_v1_1(
    data
):

    canonical = json.dumps(

        data,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

    ).encode(
        "utf-8"
    )


    return hashlib.sha256(
        canonical
    ).hexdigest()


cell8_v1_1_semantic_content = {

    key: value

    for key, value
    in cell8_v1_1_artifact.items()

    if key != "created_utc"
}


cell8_v1_1_semantic_sha = (

    canonical_json_sha256_cell8_v1_1(

        cell8_v1_1_semantic_content
    )
)


cell8_v1_1_artifact[
    "semantic_sha256"
] = (
    cell8_v1_1_semantic_sha
)


# ============================================================
# 8 V1.1.25 — SAVE ARTIFACT
# ============================================================

telemetry_artifact_dir = (

    PATHS["kpi"].parent

    / "telemetry"
)


telemetry_artifact_dir.mkdir(

    parents=True,

    exist_ok=True,
)


cell8_v1_1_path = (

    telemetry_artifact_dir

    / (
        "module5_synthetic_"
        "5g_ran_environment_v1_1.json"
    )
)


if cell8_v1_1_path.exists():

    raise FileExistsError(

        "Cell 8 V1.1 artifact already exists. "
        "Create an explicit later revision rather "
        "than silently overwriting it."
    )


with open(

    cell8_v1_1_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        cell8_v1_1_artifact,

        f,

        indent=2,

        ensure_ascii=False,
    )


cell8_v1_1_artifact_sha = (

    sha256_file(
        cell8_v1_1_path
    )
)


# ============================================================
# 8 V1.1.26 — SUMMARY
# ============================================================

print(
    "=" * 108
)

print(
    "MODULE 5 — CELL 11 "
    "SYNTHETIC 5G RAN CAUSAL TOPOLOGY EXTENSION"
)

print(
    "=" * 108
)


print(
    "\nRevision provenance"
)

print("-" * 108)

print(
    "Parent topology                  : "
    "Cell 8 V1"
)

print(
    "Validation source                : "
    "Cell 9 V1"
)

print(
    "Parent Cell 8 V1 preserved        : "
    "YES"
)

print(
    "Parent semantic SHA              : "
    f"{cell8_semantic_sha}"
)

print(
    "Cell 9 semantic SHA              : "
    f"{cell9_semantic_sha}"
)

print(
    "Cell 5 V1.2 semantic SHA         : "
    f"{ran_accessibility_v1_2_semantic_sha}"
)


print(
    "\nPreserved RAN topology"
)

print("-" * 108)

print(
    "Shared CU domains                : "
    f"{len(SYNTHETIC_RAN_TOPOLOGY_V1_1.cu_domains)}"
)

print(
    "Sites / DUs                      : "
    f"{len(SYNTHETIC_RAN_SITES_V1_1)}"
)

print(
    "Logical NR cells                 : "
    f"{len(SYNTHETIC_RAN_CELL_LOOKUP_V1_1)}"
)

print(
    "NRCellCU entities                : "
    f"{len(SYNTHETIC_NR_CELL_CU_LOOKUP_V1_1)}"
)

print(
    "NRCellDU entities                : "
    f"{len(SYNTHETIC_NR_CELL_DU_LOOKUP_V1_1)}"
)


print(
    "\nAdded causal topology"
)

print("-" * 108)

print(
    "CU-CP entities                   : "
    f"{len(SYNTHETIC_CU_CP_LOOKUP_V1_1)}"
)

print(
    "CU-UP entities                   : "
    f"{len(SYNTHETIC_CU_UP_LOOKUP_V1_1)}"
)

print(
    "N2 logical paths                 : "
    f"{len(SYNTHETIC_N2_PATH_LOOKUP_V1_1)}"
)

print(
    "AMF-facing boundary peers        : "
    f"{len(SYNTHETIC_AMF_PEER_LOOKUP_V1_1)}"
)

print(
    "F1 logical paths                 : "
    f"{len(SYNTHETIC_F1_PATH_LOOKUP_V1_1)}"
)

print(
    "E1 logical paths                 : "
    f"{len(SYNTHETIC_E1_PATH_LOOKUP_V1_1)}"
)


print(
    "\nCorrelation structure"
)

print("-" * 108)

print(
    "Sites per CU domain              : "
    "5"
)

print(
    "Shared CU-CP per CU domain       : "
    "1"
)

print(
    "Shared CU-UP per CU domain       : "
    "1"
)

print(
    "Shared N2 path per CU domain     : "
    "1"
)

print(
    "Shared AMF peer per CU domain    : "
    "1"
)

print(
    "Shared E1 path per CU domain     : "
    "1"
)

print(
    "Independent F1 path per DU       : "
    "1"
)


print(
    "\nTopology readiness"
)

print("-" * 108)

for scope in sorted(

    required_topology_scopes_global,

    key=lambda x:
        x.value,
):

    print(
        f"  OK      {scope.value}"
    )


print(
    "\nMissing required scopes          : "
    f"{len(missing_cell8_v1_1_topology_scopes)}"
)

print(
    "Causal topology ready             : "
    f"{CELL8_V1_1_CAUSAL_TOPOLOGY_READY}"
)

print(
    "Healthy telemetry generation ready: "
    f"{TELEMETRY_GENERATION_READY_V1_1}"
)


print(
    "\nGeneration boundaries"
)

print("-" * 108)

print(
    "Telemetry generated               : "
    "NO"
)

print(
    "Incidents generated               : "
    "NO"
)

print(
    "Faults injected                   : "
    "NO"
)

print(
    "Hidden ground truth exposed       : "
    "NO"
)

print(
    "Internal 5GC modelled             : "
    "NO"
)

print(
    "ML inference included             : "
    "NO"
)


print(
    "\nArtifact"
)

print("-" * 108)

print(
    f"Path                               : "
    f"{cell8_v1_1_path}"
)

print(
    f"Artifact SHA                       : "
    f"{cell8_v1_1_artifact_sha}"
)

print(
    f"Semantic SHA                       : "
    f"{cell8_v1_1_semantic_sha}"
)


print(
    "\n" + "=" * 108
)

print(
    "CELL 11 COMPLETE — "
    "CAUSAL TOPOLOGY READY FOR "
    "HEALTHY TELEMETRY GENERATION"
)

print(
    "=" * 108
)
````

##### Cell 12 — Operational Entity Observation Contract Extension V1.1

Extend the observation contract with a separate operational entity namespace rather than expanding the standardized `EntityType` enum. Build a unified registry across managed objects and causal-topology scopes.


````python
# ============================================================
# MODULE 5 — CELL 12
# OPERATIONAL ENTITY OBSERVATION CONTRACT EXTENSION
# ============================================================
#
# Purpose:
#
#   1. Extend the Cell 07 telemetry observation contract
#      so operational telemetry can reference the additional
#      causal-topology scopes introduced by Cell 11 V1.1.
#
#   2. Preserve the distinction between:
#
#          STANDARD MANAGED OBJECT
#              EntityType.NR_CELL_CU / NR_CELL_DU
#
#      and:
#
#          OPERATIONAL OBSERVABILITY SCOPE
#              CU_CP / CU_UP / N2_PATH / etc.
#
#   3. Build a deterministic observation-entity registry for
#      future healthy telemetry generation.
#
#   4. Validate that all Cell 07 standardized bindings and all
#      Cell 9 operational bindings can resolve to an available
#      topology scope.
#
#
# IMPORTANT:
#
#   - Cell 07 V1 is NOT mutated.
#
#   - EntityType is NOT expanded with synthetic operational
#     topology concepts.
#
#   - Operational metrics observed on NRCellCU / NRCellDU still
#     use STANDARD_MANAGED_OBJECT entity references.
#
#   - NO telemetry values are generated.
#
#   - NO incidents or faults are generated.
#
# ============================================================


from enum import Enum

from typing import (
    Any,
    Dict,
    List,
    Optional,
)

from datetime import (
    datetime,
    timezone,
)

from pydantic import (
    BaseModel,
    Field,
    ConfigDict,
    model_validator,
)

import hashlib
import json


# ============================================================
# 7 V1.1.1 — DEPENDENCY VALIDATION
# ============================================================

REQUIRED_CELL7_V1_1_OBJECTS = [

    # Cell 4
    "EntityType",
    "PM_INTERVAL_MINUTES",
    "PATHS",
    "sha256_file",

    # Cell 7 V1
    "RANTelemetryEntityRef",
    "RANTelemetryObservation",
    "RANTelemetryBatch",
    "RAN_TELEMETRY_METRIC_BINDINGS",
    "cell7_semantic_sha",

    # Cell 9
    "OperationalEntityScope",
    "RAN_OPERATIONAL_METRIC_BINDINGS",
    "required_topology_scopes_global",
    "cell9_semantic_sha",

    # Cell 8 V1.1
    "SYNTHETIC_RAN_TOPOLOGY_V1_1",
    "SYNTHETIC_RAN_CELL_LOOKUP_V1_1",
    "SYNTHETIC_NR_CELL_CU_LOOKUP_V1_1",
    "SYNTHETIC_NR_CELL_DU_LOOKUP_V1_1",
    "SYNTHETIC_CU_CP_LOOKUP_V1_1",
    "SYNTHETIC_CU_UP_LOOKUP_V1_1",
    "SYNTHETIC_N2_PATH_LOOKUP_V1_1",
    "SYNTHETIC_AMF_PEER_LOOKUP_V1_1",
    "SYNTHETIC_F1_PATH_LOOKUP_V1_1",
    "SYNTHETIC_E1_PATH_LOOKUP_V1_1",
    "SYNTHETIC_RAN_SITES_V1_1",
    "cell8_v1_1_semantic_sha",

    # Cell 5 V1.2
    "ran_accessibility_v1_2_semantic_sha",
]


missing_cell7_v1_1_objects = [

    name

    for name
    in REQUIRED_CELL7_V1_1_OBJECTS

    if name not in globals()
]


if missing_cell7_v1_1_objects:

    raise RuntimeError(

        "Cell 12 requires the V1 observation contract, "
        "Cell 11, Cell 09 and Cell 10. "
        f"Missing objects: "
        f"{missing_cell7_v1_1_objects}"
    )


# ============================================================
# 7 V1.1.2 — VERIFY FROZEN PARENT BASELINES
# ============================================================

EXPECTED_CELL7_V1_SEMANTIC_SHA = (
    "cc735d9a189f3346bfe7591f1f8e74ba"
    "c0abd35f2c6e7235cb91f4bdb488a53e"
)


EXPECTED_CELL8_V1_1_SEMANTIC_SHA = (
    "b12fd1cc5b208963a13caecc33d03016"
    "a612ff1f1f112686f4db435db66b421d"
)


EXPECTED_CELL9_SEMANTIC_SHA = (
    "8ccce01ee1375bbef52611a4b008ea01"
    "38bae88874993be67f7577aa0838f42a"
)


EXPECTED_CELL5_V1_2_SEMANTIC_SHA = (
    "aaeb1acf5739a9d80d7745469faa71fe"
    "142181c7e96df6bc2c9e5f8b8b031b68"
)


assert (
    cell7_semantic_sha
    == EXPECTED_CELL7_V1_SEMANTIC_SHA
)


assert (
    cell8_v1_1_semantic_sha
    == EXPECTED_CELL8_V1_1_SEMANTIC_SHA
)


assert (
    cell9_semantic_sha
    == EXPECTED_CELL9_SEMANTIC_SHA
)


assert (
    ran_accessibility_v1_2_semantic_sha
    == EXPECTED_CELL5_V1_2_SEMANTIC_SHA
)


assert (
    PM_INTERVAL_MINUTES
    == 15
)


# ============================================================
# 7 V1.1.3 — ENTITY NAMESPACE
# ============================================================

class TelemetryEntityNamespace(
    str,
    Enum,
):

    STANDARD_MANAGED_OBJECT = (
        "STANDARD_MANAGED_OBJECT"
    )

    OPERATIONAL_SCOPE = (
        "OPERATIONAL_SCOPE"
    )


# ============================================================
# 7 V1.1.4 — UNIFIED ENTITY REFERENCE
# ============================================================
#
# Exactly ONE entity namespace is active:
#
#   STANDARD_MANAGED_OBJECT
#       -> entity_type required
#
#   OPERATIONAL_SCOPE
#       -> operational_scope required
#
# ============================================================

class RANTelemetryEntityRefV1_1(
    BaseModel
):

    model_config = ConfigDict(
        extra="forbid"
    )

    entity_id: str

    namespace: (
        TelemetryEntityNamespace
    )

    entity_type: Optional[
        EntityType
    ] = None

    operational_scope: Optional[
        OperationalEntityScope
    ] = None

    parent_entity_id: Optional[
        str
    ] = None

    site_id: Optional[
        str
    ] = None

    cu_domain_id: Optional[
        str
    ] = None

    metadata: Dict[
        str,
        Any,
    ] = Field(
        default_factory=dict
    )


    @model_validator(
        mode="after"
    )
    def validate_namespace(
        self
    ):

        if (
            self.namespace
            == TelemetryEntityNamespace
            .STANDARD_MANAGED_OBJECT
        ):

            if self.entity_type is None:

                raise ValueError(

                    "STANDARD_MANAGED_OBJECT "
                    "requires entity_type."
                )


            if (
                self.operational_scope
                is not None
            ):

                raise ValueError(

                    "STANDARD_MANAGED_OBJECT "
                    "must not define "
                    "operational_scope."
                )


        elif (
            self.namespace
            == TelemetryEntityNamespace
            .OPERATIONAL_SCOPE
        ):

            if (
                self.operational_scope
                is None
            ):

                raise ValueError(

                    "OPERATIONAL_SCOPE requires "
                    "operational_scope."
                )


            if self.entity_type is not None:

                raise ValueError(

                    "OPERATIONAL_SCOPE must not "
                    "define entity_type."
                )


        return self


# ============================================================
# 7 V1.1.5 — EXTENDED OBSERVATION CONTRACT
# ============================================================
#
# All Cell 7 V1 validation behaviour is inherited.
#
# Only the entity reference is widened.
#
# ============================================================

class RANTelemetryObservationV1_1(
    RANTelemetryObservation
):

    entity: (
        RANTelemetryEntityRefV1_1
    )


class RANTelemetryBatchV1_1(
    RANTelemetryBatch
):

    observations: List[
        RANTelemetryObservationV1_1
    ]


# ============================================================
# 7 V1.1.6 — STANDARD SCOPE MAPPING
# ============================================================
#
# Cell 9 uses one common scope vocabulary for observability.
#
# NR_CELL_CU and NR_CELL_DU nevertheless remain standardized
# managed-object entities when observations are instantiated.
#
# ============================================================

STANDARD_SCOPE_TO_ENTITY_TYPE = {

    OperationalEntityScope
    .NR_CELL_CU:
        EntityType.NR_CELL_CU,

    OperationalEntityScope
    .NR_CELL_DU:
        EntityType.NR_CELL_DU,
}


# ============================================================
# 7 V1.1.7 — ENTITY REGISTRY
# ============================================================

RAN_TELEMETRY_ENTITY_REGISTRY_V1_1 = {}


def register_telemetry_entity_v1_1(
    entity
):

    if (
        entity.entity_id
        in RAN_TELEMETRY_ENTITY_REGISTRY_V1_1
    ):

        raise ValueError(

            "Duplicate telemetry entity ID: "
            f"{entity.entity_id}"
        )


    RAN_TELEMETRY_ENTITY_REGISTRY_V1_1[
        entity.entity_id
    ] = entity


# ============================================================
# 7 V1.1.8 — REGISTER STANDARD NRCellCU / NRCellDU ENTITIES
# ============================================================

for (
    entity_id,
    entity_ref
) in sorted(
    SYNTHETIC_NR_CELL_CU_LOOKUP_V1_1
    .items()
):

    logical_cell_id = (

        entity_ref.metadata.get(
            "logical_cell_id"
        )
    )


    cell = (

        SYNTHETIC_RAN_CELL_LOOKUP_V1_1[
            logical_cell_id
        ]
    )


    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=entity_id,

            namespace=(
                TelemetryEntityNamespace
                .STANDARD_MANAGED_OBJECT
            ),

            entity_type=(
                EntityType.NR_CELL_CU
            ),

            parent_entity_id=(
                entity_ref.parent_entity_id
            ),

            site_id=(
                entity_ref.site_id
            ),

            cu_domain_id=(
                cell.cu_domain_id
            ),

            metadata=dict(
                entity_ref.metadata
            ),
        )
    )


for (
    entity_id,
    entity_ref
) in sorted(
    SYNTHETIC_NR_CELL_DU_LOOKUP_V1_1
    .items()
):

    logical_cell_id = (

        entity_ref.metadata.get(
            "logical_cell_id"
        )
    )


    cell = (

        SYNTHETIC_RAN_CELL_LOOKUP_V1_1[
            logical_cell_id
        ]
    )


    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=entity_id,

            namespace=(
                TelemetryEntityNamespace
                .STANDARD_MANAGED_OBJECT
            ),

            entity_type=(
                EntityType.NR_CELL_DU
            ),

            parent_entity_id=(
                entity_ref.parent_entity_id
            ),

            site_id=(
                entity_ref.site_id
            ),

            cu_domain_id=(
                cell.cu_domain_id
            ),

            metadata=dict(
                entity_ref.metadata
            ),
        )
    )


# ============================================================
# 7 V1.1.9 — REGISTER LOGICAL CELLS
# ============================================================

for (
    logical_cell_id,
    cell
) in sorted(
    SYNTHETIC_RAN_CELL_LOOKUP_V1_1
    .items()
):

    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=(
                logical_cell_id
            ),

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .LOGICAL_CELL
            ),

            parent_entity_id=(
                cell.site_id
            ),

            site_id=(
                cell.site_id
            ),

            cu_domain_id=(
                cell.cu_domain_id
            ),

            metadata={

                "nr_cell_cu_id":
                    cell.nr_cell_cu.entity_id,

                "nr_cell_du_id":
                    cell.nr_cell_du.entity_id,

                "du_id":
                    cell.du_id,

                "cu_cp_id":
                    cell.cu_cp_id,

                "cu_up_id":
                    cell.cu_up_id,

                "n2_path_id":
                    cell.n2_path_id,

                "amf_peer_id":
                    cell.amf_peer_id,

                "f1_path_id":
                    cell.f1_path_id,

                "e1_path_id":
                    cell.e1_path_id,
            },
        )
    )


# ============================================================
# 7 V1.1.10 — REGISTER SITE + DU
# ============================================================

for site in sorted(

    SYNTHETIC_RAN_SITES_V1_1,

    key=lambda item:
        item.site_id,
):

    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=(
                site.site_id
            ),

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .SITE
            ),

            parent_entity_id=(
                site.cu_domain_id
            ),

            site_id=(
                site.site_id
            ),

            cu_domain_id=(
                site.cu_domain_id
            ),

            metadata={

                "du_id":
                    site.du_id,

                "cu_cp_id":
                    site.cu_cp_id,

                "cu_up_id":
                    site.cu_up_id,

                "n2_path_id":
                    site.n2_path_id,

                "amf_peer_id":
                    site.amf_peer_id,

                "f1_path_id":
                    site.f1_path_id,

                "e1_path_id":
                    site.e1_path_id,
            },
        )
    )


    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=(
                site.du_id
            ),

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .DU
            ),

            parent_entity_id=(
                site.site_id
            ),

            site_id=(
                site.site_id
            ),

            cu_domain_id=(
                site.cu_domain_id
            ),

            metadata={

                "f1_path_id":
                    site.f1_path_id,

                "cu_cp_id":
                    site.cu_cp_id,
            },
        )
    )


# ============================================================
# 7 V1.1.11 — REGISTER CU DOMAINS
# ============================================================

for cu_domain_id in sorted(

    SYNTHETIC_RAN_TOPOLOGY_V1_1
    .cu_domains
):

    domain_sites = [

        site.site_id

        for site
        in SYNTHETIC_RAN_SITES_V1_1

        if (
            site.cu_domain_id
            == cu_domain_id
        )
    ]


    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=(
                cu_domain_id
            ),

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .CU_DOMAIN
            ),

            cu_domain_id=(
                cu_domain_id
            ),

            metadata={

                "site_ids":
                    sorted(
                        domain_sites
                    ),
            },
        )
    )


# ============================================================
# 7 V1.1.12 — REGISTER CU-CP + CU-UP
# ============================================================

for (
    entity_id,
    entity
) in sorted(
    SYNTHETIC_CU_CP_LOOKUP_V1_1
    .items()
):

    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=entity_id,

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .CU_CP
            ),

            parent_entity_id=(
                entity.cu_domain_id
            ),

            cu_domain_id=(
                entity.cu_domain_id
            ),

            metadata={

                "site_ids":
                    list(
                        entity.site_ids
                    ),

                "du_ids":
                    list(
                        entity.du_ids
                    ),

                **dict(
                    entity.metadata
                ),
            },
        )
    )


for (
    entity_id,
    entity
) in sorted(
    SYNTHETIC_CU_UP_LOOKUP_V1_1
    .items()
):

    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=entity_id,

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .CU_UP
            ),

            parent_entity_id=(
                entity.cu_domain_id
            ),

            cu_domain_id=(
                entity.cu_domain_id
            ),

            metadata={

                "site_ids":
                    list(
                        entity.site_ids
                    ),

                "du_ids":
                    list(
                        entity.du_ids
                    ),

                **dict(
                    entity.metadata
                ),
            },
        )
    )


# ============================================================
# 7 V1.1.13 — REGISTER N2 PATHS
# ============================================================

for (
    path_id,
    path
) in sorted(
    SYNTHETIC_N2_PATH_LOOKUP_V1_1
    .items()
):

    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=path_id,

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .N2_PATH
            ),

            parent_entity_id=(
                path.cu_domain_id
            ),

            cu_domain_id=(
                path.cu_domain_id
            ),

            metadata={

                "source_entity_id":
                    path.source_entity_id,

                "target_entity_id":
                    path.target_entity_id,

                **dict(
                    path.metadata
                ),
            },
        )
    )


# ============================================================
# 7 V1.1.14 — REGISTER AMF-FACING BOUNDARY PEERS
# ============================================================

for (
    amf_peer_id,
    peer
) in sorted(
    SYNTHETIC_AMF_PEER_LOOKUP_V1_1
    .items()
):

    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=(
                amf_peer_id
            ),

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .AMF_PEER
            ),

            parent_entity_id=(
                peer.cu_domain_id
            ),

            cu_domain_id=(
                peer.cu_domain_id
            ),

            metadata={

                "n2_path_id":
                    peer.n2_path_id,

                **dict(
                    peer.metadata
                ),
            },
        )
    )


# ============================================================
# 7 V1.1.15 — REGISTER F1 PATHS
# ============================================================

for (
    path_id,
    path
) in sorted(
    SYNTHETIC_F1_PATH_LOOKUP_V1_1
    .items()
):

    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=path_id,

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .F1_PATH
            ),

            parent_entity_id=(
                path.site_id
            ),

            site_id=(
                path.site_id
            ),

            cu_domain_id=(
                path.cu_domain_id
            ),

            metadata={

                "du_id":
                    path.du_id,

                "source_entity_id":
                    path.source_entity_id,

                "target_entity_id":
                    path.target_entity_id,

                **dict(
                    path.metadata
                ),
            },
        )
    )


# ============================================================
# 7 V1.1.16 — REGISTER E1 PATHS
# ============================================================

for (
    path_id,
    path
) in sorted(
    SYNTHETIC_E1_PATH_LOOKUP_V1_1
    .items()
):

    register_telemetry_entity_v1_1(

        RANTelemetryEntityRefV1_1(

            entity_id=path_id,

            namespace=(
                TelemetryEntityNamespace
                .OPERATIONAL_SCOPE
            ),

            operational_scope=(
                OperationalEntityScope
                .E1_PATH
            ),

            parent_entity_id=(
                path.cu_domain_id
            ),

            cu_domain_id=(
                path.cu_domain_id
            ),

            metadata={

                "source_entity_id":
                    path.source_entity_id,

                "target_entity_id":
                    path.target_entity_id,

                **dict(
                    path.metadata
                ),
            },
        )
    )


# ============================================================
# 7 V1.1.17 — ENTITY SCOPE RESOLUTION
# ============================================================

def telemetry_entity_scope_v1_1(
    entity
):

    if (
        entity.namespace
        == TelemetryEntityNamespace
        .OPERATIONAL_SCOPE
    ):

        return (
            entity.operational_scope
        )


    if (
        entity.entity_type
        == EntityType.NR_CELL_CU
    ):

        return (
            OperationalEntityScope
            .NR_CELL_CU
        )


    if (
        entity.entity_type
        == EntityType.NR_CELL_DU
    ):

        return (
            OperationalEntityScope
            .NR_CELL_DU
        )


    raise ValueError(

        "No Cell 9 operational-scope mapping "
        "defined for standardized entity type: "
        f"{entity.entity_type}"
    )


# ============================================================
# 7 V1.1.18 — ENTITY COUNTS BY SCOPE
# ============================================================

ENTITY_COUNTS_BY_SCOPE_V1_1 = {}


for entity in (
    RAN_TELEMETRY_ENTITY_REGISTRY_V1_1
    .values()
):

    scope = (
        telemetry_entity_scope_v1_1(
            entity
        )
    )


    ENTITY_COUNTS_BY_SCOPE_V1_1[
        scope.value
    ] = (

        ENTITY_COUNTS_BY_SCOPE_V1_1.get(
            scope.value,
            0,
        )

        + 1
    )


EXPECTED_ENTITY_COUNTS_BY_SCOPE_V1_1 = {

    "LOGICAL_CELL": 30,

    "NR_CELL_CU": 30,

    "NR_CELL_DU": 30,

    "SITE": 10,

    "DU": 10,

    "CU_DOMAIN": 2,

    "CU_CP": 2,

    "CU_UP": 2,

    "N2_PATH": 2,

    "AMF_PEER": 2,

    "F1_PATH": 10,

    "E1_PATH": 2,
}


assert (
    ENTITY_COUNTS_BY_SCOPE_V1_1
    == EXPECTED_ENTITY_COUNTS_BY_SCOPE_V1_1
)


assert (
    len(
        RAN_TELEMETRY_ENTITY_REGISTRY_V1_1
    )
    == 132
)


# ============================================================
# 7 V1.1.19 — CELL 9 REQUIRED-SCOPE COVERAGE
# ============================================================

registered_scopes_v1_1 = {

    telemetry_entity_scope_v1_1(
        entity
    )

    for entity
    in (
        RAN_TELEMETRY_ENTITY_REGISTRY_V1_1
        .values()
    )
}


missing_required_scopes_v1_1 = (

    set(
        required_topology_scopes_global
    )

    - registered_scopes_v1_1
)


assert (
    len(
        missing_required_scopes_v1_1
    )
    == 0
)


# ============================================================
# 7 V1.1.20 — STANDARDIZED BINDING RESOLUTION
# ============================================================

standard_entity_type_counts_v1_1 = {}


for entity in (
    RAN_TELEMETRY_ENTITY_REGISTRY_V1_1
    .values()
):

    if (
        entity.namespace
        == TelemetryEntityNamespace
        .STANDARD_MANAGED_OBJECT
    ):

        standard_entity_type_counts_v1_1[
            entity.entity_type
        ] = (

            standard_entity_type_counts_v1_1
            .get(
                entity.entity_type,
                0,
            )

            + 1
        )


unresolved_standard_bindings_v1_1 = []


for (
    metric_name,
    binding
) in (
    RAN_TELEMETRY_METRIC_BINDINGS
    .items()
):

    if (
        standard_entity_type_counts_v1_1
        .get(
            binding.entity_type,
            0,
        )
        <= 0
    ):

        unresolved_standard_bindings_v1_1.append(
            metric_name
        )


assert (
    len(
        unresolved_standard_bindings_v1_1
    )
    == 0
)


# ============================================================
# 7 V1.1.21 — OPERATIONAL BINDING RESOLUTION
# ============================================================

unresolved_operational_bindings_v1_1 = []


for (
    metric_name,
    binding
) in (
    RAN_OPERATIONAL_METRIC_BINDINGS
    .items()
):

    if (
        binding.preferred_scope
        not in registered_scopes_v1_1
    ):

        unresolved_operational_bindings_v1_1.append(
            metric_name
        )


assert (
    len(
        unresolved_operational_bindings_v1_1
    )
    == 0
)


# ============================================================
# 7 V1.1.22 — ALLOWED-SCOPE COVERAGE
# ============================================================

unresolved_allowed_scopes_v1_1 = set()


for binding in (
    RAN_OPERATIONAL_METRIC_BINDINGS
    .values()
):

    for scope in (
        binding.allowed_scopes
    ):

        if (
            scope
            not in registered_scopes_v1_1
        ):

            unresolved_allowed_scopes_v1_1.add(
                scope
            )


assert (
    len(
        unresolved_allowed_scopes_v1_1
    )
    == 0
)


# ============================================================
# 7 V1.1.23 — RESOLUTION HELPER
# ============================================================

def get_telemetry_entity_v1_1(
    entity_id
):

    try:

        return (
            RAN_TELEMETRY_ENTITY_REGISTRY_V1_1[
                entity_id
            ]
        )


    except KeyError as exc:

        raise KeyError(

            "Unknown telemetry entity ID: "
            f"{entity_id}"
        ) from exc


# ============================================================
# 7 V1.1.24 — NAMESPACE SAFETY TESTS
# ============================================================

sample_standard_entity = (

    get_telemetry_entity_v1_1(
        "NRCELL_CU_001_1"
    )
)


sample_operational_entity = (

    get_telemetry_entity_v1_1(
        "CU_CP_01"
    )
)


assert (
    sample_standard_entity.namespace
    == TelemetryEntityNamespace
    .STANDARD_MANAGED_OBJECT
)


assert (
    sample_standard_entity.entity_type
    == EntityType.NR_CELL_CU
)


assert (
    sample_standard_entity.operational_scope
    is None
)


assert (
    sample_operational_entity.namespace
    == TelemetryEntityNamespace
    .OPERATIONAL_SCOPE
)


assert (
    sample_operational_entity.operational_scope
    == OperationalEntityScope.CU_CP
)


assert (
    sample_operational_entity.entity_type
    is None
)


# ============================================================
# 7 V1.1.25 — OBSERVATION CONTRACT INHERITANCE CHECK
# ============================================================

assert (
    "entity"
    in RANTelemetryObservationV1_1
    .model_fields
)


assert (
    "observations"
    in RANTelemetryBatchV1_1
    .model_fields
)


# ============================================================
# 7 V1.1.26 — HIDDEN-TRUTH SAFETY
# ============================================================

PROHIBITED_OBSERVATION_FIELDS_V1_1 = {

    "root_cause",

    "ground_truth",

    "fault_label",

    "expected_diagnosis",

    "correct_action",

    "benchmark_answer",
}


for model in [

    RANTelemetryEntityRefV1_1,

    RANTelemetryObservationV1_1,

    RANTelemetryBatchV1_1,
]:

    assert (
        len(
            set(
                model.model_fields.keys()
            )

            & PROHIBITED_OBSERVATION_FIELDS_V1_1
        )
        == 0
    )


# ============================================================
# 7 V1.1.27 — GENERATION READINESS
# ============================================================

TELEMETRY_OBSERVATION_CONTRACT_READY_V1_1 = (

    len(
        missing_required_scopes_v1_1
    )
    == 0

    and

    len(
        unresolved_standard_bindings_v1_1
    )
    == 0

    and

    len(
        unresolved_operational_bindings_v1_1
    )
    == 0

    and

    len(
        unresolved_allowed_scopes_v1_1
    )
    == 0
)


assert (
    TELEMETRY_OBSERVATION_CONTRACT_READY_V1_1
    is True
)


# ============================================================
# 7 V1.1.28 — BUILD ARTIFACT
# ============================================================

cell7_v1_1_artifact = {

    "artifact_id":
        (
            "module5_5g_ran_telemetry_"
            "observation_contract_v1_1"
        ),

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "revision":
        "V1.1",

    "parent_revision":
        "CELL_7_V1",

    "revision_role":
        (
            "OPERATIONAL_ENTITY_"
            "OBSERVATION_EXTENSION"
        ),

    "reference_domain":
        "5G RAN Accessibility",

    "entity_namespace_model": {

        "STANDARD_MANAGED_OBJECT":
            (
                "Entity uses canonical "
                "Cell 4 EntityType."
            ),

        "OPERATIONAL_SCOPE":
            (
                "Entity uses Cell 9 "
                "OperationalEntityScope "
                "without changing EntityType."
            ),
    },

    "entity_counts_by_scope":
        ENTITY_COUNTS_BY_SCOPE_V1_1,

    "entity_registry": {

        entity_id:
            entity.model_dump(
                mode="json"
            )

        for (
            entity_id,
            entity
        ) in sorted(
            RAN_TELEMETRY_ENTITY_REGISTRY_V1_1
            .items()
        )
    },

    "binding_readiness": {

        "standardized_metric_bindings":
            len(
                RAN_TELEMETRY_METRIC_BINDINGS
            ),

        "operational_metric_bindings":
            len(
                RAN_OPERATIONAL_METRIC_BINDINGS
            ),

        "unresolved_standardized_bindings":
            list(
                unresolved_standard_bindings_v1_1
            ),

        "unresolved_operational_bindings":
            list(
                unresolved_operational_bindings_v1_1
            ),

        "unresolved_allowed_scopes": [

            scope.value

            for scope
            in sorted(

                unresolved_allowed_scopes_v1_1,

                key=lambda item:
                    item.value,
            )
        ],

        "observation_contract_ready":
            (
                TELEMETRY_OBSERVATION_CONTRACT_READY_V1_1
            ),
    },

    "provenance": {

        "parent_cell7_v1_semantic_sha256":
            cell7_semantic_sha,

        "cell8_v1_1_semantic_sha256":
            cell8_v1_1_semantic_sha,

        "cell9_semantic_sha256":
            cell9_semantic_sha,

        "cell5_v1_2_semantic_sha256":
            (
                ran_accessibility_v1_2_semantic_sha
            ),
    },

    "design_constraints": {

        "parent_cell7_v1_mutated":
            False,

        "entity_type_enum_modified":
            False,

        "standardized_and_operational_entity_namespaces_separated":
            True,

        "telemetry_generated":
            False,

        "incident_generated":
            False,

        "fault_injected":
            False,

        "hidden_ground_truth_exposed":
            False,

        "ml_inference_included":
            False,

        "internal_5gc_modelled":
            False,
    },
}


# ============================================================
# 7 V1.1.29 — SEMANTIC HASH
# ============================================================

def canonical_json_sha256_cell7_v1_1(
    data
):

    canonical = json.dumps(

        data,

        sort_keys=True,

        separators=(",", ":"),

        ensure_ascii=False,

    ).encode(
        "utf-8"
    )


    return hashlib.sha256(
        canonical
    ).hexdigest()


cell7_v1_1_semantic_content = {

    key: value

    for key, value
    in cell7_v1_1_artifact.items()

    if key != "created_utc"
}


cell7_v1_1_semantic_sha = (

    canonical_json_sha256_cell7_v1_1(

        cell7_v1_1_semantic_content
    )
)


cell7_v1_1_artifact[
    "semantic_sha256"
] = (
    cell7_v1_1_semantic_sha
)


# ============================================================
# 7 V1.1.30 — SAVE ARTIFACT
# ============================================================

telemetry_artifact_dir = (

    PATHS["kpi"].parent

    / "telemetry"
)


telemetry_artifact_dir.mkdir(

    parents=True,

    exist_ok=True,
)


cell7_v1_1_path = (

    telemetry_artifact_dir

    / (
        "module5_5g_ran_telemetry_"
        "observation_contract_v1_1.json"
    )
)


if cell7_v1_1_path.exists():

    raise FileExistsError(

        "Cell 7 V1.1 artifact already exists. "
        "Create an explicit later revision rather "
        "than silently overwriting it."
    )


with open(

    cell7_v1_1_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        cell7_v1_1_artifact,

        f,

        indent=2,

        ensure_ascii=False,
    )


cell7_v1_1_artifact_sha = (

    sha256_file(
        cell7_v1_1_path
    )
)


# ============================================================
# 7 V1.1.31 — SUMMARY
# ============================================================

print(
    "=" * 108
)

print(
    "MODULE 5 — CELL 12 "
    "OPERATIONAL ENTITY OBSERVATION CONTRACT EXTENSION"
)

print(
    "=" * 108
)


print(
    "\nRevision provenance"
)

print("-" * 108)

print(
    "Parent observation contract       : "
    "Cell 7 V1"
)

print(
    "Parent Cell 7 V1 preserved         : "
    "YES"
)

print(
    "Cell 7 V1 semantic SHA            : "
    f"{cell7_semantic_sha}"
)

print(
    "Cell 8 V1.1 semantic SHA          : "
    f"{cell8_v1_1_semantic_sha}"
)

print(
    "Cell 9 semantic SHA               : "
    f"{cell9_semantic_sha}"
)

print(
    "Cell 5 V1.2 semantic SHA          : "
    f"{ran_accessibility_v1_2_semantic_sha}"
)


print(
    "\nEntity namespace"
)

print("-" * 108)

print(
    "STANDARD_MANAGED_OBJECT           : "
    "Cell 4 EntityType"
)

print(
    "OPERATIONAL_SCOPE                 : "
    "Cell 9 OperationalEntityScope"
)

print(
    "EntityType enum modified           : "
    "NO"
)


print(
    "\nObservation entity registry"
)

print("-" * 108)

for (
    scope_name,
    count
) in sorted(
    ENTITY_COUNTS_BY_SCOPE_V1_1
    .items()
):

    print(
        f"{scope_name:<32} : "
        f"{count}"
    )


print(
    "\nTotal observation entities         : "
    f"{len(RAN_TELEMETRY_ENTITY_REGISTRY_V1_1)}"
)


print(
    "\nMetric binding resolution"
)

print("-" * 108)

print(
    "Standardized metric bindings      : "
    f"{len(RAN_TELEMETRY_METRIC_BINDINGS)}"
)

print(
    "Operational metric bindings       : "
    f"{len(RAN_OPERATIONAL_METRIC_BINDINGS)}"
)

print(
    "Unresolved standardized bindings  : "
    f"{len(unresolved_standard_bindings_v1_1)}"
)

print(
    "Unresolved operational bindings   : "
    f"{len(unresolved_operational_bindings_v1_1)}"
)

print(
    "Unresolved allowed scopes         : "
    f"{len(unresolved_allowed_scopes_v1_1)}"
)


print(
    "\nReadiness decision"
)

print("-" * 108)

print(
    "Required topology scopes missing  : "
    f"{len(missing_required_scopes_v1_1)}"
)

print(
    "Observation contract ready        : "
    f"{TELEMETRY_OBSERVATION_CONTRACT_READY_V1_1}"
)


print(
    "\nGeneration boundaries"
)

print("-" * 108)

print(
    "Telemetry generated               : "
    "NO"
)

print(
    "Incidents generated               : "
    "NO"
)

print(
    "Faults injected                   : "
    "NO"
)

print(
    "Hidden ground truth exposed       : "
    "NO"
)

print(
    "Internal 5GC modelled             : "
    "NO"
)

print(
    "ML inference included             : "
    "NO"
)


print(
    "\nArtifact"
)

print("-" * 108)

print(
    f"Path                               : "
    f"{cell7_v1_1_path}"
)

print(
    f"Artifact SHA                       : "
    f"{cell7_v1_1_artifact_sha}"
)

print(
    f"Semantic SHA                       : "
    f"{cell7_v1_1_semantic_sha}"
)


print(
    "\n" + "=" * 108
)

print(
    "CELL 12 COMPLETE — "
    "UNIFIED OBSERVATION CONTRACT READY "
    "FOR HEALTHY TELEMETRY GENERATION"
)

print(
    "=" * 108
)
````

<a id="cell-13-healthy-telemetry-generator"></a>
##### Cell 13 — Pilot / Canonical Healthy 15-Minute 5G RAN Telemetry Generator V1

Generate causally correlated **healthy 15-minute 5G RAN telemetry** using one deterministic generation engine with two execution profiles:

```text
PILOT_7D
→ 7 days / 672 intervals
→ Generator qualification dataset

CANONICAL_92D
→ 92 days / 8,832 intervals
→ Canonical healthy historical baseline
```

During a clean rebuild, Cell 13 is executed twice. The first pass generates `PILOT_7D`; after Cell 14 qualifies the pilot, the same generation engine is rerun as `CANONICAL_92D`.

Standardized PM counters and operational telemetry emerge from the same underlying synthetic radio, load, processing, topology and transport state rather than being independently sampled.

Scope:

```text
21 standardized PM families
45 continuous operational families
4 event-driven families with zero healthy events
66 continuously observed metric families
```

The generator introduces **no incidents, injected faults, hidden root-cause truth or ML inference**.

Formal TS 28.554 Accessibility KPI values are deliberately **deferred** for deterministic derivation and validation in M05_02.

> **Scaling from the 7-day pilot to the 92-day canonical baseline changes the execution window only; the qualified generation model remains unchanged.**


````python
# ============================================================
# MODULE 5 — CELL 13
# PILOT / CANONICAL HEALTHY 15-MINUTE 5G RAN TELEMETRY GENERATOR V1
# ============================================================
#
# Purpose:
#
#   Generate the first deterministic healthy network-observation
#   dataset using:
#
#       Cell 10 — causal investigation graph V1.2
#           causal investigation graph
#
#       Cell 12 — unified observation contract V1.1
#           unified observation contract
#
#       Cell 11 — causal synthetic topology V1.1
#           causal synthetic topology
#
#       Cell 09 — causal observability configuration
#           causal observability configuration
#
#
# CRITICAL DESIGN PRINCIPLE
#
#   Standardized PM counters are NOT independently sampled.
#
#   They are generated from the same underlying healthy
#   radio/load/processing/transport conditions that are exposed
#   as operational evidence.
#
#
# DATASET EXECUTION PROFILES
#
#   PILOT_7D
#       2026-08-25 00:00 UTC → 2026-09-01 00:00 UTC
#       7 complete days / 672 intervals
#
#   CANONICAL_92D
#       2026-06-01 00:00 UTC → 2026-09-01 00:00 UTC
#       92 complete days / 8,832 intervals
#
#   Saved frozen execution: CANONICAL_92D
#
#
# EXPECTED OBSERVABILITY
#
#   21 standardized metric families
#   49 operational metric families
#
#   Of the 49 operational families:
#
#       45 are continuous/snapshot telemetry
#        4 are event-driven
#
#   Healthy baseline intentionally contains zero active
#   event records. Event channels remain configured and will
#   be populated later by controlled incident generation.
#
#
# IMPORTANT
#
#   - NO fault injected.
#   - NO incident labels.
#   - NO root-cause labels.
#   - NO ML inference.
#   - NO formal TS 28.554 KPI is independently generated.
#
#   Formal KPI values will later be deterministically derived
#   from these standardized PM counters.
#
# ============================================================


from pathlib import Path

from datetime import (
    datetime,
    timezone,
)

import hashlib
import json
import math
import shutil

import numpy as np
import pandas as pd


# ============================================================
# 13.1 — DEPENDENCY VALIDATION
# ============================================================

REQUIRED_CELL13_OBJECTS = [

    # Cell 7 V1 / V1.1
    "TelemetryEnvironment",
    "TelemetryDataQuality",
    "TelemetryEntityNamespace",
    "RANTelemetryObservationV1_1",
    "RAN_TELEMETRY_METRIC_BINDINGS",
    "RAN_TELEMETRY_ENTITY_REGISTRY_V1_1",
    "get_telemetry_entity_v1_1",
    "cell7_v1_1_semantic_sha",

    # Cell 8 V1.1
    "SYNTHETIC_RAN_TOPOLOGY_V1_1",
    "SYNTHETIC_RAN_CELL_LOOKUP_V1_1",
    "SYNTHETIC_RAN_SITES_V1_1",
    "SYNTHETIC_CU_CP_LOOKUP_V1_1",
    "SYNTHETIC_CU_UP_LOOKUP_V1_1",
    "SYNTHETIC_N2_PATH_LOOKUP_V1_1",
    "SYNTHETIC_AMF_PEER_LOOKUP_V1_1",
    "SYNTHETIC_F1_PATH_LOOKUP_V1_1",
    "SYNTHETIC_E1_PATH_LOOKUP_V1_1",
    "cell8_v1_1_semantic_sha",

    # Cell 9
    "OperationalEntityScope",
    "OperationalMetricCategory",
    "OperationalMetricCadence",
    "OperationalMetricValueKind",
    "RAN_OPERATIONAL_METRIC_BINDINGS",
    "cell9_semantic_sha",

    # Cell 5 V1.2
    "ran_accessibility_v1_2_semantic_sha",

    # shared
    "PM_INTERVAL_MINUTES",
    "PATHS",
    "sha256_file",
]


missing_cell13_objects = [

    name

    for name
    in REQUIRED_CELL13_OBJECTS

    if name not in globals()
]


if missing_cell13_objects:

    raise RuntimeError(

        "Cell 13 requires Cell 10, "
        "Cell 12, Cell 11 and "
        "Cell 09 to be executed first. "
        f"Missing objects: "
        f"{missing_cell13_objects}"
    )


# ============================================================
# 13.2 — VERIFY FROZEN BASELINES
# ============================================================

EXPECTED_CELL5_V1_2_SHA = (
    "aaeb1acf5739a9d80d7745469faa71fe"
    "142181c7e96df6bc2c9e5f8b8b031b68"
)

EXPECTED_CELL7_V1_1_SHA = (
    "07ba4e0cc0f672cc03268e96bb53aa0f"
    "322241be43a4f957e7065b0d6ebae312"
)

EXPECTED_CELL8_V1_1_SHA = (
    "b12fd1cc5b208963a13caecc33d03016"
    "a612ff1f1f112686f4db435db66b421d"
)

EXPECTED_CELL9_SHA = (
    "8ccce01ee1375bbef52611a4b008ea01"
    "38bae88874993be67f7577aa0838f42a"
)


assert (
    ran_accessibility_v1_2_semantic_sha
    == EXPECTED_CELL5_V1_2_SHA
)

assert (
    cell7_v1_1_semantic_sha
    == EXPECTED_CELL7_V1_1_SHA
)

assert (
    cell8_v1_1_semantic_sha
    == EXPECTED_CELL8_V1_1_SHA
)

assert (
    cell9_semantic_sha
    == EXPECTED_CELL9_SHA
)

assert PM_INTERVAL_MINUTES == 15


# ============================================================
# 13.3 — DATASET CONTROLS
# ============================================================

BASELINE_MODE = "CANONICAL_92D"

VALID_BASELINE_MODES = {
    "PILOT_7D",
    "CANONICAL_92D",
}

assert BASELINE_MODE in VALID_BASELINE_MODES

# The canonical run is intentionally locked until the pilot
# dataset has passed Cell 14 generator qualification.
CANONICAL_GENERATION_APPROVED = True

BASELINE_CONFIGS = {

    "PILOT_7D": {
        "dataset_version":
            "MODULE5_HEALTHY_TELEMETRY_PILOT_7D_V1",
        "artifact_id":
            "module5_healthy_5g_ran_telemetry_pilot_7d_v1",
        "role":
            "GENERATOR_QUALIFICATION_DATASET",
        "start_utc":
            "2026-08-25T00:00:00Z",
        "end_exclusive_utc":
            "2026-09-01T00:00:00Z",
        "dataset_dirname":
            "healthy_baseline_pilot_7d_v1",
        "expected_days":
            7,
        "expected_intervals":
            672,
        "canonical":
            False,
    },

    "CANONICAL_92D": {
        "dataset_version":
            "MODULE5_HEALTHY_TELEMETRY_V1",
        "artifact_id":
            "module5_healthy_5g_ran_telemetry_baseline_v1",
        "role":
            "HEALTHY_NETWORK_OBSERVATION_BASELINE",
        "start_utc":
            "2026-06-01T00:00:00Z",
        "end_exclusive_utc":
            "2026-09-01T00:00:00Z",
        "dataset_dirname":
            "healthy_baseline_v1",
        "expected_days":
            92,
        "expected_intervals":
            8_832,
        "canonical":
            True,
    },
}

if (
    BASELINE_MODE == "CANONICAL_92D"
    and not CANONICAL_GENERATION_APPROVED
):
    raise RuntimeError(
        "CANONICAL_92D generation is locked. "
        "Validate PILOT_7D in M05_02 first, then set "
        "CANONICAL_GENERATION_APPROVED = True."
    )

BASELINE_CONFIG = BASELINE_CONFIGS[
    BASELINE_MODE
]

HEALTHY_DATASET_VERSION = (
    BASELINE_CONFIG[
        "dataset_version"
    ]
)

BASELINE_ARTIFACT_ID = (
    BASELINE_CONFIG[
        "artifact_id"
    ]
)

BASELINE_ROLE = (
    BASELINE_CONFIG[
        "role"
    ]
)

BASELINE_DATASET_DIRNAME = (
    BASELINE_CONFIG[
        "dataset_dirname"
    ]
)

HEALTHY_GENERATOR_SEED = 20260910

HEALTHY_START_UTC = pd.Timestamp(
    BASELINE_CONFIG[
        "start_utc"
    ]
)

HEALTHY_END_EXCLUSIVE_UTC = pd.Timestamp(
    BASELINE_CONFIG[
        "end_exclusive_utc"
    ]
)

BASELINE_DAY_COUNT = int(
    (
        HEALTHY_END_EXCLUSIVE_UTC
        - HEALTHY_START_UTC
    ).days
)

EXPECTED_INTERVAL_COUNT = int(
    BASELINE_CONFIG[
        "expected_intervals"
    ]
)

assert (
    BASELINE_DAY_COUNT
    == BASELINE_CONFIG[
        "expected_days"
    ]
)

CAUSE_VALUES = [
    "mo-Data",
    "mo-Signalling",
    "mt-Access",
]

FIVE_QI_VALUES = [
    "1",
    "5",
    "9",
]

SNSSAI_VALUES = [
    "1-010203",
]

interval_starts = pd.date_range(
    start=HEALTHY_START_UTC,
    end=HEALTHY_END_EXCLUSIVE_UTC,
    freq=f"{PM_INTERVAL_MINUTES}min",
    inclusive="left",
)

assert (
    len(interval_starts)
    == EXPECTED_INTERVAL_COUNT
)


# ============================================================
# 13.4 — HEALTHY SYNTHETIC OPERATING PROFILES
# ============================================================
#
# These reproduce the Cell 08 operating assumptions.
#
# They are synthetic workload assumptions.
#
# They are NOT:
#
#   - 3GPP thresholds
#   - vendor limits
#   - SLA thresholds
#   - KPI acceptance criteria
#
# ============================================================

HEALTHY_PROFILE_BY_ARCHETYPE = {

    "URBAN_HIGH": {

        "mean_connected_ues": 420,

        "base_rrc_attempts_per_15m": 1800,

        "base_resume_attempts_per_15m": 800,

        "relative_traffic_scale": 1.00,

        "daily_load_amplitude": 0.45,

        "base_rsrp_dbm": -91.0,

        "base_sinr_db": 17.0,
    },


    "URBAN_MEDIUM": {

        "mean_connected_ues": 260,

        "base_rrc_attempts_per_15m": 1200,

        "base_resume_attempts_per_15m": 500,

        "relative_traffic_scale": 0.70,

        "daily_load_amplitude": 0.35,

        "base_rsrp_dbm": -89.0,

        "base_sinr_db": 19.0,
    },


    "SUBURBAN": {

        "mean_connected_ues": 140,

        "base_rrc_attempts_per_15m": 700,

        "base_resume_attempts_per_15m": 250,

        "relative_traffic_scale": 0.45,

        "daily_load_amplitude": 0.25,

        "base_rsrp_dbm": -94.0,

        "base_sinr_db": 16.0,
    },
}


# ============================================================
# 13.5 — SITE ARCHETYPE RESOLUTION
# ============================================================

def resolve_site_archetype(
    site
):

    raw = getattr(
        site,
        "archetype",
        None,
    )


    if raw is not None:

        if hasattr(
            raw,
            "value",
        ):

            raw = raw.value


        raw = str(
            raw
        )


        if (
            raw
            in HEALTHY_PROFILE_BY_ARCHETYPE
        ):

            return raw


    # Deterministic fallback matching frozen Cell 8 design.

    site_number = int(

        site.site_id.split(
            "_"
        )[-1]
    )


    if site_number <= 3:

        return "URBAN_HIGH"


    if site_number <= 7:

        return "URBAN_MEDIUM"


    return "SUBURBAN"


site_archetypes = {

    site.site_id:
        resolve_site_archetype(
            site
        )

    for site
    in SYNTHETIC_RAN_SITES_V1_1
}


assert (
    {
        archetype:
            list(
                site_archetypes.values()
            ).count(
                archetype
            )

        for archetype
        in HEALTHY_PROFILE_BY_ARCHETYPE
    }
    == {

        "URBAN_HIGH": 3,

        "URBAN_MEDIUM": 4,

        "SUBURBAN": 3,
    }
)


# ============================================================
# 13.6 — TOPOLOGY LOOKUPS
# ============================================================

SITE_LOOKUP_V1_1 = {

    site.site_id:
        site

    for site
    in SYNTHETIC_RAN_SITES_V1_1
}


CELLS_BY_SITE_V1_1 = {

    site.site_id:
        list(
            site.cells
        )

    for site
    in SYNTHETIC_RAN_SITES_V1_1
}


CELLS_BY_CU_DOMAIN_V1_1 = {

    cu_domain_id: []

    for cu_domain_id
    in (
        SYNTHETIC_RAN_TOPOLOGY_V1_1
        .cu_domains
    )
}


for cell in (
    SYNTHETIC_RAN_CELL_LOOKUP_V1_1
    .values()
):

    CELLS_BY_CU_DOMAIN_V1_1[
        cell.cu_domain_id
    ].append(
        cell
    )


assert (
    {
        key:
            len(value)

        for key, value
        in CELLS_BY_CU_DOMAIN_V1_1.items()
    }
    == {

        "CU_DOMAIN_01": 15,

        "CU_DOMAIN_02": 15,
    }
)


# ============================================================
# 13.7 — DETERMINISTIC CELL OFFSETS
# ============================================================

def stable_unit_interval(
    text
):

    digest = hashlib.sha256(
        text.encode(
            "utf-8"
        )
    ).digest()


    integer = int.from_bytes(

        digest[:8],

        byteorder="big",

        signed=False,
    )


    return (
        integer
        / float(
            2**64 - 1
        )
    )


def stable_range(
    text,
    low,
    high,
):

    return (
        low
        + (
            high - low
        )
        * stable_unit_interval(
            text
        )
    )


CELL_STATIC_OFFSETS = {}


for (
    logical_cell_id,
    cell
) in (
    SYNTHETIC_RAN_CELL_LOOKUP_V1_1
    .items()
):

    CELL_STATIC_OFFSETS[
        logical_cell_id
    ] = {

        "load":
            stable_range(
                logical_cell_id
                + "::load",
                0.92,
                1.08,
            ),

        "rsrp":
            stable_range(
                logical_cell_id
                + "::rsrp",
                -3.0,
                3.0,
            ),

        "sinr":
            stable_range(
                logical_cell_id
                + "::sinr",
                -2.0,
                2.0,
            ),

        "rach":
            stable_range(
                logical_cell_id
                + "::rach",
                0.92,
                1.08,
            ),
    }


# ============================================================
# 13.8 — TIME-OF-DAY ACTIVITY MODEL
# ============================================================

def gaussian_peak(
    hour,
    centre,
    width,
):

    return math.exp(

        -0.5

        * (
            (
                hour - centre
            )
            / width
        ) ** 2
    )


def activity_multiplier(
    timestamp,
    amplitude,
):

    hour = (

        timestamp.hour

        + timestamp.minute
        / 60.0
    )


    morning = gaussian_peak(
        hour,
        9.0,
        2.3,
    )


    midday = gaussian_peak(
        hour,
        13.0,
        3.2,
    )


    evening = gaussian_peak(
        hour,
        19.0,
        2.8,
    )


    raw = (

        0.63

        + amplitude
        * (
            0.35
            * morning

            + 0.25
            * midday

            + 0.75
            * evening
        )
    )


    # Saturday / Sunday:
    #
    # slightly lower business-period demand and
    # slightly broader leisure/evening demand.

    if timestamp.dayofweek >= 5:

        raw *= (
            0.94
            + 0.08
            * evening
        )


    return float(
        np.clip(
            raw,
            0.45,
            1.35,
        )
    )


# ============================================================
# 13.9 — NUMERIC HELPERS
# ============================================================

def clip(
    value,
    low,
    high,
):

    return float(
        np.clip(
            value,
            low,
            high,
        )
    )


def ratio_or_nan(
    numerator,
    denominator,
):

    if denominator <= 0:

        return float(
            "nan"
        )


    return (
        100.0
        * numerator
        / denominator
    )


def allocate_multinomial(
    rng,
    total,
    probabilities,
):

    total = int(
        max(
            0,
            total,
        )
    )


    if total == 0:

        return [
            0
            for _
            in probabilities
        ]


    p = np.asarray(
        probabilities,
        dtype=float,
    )


    p = (
        p
        / p.sum()
    )


    return (
        rng.multinomial(
            total,
            p,
        )
        .astype(
            int
        )
        .tolist()
    )


# ============================================================
# 13.10 — OPERATIONAL GRAPH-NODE MAPPING
# ============================================================

OPERATIONAL_CATEGORY_GRAPH_NODE = {

    OperationalMetricCategory.RADIO:
        "diagnostic::radio_condition",

    OperationalMetricCategory.RESOURCE:
        "diagnostic::cell_resource_load",

    OperationalMetricCategory.RACH:
        "diagnostic::rach_access_success",

    OperationalMetricCategory.PAGING:
        "diagnostic::paging_behaviour",

    OperationalMetricCategory.PROCESSING:
        "diagnostic::ran_processing_state",

    OperationalMetricCategory.N2_TRANSPORT:
        "diagnostic::transport_boundary",

    OperationalMetricCategory.SCTP:
        "diagnostic::transport_boundary",

    OperationalMetricCategory.AMF_BOUNDARY:
        "diagnostic::transport_boundary",

    OperationalMetricCategory.INTERNAL_TRANSPORT:
        (
            "diagnostic::"
            "internal_ran_interface_state"
        ),

    OperationalMetricCategory.CONTEXT:
        (
            "diagnostic::"
            "ue_bearer_context_state"
        ),

    OperationalMetricCategory.CONFIGURATION:
        "diagnostic::ran_configuration",

    OperationalMetricCategory.EVENT:
        "diagnostic::ran_alarms_events",
}


# ============================================================
# 13.11 — OBSERVATION STORAGE SCHEMA
# ============================================================
#
# Long-form Parquet representation of the Cell 7 V1.1
# observation contract.
#
# Mixed telemetry value types are kept in separate columns:
#
#     value_int
#     value_float
#     value_bool
#     value_text
#
# ============================================================

OBSERVATION_COLUMNS = [

    "observation_id",

    "environment",

    "observation_type",

    "evidence_origin",

    "metric_name",

    "is_standardized_metric",

    "operational_category",

    "entity_id",

    "entity_namespace",

    "entity_type",

    "operational_scope",

    "parent_entity_id",

    "site_id",

    "cu_domain_id",

    "interval_start_utc",

    "interval_end_utc",

    "pm_interval_minutes",

    "value_kind",

    "value_int",

    "value_float",

    "value_bool",

    "value_text",

    "unit",

    "dimension_cause",

    "dimension_5qi",

    "dimension_snssai",

    "dimensions_json",

    "quality",

    "graph_node_id",

    "investigation_stage",

    "source_system",

    "metadata_json",
]


# ============================================================
# 13.12 — STABLE OBSERVATION ID
# ============================================================

def make_observation_id(
    metric_name,
    entity_id,
    interval_start,
    dimensions,
):

    key = json.dumps(

        {

            "metric":
                metric_name,

            "entity":
                entity_id,

            "interval":
                interval_start.isoformat(),

            "dimensions":
                dimensions,
        },

        sort_keys=True,

        separators=(",", ":"),
    )


    return (
        "OBS_"
        + hashlib.sha256(
            key.encode(
                "utf-8"
            )
        ).hexdigest()[:24]
    )


# ============================================================
# 13.13 — FLAT OBSERVATION ROW BUILDER
# ============================================================

def build_observation_row(
    *,
    metric_name,
    entity_id,
    interval_start,
    value,
    dimensions=None,
    standardized,
    unit=None,
):

    dimensions = (
        dimensions
        or {}
    )


    entity = (
        get_telemetry_entity_v1_1(
            entity_id
        )
    )


    if standardized:

        binding = (

            RAN_TELEMETRY_METRIC_BINDINGS[
                metric_name
            ]
        )


        observation_type = (
            binding
            .observation_type
            .value
        )


        evidence_origin = (
            binding
            .evidence_origin
            .value
        )


        operational_category = None


        graph_node_id = (
            binding
            .graph_node_id
        )


        investigation_stage = (

            binding
            .investigation_stage
            .value

            if (
                binding
                .investigation_stage
                is not None
            )

            else None
        )


        source_system = (
            "SYNTHETIC_STANDARD_PM_V1"
        )


    else:

        binding = (

            RAN_OPERATIONAL_METRIC_BINDINGS[
                metric_name
            ]
        )


        observation_type = (
            binding
            .observation_type
            .value
        )


        evidence_origin = (
            binding
            .evidence_origin
            .value
        )


        operational_category = (
            binding
            .category
            .value
        )


        graph_node_id = (

            OPERATIONAL_CATEGORY_GRAPH_NODE[
                binding.category
            ]
        )


        investigation_stage = None


        source_system = (
            "SYNTHETIC_OPERATIONAL_TELEMETRY_V1"
        )


        if unit is None:

            unit = binding.unit


    value_int = None
    value_float = None
    value_bool = None
    value_text = None


    if isinstance(
        value,
        (bool, np.bool_),
    ):

        value_kind = "BOOLEAN"

        value_bool = bool(
            value
        )


    elif isinstance(
        value,
        (
            int,
            np.integer,
        ),
    ):

        value_kind = "INTEGER"

        value_int = int(
            value
        )


    elif isinstance(
        value,
        (
            float,
            np.floating,
        ),
    ):

        value_kind = "FLOAT"

        value_float = float(
            value
        )


    else:

        value_kind = "STATE"

        value_text = str(
            value
        )


    interval_end = (

        interval_start

        + pd.Timedelta(
            minutes=PM_INTERVAL_MINUTES
        )
    )


    dimensions_json = json.dumps(

        dimensions,

        sort_keys=True,

        separators=(",", ":"),
    )


    return {

        "observation_id":
            make_observation_id(

                metric_name,

                entity_id,

                interval_start,

                dimensions,
            ),

        "environment":
            (
                TelemetryEnvironment
                .SYNTHETIC_V1
                .value
            ),

        "observation_type":
            observation_type,

        "evidence_origin":
            evidence_origin,

        "metric_name":
            metric_name,

        "is_standardized_metric":
            bool(
                standardized
            ),

        "operational_category":
            operational_category,

        "entity_id":
            entity.entity_id,

        "entity_namespace":
            entity.namespace.value,

        "entity_type":
            (
                entity.entity_type.value

                if (
                    entity.entity_type
                    is not None
                )

                else None
            ),

        "operational_scope":
            (
                entity.operational_scope.value

                if (
                    entity.operational_scope
                    is not None
                )

                else None
            ),

        "parent_entity_id":
            entity.parent_entity_id,

        "site_id":
            entity.site_id,

        "cu_domain_id":
            entity.cu_domain_id,

        "interval_start_utc":
            interval_start,

        "interval_end_utc":
            interval_end,

        "pm_interval_minutes":
            PM_INTERVAL_MINUTES,

        "value_kind":
            value_kind,

        "value_int":
            value_int,

        "value_float":
            value_float,

        "value_bool":
            value_bool,

        "value_text":
            value_text,

        "unit":
            unit,

        "dimension_cause":
            dimensions.get(
                "Cause"
            ),

        "dimension_5qi":
            dimensions.get(
                "5QI"
            ),

        "dimension_snssai":
            dimensions.get(
                "S-NSSAI"
            ),

        "dimensions_json":
            dimensions_json,

        "quality":
            (
                TelemetryDataQuality
                .VALID
                .value
            ),

        "graph_node_id":
            graph_node_id,

        "investigation_stage":
            investigation_stage,

        "source_system":
            source_system,

        "metadata_json":
            "{}",
    }


# ============================================================
# 13.14 — SEMANTIC ROW HASH UPDATE
# ============================================================

def update_semantic_hasher(
    hasher,
    row,
):

    semantic_record = {

        key:
            (
                value.isoformat()

                if isinstance(
                    value,
                    pd.Timestamp,
                )

                else value
            )

        for key, value
        in row.items()

        if key not in {

            "observation_id",
            "metadata_json",
        }
    }


    hasher.update(

        json.dumps(

            semantic_record,

            sort_keys=True,

            separators=(",", ":"),

            default=str,

        ).encode(
            "utf-8"
        )
    )


    hasher.update(
        b"\n"
    )


# ============================================================
# 13.15 — CELL HEALTHY STATE GENERATION
# ============================================================

def generate_cell_state(
    *,
    rng,
    cell,
    site,
    timestamp,
):

    archetype = (
        site_archetypes[
            site.site_id
        ]
    )


    profile = (

        HEALTHY_PROFILE_BY_ARCHETYPE[
            archetype
        ]
    )


    static = (

        CELL_STATIC_OFFSETS[
            cell.logical_cell_id
        ]
    )


    activity = activity_multiplier(

        timestamp,

        profile[
            "daily_load_amplitude"
        ],
    )


    load_factor = clip(

        activity
        * static[
            "load"
        ]
        * (
            1.0
            + rng.normal(
                0.0,
                0.025,
            )
        ),

        0.40,

        1.45,
    )


    mean_connected = (

        profile[
            "mean_connected_ues"
        ]

        * load_factor
    )


    connected_ues = int(

        max(

            1,

            rng.poisson(
                max(
                    1.0,
                    mean_connected,
                )
            ),
        )
    )


    active_probability = clip(

        0.34
        + 0.20
        * load_factor,

        0.25,

        0.68,
    )


    active_ues = int(

        rng.binomial(

            connected_ues,

            active_probability,
        )
    )


    traffic_pressure = clip(

        (
            active_ues
            / max(
                1.0,
                profile[
                    "mean_connected_ues"
                ],
            )
        )

        * profile[
            "relative_traffic_scale"
        ],

        0.0,

        1.6,
    )


    rsrp_dbm = clip(

        profile[
            "base_rsrp_dbm"
        ]

        + static[
            "rsrp"
        ]

        + rng.normal(
            0.0,
            1.1,
        ),

        -108.0,

        -75.0,
    )


    sinr_db = clip(

        profile[
            "base_sinr_db"
        ]

        + static[
            "sinr"
        ]

        - 2.2
        * max(
            0.0,
            traffic_pressure - 0.65,
        )

        + rng.normal(
            0.0,
            0.9,
        ),

        5.0,

        30.0,
    )


    dl_prb = clip(

        12.0

        + 54.0
        * traffic_pressure

        + rng.normal(
            0.0,
            3.0,
        ),

        4.0,

        82.0,
    )


    ul_prb = clip(

        8.0

        + 38.0
        * traffic_pressure

        + rng.normal(
            0.0,
            2.5,
        ),

        3.0,

        70.0,
    )


    scheduler_pressure = clip(

        0.72
        * dl_prb

        + 0.28
        * ul_prb

        + rng.normal(
            0.0,
            1.5,
        ),

        2.0,

        85.0,
    )


    admission_pressure = clip(

        0.58
        * scheduler_pressure

        + 18.0
        * traffic_pressure

        + rng.normal(
            0.0,
            1.5,
        ),

        3.0,

        82.0,
    )


    radio_quality_pressure = clip(

        (
            22.0 - sinr_db
        )
        / 22.0,

        0.0,

        1.0,
    )


    dl_bler = clip(

        0.25

        + 2.4
        * radio_quality_pressure

        + 0.012
        * dl_prb

        + rng.normal(
            0.0,
            0.15,
        ),

        0.1,

        5.5,
    )


    ul_bler = clip(

        0.30

        + 2.8
        * radio_quality_pressure

        + 0.013
        * ul_prb

        + rng.normal(
            0.0,
            0.18,
        ),

        0.1,

        6.0,
    )


    expected_rrc_attempts = (

        profile[
            "base_rrc_attempts_per_15m"
        ]

        * load_factor

        * static[
            "rach"
        ]
    )


    rrc_total_attempts = int(

        max(

            1,

            rng.poisson(
                expected_rrc_attempts
            ),
        )
    )


    expected_resume_attempts = (

        profile[
            "base_resume_attempts_per_15m"
        ]

        * (
            0.85
            + 0.25
            * load_factor
        )
    )


    resume_total_attempts = int(

        max(

            1,

            rng.poisson(
                expected_resume_attempts
            ),
        )
    )


    rach_activity = (

        12.0

        + 0.030
        * rrc_total_attempts

        + rng.normal(
            0.0,
            1.0,
        )
    )


    rach_contention_index = clip(

        0.12

        + 0.40
        * (
            rach_activity
            / 100.0
        )

        + 0.18
        * (
            ul_prb
            / 100.0
        )

        + rng.normal(
            0.0,
            0.025,
        ),

        0.05,

        0.70,
    )


    rach_access_delay_ms = clip(

        6.0

        + 24.0
        * rach_contention_index

        + 4.0
        * radio_quality_pressure

        + rng.normal(
            0.0,
            0.8,
        ),

        5.0,

        35.0,
    )


    rach_ded = max(

        0.0,

        0.10
        * rach_activity

        + rng.normal(
            0.0,
            0.4,
        ),
    )


    rach_a = max(

        0.0,

        0.63
        * rach_activity

        + rng.normal(
            0.0,
            0.8,
        ),
    )


    rach_b = max(

        0.0,

        0.27
        * rach_activity

        + rng.normal(
            0.0,
            0.6,
        ),
    )


    paging_attempts = int(

        max(

            0,

            rng.poisson(

                max(

                    1.0,

                    connected_ues
                    * (
                        0.08
                        + 0.04
                        * load_factor
                    ),
                )
            ),
        )
    )


    paging_response_probability = clip(

        0.997

        - 0.003
        * radio_quality_pressure,

        0.985,

        0.9995,
    )


    paging_responses = int(

        rng.binomial(

            paging_attempts,

            paging_response_probability,
        )
    )


    paging_load_index = clip(

        paging_attempts

        / max(
            1.0,
            profile[
                "mean_connected_ues"
            ]
            * 0.15,
        ),

        0.0,

        2.0,
    )


    return {

        "archetype":
            archetype,

        "load_factor":
            load_factor,

        "traffic_pressure":
            traffic_pressure,

        "connected_ues":
            connected_ues,

        "active_ues":
            active_ues,

        "rsrp_dbm":
            rsrp_dbm,

        "sinr_db":
            sinr_db,

        "dl_bler_pct":
            dl_bler,

        "ul_bler_pct":
            ul_bler,

        "dl_prb_util_pct":
            dl_prb,

        "ul_prb_util_pct":
            ul_prb,

        "scheduler_pressure_pct":
            scheduler_pressure,

        "admission_pressure_pct":
            admission_pressure,

        "rrc_total_attempts":
            rrc_total_attempts,

        "resume_total_attempts":
            resume_total_attempts,

        "rach_contention_index":
            rach_contention_index,

        "rach_access_delay_ms":
            rach_access_delay_ms,

        "rach_ded":
            float(
                rach_ded
            ),

        "rach_a":
            float(
                rach_a
            ),

        "rach_b":
            float(
                rach_b
            ),

        "paging_attempts":
            paging_attempts,

        "paging_responses":
            paging_responses,

        "paging_load_index":
            paging_load_index,
    }


# ============================================================
# 13.16 — SITE / DU STATE
# ============================================================

def generate_du_state(
    *,
    rng,
    cell_states,
):

    mean_load = float(
        np.mean(
            [
                state[
                    "traffic_pressure"
                ]

                for state
                in cell_states
            ]
        )
    )


    du_cpu = clip(

        18.0

        + 45.0
        * mean_load

        + rng.normal(
            0.0,
            2.0,
        ),

        12.0,

        78.0,
    )


    du_queue = int(

        max(

            0,

            round(

                1.5

                + 12.0
                * (
                    du_cpu
                    / 100.0
                ) ** 2

                + rng.normal(
                    0.0,
                    0.8,
                )
            ),
        )
    )


    f1_latency = clip(

        1.2

        + 1.6
        * mean_load

        + rng.normal(
            0.0,
            0.15,
        ),

        0.7,

        5.0,
    )


    f1_loss = clip(

        0.004

        + 0.025
        * mean_load

        + rng.normal(
            0.0,
            0.004,
        ),

        0.0,

        0.10,
    )


    return {

        "du_cpu_pct":
            du_cpu,

        "du_queue_depth":
            du_queue,

        "f1_latency_ms":
            f1_latency,

        "f1_packet_loss_pct":
            f1_loss,

        "f1_state":
            "UP",
    }


# ============================================================
# 13.17 — SHARED CU / N2 / AMF / E1 STATE
# ============================================================

def generate_cu_domain_state(
    *,
    rng,
    cell_states,
):

    mean_load = float(
        np.mean(
            [
                state[
                    "traffic_pressure"
                ]

                for state
                in cell_states
            ]
        )
    )


    mean_active = float(
        np.mean(
            [
                state[
                    "active_ues"
                ]

                for state
                in cell_states
            ]
        )
    )


    cu_cp_cpu = clip(

        17.0

        + 40.0
        * mean_load

        + rng.normal(
            0.0,
            1.7,
        ),

        12.0,

        74.0,
    )


    cu_cp_memory = clip(

        28.0

        + 25.0
        * mean_load

        + rng.normal(
            0.0,
            1.4,
        ),

        20.0,

        72.0,
    )


    cu_cp_queue = int(

        max(

            0,

            round(

                2.0

                + 16.0
                * (
                    cu_cp_cpu
                    / 100.0
                ) ** 2

                + rng.normal(
                    0.0,
                    0.8,
                )
            ),
        )
    )


    cu_cp_delay = clip(

        0.8

        + 4.0
        * (
            cu_cp_cpu
            / 100.0
        ) ** 2

        + rng.normal(
            0.0,
            0.15,
        ),

        0.5,

        6.0,
    )


    cu_up_cpu = clip(

        19.0

        + 44.0
        * mean_load

        + rng.normal(
            0.0,
            1.8,
        ),

        13.0,

        78.0,
    )


    cu_up_memory = clip(

        30.0

        + 27.0
        * mean_load

        + rng.normal(
            0.0,
            1.5,
        ),

        22.0,

        75.0,
    )


    n2_util = clip(

        12.0

        + 48.0
        * mean_load

        + rng.normal(
            0.0,
            1.8,
        ),

        8.0,

        72.0,
    )


    n2_latency = clip(

        2.5

        + 3.0
        * (
            n2_util
            / 100.0
        )

        + rng.normal(
            0.0,
            0.20,
        ),

        1.5,

        7.0,
    )


    n2_loss = clip(

        0.005

        + 0.045
        * (
            n2_util
            / 100.0
        )

        + rng.normal(
            0.0,
            0.004,
        ),

        0.0,

        0.12,
    )


    sctp_retransmissions = int(

        rng.poisson(

            max(

                0.05,

                0.20
                + 1.5
                * n2_loss,
            )
        )
    )


    e1_latency = clip(

        0.6

        + 1.0
        * mean_load

        + rng.normal(
            0.0,
            0.10,
        ),

        0.4,

        3.0,
    )


    ue_context_availability = clip(

        99.98

        - 0.20
        * (
            cu_cp_cpu
            / 100.0
        )

        + rng.normal(
            0.0,
            0.015,
        ),

        99.50,

        100.0,
    )


    bearer_context_availability = clip(

        99.97

        - 0.20
        * (
            cu_up_cpu
            / 100.0
        )

        + rng.normal(
            0.0,
            0.018,
        ),

        99.45,

        100.0,
    )


    return {

        "mean_load":
            mean_load,

        "mean_active":
            mean_active,

        "cu_cp_cpu_pct":
            cu_cp_cpu,

        "cu_cp_memory_pct":
            cu_cp_memory,

        "cu_cp_queue_depth":
            cu_cp_queue,

        "cu_cp_delay_ms":
            cu_cp_delay,

        "cu_up_cpu_pct":
            cu_up_cpu,

        "cu_up_memory_pct":
            cu_up_memory,

        "n2_util_pct":
            n2_util,

        "n2_latency_ms":
            n2_latency,

        "n2_packet_loss_pct":
            n2_loss,

        "sctp_association_state":
            "UP",

        "sctp_retransmissions":
            sctp_retransmissions,

        "sctp_heartbeat_failures":
            0,

        "amf_reachable":
            True,

        "amf_peer_state":
            "AVAILABLE",

        "amf_overload_indication":
            False,

        "e1_latency_ms":
            e1_latency,

        "e1_state":
            "UP",

        "ue_context_availability_pct":
            ue_context_availability,

        "bearer_context_availability_pct":
            bearer_context_availability,
    }


# ============================================================
# 13.18 — HEALTHY SUCCESS-PROBABILITY MODEL
# ============================================================
#
# Smooth synthetic relationships only.
#
# These are NOT standardized thresholds.
#
# ============================================================

def success_probabilities(
    *,
    cell_state,
    du_state,
    domain_state,
):

    coverage_pressure = clip(

        (
            -80.0
            - cell_state[
                "rsrp_dbm"
            ]
        )
        / 35.0,

        0.0,

        1.0,
    )


    quality_pressure = clip(

        (
            24.0
            - cell_state[
                "sinr_db"
            ]
        )
        / 30.0,

        0.0,

        1.0,
    )


    resource_pressure = clip(

        max(

            cell_state[
                "dl_prb_util_pct"
            ],

            cell_state[
                "ul_prb_util_pct"
            ],
        )
        / 100.0,

        0.0,

        1.0,
    )


    du_pressure = clip(

        du_state[
            "du_cpu_pct"
        ]
        / 100.0,

        0.0,

        1.0,
    )


    cu_cp_pressure = clip(

        domain_state[
            "cu_cp_cpu_pct"
        ]
        / 100.0,

        0.0,

        1.0,
    )


    cu_up_pressure = clip(

        domain_state[
            "cu_up_cpu_pct"
        ]
        / 100.0,

        0.0,

        1.0,
    )


    n2_pressure = clip(

        domain_state[
            "n2_util_pct"
        ]
        / 100.0,

        0.0,

        1.0,
    )


    admission_pressure = clip(

        cell_state[
            "admission_pressure_pct"
        ]
        / 100.0,

        0.0,

        1.0,
    )


    context_pressure = clip(

        (
            100.0
            - domain_state[
                "ue_context_availability_pct"
            ]
        )
        / 2.0,

        0.0,

        1.0,
    )


    p_rrc = clip(

        0.9995

        - 0.0030
        * coverage_pressure

        - 0.0024
        * quality_pressure

        - 0.0015
        * resource_pressure

        - 0.0010
        * du_pressure,

        0.985,

        0.9997,
    )


    p_ng = clip(

        0.9996

        - 0.0015
        * cu_cp_pressure

        - 0.0011
        * n2_pressure

        - 0.0015
        * (
            domain_state[
                "n2_latency_ms"
            ]
            / 20.0
        )

        - 0.012
        * domain_state[
            "n2_packet_loss_pct"
        ],

        0.985,

        0.9997,
    )


    p_initial_drb = clip(

        0.9994

        - 0.0020
        * quality_pressure

        - 0.0025
        * resource_pressure

        - 0.0018
        * admission_pressure

        - 0.0012
        * cu_up_pressure,

        0.982,

        0.9996,
    )


    p_drb = clip(

        0.9993

        - 0.0022
        * quality_pressure

        - 0.0028
        * resource_pressure

        - 0.0020
        * admission_pressure

        - 0.0013
        * cu_up_pressure,

        0.980,

        0.9996,
    )


    p_resume = clip(

        0.9990

        - 0.0022
        * coverage_pressure

        - 0.0022
        * quality_pressure

        - 0.0013
        * cu_cp_pressure

        - 0.0030
        * context_pressure,

        0.975,

        0.9995,
    )


    p_drb_resume = clip(

        0.9991

        - 0.0020
        * quality_pressure

        - 0.0025
        * resource_pressure

        - 0.0015
        * cu_up_pressure

        - 0.0030
        * (
            (
                100.0
                - domain_state[
                    "bearer_context_availability_pct"
                ]
            )
            / 2.0
        ),

        0.975,

        0.9995,
    )


    return {

        "rrc":
            p_rrc,

        "ng":
            p_ng,

        "initial_drb":
            p_initial_drb,

        "drb":
            p_drb,

        "resume":
            p_resume,

        "drb_resume":
            p_drb_resume,
    }


# ============================================================
# 13.19 — GLOBAL CONSTRUCTION SANITY COUNTERS
# ============================================================

SANITY_COUNTS = {

    "rrc_att": 0,
    "rrc_succ": 0,

    "ng_att": 0,
    "ng_succ": 0,

    "initial_drb_att": 0,
    "initial_drb_succ": 0,

    "drb_att": 0,
    "drb_succ": 0,

    "resume_att": 0,
    "resume_succ": 0,

    "drb_resume_att": 0,
    "drb_resume_succ": 0,

    "resume_fallback_att": 0,
    "resume_fallback_succ": 0,
}


PAIR_CONSTRAINT_VIOLATIONS = 0


# ============================================================
# 13.20 — STANDARDIZED PM GENERATION
# ============================================================

def emit_standardized_pm_rows(
    *,
    rows,
    hasher,
    rng,
    timestamp,
    cell,
    cell_state,
    du_state,
    domain_state,
):

    global PAIR_CONSTRAINT_VIOLATIONS


    cu_entity_id = (
        cell.nr_cell_cu.entity_id
    )


    du_entity_id = (
        cell.nr_cell_du.entity_id
    )


    probabilities = success_probabilities(

        cell_state=cell_state,

        du_state=du_state,

        domain_state=domain_state,
    )


    # --------------------------------------------------------
    # RRC CONNECTION ESTABLISHMENT
    # --------------------------------------------------------

    rrc_attempts_by_cause = allocate_multinomial(

        rng,

        cell_state[
            "rrc_total_attempts"
        ],

        [
            0.65,
            0.20,
            0.15,
        ],
    )


    cause_adjustment = {

        "mo-Data": 0.0000,

        "mo-Signalling": 0.0004,

        "mt-Access": -0.0004,
    }


    rrc_success_by_cause = []


    for (
        cause,
        attempts
    ) in zip(
        CAUSE_VALUES,
        rrc_attempts_by_cause,
    ):

        p = clip(

            probabilities[
                "rrc"
            ]
            + cause_adjustment[
                cause
            ],

            0.0,

            1.0,
        )


        successes = int(

            rng.binomial(
                attempts,
                p,
            )
        )


        rrc_success_by_cause.append(
            successes
        )


        for (
            metric_name,
            value
        ) in [

            (
                "RRC.ConnEstabAtt.Cause",
                attempts,
            ),

            (
                "RRC.ConnEstabSucc.Cause",
                successes,
            ),
        ]:

            row = build_observation_row(

                metric_name=metric_name,

                entity_id=cu_entity_id,

                interval_start=timestamp,

                value=value,

                dimensions={
                    "Cause":
                        cause
                },

                standardized=True,

                unit="count",
            )


            rows.append(
                row
            )

            update_semantic_hasher(
                hasher,
                row,
            )


    SANITY_COUNTS[
        "rrc_att"
    ] += sum(
        rrc_attempts_by_cause
    )

    SANITY_COUNTS[
        "rrc_succ"
    ] += sum(
        rrc_success_by_cause
    )


    # --------------------------------------------------------
    # UE-ASSOCIATED LOGICAL NG CONNECTION
    # --------------------------------------------------------

    ng_attempts_by_cause = []

    ng_success_by_cause = []


    for (
        cause,
        upstream_success
    ) in zip(
        CAUSE_VALUES,
        rrc_success_by_cause,
    ):

        attempts = int(

            rng.binomial(

                upstream_success,

                0.997,
            )
        )


        p = clip(

            probabilities[
                "ng"
            ]
            + (
                0.0002

                if (
                    cause
                    == "mo-Signalling"
                )

                else 0.0
            ),

            0.0,

            1.0,
        )


        successes = int(

            rng.binomial(
                attempts,
                p,
            )
        )


        ng_attempts_by_cause.append(
            attempts
        )

        ng_success_by_cause.append(
            successes
        )


        for (
            metric_name,
            value
        ) in [

            (
                "UECNTX.ConnEstabAtt.Cause",
                attempts,
            ),

            (
                "UECNTX.ConnEstabSucc.Cause",
                successes,
            ),
        ]:

            row = build_observation_row(

                metric_name=metric_name,

                entity_id=cu_entity_id,

                interval_start=timestamp,

                value=value,

                dimensions={
                    "Cause":
                        cause
                },

                standardized=True,

                unit="count",
            )


            rows.append(
                row
            )

            update_semantic_hasher(
                hasher,
                row,
            )


    SANITY_COUNTS[
        "ng_att"
    ] += sum(
        ng_attempts_by_cause
    )

    SANITY_COUNTS[
        "ng_succ"
    ] += sum(
        ng_success_by_cause
    )


    # --------------------------------------------------------
    # SERVICE-ELIGIBLE NG SUCCESSES
    # --------------------------------------------------------
    #
    # mo-Signalling is retained in the PM data but not used as
    # the service-demand base here.
    #
    # --------------------------------------------------------

    ng_service_successes = (

        ng_success_by_cause[
            0
        ]

        + ng_success_by_cause[
            2
        ]
    )


    # --------------------------------------------------------
    # INITIAL DRB ESTABLISHMENT — 5QI
    # --------------------------------------------------------

    initial_drb_total_attempts = int(

        rng.poisson(

            max(

                1.0,

                ng_service_successes
                * 0.90,
            )
        )
    )


    initial_att_by_5qi = (
        allocate_multinomial(

            rng,

            initial_drb_total_attempts,

            [
                0.18,
                0.12,
                0.70,
            ],
        )
    )


    initial_success_by_5qi = []


    qci_adjustment = {

        "1": -0.0003,

        "5": 0.0000,

        "9": 0.0002,
    }


    for (
        five_qi,
        attempts
    ) in zip(
        FIVE_QI_VALUES,
        initial_att_by_5qi,
    ):

        p = clip(

            probabilities[
                "initial_drb"
            ]
            + qci_adjustment[
                five_qi
            ],

            0.0,

            1.0,
        )


        successes = int(

            rng.binomial(
                attempts,
                p,
            )
        )


        initial_success_by_5qi.append(
            successes
        )


        for (
            metric_name,
            value
        ) in [

            (
                "DRB.InitialEstabAtt.5QI",
                attempts,
            ),

            (
                "DRB.InitialEstabSucc.5QI",
                successes,
            ),
        ]:

            row = build_observation_row(

                metric_name=metric_name,

                entity_id=cu_entity_id,

                interval_start=timestamp,

                value=value,

                dimensions={
                    "5QI":
                        five_qi
                },

                standardized=True,

                unit="count",
            )


            rows.append(
                row
            )

            update_semantic_hasher(
                hasher,
                row,
            )


    SANITY_COUNTS[
        "initial_drb_att"
    ] += sum(
        initial_att_by_5qi
    )

    SANITY_COUNTS[
        "initial_drb_succ"
    ] += sum(
        initial_success_by_5qi
    )


    # --------------------------------------------------------
    # DRB ESTABLISHMENT — 5QI
    # --------------------------------------------------------

    drb_total_attempts = int(

        rng.poisson(

            max(

                1.0,

                ng_service_successes
                * 1.12,
            )
        )
    )


    drb_att_by_5qi = (
        allocate_multinomial(

            rng,

            drb_total_attempts,

            [
                0.18,
                0.12,
                0.70,
            ],
        )
    )


    drb_success_by_5qi = []


    for (
        five_qi,
        attempts
    ) in zip(
        FIVE_QI_VALUES,
        drb_att_by_5qi,
    ):

        p = clip(

            probabilities[
                "drb"
            ]
            + qci_adjustment[
                five_qi
            ],

            0.0,

            1.0,
        )


        successes = int(

            rng.binomial(
                attempts,
                p,
            )
        )


        drb_success_by_5qi.append(
            successes
        )


        for (
            metric_name,
            value
        ) in [

            (
                "DRB.EstabAtt.5QI",
                attempts,
            ),

            (
                "DRB.EstabSucc.5QI",
                successes,
            ),
        ]:

            row = build_observation_row(

                metric_name=metric_name,

                entity_id=cu_entity_id,

                interval_start=timestamp,

                value=value,

                dimensions={
                    "5QI":
                        five_qi
                },

                standardized=True,

                unit="count",
            )


            rows.append(
                row
            )

            update_semantic_hasher(
                hasher,
                row,
            )


    SANITY_COUNTS[
        "drb_att"
    ] += sum(
        drb_att_by_5qi
    )

    SANITY_COUNTS[
        "drb_succ"
    ] += sum(
        drb_success_by_5qi
    )


    # --------------------------------------------------------
    # DRB ESTABLISHMENT — S-NSSAI
    # --------------------------------------------------------

    for (
        metric_name,
        value
    ) in [

        (
            "DRB.EstabAtt.SNSSAI",
            sum(
                drb_att_by_5qi
            ),
        ),

        (
            "DRB.EstabSucc.SNSSAI",
            sum(
                drb_success_by_5qi
            ),
        ),
    ]:

        row = build_observation_row(

            metric_name=metric_name,

            entity_id=cu_entity_id,

            interval_start=timestamp,

            value=value,

            dimensions={
                "S-NSSAI":
                    SNSSAI_VALUES[
                        0
                    ]
            },

            standardized=True,

            unit="count",
        )


        rows.append(
            row
        )

        update_semantic_hasher(
            hasher,
            row,
        )


    # --------------------------------------------------------
    # RRC RESUME
    # --------------------------------------------------------

    resume_att_by_cause = (
        allocate_multinomial(

            rng,

            cell_state[
                "resume_total_attempts"
            ],

            [
                0.70,
                0.12,
                0.18,
            ],
        )
    )


    resume_success_by_cause = []

    fallback_att_by_cause = []

    fallback_success_by_cause = []


    for (
        cause,
        attempts
    ) in zip(
        CAUSE_VALUES,
        resume_att_by_cause,
    ):

        p = clip(

            probabilities[
                "resume"
            ]
            + cause_adjustment[
                cause
            ],

            0.0,

            1.0,
        )


        direct_success = int(

            rng.binomial(
                attempts,
                p,
            )
        )


        direct_failures = (

            attempts

            - direct_success
        )


        fallback_attempts = int(

            rng.binomial(

                direct_failures,

                0.88,
            )
        )


        fallback_success = int(

            rng.binomial(

                fallback_attempts,

                0.985,
            )
        )


        resume_success_by_cause.append(
            direct_success
        )

        fallback_att_by_cause.append(
            fallback_attempts
        )

        fallback_success_by_cause.append(
            fallback_success
        )


        metric_values = [

            (
                "RRC.ResumeAtt.Cause",
                attempts,
            ),

            (
                "RRC.ResumeSucc.Cause",
                direct_success,
            ),

            (
                "RRC.ResumeFallbackToSetupAtt.cause",
                fallback_attempts,
            ),

            (
                "RRC.ResumeSuccByFallback.cause",
                fallback_success,
            ),
        ]


        for (
            metric_name,
            value
        ) in metric_values:

            row = build_observation_row(

                metric_name=metric_name,

                entity_id=cu_entity_id,

                interval_start=timestamp,

                value=value,

                dimensions={
                    "Cause":
                        cause
                },

                standardized=True,

                unit="count",
            )


            rows.append(
                row
            )

            update_semantic_hasher(
                hasher,
                row,
            )


    SANITY_COUNTS[
        "resume_att"
    ] += sum(
        resume_att_by_cause
    )

    SANITY_COUNTS[
        "resume_succ"
    ] += sum(
        resume_success_by_cause
    )

    SANITY_COUNTS[
        "resume_fallback_att"
    ] += sum(
        fallback_att_by_cause
    )

    SANITY_COUNTS[
        "resume_fallback_succ"
    ] += sum(
        fallback_success_by_cause
    )


    # --------------------------------------------------------
    # RRC RE-ESTABLISHMENT SUPPORTING DEPENDENCIES
    # --------------------------------------------------------

    residual_resume_failures = int(

        max(

            0,

            sum(
                resume_att_by_cause
            )

            - sum(
                resume_success_by_cause
            )

            - sum(
                fallback_success_by_cause
            ),
        )
    )


    reestab_without_context = int(

        rng.binomial(

            residual_resume_failures,

            0.45,
        )
    )


    remaining_reestab = (

        residual_resume_failures

        - reestab_without_context
    )


    reestab_fallback_setup = int(

        rng.binomial(

            remaining_reestab,

            0.70,
        )
    )


    for (
        metric_name,
        value
    ) in [

        (
            "RRC.ReEstabSuccWithoutUeContext",
            reestab_without_context,
        ),

        (
            "RRC.ReEstabFallbackToSetupAtt",
            reestab_fallback_setup,
        ),
    ]:

        row = build_observation_row(

            metric_name=metric_name,

            entity_id=cu_entity_id,

            interval_start=timestamp,

            value=value,

            standardized=True,

            unit="count",
        )


        rows.append(
            row
        )

        update_semantic_hasher(
            hasher,
            row,
        )


    # --------------------------------------------------------
    # DRB RESUME
    # --------------------------------------------------------

    successful_resume_events = (

        sum(
            resume_success_by_cause
        )

        + sum(
            fallback_success_by_cause
        )
    )


    drb_resume_total_attempts = int(

        rng.poisson(

            max(

                1.0,

                successful_resume_events
                * 1.05,
            )
        )
    )


    drb_resume_att_by_5qi = (
        allocate_multinomial(

            rng,

            drb_resume_total_attempts,

            [
                0.18,
                0.12,
                0.70,
            ],
        )
    )


    drb_resume_success_by_5qi = []


    for (
        five_qi,
        attempts
    ) in zip(
        FIVE_QI_VALUES,
        drb_resume_att_by_5qi,
    ):

        p = clip(

            probabilities[
                "drb_resume"
            ]
            + qci_adjustment[
                five_qi
            ],

            0.0,

            1.0,
        )


        successes = int(

            rng.binomial(
                attempts,
                p,
            )
        )


        drb_resume_success_by_5qi.append(
            successes
        )


        for (
            metric_name,
            value
        ) in [

            (
                "DRB.ResumeAtt.5QI",
                attempts,
            ),

            (
                "DRB.ResumeSucc.5QI",
                successes,
            ),
        ]:

            row = build_observation_row(

                metric_name=metric_name,

                entity_id=cu_entity_id,

                interval_start=timestamp,

                value=value,

                dimensions={
                    "5QI":
                        five_qi
                },

                standardized=True,

                unit="count",
            )


            rows.append(
                row
            )

            update_semantic_hasher(
                hasher,
                row,
            )


    SANITY_COUNTS[
        "drb_resume_att"
    ] += sum(
        drb_resume_att_by_5qi
    )

    SANITY_COUNTS[
        "drb_resume_succ"
    ] += sum(
        drb_resume_success_by_5qi
    )


    # --------------------------------------------------------
    # STANDARDIZED SUPPORTING RACH PM
    # --------------------------------------------------------

    for (
        metric_name,
        value
    ) in [

        (
            "RACH.PreambleDedCell",
            cell_state[
                "rach_ded"
            ],
        ),

        (
            "RACH.PreambleACell",
            cell_state[
                "rach_a"
            ],
        ),

        (
            "RACH.PreambleBCell",
            cell_state[
                "rach_b"
            ],
        ),
    ]:

        row = build_observation_row(

            metric_name=metric_name,

            entity_id=du_entity_id,

            interval_start=timestamp,

            value=float(
                value
            ),

            standardized=True,

            unit="average_count",
        )


        rows.append(
            row
        )

        update_semantic_hasher(
            hasher,
            row,
        )


    # --------------------------------------------------------
    # LOCAL PAIR VALIDATION
    # --------------------------------------------------------

    pair_checks = [

        (
            sum(
                rrc_success_by_cause
            ),

            sum(
                rrc_attempts_by_cause
            ),
        ),

        (
            sum(
                ng_success_by_cause
            ),

            sum(
                ng_attempts_by_cause
            ),
        ),

        (
            sum(
                initial_success_by_5qi
            ),

            sum(
                initial_att_by_5qi
            ),
        ),

        (
            sum(
                drb_success_by_5qi
            ),

            sum(
                drb_att_by_5qi
            ),
        ),

        (
            sum(
                resume_success_by_cause
            ),

            sum(
                resume_att_by_cause
            ),
        ),

        (
            sum(
                fallback_success_by_cause
            ),

            sum(
                fallback_att_by_cause
            ),
        ),

        (
            sum(
                drb_resume_success_by_5qi
            ),

            sum(
                drb_resume_att_by_5qi
            ),
        ),
    ]


    for (
        successes,
        attempts
    ) in pair_checks:

        if successes > attempts:

            PAIR_CONSTRAINT_VIOLATIONS += 1


# ============================================================
# 13.21 — OPERATIONAL CELL OBSERVATIONS
# ============================================================

def emit_cell_operational_rows(
    *,
    rows,
    hasher,
    timestamp,
    cell,
    state,
):

    cu_id = (
        cell.nr_cell_cu.entity_id
    )

    du_id = (
        cell.nr_cell_du.entity_id
    )


    metric_values = [

        # RADIO
        (
            "radio.rsrp_dbm",
            du_id,
            state["rsrp_dbm"],
        ),

        (
            "radio.sinr_db",
            du_id,
            state["sinr_db"],
        ),

        (
            "radio.dl_bler_pct",
            du_id,
            state["dl_bler_pct"],
        ),

        (
            "radio.ul_bler_pct",
            du_id,
            state["ul_bler_pct"],
        ),

        # RESOURCE
        (
            "resource.dl_prb_util_pct",
            du_id,
            state["dl_prb_util_pct"],
        ),

        (
            "resource.ul_prb_util_pct",
            du_id,
            state["ul_prb_util_pct"],
        ),

        (
            "resource.scheduler_pressure_pct",
            du_id,
            state["scheduler_pressure_pct"],
        ),

        (
            "load.connected_ues",
            cu_id,
            state["connected_ues"],
        ),

        (
            "load.active_ues",
            cu_id,
            state["active_ues"],
        ),

        (
            "resource.bearer_admission_pressure_pct",
            cu_id,
            state["admission_pressure_pct"],
        ),

        # RACH
        (
            "rach.contention_index",
            du_id,
            state["rach_contention_index"],
        ),

        (
            "rach.access_delay_ms",
            du_id,
            state["rach_access_delay_ms"],
        ),

        # PAGING
        (
            "paging.attempts",
            cu_id,
            state["paging_attempts"],
        ),

        (
            "paging.responses",
            cu_id,
            state["paging_responses"],
        ),

        (
            "paging.load_index",
            cu_id,
            state["paging_load_index"],
        ),

        # CONFIGURATION SNAPSHOTS
        (
            "config.prach_profile_state",
            du_id,
            "BASELINE",
        ),

        (
            "config.rrc_profile_state",
            cu_id,
            "BASELINE",
        ),

        (
            "config.qos_profile_state",
            cu_id,
            "BASELINE",
        ),

        (
            "config.slice_profile_state",
            cu_id,
            "BASELINE",
        ),
    ]


    for (
        metric_name,
        entity_id,
        value
    ) in metric_values:

        row = build_observation_row(

            metric_name=metric_name,

            entity_id=entity_id,

            interval_start=timestamp,

            value=value,

            standardized=False,
        )


        rows.append(
            row
        )

        update_semantic_hasher(
            hasher,
            row,
        )


# ============================================================
# 13.22 — SITE / DU + F1 OBSERVATIONS
# ============================================================

def emit_site_operational_rows(
    *,
    rows,
    hasher,
    timestamp,
    site,
    state,
):

    metric_values = [

        (
            "processing.du_cpu_pct",
            site.du_id,
            state["du_cpu_pct"],
        ),

        (
            "processing.du_queue_depth",
            site.du_id,
            state["du_queue_depth"],
        ),

        (
            "transport.f1_latency_ms",
            site.f1_path_id,
            state["f1_latency_ms"],
        ),

        (
            "transport.f1_packet_loss_pct",
            site.f1_path_id,
            state["f1_packet_loss_pct"],
        ),

        (
            "transport.f1_state",
            site.f1_path_id,
            state["f1_state"],
        ),
    ]


    for (
        metric_name,
        entity_id,
        value
    ) in metric_values:

        row = build_observation_row(

            metric_name=metric_name,

            entity_id=entity_id,

            interval_start=timestamp,

            value=value,

            standardized=False,
        )


        rows.append(
            row
        )

        update_semantic_hasher(
            hasher,
            row,
        )


# ============================================================
# 13.23 — SHARED CU / N2 / AMF / E1 OBSERVATIONS
# ============================================================

def emit_domain_operational_rows(
    *,
    rows,
    hasher,
    timestamp,
    domain_sites,
    state,
):

    first_site = (
        domain_sites[
            0
        ]
    )


    cu_cp_id = (
        first_site.cu_cp_id
    )

    cu_up_id = (
        first_site.cu_up_id
    )

    n2_id = (
        first_site.n2_path_id
    )

    amf_id = (
        first_site.amf_peer_id
    )

    e1_id = (
        first_site.e1_path_id
    )


    metric_values = [

        # CU-CP
        (
            "processing.cu_cp_cpu_pct",
            cu_cp_id,
            state["cu_cp_cpu_pct"],
        ),

        (
            "processing.cu_cp_memory_pct",
            cu_cp_id,
            state["cu_cp_memory_pct"],
        ),

        (
            "processing.cu_cp_control_queue_depth",
            cu_cp_id,
            state["cu_cp_queue_depth"],
        ),

        (
            "processing.cu_cp_control_delay_ms",
            cu_cp_id,
            state["cu_cp_delay_ms"],
        ),

        # CU-UP
        (
            "processing.cu_up_cpu_pct",
            cu_up_id,
            state["cu_up_cpu_pct"],
        ),

        (
            "processing.cu_up_memory_pct",
            cu_up_id,
            state["cu_up_memory_pct"],
        ),

        # N2
        (
            "transport.n2_util_pct",
            n2_id,
            state["n2_util_pct"],
        ),

        (
            "transport.n2_latency_ms",
            n2_id,
            state["n2_latency_ms"],
        ),

        (
            "transport.n2_packet_loss_pct",
            n2_id,
            state["n2_packet_loss_pct"],
        ),

        # SCTP
        (
            "sctp.association_state",
            n2_id,
            state["sctp_association_state"],
        ),

        (
            "sctp.retransmissions",
            n2_id,
            state["sctp_retransmissions"],
        ),

        (
            "sctp.heartbeat_failures",
            n2_id,
            state["sctp_heartbeat_failures"],
        ),

        # AMF-FACING BOUNDARY
        (
            "boundary.amf_reachable",
            amf_id,
            state["amf_reachable"],
        ),

        (
            "boundary.amf_peer_state",
            amf_id,
            state["amf_peer_state"],
        ),

        (
            "boundary.amf_overload_indication",
            amf_id,
            state["amf_overload_indication"],
        ),

        (
            "config.n2_peer_state",
            amf_id,
            "BASELINE",
        ),

        # E1
        (
            "transport.e1_latency_ms",
            e1_id,
            state["e1_latency_ms"],
        ),

        (
            "transport.e1_state",
            e1_id,
            state["e1_state"],
        ),

        # CONTEXT
        (
            "context.ue_context_availability_pct",
            cu_cp_id,
            state["ue_context_availability_pct"],
        ),

        (
            "context.bearer_context_availability_pct",
            cu_up_id,
            state["bearer_context_availability_pct"],
        ),

        # RESUME CONFIG
        (
            "config.resume_profile_state",
            cu_cp_id,
            "BASELINE",
        ),
    ]


    for (
        metric_name,
        entity_id,
        value
    ) in metric_values:

        row = build_observation_row(

            metric_name=metric_name,

            entity_id=entity_id,

            interval_start=timestamp,

            value=value,

            standardized=False,
        )


        rows.append(
            row
        )

        update_semantic_hasher(
            hasher,
            row,
        )


# ============================================================
# 13.24 — EXPECTED ROW COUNTS
# ============================================================

STANDARD_ROWS_PER_CELL_INTERVAL = 49

STANDARD_ROWS_PER_INTERVAL = (

    30
    * STANDARD_ROWS_PER_CELL_INTERVAL
)


continuous_operational_bindings = {

    metric_name:
        binding

    for (
        metric_name,
        binding
    ) in (
        RAN_OPERATIONAL_METRIC_BINDINGS
        .items()
    )

    if (
        binding.cadence
        != OperationalMetricCadence
        .EVENT_DRIVEN
    )
}


event_driven_bindings = {

    metric_name:
        binding

    for (
        metric_name,
        binding
    ) in (
        RAN_OPERATIONAL_METRIC_BINDINGS
        .items()
    )

    if (
        binding.cadence
        == OperationalMetricCadence
        .EVENT_DRIVEN
    )
}


assert (
    len(
        continuous_operational_bindings
    )
    == 45
)


assert (
    len(
        event_driven_bindings
    )
    == 4
)


ENTITY_COUNT_BY_SCOPE = {

    "NR_CELL_DU": 30,

    "NR_CELL_CU": 30,

    "DU": 10,

    "CU_CP": 2,

    "CU_UP": 2,

    "N2_PATH": 2,

    "AMF_PEER": 2,

    "F1_PATH": 10,

    "E1_PATH": 2,

    "CU_DOMAIN": 2,

    "SITE": 10,

    "LOGICAL_CELL": 30,
}


EXPECTED_OPERATIONAL_ROWS_PER_INTERVAL = sum(

    ENTITY_COUNT_BY_SCOPE[
        binding.preferred_scope.value
    ]

    for binding
    in continuous_operational_bindings.values()
)


assert (
    EXPECTED_OPERATIONAL_ROWS_PER_INTERVAL
    == 662
)


EXPECTED_ROWS_PER_INTERVAL = (

    STANDARD_ROWS_PER_INTERVAL

    + EXPECTED_OPERATIONAL_ROWS_PER_INTERVAL
)


assert (
    EXPECTED_ROWS_PER_INTERVAL
    == 2132
)


EXPECTED_ROWS_PER_DAY = (

    EXPECTED_ROWS_PER_INTERVAL
    * 96
)


assert (
    EXPECTED_ROWS_PER_DAY
    == 204672
)


EXPECTED_STANDARD_ROWS = (
    141_120
    * BASELINE_DAY_COUNT
)

EXPECTED_OPERATIONAL_ROWS = (
    63_552
    * BASELINE_DAY_COUNT
)

EXPECTED_TOTAL_ROWS = (
    EXPECTED_ROWS_PER_DAY
    * BASELINE_DAY_COUNT
)

assert (
    EXPECTED_TOTAL_ROWS
    == (
        EXPECTED_STANDARD_ROWS
        + EXPECTED_OPERATIONAL_ROWS
    )
)


# ============================================================
# 13.25 — OUTPUT PATHS
# ============================================================

telemetry_root = (

    PATHS["kpi"].parent

    / "telemetry"
)


FINAL_DATASET_DIR = (
    telemetry_root
    / BASELINE_DATASET_DIRNAME
)

STAGING_DATASET_DIR = (
    telemetry_root
    / f".{BASELINE_DATASET_DIRNAME}_staging"
)

if FINAL_DATASET_DIR.exists():
    raise FileExistsError(
        f"{BASELINE_MODE} dataset already exists. "
        "Do not overwrite a completed dataset. "
        f"Path: {FINAL_DATASET_DIR}"
    )


if STAGING_DATASET_DIR.exists():

    shutil.rmtree(
        STAGING_DATASET_DIR
    )


STAGING_DATASET_DIR.mkdir(

    parents=True,

    exist_ok=False,
)


# ============================================================
# 13.26 — GENERATOR
# ============================================================

rng = np.random.default_rng(
    HEALTHY_GENERATOR_SEED
)


daily_manifest_entries = []

generated_metric_names = set()

total_rows_generated = 0

standard_rows_generated = 0

operational_rows_generated = 0


for day_start in pd.date_range(

    HEALTHY_START_UTC,

    HEALTHY_END_EXCLUSIVE_UTC
    - pd.Timedelta(
        days=1
    ),

    freq="1D",
):

    day_end = (

        day_start

        + pd.Timedelta(
            days=1
        )
    )


    day_intervals = [

        timestamp

        for timestamp
        in interval_starts

        if (
            timestamp >= day_start

            and timestamp < day_end
        )
    ]


    assert (
        len(
            day_intervals
        )
        == 96
    )


    day_rows = []

    day_semantic_hasher = (
        hashlib.sha256()
    )


    for timestamp in day_intervals:

        # ----------------------------------------------------
        # 1. CELL STATES
        # ----------------------------------------------------

        cell_states = {}


        for site in sorted(

            SYNTHETIC_RAN_SITES_V1_1,

            key=lambda item:
                item.site_id,
        ):

            for cell in sorted(

                site.cells,

                key=lambda item:
                    item.logical_cell_id,
            ):

                cell_states[
                    cell.logical_cell_id
                ] = (
                    generate_cell_state(

                        rng=rng,

                        cell=cell,

                        site=site,

                        timestamp=timestamp,
                    )
                )


        # ----------------------------------------------------
        # 2. SITE / DU STATES
        # ----------------------------------------------------

        du_states = {}


        for site in sorted(

            SYNTHETIC_RAN_SITES_V1_1,

            key=lambda item:
                item.site_id,
        ):

            states = [

                cell_states[
                    cell.logical_cell_id
                ]

                for cell
                in site.cells
            ]


            du_states[
                site.site_id
            ] = (
                generate_du_state(

                    rng=rng,

                    cell_states=states,
                )
            )


        # ----------------------------------------------------
        # 3. SHARED CU-DOMAIN STATES
        # ----------------------------------------------------

        domain_states = {}


        for cu_domain_id in sorted(

            SYNTHETIC_RAN_TOPOLOGY_V1_1
            .cu_domains
        ):

            domain_cells = (

                CELLS_BY_CU_DOMAIN_V1_1[
                    cu_domain_id
                ]
            )


            states = [

                cell_states[
                    cell.logical_cell_id
                ]

                for cell
                in domain_cells
            ]


            domain_states[
                cu_domain_id
            ] = (
                generate_cu_domain_state(

                    rng=rng,

                    cell_states=states,
                )
            )


        # ----------------------------------------------------
        # 4. CELL OPERATIONAL TELEMETRY
        # ----------------------------------------------------

        for site in sorted(

            SYNTHETIC_RAN_SITES_V1_1,

            key=lambda item:
                item.site_id,
        ):

            for cell in sorted(

                site.cells,

                key=lambda item:
                    item.logical_cell_id,
            ):

                emit_cell_operational_rows(

                    rows=day_rows,

                    hasher=day_semantic_hasher,

                    timestamp=timestamp,

                    cell=cell,

                    state=(
                        cell_states[
                            cell.logical_cell_id
                        ]
                    ),
                )


        # ----------------------------------------------------
        # 5. SITE / DU / F1 TELEMETRY
        # ----------------------------------------------------

        for site in sorted(

            SYNTHETIC_RAN_SITES_V1_1,

            key=lambda item:
                item.site_id,
        ):

            emit_site_operational_rows(

                rows=day_rows,

                hasher=day_semantic_hasher,

                timestamp=timestamp,

                site=site,

                state=(
                    du_states[
                        site.site_id
                    ]
                ),
            )


        # ----------------------------------------------------
        # 6. SHARED CU / N2 / AMF / E1 TELEMETRY
        # ----------------------------------------------------

        for cu_domain_id in sorted(

            SYNTHETIC_RAN_TOPOLOGY_V1_1
            .cu_domains
        ):

            domain_sites = [

                site

                for site
                in SYNTHETIC_RAN_SITES_V1_1

                if (
                    site.cu_domain_id
                    == cu_domain_id
                )
            ]


            emit_domain_operational_rows(

                rows=day_rows,

                hasher=day_semantic_hasher,

                timestamp=timestamp,

                domain_sites=domain_sites,

                state=(
                    domain_states[
                        cu_domain_id
                    ]
                ),
            )


        # ----------------------------------------------------
        # 7. STANDARDIZED PM COUNTERS
        # ----------------------------------------------------

        for site in sorted(

            SYNTHETIC_RAN_SITES_V1_1,

            key=lambda item:
                item.site_id,
        ):

            for cell in sorted(

                site.cells,

                key=lambda item:
                    item.logical_cell_id,
            ):

                emit_standardized_pm_rows(

                    rows=day_rows,

                    hasher=day_semantic_hasher,

                    rng=rng,

                    timestamp=timestamp,

                    cell=cell,

                    cell_state=(
                        cell_states[
                            cell.logical_cell_id
                        ]
                    ),

                    du_state=(
                        du_states[
                            site.site_id
                        ]
                    ),

                    domain_state=(
                        domain_states[
                            site.cu_domain_id
                        ]
                    ),
                )


    # --------------------------------------------------------
    # DAILY DATAFRAME
    # --------------------------------------------------------

    day_df = pd.DataFrame(

        day_rows,

        columns=OBSERVATION_COLUMNS,
    )


    assert (
        len(
            day_df
        )
        == EXPECTED_ROWS_PER_DAY
    )


    assert (
        day_df[
            "observation_id"
        ].is_unique
    )


    assert (
        not day_df[
            "metric_name"
        ].isna().any()
    )


    generated_metric_names.update(

        day_df[
            "metric_name"
        ].unique()
    )


    day_standard_rows = int(

        day_df[
            "is_standardized_metric"
        ].sum()
    )


    day_operational_rows = (

        len(
            day_df
        )

        - day_standard_rows
    )


    assert (
        day_standard_rows
        == 141_120
    )


    assert (
        day_operational_rows
        == 63_552
    )


    # --------------------------------------------------------
    # PERSIST DAILY PARTITION
    # --------------------------------------------------------

    date_label = (
        day_start.strftime(
            "%Y-%m-%d"
        )
    )


    partition_dir = (

        STAGING_DATASET_DIR

        / f"date={date_label}"
    )


    partition_dir.mkdir(

        parents=True,

        exist_ok=False,
    )


    parquet_path = (

        partition_dir

        / "observations.parquet"
    )


    day_df.to_parquet(

        parquet_path,

        index=False,

        compression="snappy",
    )


    parquet_sha = (

        sha256_file(
            parquet_path
        )
    )


    daily_manifest_entries.append({

        "date":
            date_label,

        "relative_path":
            (
                f"date={date_label}/"
                "observations.parquet"
            ),

        "rows":
            int(
                len(
                    day_df
                )
            ),

        "standardized_rows":
            day_standard_rows,

        "operational_rows":
            day_operational_rows,

        "parquet_sha256":
            parquet_sha,

        "semantic_sha256":
            (
                day_semantic_hasher
                .hexdigest()
            ),
    })


    total_rows_generated += (
        len(
            day_df
        )
    )


    standard_rows_generated += (
        day_standard_rows
    )


    operational_rows_generated += (
        day_operational_rows
    )


    del day_df
    del day_rows


# ============================================================
# 13.27 — EVENT-DRIVEN HEALTHY CHANNEL
# ============================================================
#
# No event occurrence is fabricated.
#
# Absence of active alarms/config/software events is itself
# represented by zero event records during the healthy period.
#
# ============================================================

event_channel = {

    "dataset":
        HEALTHY_DATASET_VERSION,

    "event_driven_metric_families":
        sorted(
            event_driven_bindings.keys()
        ),

    "healthy_event_rows":
        0,

    "interpretation":
        (
            "No active synthetic alarm, configuration-change "
            "or software-event occurrence is generated in the "
            "healthy baseline. Event families remain available "
            "for later controlled incident overlays."
        ),
}


event_channel_path = (

    STAGING_DATASET_DIR

    / "healthy_event_channel.json"
)


with open(

    event_channel_path,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        event_channel,

        f,

        indent=2,

        ensure_ascii=False,
    )


event_channel_sha = (

    sha256_file(
        event_channel_path
    )
)


# ============================================================
# 13.28 — GLOBAL ROW VALIDATION
# ============================================================

assert (
    total_rows_generated
    == EXPECTED_TOTAL_ROWS
)


assert (
    standard_rows_generated
    == EXPECTED_STANDARD_ROWS
)

assert (
    operational_rows_generated
    == EXPECTED_OPERATIONAL_ROWS
)


assert (
    PAIR_CONSTRAINT_VIOLATIONS
    == 0
)


# ============================================================
# 13.29 — METRIC FAMILY COVERAGE
# ============================================================

expected_continuous_metric_names = (

    set(
        RAN_TELEMETRY_METRIC_BINDINGS
        .keys()
    )

    | set(
        continuous_operational_bindings
        .keys()
    )
)


missing_continuous_metrics = (

    expected_continuous_metric_names

    - generated_metric_names
)


unexpected_generated_metrics = (

    generated_metric_names

    - expected_continuous_metric_names
)


assert (
    len(
        missing_continuous_metrics
    )
    == 0
)


assert (
    len(
        unexpected_generated_metrics
    )
    == 0
)


assert (
    len(
        generated_metric_names
    )
    == 66
)


# ============================================================
# 13.30 — CONSTRUCTION SANITY RATIOS
# ============================================================
#
# These are generator QA ratios.
#
# They are NOT persisted as standardized KPI observations and
# are NOT the formal TS 28.554 KPI calculation.
#
# ============================================================

SANITY_RATIOS = {

    "RRC":
        ratio_or_nan(

            SANITY_COUNTS[
                "rrc_succ"
            ],

            SANITY_COUNTS[
                "rrc_att"
            ],
        ),

    "NG":
        ratio_or_nan(

            SANITY_COUNTS[
                "ng_succ"
            ],

            SANITY_COUNTS[
                "ng_att"
            ],
        ),

    "INITIAL_DRB":
        ratio_or_nan(

            SANITY_COUNTS[
                "initial_drb_succ"
            ],

            SANITY_COUNTS[
                "initial_drb_att"
            ],
        ),

    "DRB":
        ratio_or_nan(

            SANITY_COUNTS[
                "drb_succ"
            ],

            SANITY_COUNTS[
                "drb_att"
            ],
        ),

    "RRC_RESUME":
        ratio_or_nan(

            SANITY_COUNTS[
                "resume_succ"
            ],

            SANITY_COUNTS[
                "resume_att"
            ],
        ),

    "DRB_RESUME":
        ratio_or_nan(

            SANITY_COUNTS[
                "drb_resume_succ"
            ],

            SANITY_COUNTS[
                "drb_resume_att"
            ],
        ),
}


# ============================================================
# 13.31 — SAMPLE CONTRACT VALIDATION
# ============================================================

first_partition_date = (
    HEALTHY_START_UTC.strftime(
        "%Y-%m-%d"
    )
)

first_partition = (
    STAGING_DATASET_DIR
    / f"date={first_partition_date}"
    / "observations.parquet"
)


sample_df = pd.read_parquet(

    first_partition
)


sample_standard = (

    sample_df[
        sample_df[
            "is_standardized_metric"
        ]
    ]
    .iloc[
        0
    ]
)


sample_operational = (

    sample_df[
        ~sample_df[
            "is_standardized_metric"
        ]
    ]
    .iloc[
        0
    ]
)


def row_value_for_contract(
    row
):

    if (
        row[
            "value_kind"
        ]
        == "INTEGER"
    ):

        return int(
            row[
                "value_int"
            ]
        )


    if (
        row[
            "value_kind"
        ]
        == "FLOAT"
    ):

        return float(
            row[
                "value_float"
            ]
        )


    if (
        row[
            "value_kind"
        ]
        == "BOOLEAN"
    ):

        return bool(
            row[
                "value_bool"
            ]
        )


    return str(
        row[
            "value_text"
        ]
    )


for sample_row in [

    sample_standard,

    sample_operational,
]:

    entity = (

        get_telemetry_entity_v1_1(

            sample_row[
                "entity_id"
            ]
        )
    )


    RANTelemetryObservationV1_1(

        observation_id=(
            sample_row[
                "observation_id"
            ]
        ),

        environment=(
            sample_row[
                "environment"
            ]
        ),

        observation_type=(
            sample_row[
                "observation_type"
            ]
        ),

        evidence_origin=(
            sample_row[
                "evidence_origin"
            ]
        ),

        metric_name=(
            sample_row[
                "metric_name"
            ]
        ),

        entity=entity,

        interval_start=(
            sample_row[
                "interval_start_utc"
            ]
        ),

        interval_end=(
            sample_row[
                "interval_end_utc"
            ]
        ),

        pm_interval_minutes=(
            int(
                sample_row[
                    "pm_interval_minutes"
                ]
            )
        ),

        value=(
            row_value_for_contract(
                sample_row
            )
        ),

        unit=(
            None

            if pd.isna(
                sample_row[
                    "unit"
                ]
            )

            else sample_row[
                "unit"
            ]
        ),

        dimensions=(
            json.loads(

                sample_row[
                    "dimensions_json"
                ]
            )
        ),

        quality=(
            sample_row[
                "quality"
            ]
        ),

        graph_node_id=(
            None

            if pd.isna(
                sample_row[
                    "graph_node_id"
                ]
            )

            else sample_row[
                "graph_node_id"
            ]
        ),

        investigation_stage=(
            None

            if pd.isna(
                sample_row[
                    "investigation_stage"
                ]
            )

            else sample_row[
                "investigation_stage"
            ]
        ),

        source_system=(
            sample_row[
                "source_system"
            ]
        ),

        metadata={},
    )


del sample_df


# ============================================================
# 13.32 — LEAKAGE / SAFETY VALIDATION
# ============================================================

PROHIBITED_DATASET_FIELDS = {

    "root_cause",

    "ground_truth",

    "fault_label",

    "expected_diagnosis",

    "correct_action",

    "benchmark_answer",
}


assert (
    len(
        set(
            OBSERVATION_COLUMNS
        )

        & PROHIBITED_DATASET_FIELDS
    )
    == 0
)


# ============================================================
# 13.33 — DATASET MANIFEST
# ============================================================

dataset_manifest = {

    "artifact_id":
        BASELINE_ARTIFACT_ID,

    "dataset_role":
        BASELINE_ROLE,

    "baseline_mode":
        BASELINE_MODE,

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "dataset_version":
        HEALTHY_DATASET_VERSION,

    "environment":
        (
            TelemetryEnvironment
            .SYNTHETIC_V1
            .value
        ),

    "reference_domain":
        "5G RAN Accessibility",

    "generator_seed":
        HEALTHY_GENERATOR_SEED,

    "window": {

        "start_utc":
            HEALTHY_START_UTC.isoformat(),

        "end_exclusive_utc":
            (
                HEALTHY_END_EXCLUSIVE_UTC
                .isoformat()
            ),

        "days":
            BASELINE_DAY_COUNT,

        "pm_interval_minutes":
            PM_INTERVAL_MINUTES,

        "interval_count":
            len(
                interval_starts
            ),
    },

    "topology": {

        "logical_cells":
            30,

        "sites":
            10,

        "cu_domains":
            2,

        "cu_cp_entities":
            2,

        "cu_up_entities":
            2,

        "n2_paths":
            2,

        "amf_boundary_peers":
            2,

        "f1_paths":
            10,

        "e1_paths":
            2,
    },

    "observability": {

        "standardized_metric_families":
            21,

        "operational_metric_families":
            49,

        "continuous_operational_families":
            45,

        "event_driven_operational_families":
            4,

        "continuously_observed_metric_families":
            66,

        "healthy_event_rows":
            0,
    },

    "rows": {

        "rows_per_interval":
            EXPECTED_ROWS_PER_INTERVAL,

        "rows_per_day":
            EXPECTED_ROWS_PER_DAY,

        "standardized_rows":
            standard_rows_generated,

        "operational_rows":
            operational_rows_generated,

        "total_rows":
            total_rows_generated,
    },

    "daily_partitions":
        daily_manifest_entries,

    "event_channel": {

        "relative_path":
            "healthy_event_channel.json",

        "sha256":
            event_channel_sha,
    },

    "construction_sanity_ratios_pct":
        SANITY_RATIOS,

    "provenance": {

        "cell5_v1_2_semantic_sha256":
            (
                ran_accessibility_v1_2_semantic_sha
            ),

        "cell7_v1_1_semantic_sha256":
            cell7_v1_1_semantic_sha,

        "cell8_v1_1_semantic_sha256":
            cell8_v1_1_semantic_sha,

        "cell9_semantic_sha256":
            cell9_semantic_sha,
    },

    "design_constraints": {

        "healthy_baseline_only":
            True,

        "qualification_dataset":
            (
                BASELINE_MODE
                == "PILOT_7D"
            ),

        "canonical_dataset":
            (
                BASELINE_MODE
                == "CANONICAL_92D"
            ),

        "downstream_validation_required_before_canonical":
            True,

        "standard_pm_independently_sampled":
            False,

        "standard_pm_generated_from_shared_network_state":
            True,

        "formal_ts_28_554_kpi_generated_independently":
            False,

        "formal_kpi_derivation_deferred":
            True,

        "event_occurrences_fabricated":
            False,

        "fault_injected":
            False,

        "incident_generated":
            False,

        "hidden_ground_truth_exposed":
            False,

        "ml_inference_included":
            False,

        "synthetic_operating_ranges_are_3gpp_thresholds":
            False,

        "synthetic_operating_ranges_are_vendor_thresholds":
            False,
    },
}


# ============================================================
# 13.34 — DATASET SEMANTIC HASH
# ============================================================

semantic_manifest_content = {

    "dataset_version":
        dataset_manifest[
            "dataset_version"
        ],

    "generator_seed":
        dataset_manifest[
            "generator_seed"
        ],

    "window":
        dataset_manifest[
            "window"
        ],

    "topology":
        dataset_manifest[
            "topology"
        ],

    "observability":
        dataset_manifest[
            "observability"
        ],

    "rows":
        dataset_manifest[
            "rows"
        ],

    "daily_semantic_hashes": [

        {
            "date":
                item[
                    "date"
                ],

            "semantic_sha256":
                item[
                    "semantic_sha256"
                ],
        }

        for item
        in daily_manifest_entries
    ],

    "provenance":
        dataset_manifest[
            "provenance"
        ],

    "design_constraints":
        dataset_manifest[
            "design_constraints"
        ],
}


dataset_semantic_sha = (

    hashlib.sha256(

        json.dumps(

            semantic_manifest_content,

            sort_keys=True,

            separators=(",", ":"),

            ensure_ascii=False,

        ).encode(
            "utf-8"
        )

    ).hexdigest()
)


dataset_manifest[
    "semantic_sha256"
] = dataset_semantic_sha


# ============================================================
# 13.35 — WRITE MANIFEST
# ============================================================

manifest_path_staging = (

    STAGING_DATASET_DIR

    / "manifest.json"
)


with open(

    manifest_path_staging,

    "w",

    encoding="utf-8",

) as f:

    json.dump(

        dataset_manifest,

        f,

        indent=2,

        ensure_ascii=False,
    )


# ============================================================
# 13.36 — ATOMIC PROMOTION OF COMPLETE DATASET
# ============================================================

STAGING_DATASET_DIR.rename(
    FINAL_DATASET_DIR
)


manifest_path = (

    FINAL_DATASET_DIR

    / "manifest.json"
)


dataset_manifest_sha = (

    sha256_file(
        manifest_path
    )
)


# ============================================================
# 13.37 — FINAL SUMMARY
# ============================================================

print(
    "=" * 108
)

print(
    "MODULE 5 — CELL 13 "
    "PILOT / CANONICAL HEALTHY 15-MINUTE "
    "5G RAN TELEMETRY GENERATOR V1"
)

print(
    "=" * 108
)


print(
    "\nExecution profile"
)

print("-" * 108)

print(
    "Baseline mode                     : "
    f"{BASELINE_MODE}"
)

print(
    "Dataset role                      : "
    f"{BASELINE_ROLE}"
)

print(
    "Canonical generation approved     : "
    f"{CANONICAL_GENERATION_APPROVED}"
)

print(
    "\nDataset window"
)

print("-" * 108)

print(
    "Start UTC                         : "
    f"{HEALTHY_START_UTC}"
)

print(
    "End exclusive UTC                 : "
    f"{HEALTHY_END_EXCLUSIVE_UTC}"
)

print(
    "Duration                          : "
    f"{BASELINE_DAY_COUNT} days"
)

print(
    "PM interval                       : "
    f"{PM_INTERVAL_MINUTES} minutes"
)

print(
    "Intervals                         : "
    f"{len(interval_starts)}"
)


print(
    "\nObservability"
)

print("-" * 108)

print(
    "Standardized metric families      : "
    f"{len(RAN_TELEMETRY_METRIC_BINDINGS)}"
)

print(
    "Operational metric families       : "
    f"{len(RAN_OPERATIONAL_METRIC_BINDINGS)}"
)

print(
    "Continuous operational families   : "
    f"{len(continuous_operational_bindings)}"
)

print(
    "Event-driven families             : "
    f"{len(event_driven_bindings)}"
)

print(
    "Continuously observed families    : "
    f"{len(generated_metric_names)}"
)

print(
    "Healthy event rows                : "
    "0"
)


print(
    "\nRow counts"
)

print("-" * 108)

print(
    "Rows per 15-minute interval       : "
    f"{EXPECTED_ROWS_PER_INTERVAL:,}"
)

print(
    "Rows per day                      : "
    f"{EXPECTED_ROWS_PER_DAY:,}"
)

print(
    "Standardized PM rows              : "
    f"{standard_rows_generated:,}"
)

print(
    "Operational telemetry rows        : "
    f"{operational_rows_generated:,}"
)

print(
    "Total observation rows            : "
    f"{total_rows_generated:,}"
)


print(
    "\nConstruction validation"
)

print("-" * 108)

print(
    "Missing continuous metrics        : "
    f"{len(missing_continuous_metrics)}"
)

print(
    "Unexpected metrics                : "
    f"{len(unexpected_generated_metrics)}"
)

print(
    "Success > attempt violations      : "
    f"{PAIR_CONSTRAINT_VIOLATIONS}"
)

print(
    "Observation contract samples      : "
    "PASS"
)


print(
    "\nHealthy construction sanity ratios"
)

print("-" * 108)

for (
    component,
    ratio
) in (
    SANITY_RATIOS.items()
):

    print(
        f"{component:<32} : "
        f"{ratio:.4f}%"
    )


print(
    "\nGeneration boundaries"
)

print("-" * 108)

print(
    "Faults injected                   : "
    "NO"
)

print(
    "Incidents generated               : "
    "NO"
)

print(
    "Hidden ground truth exposed       : "
    "NO"
)

print(
    "ML inference included             : "
    "NO"
)

print(
    "Formal TS 28.554 KPI persisted    : "
    "NO"
)

print(
    "KPI derivation from PM counters   : "
    "DEFERRED"
)


print(
    "\nDataset"
)

print("-" * 108)

print(
    "Path                               : "
    f"{FINAL_DATASET_DIR}"
)

print(
    "Daily Parquet partitions          : "
    f"{len(daily_manifest_entries)}"
)

print(
    "Manifest SHA                      : "
    f"{dataset_manifest_sha}"
)

print(
    "Dataset semantic SHA              : "
    f"{dataset_semantic_sha}"
)


print(
    "\n" + "=" * 108
)

print(
    "CELL 13 COMPLETE — "
    f"{BASELINE_MODE} CAUSALLY-CORRELATED "
    "NETWORK OBSERVATION DATASET READY"
)

print(
    "=" * 108
)
````

##### Cell 14 — 7-Day Pilot Telemetry Generator Qualification Gate V1

Validate the **7-day healthy telemetry pilot dataset** before authorizing large-scale canonical generation.

The qualification gate checks dataset integrity, temporal continuity, metric and entity coverage, healthy-state invariants, standardized PM consistency, site-archetype differentiation, and the causal relationships intentionally encoded in the synthetic environment.

> **The objective is to confirm that the generator produces structurally complete, internally consistent and causally credible healthy network observations before authorizing the 92-day canonical baseline.**

The checks in this cell are **generator-quality controls only** and must not be interpreted as 3GPP thresholds, vendor limits, SLA criteria or formal telecom KPI acceptance thresholds.


````python
# ============================================================
# MODULE 5 — CELL 14
# 7-DAY PILOT TELEMETRY GENERATOR QUALIFICATION GATE V1
# ============================================================
#
# Generator-QA only.
#
# Correlation floors and synthetic ranges below are NOT:
#
#   - 3GPP thresholds
#   - vendor limits
#   - SLA targets
#   - operational KPI acceptance criteria
#
# No formal TS 28.554 KPI is persisted here.
#
# Exact Partial / Total DRB Accessibility derivation remains
# a downstream M05_02 responsibility.
#
# ============================================================


from pathlib import Path
from datetime import datetime, timezone

import hashlib
import json

import numpy as np
import pandas as pd
import pyarrow.parquet as pq


# ============================================================
# 14.1 — DEPENDENCIES / PILOT IDENTITY
# ============================================================

required = [

    "BASELINE_MODE",
    "BASELINE_ROLE",
    "FINAL_DATASET_DIR",

    "OBSERVATION_COLUMNS",
    "PM_INTERVAL_MINUTES",

    "EXPECTED_ROWS_PER_INTERVAL",
    "EXPECTED_ROWS_PER_DAY",
    "EXPECTED_TOTAL_ROWS",

    "STANDARD_ROWS_PER_INTERVAL",
    "EXPECTED_OPERATIONAL_ROWS_PER_INTERVAL",

    "expected_continuous_metric_names",

    "RAN_TELEMETRY_METRIC_BINDINGS",

    "continuous_operational_bindings",
    "event_driven_bindings",

    "ENTITY_COUNT_BY_SCOPE",

    "SANITY_RATIOS",

    "site_archetypes",

    "dataset_manifest_sha",
    "dataset_semantic_sha",

    "PATHS",
    "sha256_file",
]


missing = [

    name

    for name in required

    if name not in globals()
]


if missing:

    raise RuntimeError(

        "Cell 14 requires completed Cell 13. "
        f"Missing: {missing}"
    )


if BASELINE_MODE != "PILOT_7D":

    raise RuntimeError(

        "Cell 14 qualifies PILOT_7D only; "
        f"got {BASELINE_MODE!r}"
    )


if (
    BASELINE_ROLE
    != "GENERATOR_QUALIFICATION_DATASET"
):

    raise RuntimeError(

        "Unexpected pilot role: "
        f"{BASELINE_ROLE!r}"
    )


PILOT_DIR = Path(
    FINAL_DATASET_DIR
)

MANIFEST_PATH = (
    PILOT_DIR
    / "manifest.json"
)

EVENT_PATH = (
    PILOT_DIR
    / "healthy_event_channel.json"
)


for path in [

    PILOT_DIR,
    MANIFEST_PATH,
    EVENT_PATH,

]:

    if not path.exists():

        raise FileNotFoundError(
            path
        )


with MANIFEST_PATH.open(
    "r",
    encoding="utf-8",
) as f:

    manifest = json.load(
        f
    )


with EVENT_PATH.open(
    "r",
    encoding="utf-8",
) as f:

    event_channel = json.load(
        f
    )


# ============================================================
# 14.2 — HELPERS
# ============================================================

checks = {}


def check(
    name,
    passed,
    detail="",
):

    checks[
        name
    ] = {

        "passed":
            bool(
                passed
            ),

        "detail":
            str(
                detail
            ),
    }


def numeric_value(
    df
):

    return (

        df[
            "value_float"
        ]

        .combine_first(
            df[
                "value_int"
            ]
        )

        .astype(
            float
        )
    )


def corr(
    a,
    b,
):

    z = pd.concat(
        [
            a,
            b,
        ],
        axis=1,
    ).dropna()


    if (
        len(z) < 3

        or z.iloc[
            :,
            0
        ].nunique() < 2

        or z.iloc[
            :,
            1
        ].nunique() < 2
    ):

        return float(
            "nan"
        )


    return float(

        z.iloc[
            :,
            0
        ].corr(

            z.iloc[
                :,
                1
            ]
        )
    )


def median_group_corr(
    wide,
    group_col,
    x,
    y,
):

    vals = []


    for _, g in wide.groupby(
        group_col
    ):

        r = (

            corr(
                g[x],
                g[y],
            )

            if (
                x in g
                and y in g
            )

            else float(
                "nan"
            )
        )


        if np.isfinite(
            r
        ):

            vals.append(
                r
            )


    return (

        float(
            np.median(
                vals
            )
        )

        if vals

        else float(
            "nan"
        ),

        vals,
    )


def semantic_sha(
    obj
):

    raw = json.dumps(

        obj,

        sort_keys=True,

        separators=(
            ",",
            ":",
        ),

        default=str,

    ).encode()


    return (

        hashlib
        .sha256(
            raw
        )
        .hexdigest()
    )


# ============================================================
# 14.3 — QUALIFICATION CONFIGURATION
# ============================================================
#
# These are synthetic generator QA criteria only.
#
# They are NOT telecom performance thresholds.
#
# ============================================================

MIN_ABS_DIRECTIONAL_CORR = 0.20

RATIO_MATCH_TOL_PCT = 1e-9


# ------------------------------------------------------------
# Exact clipping / construction ranges encoded by Cell 13
# ------------------------------------------------------------

QA_RANGES = {

    "radio.rsrp_dbm":
        (
            -108.0,
            -75.0,
        ),

    "radio.sinr_db":
        (
            5.0,
            30.0,
        ),

    "radio.dl_bler_pct":
        (
            0.1,
            5.5,
        ),

    "radio.ul_bler_pct":
        (
            0.1,
            6.0,
        ),

    "resource.dl_prb_util_pct":
        (
            4.0,
            82.0,
        ),

    "resource.ul_prb_util_pct":
        (
            3.0,
            70.0,
        ),

    "resource.scheduler_pressure_pct":
        (
            2.0,
            85.0,
        ),

    "resource.bearer_admission_pressure_pct":
        (
            3.0,
            82.0,
        ),

    "rach.contention_index":
        (
            0.05,
            0.70,
        ),

    "rach.access_delay_ms":
        (
            5.0,
            35.0,
        ),

    "processing.du_cpu_pct":
        (
            12.0,
            78.0,
        ),

    "transport.f1_latency_ms":
        (
            0.7,
            5.0,
        ),

    "transport.f1_packet_loss_pct":
        (
            0.0,
            0.10,
        ),

    "processing.cu_cp_cpu_pct":
        (
            12.0,
            74.0,
        ),

    "processing.cu_up_cpu_pct":
        (
            13.0,
            78.0,
        ),

    "transport.n2_util_pct":
        (
            8.0,
            72.0,
        ),

    "transport.n2_latency_ms":
        (
            1.5,
            7.0,
        ),

    "transport.n2_packet_loss_pct":
        (
            0.0,
            0.12,
        ),

    "transport.e1_latency_ms":
        (
            0.4,
            3.0,
        ),

    "context.ue_context_availability_pct":
        (
            99.50,
            100.0,
        ),

    "context.bearer_context_availability_pct":
        (
            99.45,
            100.0,
        ),
}


CAUSAL_METRICS = {

    "load.active_ues",

    "radio.sinr_db",
    "radio.dl_bler_pct",

    "resource.dl_prb_util_pct",
    "resource.scheduler_pressure_pct",
    "resource.bearer_admission_pressure_pct",

    "rach.contention_index",
    "rach.access_delay_ms",

    "processing.du_cpu_pct",
    "transport.f1_latency_ms",

    "processing.cu_cp_cpu_pct",
    "processing.cu_cp_control_delay_ms",

    "processing.cu_up_cpu_pct",

    "transport.n2_util_pct",
    "transport.n2_latency_ms",
}


NUMERIC_QA_METRICS = (

    set(
        QA_RANGES
    )

    | CAUSAL_METRICS
)


TEXT_STATES = {

    "transport.f1_state":
        "UP",

    "transport.e1_state":
        "UP",

    "sctp.association_state":
        "UP",

    "boundary.amf_peer_state":
        "AVAILABLE",

    "config.prach_profile_state":
        "BASELINE",

    "config.rrc_profile_state":
        "BASELINE",

    "config.qos_profile_state":
        "BASELINE",

    "config.slice_profile_state":
        "BASELINE",

    "config.n2_peer_state":
        "BASELINE",

    "config.resume_profile_state":
        "BASELINE",
}


BOOL_STATES = {

    "boundary.amf_reachable":
        True,

    "boundary.amf_overload_indication":
        False,
}


INT_STATES = {

    "sctp.heartbeat_failures":
        0,
}


STATE_METRICS = (

    set(
        TEXT_STATES
    )

    | set(
        BOOL_STATES
    )

    | set(
        INT_STATES
    )
)


PAIRS = {

    "RRC": (

        "RRC.ConnEstabAtt.Cause",
        "RRC.ConnEstabSucc.Cause",
    ),

    "NG": (

        "UECNTX.ConnEstabAtt.Cause",
        "UECNTX.ConnEstabSucc.Cause",
    ),

    "INITIAL_DRB": (

        "DRB.InitialEstabAtt.5QI",
        "DRB.InitialEstabSucc.5QI",
    ),

    "DRB": (

        "DRB.EstabAtt.5QI",
        "DRB.EstabSucc.5QI",
    ),

    "RRC_RESUME": (

        "RRC.ResumeAtt.Cause",
        "RRC.ResumeSucc.Cause",
    ),

    "DRB_RESUME": (

        "DRB.ResumeAtt.5QI",
        "DRB.ResumeSucc.5QI",
    ),
}


PAIR_METRICS = {

    metric

    for pair
    in PAIRS.values()

    for metric
    in pair
}


# ============================================================
# 14.4 — MANIFEST / ARTIFACT INTEGRITY
# ============================================================

observed_manifest_sha = (
    sha256_file(
        MANIFEST_PATH
    )
)


check(

    "manifest_byte_sha",

    observed_manifest_sha
    == dataset_manifest_sha,

    (
        f"observed={observed_manifest_sha} "
        f"expected={dataset_manifest_sha}"
    ),
)


check(

    "manifest_semantic_sha",

    manifest.get(
        "semantic_sha256"
    )
    == dataset_semantic_sha,

    (
        f"observed="
        f"{manifest.get('semantic_sha256')} "
        f"expected={dataset_semantic_sha}"
    ),
)


check(

    "manifest_mode",

    manifest.get(
        "baseline_mode"
    )
    == "PILOT_7D",

    manifest.get(
        "baseline_mode"
    ),
)


check(

    "manifest_role",

    manifest.get(
        "dataset_role"
    )
    == "GENERATOR_QUALIFICATION_DATASET",

    manifest.get(
        "dataset_role"
    ),
)


partitions = manifest.get(
    "daily_partitions",
    [],
)


check(

    "manifest_partition_count",

    len(
        partitions
    )
    == 7,

    f"observed={len(partitions)}",
)


# ============================================================
# 14.5 — PARTITION SCAN
# ============================================================

scan_cols = [

    "observation_id",
    "environment",

    "metric_name",
    "is_standardized_metric",

    "entity_id",
    "site_id",
    "cu_domain_id",

    "interval_start_utc",
    "interval_end_utc",

    "pm_interval_minutes",

    "value_kind",

    "value_int",
    "value_float",
    "value_bool",
    "value_text",

    "quality",

    "dimensions_json",
]


missing_cols = (

    set(
        scan_cols
    )

    - set(
        OBSERVATION_COLUMNS
    )
)


if missing_cols:

    raise RuntimeError(

        "Cell 13 schema missing "
        "Cell 14 columns: "
        f"{sorted(missing_cols)}"
    )


key_frames = []
state_frames = []
pair_frames = []


row_counts = []
dates = []


sha_fail = 0
dup_fail = 0

interval_fail = 0

standard_fail = 0
operational_fail = 0

family_fail = 0
multiplicity_fail = 0

value_fail = 0
contract_fail = 0


observed_standard = set()
observed_operational = set()


expected_standard = set(

    RAN_TELEMETRY_METRIC_BINDINGS
)


expected_operational = set(

    continuous_operational_bindings
)


for entry in partitions:

    path = (

        PILOT_DIR
        / entry[
            "relative_path"
        ]
    )


    if not path.exists():

        raise FileNotFoundError(
            path
        )


    sha_fail += int(

        sha256_file(
            path
        )

        != entry[
            "parquet_sha256"
        ]
    )


    rows = int(

        pq.ParquetFile(
            path
        )
        .metadata
        .num_rows
    )


    row_counts.append(
        rows
    )


    dates.append(
        entry[
            "date"
        ]
    )


    df = pd.read_parquet(

        path,

        columns=scan_cols,
    )


    df[
        "interval_start_utc"
    ] = pd.to_datetime(

        df[
            "interval_start_utc"
        ],

        utc=True,
    )


    df[
        "interval_end_utc"
    ] = pd.to_datetime(

        df[
            "interval_end_utc"
        ],

        utc=True,
    )


    # --------------------------------------------------------
    # Observation identity
    # --------------------------------------------------------

    dup_fail += int(

        df[
            "observation_id"
        ]
        .duplicated()
        .sum()
    )


    # --------------------------------------------------------
    # Interval completeness
    # --------------------------------------------------------

    per_interval = (

        df.groupby(
            "interval_start_utc"
        )
        .size()
    )


    interval_fail += int(

        len(
            per_interval
        )
        != 96
    )


    interval_fail += int(

        (
            per_interval
            != EXPECTED_ROWS_PER_INTERVAL
        )
        .sum()
    )


    standard_counts = (

        df[
            df[
                "is_standardized_metric"
            ]
        ]

        .groupby(
            "interval_start_utc"
        )

        .size()
    )


    operational_counts = (

        df[
            ~df[
                "is_standardized_metric"
            ]
        ]

        .groupby(
            "interval_start_utc"
        )

        .size()
    )


    standard_fail += int(

        (
            standard_counts
            != STANDARD_ROWS_PER_INTERVAL
        )
        .sum()
    )


    operational_fail += int(

        (
            operational_counts
            != EXPECTED_OPERATIONAL_ROWS_PER_INTERVAL
        )
        .sum()
    )


    # --------------------------------------------------------
    # Metric-family completeness
    # --------------------------------------------------------

    family_fail += int(

        (
            df
            .groupby(
                "interval_start_utc"
            )[
                "metric_name"
            ]
            .nunique()

            != 66
        )
        .sum()
    )


    # --------------------------------------------------------
    # Operational metric multiplicity
    # --------------------------------------------------------

    op_group = (

        df[
            ~df[
                "is_standardized_metric"
            ]
        ]

        .groupby(
            [
                "metric_name",
                "interval_start_utc",
            ]
        )

        .size()
    )


    for (
        metric,
        binding
    ) in (
        continuous_operational_bindings
        .items()
    ):

        expected_n = (

            ENTITY_COUNT_BY_SCOPE[

                binding
                .preferred_scope
                .value
            ]
        )


        try:

            counts = (
                op_group.loc[
                    metric
                ]
            )


            multiplicity_fail += int(

                (
                    counts
                    != expected_n
                )
                .sum()
            )


            multiplicity_fail += (

                96
                - len(
                    counts
                )
            )


        except KeyError:

            multiplicity_fail += 96


    # --------------------------------------------------------
    # Value-slot integrity
    # --------------------------------------------------------

    slots = (

        df[
            [
                "value_int",
                "value_float",
                "value_bool",
                "value_text",
            ]
        ]

        .notna()

        .sum(
            axis=1
        )
    )


    value_fail += int(

        (
            slots
            != 1
        )
        .sum()
    )


    value_fail += int(

        (

            (
                (
                    df[
                        "value_kind"
                    ]
                    == "INTEGER"
                )

                & df[
                    "value_int"
                ].isna()
            )

            |

            (
                (
                    df[
                        "value_kind"
                    ]
                    == "FLOAT"
                )

                & df[
                    "value_float"
                ].isna()
            )

            |

            (
                (
                    df[
                        "value_kind"
                    ]
                    == "BOOLEAN"
                )

                & df[
                    "value_bool"
                ].isna()
            )

            |

            (
                (
                    df[
                        "value_kind"
                    ]
                    == "STATE"
                )

                & df[
                    "value_text"
                ].isna()
            )
        )
        .sum()
    )


    # --------------------------------------------------------
    # Core observation-contract invariants
    # --------------------------------------------------------

    duration_ok = (

        (
            df[
                "interval_end_utc"
            ]

            - df[
                "interval_start_utc"
            ]
        )

        == pd.Timedelta(
            minutes=15
        )
    )


    contract_fail += int(

        (
            ~duration_ok
        )
        .sum()
    )


    contract_fail += int(

        (
            df[
                "pm_interval_minutes"
            ]
            != 15
        )
        .sum()
    )


    contract_fail += int(

        (
            df[
                "quality"
            ]
            != "VALID"
        )
        .sum()
    )


    contract_fail += int(

        (
            df[
                "environment"
            ]
            != "SYNTHETIC_V1"
        )
        .sum()
    )


    # --------------------------------------------------------
    # Metric registry coverage
    # --------------------------------------------------------

    observed_standard.update(

        df.loc[

            df[
                "is_standardized_metric"
            ],

            "metric_name",

        ].unique()
    )


    observed_operational.update(

        df.loc[

            ~df[
                "is_standardized_metric"
            ],

            "metric_name",

        ].unique()
    )


    # --------------------------------------------------------
    # Compact numeric QA frame
    # --------------------------------------------------------

    q = df[

        df[
            "metric_name"
        ]
        .isin(
            NUMERIC_QA_METRICS
        )

    ].copy()


    q[
        "numeric_value"
    ] = numeric_value(
        q
    )


    key_frames.append(

        q[
            [
                "metric_name",
                "entity_id",
                "site_id",
                "cu_domain_id",
                "interval_start_utc",
                "numeric_value",
            ]
        ]
    )


    # --------------------------------------------------------
    # Healthy state frame
    # --------------------------------------------------------

    s = df[

        df[
            "metric_name"
        ]
        .isin(
            STATE_METRICS
        )

    ].copy()


    state_frames.append(

        s[
            [
                "metric_name",
                "value_int",
                "value_bool",
                "value_text",
            ]
        ]
    )


    # --------------------------------------------------------
    # Standard PM attempt/success frame
    # --------------------------------------------------------

    p = df[

        df[
            "is_standardized_metric"
        ]

        & df[
            "metric_name"
        ]
        .isin(
            PAIR_METRICS
        )

    ].copy()


    p[
        "numeric_value"
    ] = numeric_value(
        p
    )


    pair_frames.append(

        p[
            [
                "metric_name",
                "entity_id",
                "interval_start_utc",
                "dimensions_json",
                "numeric_value",
            ]
        ]
    )


    del df


# ============================================================
# 14.6 — STRUCTURAL QUALIFICATION
# ============================================================

check(

    "partition_sha_integrity",

    sha_fail == 0,

    f"failures={sha_fail}",
)


check(

    "rows_per_partition",

    all(

        x
        == EXPECTED_ROWS_PER_DAY

        for x
        in row_counts
    ),

    (
        f"observed="
        f"{sorted(set(row_counts))} "
        f"expected="
        f"{EXPECTED_ROWS_PER_DAY}"
    ),
)


check(

    "total_rows",

    sum(
        row_counts
    )
    == EXPECTED_TOTAL_ROWS,

    (
        f"observed="
        f"{sum(row_counts):,} "
        f"expected="
        f"{EXPECTED_TOTAL_ROWS:,}"
    ),
)


check(

    "duplicate_observation_ids",

    dup_fail == 0,

    f"duplicates={dup_fail}",
)


check(

    "interval_structure",

    interval_fail == 0,

    f"failures={interval_fail}",
)


check(

    "standard_rows_per_interval",

    standard_fail == 0,

    f"failures={standard_fail}",
)


check(

    "operational_rows_per_interval",

    operational_fail == 0,

    f"failures={operational_fail}",
)


check(

    "metric_families_per_interval",

    family_fail == 0,

    f"failures={family_fail}",
)


check(

    "operational_scope_multiplicity",

    multiplicity_fail == 0,

    f"failures={multiplicity_fail}",
)


check(

    "value_slot_integrity",

    value_fail == 0,

    f"failures={value_fail}",
)


check(

    "observation_contract_invariants",

    contract_fail == 0,

    f"failures={contract_fail}",
)


check(

    "standard_metric_coverage",

    observed_standard
    == expected_standard,

    (
        f"observed="
        f"{len(observed_standard)} "
        f"expected="
        f"{len(expected_standard)}"
    ),
)


check(

    "operational_metric_coverage",

    observed_operational
    == expected_operational,

    (
        f"observed="
        f"{len(observed_operational)} "
        f"expected="
        f"{len(expected_operational)}"
    ),
)


# ------------------------------------------------------------
# Daily partition continuity
# ------------------------------------------------------------

obs_dates = pd.to_datetime(

    sorted(
        dates
    ),

    utc=True,
)


exp_dates = pd.date_range(

    "2026-08-25",

    periods=7,

    freq="1D",

    tz="UTC",
)


check(

    "daily_partition_continuity",

    obs_dates.equals(
        exp_dates
    ),

    (
        f"start="
        f"{obs_dates.min()} "
        f"end="
        f"{obs_dates.max()}"
    ),
)


# ============================================================
# 14.7 — EVENT CHANNEL / HEALTHY STATE
# ============================================================

event_sha = sha256_file(
    EVENT_PATH
)


check(

    "event_channel_sha",

    event_sha
    == manifest[
        "event_channel"
    ][
        "sha256"
    ],

    (
        f"observed="
        f"{event_sha} "
        f"expected="
        f"{manifest['event_channel']['sha256']}"
    ),
)


check(

    "healthy_event_rows_zero",

    event_channel.get(
        "healthy_event_rows"
    )
    == 0,

    event_channel.get(
        "healthy_event_rows"
    ),
)


check(

    "event_family_registry",

    set(

        event_channel.get(
            "event_driven_metric_families",
            [],
        )

    )
    == set(
        event_driven_bindings
    ),

    (
        f"observed="
        f"{len(event_channel.get('event_driven_metric_families', []))} "
        f"expected="
        f"{len(event_driven_bindings)}"
    ),
)


state_df = pd.concat(

    state_frames,

    ignore_index=True,
)


state_fail = {}


for (
    metric,
    expected
) in (
    TEXT_STATES.items()
):

    v = state_df.loc[

        state_df[
            "metric_name"
        ]
        == metric,

        "value_text",

    ].dropna()


    state_fail[
        metric
    ] = int(

        (
            v
            != expected
        )
        .sum()
    )


for (
    metric,
    expected
) in (
    BOOL_STATES.items()
):

    v = state_df.loc[

        state_df[
            "metric_name"
        ]
        == metric,

        "value_bool",

    ].dropna().astype(
        bool
    )


    state_fail[
        metric
    ] = int(

        (
            v
            != expected
        )
        .sum()
    )


for (
    metric,
    expected
) in (
    INT_STATES.items()
):

    v = state_df.loc[

        state_df[
            "metric_name"
        ]
        == metric,

        "value_int",

    ].dropna().astype(
        float
    )


    state_fail[
        metric
    ] = int(

        (
            v
            != float(
                expected
            )
        )
        .sum()
    )


check(

    "healthy_state_invariants",

    sum(
        state_fail.values()
    )
    == 0,

    state_fail,
)


# ============================================================
# 14.8 — SYNTHETIC RANGES / VARIATION
# ============================================================

key_df = pd.concat(

    key_frames,

    ignore_index=True,
)


check(

    "numeric_values_finite",

    np.isfinite(
        key_df[
            "numeric_value"
        ]
    ).all(),

    f"rows={len(key_df):,}",
)


range_results = {}

range_fail = []
variation_fail = []


for (
    metric,
    (
        lo,
        hi,
    )
) in (
    QA_RANGES.items()
):

    v = key_df.loc[

        key_df[
            "metric_name"
        ]
        == metric,

        "numeric_value",

    ]


    mn = float(
        v.min()
    )

    mx = float(
        v.max()
    )

    sd = float(
        v.std()
    )


    in_range = (

        mn
        >= lo - 1e-12

        and mx
        <= hi + 1e-12
    )


    varies = (

        v.nunique() > 1

        and np.isfinite(
            sd
        )

        and sd > 0
    )


    range_results[
        metric
    ] = {

        "expected":
            [
                lo,
                hi,
            ],

        "observed":
            [
                mn,
                mx,
            ],

        "std":
            sd,

        "in_range":
            in_range,

        "varies":
            varies,
    }


    if not in_range:

        range_fail.append(
            metric
        )


    if not varies:

        variation_fail.append(
            metric
        )


check(

    "synthetic_generator_ranges",

    not range_fail,

    range_fail,
)


check(

    "healthy_numeric_variation",

    not variation_fail,

    variation_fail,
)


# ============================================================
# 14.9 — TEMPORAL HEALTHY BEHAVIOUR
# ============================================================

active = key_df[

    key_df[
        "metric_name"
    ]
    == "load.active_ues"

].copy()


active_t = (

    active

    .groupby(
        "interval_start_utc"
    )[
        "numeric_value"
    ]

    .mean()

    .rename(
        "active"
    )

    .reset_index()
)


active_t[
    "hour"
] = (

    active_t[
        "interval_start_utc"
    ].dt.hour

    + active_t[
        "interval_start_utc"
    ].dt.minute
    / 60
)


active_t[
    "weekend"
] = (

    active_t[
        "interval_start_utc"
    ]
    .dt
    .dayofweek

    >= 5
)


night = float(

    active_t.loc[

        active_t[
            "hour"
        ].between(
            0,
            5.999,
        ),

        "active",

    ].mean()
)


evening = float(

    active_t.loc[

        active_t[
            "hour"
        ].between(
            17,
            21.999,
        ),

        "active",

    ].mean()
)


weekday = float(

    active_t.loc[

        ~active_t[
            "weekend"
        ],

        "active",

    ].mean()
)


weekend = float(

    active_t.loc[

        active_t[
            "weekend"
        ],

        "active",

    ].mean()
)


check(

    "daily_activity_cycle",

    evening
    > night,

    (
        f"evening="
        f"{evening:.3f} "
        f"night="
        f"{night:.3f}"
    ),
)


check(

    "weekday_weekend_modulation",

    weekend
    < weekday,

    (
        f"weekday="
        f"{weekday:.3f} "
        f"weekend="
        f"{weekend:.3f}"
    ),
)


# ============================================================
# 14.10 — SITE ARCHETYPE DIFFERENTIATION
# ============================================================

site_means = (

    active

    .groupby(
        "site_id"
    )[
        "numeric_value"
    ]

    .mean()
)


arch_rows = [

    {

        "site_id":
            site_id,

        "archetype":
            site_archetypes[
                site_id
            ],

        "mean":
            float(
                value
            ),
    }

    for (
        site_id,
        value
    )
    in site_means.items()
]


arch_means = (

    pd.DataFrame(
        arch_rows
    )

    .groupby(
        "archetype"
    )

    .mean(
        numeric_only=True
    )[
        "mean"
    ]

    .to_dict()
)


arch_ok = (

    arch_means[
        "URBAN_HIGH"
    ]

    > arch_means[
        "URBAN_MEDIUM"
    ]

    > arch_means[
        "SUBURBAN"
    ]
)


check(

    "site_archetype_differentiation",

    arch_ok,

    (
        f"HIGH="
        f"{arch_means['URBAN_HIGH']:.3f} "
        f"MED="
        f"{arch_means['URBAN_MEDIUM']:.3f} "
        f"SUB="
        f"{arch_means['SUBURBAN']:.3f}"
    ),
)


# ============================================================
# 14.11 — PERSISTED STANDARD PM PAIR VALIDATION
# ============================================================

pair_df = pd.concat(

    pair_frames,

    ignore_index=True,
)


keys = [

    "entity_id",
    "interval_start_utc",
    "dimensions_json",
]


pair_results = {}

reconstructed = {}


pair_alignment_fail = 0
pair_value_fail = 0


for (
    label,
    (
        att_m,
        succ_m,
    )
) in (
    PAIRS.items()
):

    att = pair_df[

        pair_df[
            "metric_name"
        ]
        == att_m

    ][
        keys
        + [
            "numeric_value"
        ]
    ].rename(

        columns={

            "numeric_value":
                "att"
        }
    )


    succ = pair_df[

        pair_df[
            "metric_name"
        ]
        == succ_m

    ][
        keys
        + [
            "numeric_value"
        ]
    ].rename(

        columns={

            "numeric_value":
                "succ"
        }
    )


    merged = succ.merge(

        att,

        on=keys,

        how="outer",

        validate="one_to_one",

        indicator=True,
    )


    alignment = int(

        (
            merged[
                "_merge"
            ]
            != "both"
        )
        .sum()
    )


    aligned = merged[

        merged[
            "_merge"
        ]
        == "both"

    ]


    violations = int(

        (

            (
                aligned[
                    "succ"
                ]
                > aligned[
                    "att"
                ]
            )

            |

            (
                aligned[
                    "succ"
                ]
                < 0
            )

            |

            (
                aligned[
                    "att"
                ]
                < 0
            )
        )
        .sum()
    )


    att_total = float(

        aligned[
            "att"
        ]
        .sum()
    )


    succ_total = float(

        aligned[
            "succ"
        ]
        .sum()
    )


    ratio = (

        float(

            100
            * succ_total
            / att_total
        )

        if att_total > 0

        else float(
            "nan"
        )
    )


    delta = abs(

        ratio

        - float(
            SANITY_RATIOS[
                label
            ]
        )
    )


    pair_results[
        label
    ] = {

        "ratio_pct":
            ratio,

        "cell13_ratio_pct":
            float(
                SANITY_RATIOS[
                    label
                ]
            ),

        "delta_pct":
            delta,

        "alignment_failures":
            alignment,

        "constraint_violations":
            violations,
    }


    reconstructed[
        label
    ] = ratio


    pair_alignment_fail += (
        alignment
    )


    pair_value_fail += (
        violations
    )


check(

    "standard_pm_pair_alignment",

    pair_alignment_fail
    == 0,

    pair_alignment_fail,
)


check(

    "standard_pm_success_le_attempt",

    pair_value_fail
    == 0,

    pair_value_fail,
)


ratio_fail = [

    label

    for (
        label,
        result
    )
    in pair_results.items()

    if (

        not np.isfinite(
            result[
                "ratio_pct"
            ]
        )

        or result[
            "delta_pct"
        ]
        > RATIO_MATCH_TOL_PCT
    )
]


check(

    "persisted_pm_ratio_reconstruction",

    not ratio_fail,

    ratio_fail,
)


# ============================================================
# 14.12 — CAUSAL RELATIONSHIP QUALIFICATION
# ============================================================
#
# Correlations are calculated within site / CU-domain groups
# and the median group correlation is used.
#
# This avoids letting differences between site archetypes alone
# create the apparent relationship.
#
# ============================================================

site_metric_set = {

    "load.active_ues",

    "radio.sinr_db",
    "radio.dl_bler_pct",

    "resource.dl_prb_util_pct",

    "resource.scheduler_pressure_pct",
    "resource.bearer_admission_pressure_pct",

    "rach.contention_index",
    "rach.access_delay_ms",

    "processing.du_cpu_pct",

    "transport.f1_latency_ms",
}


site_wide = (

    key_df[

        key_df[
            "metric_name"
        ]
        .isin(
            site_metric_set
        )

    ]

    .groupby(
        [
            "interval_start_utc",
            "site_id",
            "metric_name",
        ],
        dropna=False,
    )[
        "numeric_value"
    ]

    .mean()

    .unstack(
        "metric_name"
    )

    .reset_index()
)


domain_metric_set = {

    "load.active_ues",

    "transport.n2_util_pct",
    "transport.n2_latency_ms",

    "processing.cu_cp_cpu_pct",
    "processing.cu_cp_control_delay_ms",
}


domain_wide = (

    key_df[

        key_df[
            "metric_name"
        ]
        .isin(
            domain_metric_set
        )

    ]

    .groupby(
        [
            "interval_start_utc",
            "cu_domain_id",
            "metric_name",
        ],
        dropna=False,
    )[
        "numeric_value"
    ]

    .mean()

    .unstack(
        "metric_name"
    )

    .reset_index()
)


specs = [

    (
        "active_ues_to_dl_prb",
        site_wide,
        "site_id",
        "load.active_ues",
        "resource.dl_prb_util_pct",
        +1,
    ),

    (
        "sinr_to_dl_bler",
        site_wide,
        "site_id",
        "radio.sinr_db",
        "radio.dl_bler_pct",
        -1,
    ),

    (
        "dl_prb_to_scheduler",
        site_wide,
        "site_id",
        "resource.dl_prb_util_pct",
        "resource.scheduler_pressure_pct",
        +1,
    ),

    (
        "scheduler_to_admission",
        site_wide,
        "site_id",
        "resource.scheduler_pressure_pct",
        "resource.bearer_admission_pressure_pct",
        +1,
    ),

    (
        "rach_contention_to_delay",
        site_wide,
        "site_id",
        "rach.contention_index",
        "rach.access_delay_ms",
        +1,
    ),

    (
        "active_ues_to_du_cpu",
        site_wide,
        "site_id",
        "load.active_ues",
        "processing.du_cpu_pct",
        +1,
    ),

    (
        "active_ues_to_f1_latency",
        site_wide,
        "site_id",
        "load.active_ues",
        "transport.f1_latency_ms",
        +1,
    ),

    (
        "domain_active_to_n2_util",
        domain_wide,
        "cu_domain_id",
        "load.active_ues",
        "transport.n2_util_pct",
        +1,
    ),

    (
        "n2_util_to_n2_latency",
        domain_wide,
        "cu_domain_id",
        "transport.n2_util_pct",
        "transport.n2_latency_ms",
        +1,
    ),

    (
        "cu_cp_cpu_to_delay",
        domain_wide,
        "cu_domain_id",
        "processing.cu_cp_cpu_pct",
        "processing.cu_cp_control_delay_ms",
        +1,
    ),
]


causal_results = {}

causal_fail = []


for (
    name,
    frame,
    group,
    x,
    y,
    direction,
) in specs:

    med, vals = (
        median_group_corr(

            frame,
            group,
            x,
            y,
        )
    )


    passed = (

        np.isfinite(
            med
        )

        and (

            med
            >= MIN_ABS_DIRECTIONAL_CORR

            if direction > 0

            else med
            <= -MIN_ABS_DIRECTIONAL_CORR
        )
    )


    causal_results[
        name
    ] = {

        "median_group_correlation":
            med,

        "group_correlations":
            [
                float(
                    value
                )
                for value
                in vals
            ],

        "expected_direction":
            (
                "POSITIVE"

                if direction > 0

                else "NEGATIVE"
            ),

        "passed":
            bool(
                passed
            ),
    }


    if not passed:

        causal_fail.append(
            name
        )


check(

    "causal_relationships",

    not causal_fail,

    causal_fail,
)


# ============================================================
# 14.13 — SHARED-DOMAIN VS LOCAL-SCOPE MULTIPLICITY
# ============================================================

scope_expected = {

    "processing.cu_cp_cpu_pct":
        2,

    "processing.cu_up_cpu_pct":
        2,

    "transport.n2_util_pct":
        2,

    "transport.e1_latency_ms":
        2,

    "processing.du_cpu_pct":
        10,

    "transport.f1_latency_ms":
        10,
}


scope_results = {}

scope_fail = []


for (
    metric,
    expected_n
) in (
    scope_expected.items()
):

    counts = (

        key_df[

            key_df[
                "metric_name"
            ]
            == metric

        ]

        .groupby(
            "interval_start_utc"
        )

        .size()
    )


    passed = (

        len(
            counts
        )
        == 672

        and (

            counts
            == expected_n

        ).all()
    )


    scope_results[
        metric
    ] = {

        "expected":
            expected_n,

        "observed":
            sorted(

                map(
                    int,
                    counts.unique()
                )
            ),

        "passed":
            bool(
                passed
            ),
    }


    if not passed:

        scope_fail.append(
            metric
        )


check(

    "shared_vs_local_scope_multiplicity",

    not scope_fail,

    scope_fail,
)


# ============================================================
# 14.14 — FINAL QUALIFICATION GATE
# ============================================================

PILOT_QUALIFICATION_PASS = all(

    result[
        "passed"
    ]

    for result
    in checks.values()
)


CANONICAL_GENERATION_AUTHORIZED_BY_CELL14 = (

    PILOT_QUALIFICATION_PASS
)


failed_checks = [

    name

    for (
        name,
        result
    )
    in checks.items()

    if not result[
        "passed"
    ]
]


# ============================================================
# 14.15 — QUALIFICATION PROVENANCE
# ============================================================

semantic_content = {

    "artifact_id":
        "module5_pilot_7d_generator_qualification_v1",

    "dataset_semantic_sha256":
        dataset_semantic_sha,

    "dataset_manifest_sha256":
        dataset_manifest_sha,

    "checks":
        checks,

    "reconstructed_sanity_ratios_pct":
        reconstructed,

    "causal_relationships":
        causal_results,

    "archetype_mean_active_ues":
        arch_means,

    "temporal": {

        "night":
            night,

        "evening":
            evening,

        "weekday":
            weekday,

        "weekend":
            weekend,
    },

    "scope_multiplicity":
        scope_results,

    "range_results":
        range_results,

    "pilot_qualification_pass":
        PILOT_QUALIFICATION_PASS,

    "canonical_generation_authorized":
        CANONICAL_GENERATION_AUTHORIZED_BY_CELL14,
}


pilot_qualification_semantic_sha = (

    semantic_sha(
        semantic_content
    )
)


report = {

    **semantic_content,

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "semantic_sha256":
        pilot_qualification_semantic_sha,

    "standards_boundary": {

        "formal_ts_28_554_kpi_persisted":
            False,

        "exact_kpi_derivation_deferred_to":
            "M05_02",

        "qa_correlations_are_telecom_thresholds":
            False,

        "qa_ranges_are_telecom_thresholds":
            False,
    },
}


qualification_path = (

    PATHS[
        "manifests"
    ]

    / "module5_pilot_7d_generator_qualification_v1.json"
)


qualification_artifact_sha = None


# Persist only a passing qualification.
#
# A failed pilot remains diagnostic output rather than becoming
# a frozen approval artifact.

if PILOT_QUALIFICATION_PASS:

    qualification_path.parent.mkdir(

        parents=True,

        exist_ok=True,
    )


    if qualification_path.exists():

        with qualification_path.open(
            "r",
            encoding="utf-8",
        ) as f:

            existing = json.load(
                f
            )


        if (

            existing.get(
                "semantic_sha256"
            )

            != pilot_qualification_semantic_sha
        ):

            raise FileExistsError(

                "Different qualification artifact "
                "already exists: "
                f"{qualification_path}"
            )


    else:

        with qualification_path.open(
            "w",
            encoding="utf-8",
        ) as f:

            json.dump(

                report,

                f,

                indent=2,

                sort_keys=True,

                default=str,
            )

            f.write(
                "\n"
            )


    qualification_artifact_sha = (

        sha256_file(
            qualification_path
        )
    )


# ============================================================
# 14.16 — FINAL SUMMARY
# ============================================================

print(
    "=" * 108
)

print(
    "MODULE 5 — CELL 14 "
    "7-DAY PILOT TELEMETRY GENERATOR QUALIFICATION GATE V1"
)

print(
    "=" * 108
)


print(
    "\nPilot dataset"
)

print(
    "-" * 108
)

print(
    f"Mode                             : "
    f"{BASELINE_MODE}"
)

print(
    f"Daily partitions                 : "
    f"{len(partitions)}"
)

print(
    f"Total rows                       : "
    f"{sum(row_counts):,}"
)

print(
    f"Dataset semantic SHA             : "
    f"{dataset_semantic_sha}"
)


print(
    "\nTemporal / archetype sanity"
)

print(
    "-" * 108
)

print(
    f"Night mean active UEs            : "
    f"{night:.3f}"
)

print(
    f"Evening mean active UEs          : "
    f"{evening:.3f}"
)

print(
    f"Weekday mean active UEs          : "
    f"{weekday:.3f}"
)

print(
    f"Weekend mean active UEs          : "
    f"{weekend:.3f}"
)


for archetype in [

    "URBAN_HIGH",
    "URBAN_MEDIUM",
    "SUBURBAN",

]:

    print(

        f"{archetype:<32} : "
        f"{arch_means[archetype]:.3f} "
        "mean active UEs"
    )


print(
    "\nPersisted PM success-ratio reconstruction"
)

print(
    "-" * 108
)


for label in [

    "RRC",
    "NG",
    "INITIAL_DRB",
    "DRB",
    "RRC_RESUME",
    "DRB_RESUME",

]:

    print(

        f"{label:<32} : "
        f"{reconstructed[label]:.4f}% "
        f"| Cell 13 "
        f"{SANITY_RATIOS[label]:.4f}%"
    )


print(
    "\nCausal relationship checks"
)

print(
    "-" * 108
)

print(

    "Generator QA floor               : "
    f"|r| >= {MIN_ABS_DIRECTIONAL_CORR:.2f} "
    "in expected direction"
)

print(
    "Telecom / SLA threshold          : NO"
)


for (
    name,
    result
) in (
    causal_results.items()
):

    status = (

        "PASS"

        if result[
            "passed"
        ]

        else "FAIL"
    )


    print(

        f"{name:<38} : "
        f"r_med="
        f"{result['median_group_correlation']:+.4f} "
        f"[{status}]"
    )


print(
    "\nQualification checks"
)

print(
    "-" * 108
)


for (
    name,
    result
) in (
    checks.items()
):

    print(

        f"{'PASS' if result['passed'] else 'FAIL':<7} "
        f"{name}"
    )


print(
    "\nDecision"
)

print(
    "-" * 108
)

print(
    f"Checks executed                   : "
    f"{len(checks)}"
)

print(
    f"Checks failed                     : "
    f"{len(failed_checks)}"
)

print(
    f"PILOT_QUALIFICATION_PASS          : "
    f"{PILOT_QUALIFICATION_PASS}"
)

print(
    "CANONICAL_GENERATION_AUTHORIZED  : "
    f"{CANONICAL_GENERATION_AUTHORIZED_BY_CELL14}"
)


if failed_checks:

    print(
        "\nFailed checks"
    )

    print(
        "-" * 108
    )


    for name in failed_checks:

        print(

            f"FAIL    {name}: "
            f"{checks[name]['detail']}"
        )


print(
    "\nStandards boundary"
)

print(
    "-" * 108
)

print(
    "Formal TS 28.554 KPI persisted   : NO"
)

print(
    "Exact KPI derivation              : DEFERRED TO M05_02"
)


print(
    "\nArtifact"
)

print(
    "-" * 108
)


print(

    f"Path                               : "
    f"{qualification_path}"

    if PILOT_QUALIFICATION_PASS

    else (
        "Path                               : "
        "NOT WRITTEN — PILOT FAILED"
    )
)


print(

    f"Artifact SHA                       : "
    f"{qualification_artifact_sha}"

    if PILOT_QUALIFICATION_PASS

    else (
        "Artifact SHA                       : "
        "N/A"
    )
)


print(
    f"Semantic SHA                       : "
    f"{pilot_qualification_semantic_sha}"
)


print(
    "\n" + "=" * 108
)


print(

    (
        "CELL 14 COMPLETE — "
        "PILOT GENERATOR QUALIFIED; "
        "92-DAY CANONICAL GENERATION AUTHORIZED"
    )

    if PILOT_QUALIFICATION_PASS

    else (

        "CELL 14 COMPLETE — "
        "PILOT QUALIFICATION FAILED; "
        "CANONICAL GENERATION NOT AUTHORIZED"
    )
)


print(
    "=" * 108
)
````

<a id="cell-15-kaggle-persistence-v1-3"></a>
##### Cell 15 — Canonical Dataset Integrity + Private Kaggle Persistence V1.3

Perform the final persistence gate for the **92-day canonical healthy telemetry baseline**.

The cell revalidates all **92 daily Parquet partitions**, row counts, manifest identity, file hashes and healthy-event-channel integrity before checking the private Kaggle persistence target.

- If the dataset is absent, Cell 15 creates it once.
- If the dataset already exists, Cell 15 executes `VERIFY_EXISTING` and **does not upload or create a new version**.

The remote inventory and canonical byte sizes are verified so that a successful CLI return code alone is never treated as proof of persistence.

```text
Local canonical dataset
        ↓
Integrity + SHA verification
        ↓
PASS
        ↓
Create once OR verify existing Kaggle dataset
        ↓
Remote inventory + canonical byte verification
        ↓
Persistence receipt + provenance
```

> **The objective is to ensure that the qualified canonical telemetry baseline is durably stored, independently identifiable and reproducibly verifiable before downstream Module 5 KPI derivation begins.**


````python
# ============================================================
# MODULE 5 — CELL 15
# CANONICAL DATASET INTEGRITY + PRIVATE KAGGLE PERSISTENCE V1.3
# ============================================================
#
# Purpose
# ------------------------------------------------------------
# 1. Verify the canonical 92-day healthy telemetry baseline.
# 2. Preserve the canonical Cell 13 artifact unchanged.
# 3. Derive an explicit canonical → Kaggle storage mapping.
# 4. Use Kaggle-safe filenames from the outset.
# 5. Create the PRIVATE Kaggle Dataset only if it is absent.
# 6. If the dataset already exists, VERIFY ONLY.
# 7. Verify remote file count, Parquet count, filenames and
#    canonical byte sizes.
# 8. Write durable persistence provenance.
#
# IMPORTANT
# ------------------------------------------------------------
# - No telemetry is regenerated.
# - No canonical Parquet file is modified.
# - Existing Kaggle datasets are NOT silently versioned.
# - Re-running this cell is therefore safe and idempotent.
#
# Canonical path:
#
#   date=2026-06-01/observations.parquet
#
# Kaggle-safe representation:
#
#   date2026-06-01__observations.parquet
#
# ============================================================


from pathlib import Path
from datetime import datetime, timezone
from importlib import metadata as importlib_metadata
from io import StringIO

import hashlib
import json
import os
import shutil
import subprocess
import sys
import time

import pandas as pd
import pyarrow.parquet as pq


# ============================================================
# 15.1 — REQUIRED UPSTREAM STATE
# ============================================================

required = [

    "BASELINE_MODE",

    "BASELINE_ROLE",

    "FINAL_DATASET_DIR",

    "EXPECTED_TOTAL_ROWS",

    "dataset_manifest_sha",

    "dataset_semantic_sha",

    "CANONICAL_GENERATION_AUTHORIZED_BY_CELL14",

    "PILOT_QUALIFICATION_PASS",

    "pilot_qualification_semantic_sha",

    "PATHS",

    "sha256_file",
]


missing = [

    name

    for name in required

    if name not in globals()
]


if missing:

    raise RuntimeError(

        "Cell 15 requires completed Cells 13 and 14. "

        f"Missing: {missing}"
    )


if BASELINE_MODE != "CANONICAL_92D":

    raise RuntimeError(

        "Cell 15 persists CANONICAL_92D only. "

        f"Current mode: {BASELINE_MODE!r}"
    )


if (

    BASELINE_ROLE

    != "HEALTHY_NETWORK_OBSERVATION_BASELINE"

):

    raise RuntimeError(

        "Unexpected canonical dataset role: "

        f"{BASELINE_ROLE!r}"
    )


if not PILOT_QUALIFICATION_PASS:

    raise RuntimeError(

        "Cell 14 pilot qualification did not pass."
    )


if not CANONICAL_GENERATION_AUTHORIZED_BY_CELL14:

    raise RuntimeError(

        "Canonical generation was not authorized "

        "by Cell 14."
    )


# ============================================================
# 15.2 — CANONICAL ARTIFACT PATHS
# ============================================================

DATASET_DIR = Path(

    FINAL_DATASET_DIR
)


MANIFEST_PATH = (

    DATASET_DIR

    / "manifest.json"
)


EVENT_PATH = (

    DATASET_DIR

    / "healthy_event_channel.json"
)


for path in [

    DATASET_DIR,

    MANIFEST_PATH,

    EVENT_PATH,

]:

    if not path.exists():

        raise FileNotFoundError(path)


# ============================================================
# 15.3 — LOAD CANONICAL MANIFEST
# ============================================================

with MANIFEST_PATH.open(

    "r",

    encoding="utf-8",

) as file:

    manifest = json.load(file)


partitions = manifest.get(

    "daily_partitions",

    [],
)


if len(partitions) != 92:

    raise RuntimeError(

        "Expected 92 canonical daily partitions; "

        f"found {len(partitions)}."
    )


# ============================================================
# 15.4 — LOCAL CANONICAL INTEGRITY GATE
# ============================================================

local_checks = {}


def record_check(

    name,

    passed,

    detail="",

):

    local_checks[name] = {

        "passed":
            bool(passed),

        "detail":
            str(detail),
    }


# ------------------------------------------------------------
# Manifest byte identity
# ------------------------------------------------------------

observed_manifest_sha = (

    sha256_file(

        MANIFEST_PATH
    )
)


record_check(

    "manifest_byte_sha",

    observed_manifest_sha

    == dataset_manifest_sha,

    (
        f"observed={observed_manifest_sha} "
        f"expected={dataset_manifest_sha}"
    ),
)


# ------------------------------------------------------------
# Manifest semantic identity
# ------------------------------------------------------------

record_check(

    "manifest_semantic_sha",

    manifest.get(

        "semantic_sha256"

    )
    == dataset_semantic_sha,

    (
        f"observed="
        f"{manifest.get('semantic_sha256')} "
        f"expected={dataset_semantic_sha}"
    ),
)


# ------------------------------------------------------------
# Dataset identity
# ------------------------------------------------------------

record_check(

    "baseline_mode",

    manifest.get(

        "baseline_mode"

    )
    == "CANONICAL_92D",

    manifest.get(

        "baseline_mode"
    ),
)


record_check(

    "dataset_role",

    manifest.get(

        "dataset_role"

    )
    == "HEALTHY_NETWORK_OBSERVATION_BASELINE",

    manifest.get(

        "dataset_role"
    ),
)


record_check(

    "partition_count",

    len(partitions) == 92,

    len(partitions),
)


# ============================================================
# 15.5 — VERIFY ALL 92 CANONICAL PARQUET PARTITIONS
# ============================================================

partition_paths = []

partition_rows = 0

partition_bytes = 0

missing_partitions = []

sha_failures = []


print(

    "Verifying 92 canonical Parquet partitions..."
)


for idx, entry in enumerate(

    partitions,

    start=1,

):


    relative_path = Path(

        entry["relative_path"]
    )


    path = (

        DATASET_DIR

        / relative_path
    )


    if not path.exists():

        missing_partitions.append(

            str(relative_path)
        )

        continue


    observed_sha = (

        sha256_file(path)
    )


    expected_sha = (

        entry["parquet_sha256"]
    )


    if observed_sha != expected_sha:

        sha_failures.append(

            {

                "file":
                    str(relative_path),

                "observed":
                    observed_sha,

                "expected":
                    expected_sha,
            }
        )


    rows = int(

        pq.ParquetFile(path)

        .metadata

        .num_rows
    )


    partition_rows += rows


    partition_bytes += (

        path.stat().st_size
    )


    partition_paths.append(

        path
    )


    if (

        idx % 10 == 0

        or idx == 92

    ):

        print(

            f"  verified "

            f"{idx:>2}/92 partitions"
        )


record_check(

    "partition_files_present",

    len(missing_partitions) == 0,

    missing_partitions,
)


record_check(

    "partition_sha_integrity",

    len(sha_failures) == 0,

    f"failures={len(sha_failures)}",
)


record_check(

    "canonical_row_count",

    partition_rows

    == EXPECTED_TOTAL_ROWS,

    (
        f"observed={partition_rows:,} "
        f"expected={EXPECTED_TOTAL_ROWS:,}"
    ),
)


# ============================================================
# 15.6 — VERIFY HEALTHY EVENT CHANNEL
# ============================================================

event_sha = (

    sha256_file(

        EVENT_PATH
    )
)


expected_event_sha = (

    manifest

    .get(

        "event_channel",

        {}

    )

    .get(

        "sha256"
    )
)


record_check(

    "event_channel_sha",

    event_sha

    == expected_event_sha,

    (
        f"observed={event_sha} "
        f"expected={expected_event_sha}"
    ),
)


# ============================================================
# 15.7 — CANONICAL DATASET SIZE
# ============================================================

canonical_files = (

    partition_paths

    + [

        MANIFEST_PATH,

        EVENT_PATH,
    ]
)


canonical_bytes = sum(

    path.stat().st_size

    for path in canonical_files
)


canonical_gib = (

    canonical_bytes

    / (1024 ** 3)
)


record_check(

    "canonical_file_count",

    len(canonical_files) == 94,

    (
        f"observed={len(canonical_files)} "
        "expected=94"
    ),
)


LOCAL_CANONICAL_INTEGRITY_PASS = all(

    result["passed"]

    for result

    in local_checks.values()
)


if not LOCAL_CANONICAL_INTEGRITY_PASS:

    failed = [

        name

        for name, result

        in local_checks.items()

        if not result["passed"]
    ]


    raise RuntimeError(

        "Canonical local integrity gate FAILED. "

        f"Failed checks: {failed}"
    )


print(

    "\n✓ Local canonical integrity passed."
)


# ============================================================
# 15.8 — EXPLICIT KAGGLE STORAGE ADAPTER
# ============================================================
#
# We deliberately generate Kaggle-safe names instead of
# allowing Kaggle to silently sanitize "=" after upload.
#
# Canonical:
#
#   date=2026-06-01/observations.parquet
#
# Kaggle:
#
#   date2026-06-01__observations.parquet
#
# ============================================================

def canonical_to_kaggle_name(

    relative_path,

):

    return (

        str(relative_path)

        .replace(

            "\\",

            "/",
        )

        .replace(

            "=",

            "",
        )

        .replace(

            "/",

            "__",
        )
    )


storage_map = []


used_kaggle_names = set()


for entry in partitions:


    canonical_relative_path = (

        str(

            entry["relative_path"]

        )

        .replace(

            "\\",

            "/",
        )
    )


    kaggle_filename = (

        canonical_to_kaggle_name(

            canonical_relative_path
        )
    )


    if kaggle_filename in used_kaggle_names:

        raise RuntimeError(

            "Kaggle-safe filename collision: "

            f"{kaggle_filename}"
        )


    used_kaggle_names.add(

        kaggle_filename
    )


    source_path = (

        DATASET_DIR

        / canonical_relative_path
    )


    storage_map.append(

        {

            "canonical_relative_path":
                canonical_relative_path,

            "kaggle_filename":
                kaggle_filename,

            "sha256":
                entry["parquet_sha256"],

            "bytes":
                int(

                    source_path.stat().st_size
                ),
        }
    )


if len(storage_map) != 92:

    raise RuntimeError(

        "Unexpected Kaggle storage-map size."
    )


# ============================================================
# 15.9 — WRITE LOCAL STORAGE-MAP PROVENANCE
# ============================================================

storage_map_semantic = {

    "artifact_id":
        "module5_kaggle_storage_map_v1_3",

    "canonical_dataset_semantic_sha256":
        dataset_semantic_sha,

    "canonical_manifest_sha256":
        dataset_manifest_sha,

    "canonical_dataset_modified":
        False,

    "adapter_rule":
        (
            'remove "=" from partition component; '
            'replace "/" with "__"'
        ),

    "partition_count":
        92,

    "files":
        storage_map,
}


storage_map_raw = json.dumps(

    storage_map_semantic,

    sort_keys=True,

    separators=(",", ":"),

).encode()


storage_map_semantic_sha = (

    hashlib

    .sha256(

        storage_map_raw

    )

    .hexdigest()
)


storage_map_payload = {

    **storage_map_semantic,

    "semantic_sha256":
        storage_map_semantic_sha,
}


STORAGE_MAP_PATH = (

    PATHS["manifests"]

    / "module5_kaggle_storage_map_v1_3.json"
)


STORAGE_MAP_PATH.parent.mkdir(

    parents=True,

    exist_ok=True,
)


with STORAGE_MAP_PATH.open(

    "w",

    encoding="utf-8",

) as file:

    json.dump(

        storage_map_payload,

        file,

        indent=2,

        sort_keys=True,
    )

    file.write("\n")


storage_map_artifact_sha = (

    sha256_file(

        STORAGE_MAP_PATH
    )
)


# ============================================================
# 15.10 — KAGGLE DATASET IDENTITY
# ============================================================

KAGGLE_OWNER = (

    "cliffordimaguezegie"
)


KAGGLE_DATASET_SLUG = (

    "module5-5g-ran-healthy-telemetry-v1"
)


KAGGLE_DATASET_HANDLE = (

    f"{KAGGLE_OWNER}/"

    f"{KAGGLE_DATASET_SLUG}"
)


KAGGLE_DATASET_TITLE = (

    "Module 5 5G RAN Healthy Telemetry V1"
)


KAGGLE_DATASET_URL = (

    "https://www.kaggle.com/datasets/"

    f"{KAGGLE_DATASET_HANDLE}"
)


# ============================================================
# 15.11 — KAGGLE CLI RUNTIME
# ============================================================

TARGET_KAGGLE_VERSION = "2.2.4"


try:

    kaggle_version = (

        importlib_metadata

        .version(

            "kaggle"
        )
    )


except importlib_metadata.PackageNotFoundError:

    kaggle_version = None


if kaggle_version != TARGET_KAGGLE_VERSION:


    print(

        "\nInstalling Kaggle CLI "

        f"{TARGET_KAGGLE_VERSION}..."
    )


    subprocess.run(

        [

            sys.executable,

            "-m",

            "pip",

            "install",

            "-q",

            (
                "kaggle=="

                f"{TARGET_KAGGLE_VERSION}"
            ),
        ],

        check=True,
    )


kaggle_version = (

    importlib_metadata

    .version(

        "kaggle"
    )
)


# ============================================================
# 15.12 — KAGGLE AUTHENTICATION
# ============================================================

kaggle_token = (

    os.environ

    .get(

        "KAGGLE_API_TOKEN",

        "",

    )

    .strip()
)


try:

    from google.colab import userdata


    if not kaggle_token:

        try:

            kaggle_token = (

                userdata

                .get(

                    "KAGGLE_API_TOKEN"

                )

                .strip()
            )


        except Exception:

            pass


except Exception:

    pass


if kaggle_token:

    os.environ[

        "KAGGLE_API_TOKEN"

    ] = kaggle_token


legacy_json = (

    Path.home()

    / ".kaggle"

    / "kaggle.json"
)


if (

    not kaggle_token

    and not legacy_json.exists()

):

    raise RuntimeError(

        "No Kaggle credential found. "

        "Add KAGGLE_API_TOKEN "

        "to Colab Secrets."
    )


# ============================================================
# 15.13 — KAGGLE COMMAND HELPERS
# ============================================================

def run_kaggle(

    args,

    *,

    check=True,

):

    result = subprocess.run(

        [

            "kaggle",

            *args,
        ],

        text=True,

        capture_output=True,
    )


    if (

        check

        and result.returncode != 0

    ):

        raise RuntimeError(

            "Kaggle command failed:\n"

            + " ".join(

                [

                    "kaggle",

                    *args,
                ]
            )

            + "\n\nSTDOUT:\n"

            + (

                result.stdout

                or "<empty>"
            )

            + "\n\nSTDERR:\n"

            + (

                result.stderr

                or "<empty>"
            )
        )


    return result


def parse_csv_output(

    text,

):


    if not text.strip():

        return pd.DataFrame()


    return pd.read_csv(

        StringIO(text)
    )


# ============================================================
# 15.14 — AUTHENTICATION + DATASET DISCOVERY
# ============================================================

def get_owned_dataset_refs():


    result = run_kaggle(

        [

            "datasets",

            "list",

            "--mine",

            "--page-size",

            "100",

            "--format",

            "csv(ref)",
        ],

        check=False,
    )


    # --------------------------------------------------------
    # Compatibility fallback
    # --------------------------------------------------------

    if result.returncode != 0:

        result = run_kaggle(

            [

                "datasets",

                "list",

                "--mine",

                "--page-size",

                "100",

                "-v",
            ]
        )


    frame = parse_csv_output(

        result.stdout
    )


    if "ref" not in frame.columns:

        raise RuntimeError(

            "Could not identify Kaggle dataset "

            "reference column."
        )


    return set(

        frame["ref"]

        .astype(str)
    )


owned_dataset_refs = (

    get_owned_dataset_refs()
)


KAGGLE_AUTHENTICATION_PASS = True


KAGGLE_DATASET_EXISTS = (

    KAGGLE_DATASET_HANDLE

    in owned_dataset_refs
)


print(

    "\nKaggle persistence target"
)

print(

    "-" * 108
)

print(

    f"Dataset handle          : "

    f"{KAGGLE_DATASET_HANDLE}"
)

print(

    "Visibility              : PRIVATE"
)

print(

    f"Dataset already exists  : "

    f"{KAGGLE_DATASET_EXISTS}"
)


# ============================================================
# 15.15 — CREATE DATASET ONLY IF ABSENT
# ============================================================
#
# IMPORTANT:
#
# Existing dataset → VERIFY ONLY.
#
# We deliberately do NOT create a new version automatically.
#
# ============================================================

CREATED_NOW = False


if not KAGGLE_DATASET_EXISTS:


    print(

        "\nDataset not found."

        "\nPreparing Kaggle staging area..."
    )


    STAGING_ROOT = (

        Path(

            "/content/module5_artifacts/"

            "kaggle_staging"
        )
    )


    STAGING_DIR = (

        STAGING_ROOT

        / KAGGLE_DATASET_SLUG
    )


    if STAGING_DIR.exists():

        shutil.rmtree(

            STAGING_DIR
        )


    STAGING_DIR.mkdir(

        parents=True,

        exist_ok=True,
    )


    hardlink_count = 0

    copy_count = 0


    # --------------------------------------------------------
    # Stage 92 Kaggle-safe Parquet files
    # --------------------------------------------------------

    for item in storage_map:


        source = (

            DATASET_DIR

            / item[

                "canonical_relative_path"

            ]
        )


        destination = (

            STAGING_DIR

            / item[

                "kaggle_filename"

            ]
        )


        try:

            os.link(

                source,

                destination,
            )


            hardlink_count += 1


        except OSError:


            shutil.copy2(

                source,

                destination,
            )


            copy_count += 1


        if (

            sha256_file(

                destination

            )

            != item["sha256"]

        ):

            raise RuntimeError(

                "Kaggle staging SHA mismatch: "

                f"{destination.name}"
            )


    # --------------------------------------------------------
    # Canonical control artifacts
    # --------------------------------------------------------

    shutil.copy2(

        MANIFEST_PATH,

        STAGING_DIR

        / "manifest.json",
    )


    shutil.copy2(

        EVENT_PATH,

        STAGING_DIR

        / "healthy_event_channel.json",
    )


    # --------------------------------------------------------
    # Storage map exposed with dataset
    # --------------------------------------------------------

    shutil.copy2(

        STORAGE_MAP_PATH,

        STAGING_DIR

        / "kaggle_file_map.json",
    )


    KAGGLE_FILE_MAP_PATH = (

        STAGING_DIR

        / "kaggle_file_map.json"
    )


    # ========================================================
    # Proven minimal Kaggle metadata
    # ========================================================

    kaggle_metadata = {

        "title":
            KAGGLE_DATASET_TITLE,

        "id":
            KAGGLE_DATASET_HANDLE,

        "licenses": [

            {

                "name":
                    "other"
            }
        ],
    }


    KAGGLE_METADATA_PATH = (

        STAGING_DIR

        / "dataset-metadata.json"
    )


    with KAGGLE_METADATA_PATH.open(

        "w",

        encoding="utf-8",

    ) as file:

        json.dump(

            kaggle_metadata,

            file,

            indent=2,
        )

        file.write("\n")


    # --------------------------------------------------------
    # Metadata validation
    # --------------------------------------------------------

    with KAGGLE_METADATA_PATH.open(

        "r",

        encoding="utf-8",

    ) as file:

        metadata_check = (

            json.load(file)
        )


    if (

        metadata_check["id"]

        != KAGGLE_DATASET_HANDLE

    ):

        raise RuntimeError(

            "Kaggle metadata ID validation failed."
        )


    if (

        metadata_check["licenses"][0]["name"]

        != "other"

    ):

        raise RuntimeError(

            "Kaggle metadata license "

            "validation failed."
        )


    print(

        "\nKaggle staging ready."
    )


    print(

        f"Hard-linked partitions  : "

        f"{hardlink_count}"
    )


    print(

        f"Copied partitions       : "

        f"{copy_count}"
    )


    print(

        "License                 : other"
    )


    # ========================================================
    # CREATE PRIVATE DATASET
    # ========================================================
    #
    # stdout/stderr are intentionally inherited by the notebook
    # so Kaggle upload progress remains visible.
    #
    # ========================================================

    print(

        "\nCreating private Kaggle Dataset..."

        "\nUpload progress will appear below.\n"
    )


    upload_result = subprocess.run(

        [

            "kaggle",

            "datasets",

            "create",

            "--path",

            str(

                STAGING_DIR
            ),

            "--keep-tabular",

            "--dir-mode",

            "skip",
        ]
    )


    if upload_result.returncode != 0:

        raise RuntimeError(

            "Kaggle dataset creation failed. "

            "Canonical local dataset remains unchanged."
        )


    CREATED_NOW = True


    # --------------------------------------------------------
    # Wait until Kaggle account inventory exposes the dataset
    # --------------------------------------------------------

    DATASET_DISCOVERY_PASS = False


    for _ in range(24):


        owned_dataset_refs = (

            get_owned_dataset_refs()
        )


        if (

            KAGGLE_DATASET_HANDLE

            in owned_dataset_refs

        ):

            DATASET_DISCOVERY_PASS = True

            break


        time.sleep(5)


    if not DATASET_DISCOVERY_PASS:

        raise RuntimeError(

            "Kaggle upload returned successfully, "

            "but the new dataset has not yet appeared "

            "in the authenticated account inventory."
        )


    print(

        "\n✓ Kaggle dataset created "

        "and discovered."
    )


else:


    DATASET_DISCOVERY_PASS = True


    print(

        "\n✓ Kaggle dataset already exists."

        "\n  No upload or version operation will be performed."

        "\n  Proceeding directly to remote verification."
    )


# ============================================================
# 15.16 — REMOTE FILE INVENTORY
# ============================================================

remote_result = run_kaggle(

    [

        "datasets",

        "files",

        KAGGLE_DATASET_HANDLE,

        "--page-size",

        "200",

        "--format",

        "csv(name,size)",
    ]
)


remote_df = parse_csv_output(

    remote_result.stdout
)


if not {

    "name",

    "size",

}.issubset(

    remote_df.columns

):

    raise RuntimeError(

        "Unexpected Kaggle remote inventory schema: "

        f"{remote_df.columns.tolist()}"
    )


remote_df["name"] = (

    remote_df["name"]

    .astype(str)
)


remote_df["size"] = (

    pd.to_numeric(

        remote_df["size"],

        errors="raise",

    )

    .astype(int)
)


remote_sizes = dict(

    zip(

        remote_df["name"],

        remote_df["size"],
    )
)


# ============================================================
# 15.17 — EXPECTED REMOTE INVENTORY
# ============================================================

expected_remote_names = {

    item["kaggle_filename"]

    for item in storage_map
}


expected_remote_names.update(

    {

        "manifest.json",

        "healthy_event_channel.json",

        "kaggle_file_map.json",
    }
)


if len(expected_remote_names) != 95:

    raise RuntimeError(

        "Unexpected expected remote file count: "

        f"{len(expected_remote_names)}"
    )


observed_remote_names = set(

    remote_df["name"]
)


remote_missing = sorted(

    expected_remote_names

    - observed_remote_names
)


remote_unexpected = sorted(

    observed_remote_names

    - expected_remote_names
)


REMOTE_FILE_NAME_PASS = (

    len(remote_missing) == 0

    and len(remote_unexpected) == 0
)


# ============================================================
# 15.18 — REMOTE STRUCTURAL VERIFICATION
# ============================================================

remote_parquet_df = remote_df[

    remote_df["name"]

    .str.endswith(

        ".parquet",

        na=False,
    )
]


REMOTE_FILE_COUNT_PASS = (

    len(remote_df) == 95
)


REMOTE_PARQUET_COUNT_PASS = (

    len(remote_parquet_df) == 92
)


# ============================================================
# 15.19 — CANONICAL BYTE-SIZE VERIFICATION
# ============================================================
#
# Exact byte verification is performed for:
#
#   92 Parquet partitions
#   manifest.json
#   healthy_event_channel.json
#
# These are canonical artifacts.
#
# kaggle_file_map.json is an adapter/provenance artifact.
# If this execution created the dataset, its bytes are also
# verified exactly. If the dataset pre-existed, its presence
# is verified but we do not require it to match a newly
# regenerated local adapter-map representation.
#
# ============================================================

expected_remote_sizes = {

    item["kaggle_filename"]:

        int(

            item["bytes"]
        )

    for item in storage_map
}


expected_remote_sizes[

    "manifest.json"

] = (

    MANIFEST_PATH

    .stat()

    .st_size
)


expected_remote_sizes[

    "healthy_event_channel.json"

] = (

    EVENT_PATH

    .stat()

    .st_size
)


if CREATED_NOW:


    expected_remote_sizes[

        "kaggle_file_map.json"

    ] = (

        KAGGLE_FILE_MAP_PATH

        .stat()

        .st_size
    )


size_mismatches = []


for (

    filename,

    expected_size,

) in expected_remote_sizes.items():


    observed_size = (

        remote_sizes.get(

            filename
        )
    )


    if observed_size != expected_size:

        size_mismatches.append(

            {

                "file":
                    filename,

                "expected_bytes":
                    expected_size,

                "observed_bytes":
                    observed_size,
            }
        )


REMOTE_CANONICAL_SIZE_PASS = (

    len(size_mismatches) == 0
)


# ============================================================
# 15.20 — FINAL PERSISTENCE GATE
# ============================================================

KAGGLE_PERSISTENCE_PASS = all(

    [

        LOCAL_CANONICAL_INTEGRITY_PASS,

        KAGGLE_AUTHENTICATION_PASS,

        DATASET_DISCOVERY_PASS,

        REMOTE_FILE_COUNT_PASS,

        REMOTE_PARQUET_COUNT_PASS,

        REMOTE_FILE_NAME_PASS,

        REMOTE_CANONICAL_SIZE_PASS,
    ]
)


if not KAGGLE_PERSISTENCE_PASS:

    raise RuntimeError(

        "Kaggle persistence verification FAILED.\n"

        f"Missing files      : {remote_missing}\n"

        f"Unexpected files   : {remote_unexpected}\n"

        f"Size mismatches    : {size_mismatches}\n"

        f"Remote files       : {len(remote_df)}\n"

        f"Remote Parquet     : {len(remote_parquet_df)}"
    )


# ============================================================
# 15.21 — FINAL PERSISTENCE RECEIPT V1.3
# ============================================================

operation = (

    "CREATE"

    if CREATED_NOW

    else "VERIFY_EXISTING"
)


receipt_semantic = {

    "artifact_id":
        "module5_canonical_kaggle_persistence_v1_3",

    "storage_provider":
        "KAGGLE_DATASET",

    "dataset_handle":
        KAGGLE_DATASET_HANDLE,

    "visibility":
        "PRIVATE",

    "operation":
        operation,

    "canonical_dataset_role":
        BASELINE_ROLE,

    "canonical_dataset_semantic_sha256":
        dataset_semantic_sha,

    "canonical_manifest_sha256":
        dataset_manifest_sha,

    "pilot_qualification_semantic_sha256":
        pilot_qualification_semantic_sha,

    "kaggle_storage_map_semantic_sha256":
        storage_map_semantic_sha,

    "kaggle_license":
        "other",

    "canonical_rows":
        int(

            partition_rows
        ),

    "canonical_bytes":
        int(

            canonical_bytes
        ),

    "canonical_daily_partitions":
        92,

    "canonical_files":
        94,

    "remote_files":
        int(

            len(remote_df)
        ),

    "remote_parquet_files":
        int(

            len(remote_parquet_df)
        ),

    "dataset_discovery_verified":
        DATASET_DISCOVERY_PASS,

    "remote_name_inventory_verified":
        REMOTE_FILE_NAME_PASS,

    "remote_canonical_byte_sizes_verified":
        REMOTE_CANONICAL_SIZE_PASS,

    "canonical_dataset_modified":
        False,

    "automatic_version_created":
        False,

    "kaggle_cli_version":
        kaggle_version,

    "persistence_pass":
        KAGGLE_PERSISTENCE_PASS,
}


receipt_raw = json.dumps(

    receipt_semantic,

    sort_keys=True,

    separators=(",", ":"),

).encode()


receipt_semantic_sha = (

    hashlib

    .sha256(

        receipt_raw

    )

    .hexdigest()
)


receipt = {

    **receipt_semantic,

    "verified_utc":
        datetime.now(

            timezone.utc

        ).isoformat(),

    "semantic_sha256":
        receipt_semantic_sha,
}


RECEIPT_PATH = (

    PATHS["manifests"]

    / "module5_canonical_kaggle_persistence_v1_3.json"
)


if RECEIPT_PATH.exists():


    with RECEIPT_PATH.open(

        "r",

        encoding="utf-8",

    ) as file:

        existing_receipt = (

            json.load(file)
        )


    if (

        existing_receipt.get(

            "semantic_sha256"

        )
        != receipt_semantic_sha

    ):

        raise FileExistsError(

            "A different V1.3 persistence receipt "

            "already exists:\n"

            f"{RECEIPT_PATH}"
        )


else:


    with RECEIPT_PATH.open(

        "w",

        encoding="utf-8",

    ) as file:

        json.dump(

            receipt,

            file,

            indent=2,

            sort_keys=True,
        )

        file.write("\n")


receipt_artifact_sha = (

    sha256_file(

        RECEIPT_PATH
    )
)


# ============================================================
# 15.22 — FINAL SUMMARY
# ============================================================

print(

    "\n" + "=" * 108
)


print(

    "MODULE 5 — CELL 15 "

    "CANONICAL DATASET INTEGRITY + "

    "PRIVATE KAGGLE PERSISTENCE V1.3"
)


print(

    "=" * 108
)


print(

    "\nCanonical dataset"
)


print(

    "-" * 108
)


print(

    f"Role                             : "

    f"{BASELINE_ROLE}"
)


print(

    "Window                           : "

    "2026-06-01 → 2026-09-01 UTC"
)


print(

    f"Daily Parquet partitions         : "

    f"{len(partitions)}"
)


print(

    f"Total rows                       : "

    f"{partition_rows:,}"
)


print(

    f"Canonical files                  : "

    f"{len(canonical_files)}"
)


print(

    f"Canonical size                   : "

    f"{canonical_gib:.3f} GiB"
)


print(

    "Canonical dataset modified       : NO"
)


print(

    "\nLocal integrity"
)


print(

    "-" * 108
)


for (

    name,

    result,

) in local_checks.items():


    print(

        f"{'PASS' if result['passed'] else 'FAIL':<7} "

        f"{name}"
    )


print(

    "\nKaggle storage adapter"
)


print(

    "-" * 108
)


print(

    "Canonical example                : "

    "date=2026-06-01/observations.parquet"
)


print(

    "Kaggle example                   : "

    "date2026-06-01__observations.parquet"
)


print(

    f"Storage-map semantic SHA         : "

    f"{storage_map_semantic_sha}"
)


print(

    f"Storage-map artifact SHA         : "

    f"{storage_map_artifact_sha}"
)


print(

    "\nKaggle persistence"
)


print(

    "-" * 108
)


print(

    f"Kaggle CLI version               : "

    f"{kaggle_version}"
)


print(

    f"Dataset handle                   : "

    f"{KAGGLE_DATASET_HANDLE}"
)


print(

    "Visibility                       : PRIVATE"
)


print(

    f"Operation                        : "

    f"{operation}"
)


print(

    f"Dataset discovery                : "

    f"{'PASS' if DATASET_DISCOVERY_PASS else 'FAIL'}"
)


print(

    f"Remote files                     : "

    f"{len(remote_df)}"
)


print(

    f"Remote Parquet partitions        : "

    f"{len(remote_parquet_df)}"
)


print(

    f"Remote file count                : "

    f"{'PASS' if REMOTE_FILE_COUNT_PASS else 'FAIL'}"
)


print(

    f"Remote Parquet count             : "

    f"{'PASS' if REMOTE_PARQUET_COUNT_PASS else 'FAIL'}"
)


print(

    f"Remote filename inventory        : "

    f"{'PASS' if REMOTE_FILE_NAME_PASS else 'FAIL'}"
)


print(

    f"Remote canonical byte sizes      : "

    f"{'PASS' if REMOTE_CANONICAL_SIZE_PASS else 'FAIL'}"
)


print(

    "\nProvenance"
)


print(

    "-" * 108
)


print(

    f"Dataset semantic SHA             : "

    f"{dataset_semantic_sha}"
)


print(

    f"Generator manifest SHA           : "

    f"{dataset_manifest_sha}"
)


print(

    f"Pilot qualification semantic SHA : "

    f"{pilot_qualification_semantic_sha}"
)


print(

    f"Kaggle storage-map semantic SHA  : "

    f"{storage_map_semantic_sha}"
)


print(

    "\nPersistence receipt"
)


print(

    "-" * 108
)


print(

    f"Path                             : "

    f"{RECEIPT_PATH}"
)


print(

    f"Artifact SHA                     : "

    f"{receipt_artifact_sha}"
)


print(

    f"Semantic SHA                     : "

    f"{receipt_semantic_sha}"
)


print(

    "\nDecision"
)


print(

    "-" * 108
)


print(

    f"LOCAL_CANONICAL_INTEGRITY_PASS   : "

    f"{LOCAL_CANONICAL_INTEGRITY_PASS}"
)


print(

    f"DATASET_DISCOVERY_PASS           : "

    f"{DATASET_DISCOVERY_PASS}"
)


print(

    f"KAGGLE_PERSISTENCE_PASS          : "

    f"{KAGGLE_PERSISTENCE_PASS}"
)


print(

    "\n" + "=" * 108
)


print(

    "CELL 15 COMPLETE — "

    "CANONICAL DATASET VERIFIED ON "

    "PRIVATE KAGGLE STORAGE"
)


print(

    "=" * 108
)
````

##### Cell 16 — M05_00 → Cell 09 Provenance Reconciliation V1

Formally reconcile the earlier M05_00 causal-reference SHA embedded in Cell 09 provenance with the final authoritative M05_00 notebook identity.

The dependency audit confirmed that the affected Cell 09 identity propagates downstream **only through provenance metadata** and does not influence telemetry values, KPI measurements, Parquet content or generated network observations.

Therefore, the reconciliation records the final M05_00 reference as authoritative while preserving all existing downstream semantic identities and persisted telemetry artifacts.

```text
Telemetry regeneration required : NO
Parquet modification required   : NO
Downstream artifact rewrite     : NO
Kaggle re-upload required       : NO
```

> **The provenance record is reconciled without altering validated Module 5 network truth.**


````python
# ============================================================
# MODULE 5 — CELL 16
# M05_00 → CELL 09 PROVENANCE RECONCILIATION V1
# ============================================================
#
# Purpose
# ------------------------------------------------------------
# Formally reconcile the Cell 09 causal-reference provenance
# pointer without modifying any previously generated artifact.
#
# Background
# ------------------------------------------------------------
# Cell 09 was originally recorded against an earlier M05_00
# notebook byte identity:
#
#   6a58fb...
#
# The final expanded M05_00 implementation / causal reference
# notebook has byte identity:
#
#   9c4db1...
#
# Read-only dependency audits confirmed that the Cell 09 SHA
# propagates downstream ONLY through provenance metadata.
#
# Therefore:
#
#   - no telemetry values are regenerated
#   - no Parquet files are modified
#   - no existing semantic SHA is rewritten
#   - no Kaggle Dataset version is created
#
# Instead, an immutable reconciliation artifact records that
# the earlier M05_00 provenance reference is superseded by the
# final M05_00 reference.
#
# ============================================================


from pathlib import Path
from datetime import datetime, timezone

import hashlib
import json


# ============================================================
# 16.1 — REQUIRED STATE
# ============================================================

required = [
    "PATHS",
    "sha256_file",
    "dataset_semantic_sha",
    "dataset_manifest_sha",
    "pilot_qualification_semantic_sha",
]


missing = [
    name
    for name in required
    if name not in globals()
]


if missing:
    raise RuntimeError(
        "Cell 16 requires completed Module 5 foundation state. "
        f"Missing: {missing}"
    )


# ============================================================
# 16.2 — AUTHORITATIVE IDENTITIES
# ============================================================

M05_00_EARLIER_SHA256 = (
    "6a58fb847fa47224c7289ccffb3126ffe"
    "76701d4d9c6b7fc5fd1e2a213858747"
)


M05_00_FINAL_SHA256 = (
    "9c4db1552791510d3237fd8c7ff93cd"
    "7ae66d614d42fed0da7a268bf77fa910e"
)


CELL09_SEMANTIC_SHA256 = (
    "8ccce01ee1375bbef52611a4b008ea013"
    "8bae88874993be67f7577aa0838f42a"
)


CELL10_SEMANTIC_SHA256 = (
    "aaeb1acf5739a9d80d7745469faa71fe"
    "142181c7e96df6bc2c9e5f8b8b031b68"
)


CELL11_SEMANTIC_SHA256 = (
    "b12fd1cc5b208963a13caecc33d03016"
    "a612ff1f1f112686f4db435db66b421d"
)


CELL12_SEMANTIC_SHA256 = (
    "07ba4e0cc0f672cc03268e96bb53aa0f"
    "322241be43a4f957e7065b0d6ebae312"
)


CANONICAL_DATASET_SEMANTIC_SHA256 = (
    dataset_semantic_sha
)


CANONICAL_MANIFEST_SHA256 = (
    dataset_manifest_sha
)


PILOT_QUALIFICATION_SEMANTIC_SHA256 = (
    pilot_qualification_semantic_sha
)


KAGGLE_DATASET_HANDLE = (
    "cliffordimaguezegie/"
    "module5-5g-ran-healthy-telemetry-v1"
)


KAGGLE_PERSISTENCE_SEMANTIC_SHA256 = (
    "0719523ca6c6907ecefbacae84f8bce4"
    "91ddf37a886b71c6a9ed3ffe729033c7"
)


# ============================================================
# 16.3 — EXPECTED AFFECTED ARTIFACTS
# ============================================================

ARTIFACT_ROOT = Path(
    "/content/module5_artifacts"
)


EXPECTED_REFERENCES = {

    (
        ARTIFACT_ROOT
        / "telemetry"
        / "module5_5g_ran_accessibility_causal_observability_config_v1.json"
    ): {
        "$.semantic_sha256",
    },

    (
        ARTIFACT_ROOT
        / "kpi"
        / "module5_5g_ran_accessibility_reference_domain_v1_2.json"
    ): {
        "$.nodes.diagnostic::rach_access_success.metadata.cell9_semantic_sha256",
        "$.nodes.diagnostic::radio_condition.metadata.cell9_semantic_sha256",
        "$.nodes.diagnostic::cell_resource_load.metadata.cell9_semantic_sha256",
        "$.nodes.diagnostic::ran_alarms_events.metadata.cell9_semantic_sha256",
        "$.nodes.diagnostic::ran_configuration.metadata.cell9_semantic_sha256",
        "$.nodes.diagnostic::transport_boundary.metadata.cell9_semantic_sha256",
        "$.provenance.cell9_causal_observability_semantic_sha256",
    },

    (
        ARTIFACT_ROOT
        / "telemetry"
        / "module5_synthetic_5g_ran_environment_v1_1.json"
    ): {
        "$.provenance.cell9_causal_observability_semantic_sha256",
    },

    (
        ARTIFACT_ROOT
        / "telemetry"
        / "module5_5g_ran_telemetry_observation_contract_v1_1.json"
    ): {
        "$.provenance.cell9_semantic_sha256",
    },

    (
        ARTIFACT_ROOT
        / "telemetry"
        / "healthy_baseline_pilot_7d_v1"
        / "manifest.json"
    ): {
        "$.provenance.cell9_semantic_sha256",
    },

    (
        ARTIFACT_ROOT
        / "telemetry"
        / "healthy_baseline_v1"
        / "manifest.json"
    ): {
        "$.provenance.cell9_semantic_sha256",
    },

    (
        ARTIFACT_ROOT
        / "kaggle_staging"
        / "module5-5g-ran-healthy-telemetry-v1"
        / "manifest.json"
    ): {
        "$.provenance.cell9_semantic_sha256",
    },
}


# ============================================================
# 16.4 — JSON PATH SEARCH
# ============================================================

def find_json_value_paths(
    obj,
    target,
    path="$",
):
    hits = []


    if isinstance(obj, dict):

        for key, value in obj.items():

            hits.extend(
                find_json_value_paths(
                    value,
                    target,
                    f"{path}.{key}",
                )
            )


    elif isinstance(obj, list):

        for index, value in enumerate(obj):

            hits.extend(
                find_json_value_paths(
                    value,
                    target,
                    f"{path}[{index}]",
                )
            )


    elif isinstance(obj, str):

        if target in obj:

            hits.append(path)


    return hits


# ============================================================
# 16.5 — REVALIDATE PROVENANCE-ONLY PROPAGATION
# ============================================================

observed_references = {}

missing_expected_artifacts = []


for artifact_path, expected_paths in EXPECTED_REFERENCES.items():


    if not artifact_path.exists():

        missing_expected_artifacts.append(
            str(artifact_path)
        )

        continue


    with artifact_path.open(
        "r",
        encoding="utf-8",
    ) as file:

        payload = json.load(file)


    observed_paths = set(
        find_json_value_paths(
            payload,
            CELL09_SEMANTIC_SHA256,
        )
    )


    observed_references[
        artifact_path
    ] = observed_paths


if missing_expected_artifacts:

    raise RuntimeError(
        "Expected provenance artifacts are missing:\n"
        + "\n".join(
            missing_expected_artifacts
        )
    )


# ============================================================
# 16.6 — EXACT REFERENCE-SET VERIFICATION
# ============================================================

reference_mismatches = []


for artifact_path, expected_paths in EXPECTED_REFERENCES.items():


    observed_paths = (
        observed_references[
            artifact_path
        ]
    )


    if observed_paths != expected_paths:

        reference_mismatches.append(
            {
                "artifact":
                    str(artifact_path),

                "expected":
                    sorted(expected_paths),

                "observed":
                    sorted(observed_paths),
            }
        )


PROVENANCE_REFERENCE_SET_PASS = (
    len(reference_mismatches) == 0
)


if not PROVENANCE_REFERENCE_SET_PASS:

    raise RuntimeError(
        "Cell 09 downstream reference set changed "
        "since the provenance audit.\n"
        f"{reference_mismatches}"
    )


# ============================================================
# 16.7 — CONFIRM NO PARQUET CONTENT REFERENCES CELL 09 SHA
# ============================================================
#
# SHA references occur only inside JSON metadata.
# The Parquet files themselves are immutable binary telemetry
# artifacts and are not rewritten by this reconciliation.
#
# ============================================================

canonical_dataset_dir = (
    ARTIFACT_ROOT
    / "telemetry"
    / "healthy_baseline_v1"
)


pilot_dataset_dir = (
    ARTIFACT_ROOT
    / "telemetry"
    / "healthy_baseline_pilot_7d_v1"
)


canonical_parquet_files = list(
    canonical_dataset_dir.rglob(
        "*.parquet"
    )
)


pilot_parquet_files = list(
    pilot_dataset_dir.rglob(
        "*.parquet"
    )
)


PARQUET_STRUCTURE_PASS = (
    len(canonical_parquet_files) == 92
    and len(pilot_parquet_files) == 7
)


if not PARQUET_STRUCTURE_PASS:

    raise RuntimeError(
        "Unexpected telemetry partition structure. "
        f"Canonical={len(canonical_parquet_files)}, "
        f"Pilot={len(pilot_parquet_files)}"
    )


# ============================================================
# 16.8 — RECONCILIATION DECISION
# ============================================================

RECONCILIATION_CLASSIFICATION = (
    "PROVENANCE_ONLY"
)


TELEMETRY_REGENERATION_REQUIRED = False

PARQUET_MODIFICATION_REQUIRED = False

DOWNSTREAM_ARTIFACT_REWRITE_REQUIRED = False

KAGGLE_REUPLOAD_REQUIRED = False

KAGGLE_VERSION_REQUIRED = False


# ============================================================
# 16.9 — BUILD IMMUTABLE RECONCILIATION ARTIFACT
# ============================================================

affected_artifacts = []


for artifact_path, paths in observed_references.items():

    affected_artifacts.append(
        {
            "artifact_path":
                str(artifact_path),

            "reference_paths":
                sorted(paths),

            "reference_count":
                len(paths),
        }
    )


reconciliation_semantic = {

    "artifact_id":
        "module5_m05_00_cell09_provenance_reconciliation_v1",

    "module":
        "M05_01",

    "reconciliation_type":
        "SUPERSEDED_UPSTREAM_REFERENCE",

    "classification":
        RECONCILIATION_CLASSIFICATION,

    "earlier_upstream_reference": {

        "artifact":
            (
                "M05_00 implementation / "
                "causal reference notebook"
            ),

        "sha256":
            M05_00_EARLIER_SHA256,

        "status":
            "SUPERSEDED",
    },

    "authoritative_upstream_reference": {

        "artifact":
            (
                "M05_00_Implementation_and_"
                "Causal_Observability_Reference_v1.ipynb"
            ),

        "sha256":
            M05_00_FINAL_SHA256,

        "status":
            "AUTHORITATIVE",
    },

    "affected_cell09_semantic_sha256":
        CELL09_SEMANTIC_SHA256,

    "downstream_semantic_identities_preserved": {

        "cell10_semantic_sha256":
            CELL10_SEMANTIC_SHA256,

        "cell11_semantic_sha256":
            CELL11_SEMANTIC_SHA256,

        "cell12_semantic_sha256":
            CELL12_SEMANTIC_SHA256,

        "canonical_dataset_semantic_sha256":
            CANONICAL_DATASET_SEMANTIC_SHA256,

        "canonical_manifest_sha256":
            CANONICAL_MANIFEST_SHA256,

        "pilot_qualification_semantic_sha256":
            PILOT_QUALIFICATION_SEMANTIC_SHA256,

        "kaggle_persistence_semantic_sha256":
            KAGGLE_PERSISTENCE_SEMANTIC_SHA256,
    },

    "affected_artifacts":
        affected_artifacts,

    "exact_reference_locations":
        sum(
            item["reference_count"]
            for item
            in affected_artifacts
        ),

    "reconciliation_policy": {

        "existing_artifacts_mutated":
            False,

        "existing_semantic_hashes_recomputed":
            False,

        "telemetry_regenerated":
            False,

        "parquet_files_modified":
            False,

        "kaggle_dataset_reuploaded":
            False,

        "kaggle_dataset_version_created":
            False,

        "final_freeze_must_reference_this_artifact":
            True,
    },
}


reconciliation_raw = json.dumps(
    reconciliation_semantic,
    sort_keys=True,
    separators=(",", ":"),
).encode()


reconciliation_semantic_sha = (
    hashlib
    .sha256(
        reconciliation_raw
    )
    .hexdigest()
)


reconciliation_artifact = {

    **reconciliation_semantic,

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "semantic_sha256":
        reconciliation_semantic_sha,
}


# ============================================================
# 16.10 — WRITE WITHOUT SILENT OVERWRITE
# ============================================================

RECONCILIATION_PATH = (

    PATHS["manifests"]

    / (
        "module5_m05_00_cell09_"
        "provenance_reconciliation_v1.json"
    )
)


RECONCILIATION_PATH.parent.mkdir(
    parents=True,
    exist_ok=True,
)


if RECONCILIATION_PATH.exists():


    with RECONCILIATION_PATH.open(
        "r",
        encoding="utf-8",
    ) as file:

        existing = json.load(file)


    if (
        existing.get(
            "semantic_sha256"
        )
        != reconciliation_semantic_sha
    ):

        raise FileExistsError(
            "A different provenance reconciliation "
            "artifact already exists:\n"
            f"{RECONCILIATION_PATH}"
        )


else:


    with RECONCILIATION_PATH.open(
        "w",
        encoding="utf-8",
    ) as file:

        json.dump(
            reconciliation_artifact,
            file,
            indent=2,
            sort_keys=True,
        )

        file.write("\n")


reconciliation_artifact_sha = (
    sha256_file(
        RECONCILIATION_PATH
    )
)


# ============================================================
# 16.11 — FINAL DECISION
# ============================================================

PROVENANCE_RECONCILIATION_PASS = all(
    [
        PROVENANCE_REFERENCE_SET_PASS,
        PARQUET_STRUCTURE_PASS,
        RECONCILIATION_CLASSIFICATION
        == "PROVENANCE_ONLY",
        not TELEMETRY_REGENERATION_REQUIRED,
        not PARQUET_MODIFICATION_REQUIRED,
        not DOWNSTREAM_ARTIFACT_REWRITE_REQUIRED,
        not KAGGLE_REUPLOAD_REQUIRED,
        not KAGGLE_VERSION_REQUIRED,
    ]
)


if not PROVENANCE_RECONCILIATION_PASS:

    raise RuntimeError(
        "M05_00 → Cell 09 provenance reconciliation FAILED."
    )


# ============================================================
# 16.12 — SUMMARY
# ============================================================

print("=" * 108)

print(
    "MODULE 5 — CELL 16 "
    "M05_00 → CELL 09 PROVENANCE RECONCILIATION V1"
)

print("=" * 108)


print("\nReference correction")
print("-" * 108)

print(
    f"Earlier M05_00 SHA               : "
    f"{M05_00_EARLIER_SHA256}"
)

print(
    f"Authoritative M05_00 SHA         : "
    f"{M05_00_FINAL_SHA256}"
)

print(
    "Reconciliation type              : "
    "SUPERSEDED_UPSTREAM_REFERENCE"
)

print(
    f"Classification                   : "
    f"{RECONCILIATION_CLASSIFICATION}"
)


print("\nAffected provenance")
print("-" * 108)

print(
    f"Affected artifacts               : "
    f"{len(affected_artifacts)}"
)

print(
    "Exact reference locations        : "
    f"{sum(item['reference_count'] for item in affected_artifacts)}"
)

print(
    f"Reference-set verification       : "
    f"{'PASS' if PROVENANCE_REFERENCE_SET_PASS else 'FAIL'}"
)


print("\nTelemetry preservation")
print("-" * 108)

print(
    f"Canonical Parquet partitions     : "
    f"{len(canonical_parquet_files)}"
)

print(
    f"Pilot Parquet partitions         : "
    f"{len(pilot_parquet_files)}"
)

print(
    "Telemetry regeneration required  : NO"
)

print(
    "Parquet modification required    : NO"
)

print(
    "Downstream artifact rewrite      : NO"
)

print(
    "Kaggle re-upload required        : NO"
)

print(
    "Kaggle new version required      : NO"
)


print("\nPreserved semantic identities")
print("-" * 108)

print(
    f"Cell 09 semantic SHA             : "
    f"{CELL09_SEMANTIC_SHA256}"
)

print(
    f"Cell 10 semantic SHA             : "
    f"{CELL10_SEMANTIC_SHA256}"
)

print(
    f"Cell 11 semantic SHA             : "
    f"{CELL11_SEMANTIC_SHA256}"
)

print(
    f"Cell 12 semantic SHA             : "
    f"{CELL12_SEMANTIC_SHA256}"
)

print(
    f"Canonical dataset semantic SHA   : "
    f"{CANONICAL_DATASET_SEMANTIC_SHA256}"
)

print(
    f"Kaggle persistence semantic SHA  : "
    f"{KAGGLE_PERSISTENCE_SEMANTIC_SHA256}"
)


print("\nReconciliation artifact")
print("-" * 108)

print(
    f"Path                             : "
    f"{RECONCILIATION_PATH}"
)

print(
    f"Artifact SHA                     : "
    f"{reconciliation_artifact_sha}"
)

print(
    f"Semantic SHA                     : "
    f"{reconciliation_semantic_sha}"
)


print("\nDecision")
print("-" * 108)

print(
    f"PROVENANCE_RECONCILIATION_PASS   : "
    f"{PROVENANCE_RECONCILIATION_PASS}"
)


print("\n" + "=" * 108)

print(
    "CELL 16 COMPLETE — "
    "AUTHORITATIVE M05_00 REFERENCE RECONCILED "
    "WITHOUT MODIFYING TELEMETRY"
)

print("=" * 108)
````

##### Cell 17 — M05_01 Foundation Freeze Manifest V1

Bind the authoritative foundation identities, pilot qualification, canonical 92-day telemetry identity, verified Kaggle persistence and Cell 16 provenance reconciliation into a single M05_01 root-of-trust freeze manifest.

````python
# ============================================================
# MODULE 5 — CELL 17
# M05_01 FOUNDATION + HEALTHY TELEMETRY FREEZE MANIFEST V1
# ============================================================
#
# Purpose
# ------------------------------------------------------------
# Freeze the validated state of:
#
#   M05_01 — Foundation + Healthy Telemetry
#
# The freeze manifest records:
#
#   - foundational schema / graph identities
#   - causal observability identities
#   - telemetry contract / topology identities
#   - pilot qualification identity
#   - canonical 92-day telemetry identity
#   - private Kaggle persistence identity
#   - M05_00 → Cell 09 provenance reconciliation
#
# No upstream artifact is modified.
# No telemetry is regenerated.
# No Kaggle upload or version is performed.
#
# ============================================================


from pathlib import Path
from datetime import datetime, timezone

import hashlib
import json


# ============================================================
# 17.1 — MODULE IDENTITY
# ============================================================

MODULE_ID = "M05_01"

MODULE_TITLE = (
    "Foundation + Healthy Telemetry"
)

MODULE_STATUS = "FROZEN"


# ============================================================
# 17.2 — AUTHORITATIVE UPSTREAM REFERENCE
# ============================================================

M05_00_FINAL_SHA256 = (
    "9c4db1552791510d3237fd8c7ff93cd"
    "7ae66d614d42fed0da7a268bf77fa910e"
)


# ============================================================
# 17.3 — VALIDATED FOUNDATION IDENTITIES
# ============================================================

CELL03_SCHEMA_ARTIFACT_SHA256 = (
    "b431dd3ffe3543d1060d42ede8411e0a"
    "b369b758e3fd2fe16dfc495a6c9ce0e"
)


CELL04_GRAPH_V1_ARTIFACT_SHA256 = (
    "92c96bb7fc2431115b2c25663729a4dbb"
    "2ee4d70b6595bb13a7615a9cb4ac9f9"
)


CELL04_GRAPH_V1_SEMANTIC_SHA256 = (
    "11e52a727942b6d25e1b907e07539a82"
    "d37ad424d93039aa765c53489ee2896a"
)


CELL05_KPI_SUPPLEMENT_SEMANTIC_SHA256 = (
    "a205b4ab672eabe90380fd2bcc7dda81e"
    "f4b125c6eb278e69ab0c462d3949348"
)


CELL06_GRAPH_V1_1_ARTIFACT_SHA256 = (
    "2d7da5933fbf82e57e8c0f25e879a453"
    "b06f3167e08d09b17f4240f51d6be570"
)


CELL06_GRAPH_V1_1_SEMANTIC_SHA256 = (
    "28af15d5d97c8ccfc8f636e3175650b67"
    "d55de8b48dfb021365d3206110aee42"
)


CELL07_OBSERVATION_CONTRACT_SEMANTIC_SHA256 = (
    "cc735d9a189f3346bfe7591f1f8e74ba"
    "c0abd35f2c6e7235cb91f4bdb488a53e"
)


CELL08_SYNTHETIC_TOPOLOGY_SEMANTIC_SHA256 = (
    "9e23832e8191e0dca58bde2c9d15631c"
    "d20539e2033e458fb39b54629f8e86d3"
)


CELL09_CAUSAL_CONFIG_SEMANTIC_SHA256 = (
    "8ccce01ee1375bbef52611a4b008ea013"
    "8bae88874993be67f7577aa0838f42a"
)


CELL10_DIAGNOSTIC_GRAPH_SEMANTIC_SHA256 = (
    "aaeb1acf5739a9d80d7745469faa71fe"
    "142181c7e96df6bc2c9e5f8b8b031b68"
)


CELL11_CAUSAL_TOPOLOGY_SEMANTIC_SHA256 = (
    "b12fd1cc5b208963a13caecc33d03016"
    "a612ff1f1f112686f4db435db66b421d"
)


CELL12_OPERATIONAL_CONTRACT_SEMANTIC_SHA256 = (
    "07ba4e0cc0f672cc03268e96bb53aa0f"
    "322241be43a4f957e7065b0d6ebae312"
)


# ============================================================
# 17.4 — HEALTHY TELEMETRY IDENTITIES
# ============================================================

PILOT_DATASET_SEMANTIC_SHA256 = (
    "0d3ced1fbd2e29f5a0a01c1598c04365"
    "b7b0929247c8382ed56f7923b9e58a71"
)


PILOT_MANIFEST_ARTIFACT_SHA256 = (
    "f36e1789ace1804edefad609c97ae39d"
    "d5f99ded03f39fd430bd2d662dab51b2"
)


PILOT_QUALIFICATION_SEMANTIC_SHA256 = (
    "e3fbe0f1226a719051830755239c1df3"
    "edbf79be4ec3fb2186cc1957260339d0"
)


PILOT_QUALIFICATION_ARTIFACT_SHA256 = (
    "1f86be858d7e617671dc8b0c8780067d"
    "92d270e0b84a02cc9de4a0c8dcef818a"
)


CANONICAL_DATASET_SEMANTIC_SHA256 = (
    "d6b75444faf53d602edf952f4620fe02"
    "f2f3a4ea68431a4889e4829adf008f4d"
)


CANONICAL_MANIFEST_ARTIFACT_SHA256 = (
    "85d035e42eaa4afe6ca2e21b1bd0d950"
    "13fff5a0ec99edfa7e4d025c23da9da0"
)


# ============================================================
# 17.5 — KAGGLE PERSISTENCE IDENTITIES
# ============================================================

KAGGLE_DATASET_HANDLE = (
    "cliffordimaguezegie/"
    "module5-5g-ran-healthy-telemetry-v1"
)


KAGGLE_STORAGE_MAP_SEMANTIC_SHA256 = (
    "9dcf2785d2c6dbf890633679c09659a6"
    "a30304c1fd5c2ca8044da4d0e171b999"
)


KAGGLE_PERSISTENCE_SEMANTIC_SHA256 = (
    "0719523ca6c6907ecefbacae84f8bce4"
    "91ddf37a886b71c6a9ed3ffe729033c7"
)


KAGGLE_PERSISTENCE_ARTIFACT_SHA256 = (
    "6938539a766e0a182fdeb37d56c1c6dc"
    "9f94a206c7bbbb20a5d832810f7e1c78"
)


# ============================================================
# 17.6 — PROVENANCE RECONCILIATION IDENTITY
# ============================================================

PROVENANCE_RECONCILIATION_SEMANTIC_SHA256 = (
    "16aaefc2282a0d43caec53a3aee33d0b"
    "12aa02e2b2d64b3a2d408d99e86cd521"
)


PROVENANCE_RECONCILIATION_ARTIFACT_SHA256 = (
    "04e7817c5936ffe16f0684fa7d1f7d6c"
    "90ceb61e21a52bbb9a9ed4640fc7fca2"
)


# ============================================================
# 17.7 — PERSISTED ARTIFACT PATHS
# ============================================================

ARTIFACT_ROOT = Path(
    "/content/module5_artifacts"
)


PILOT_MANIFEST_PATH = (
    ARTIFACT_ROOT
    / "telemetry"
    / "healthy_baseline_pilot_7d_v1"
    / "manifest.json"
)


PILOT_QUALIFICATION_PATH = (
    ARTIFACT_ROOT
    / "manifests"
    / "module5_pilot_7d_generator_qualification_v1.json"
)


CANONICAL_MANIFEST_PATH = (
    ARTIFACT_ROOT
    / "telemetry"
    / "healthy_baseline_v1"
    / "manifest.json"
)


KAGGLE_PERSISTENCE_PATH = (
    ARTIFACT_ROOT
    / "manifests"
    / "module5_canonical_kaggle_persistence_v1_3.json"
)


PROVENANCE_RECONCILIATION_PATH = (
    ARTIFACT_ROOT
    / "manifests"
    / (
        "module5_m05_00_cell09_"
        "provenance_reconciliation_v1.json"
    )
)


# ============================================================
# 17.8 — HASH HELPER
# ============================================================

def freeze_sha256_file(
    path,
):

    digest = hashlib.sha256()


    with Path(path).open(
        "rb"
    ) as file:

        for block in iter(
            lambda: file.read(
                1024 * 1024
            ),
            b"",
        ):

            digest.update(block)


    return digest.hexdigest()


# ============================================================
# 17.9 — JSON SEMANTIC VERIFICATION HELPER
# ============================================================

def verify_json_semantic(
    path,
    expected_semantic_sha,
):


    if not path.exists():

        return {
            "passed":
                False,

            "detail":
                "MISSING",
        }


    with path.open(
        "r",
        encoding="utf-8",
    ) as file:

        payload = json.load(file)


    observed = payload.get(
        "semantic_sha256"
    )


    return {

        "passed":
            observed
            == expected_semantic_sha,

        "detail":
            (
                f"observed={observed} "
                f"expected={expected_semantic_sha}"
            ),
    }


# ============================================================
# 17.10 — FREEZE GATES
# ============================================================

freeze_checks = {}


# ------------------------------------------------------------
# Pilot manifest
# ------------------------------------------------------------

freeze_checks[
    "pilot_manifest_exists"
] = {
    "passed":
        PILOT_MANIFEST_PATH.exists(),

    "detail":
        str(PILOT_MANIFEST_PATH),
}


if PILOT_MANIFEST_PATH.exists():

    observed_pilot_manifest_sha = (
        freeze_sha256_file(
            PILOT_MANIFEST_PATH
        )
    )

else:

    observed_pilot_manifest_sha = None


freeze_checks[
    "pilot_manifest_artifact_sha"
] = {
    "passed":
        observed_pilot_manifest_sha
        == PILOT_MANIFEST_ARTIFACT_SHA256,

    "detail":
        (
            f"observed={observed_pilot_manifest_sha} "
            f"expected={PILOT_MANIFEST_ARTIFACT_SHA256}"
        ),
}


freeze_checks[
    "pilot_manifest_semantic_sha"
] = verify_json_semantic(
    PILOT_MANIFEST_PATH,
    PILOT_DATASET_SEMANTIC_SHA256,
)


# ------------------------------------------------------------
# Pilot qualification
# ------------------------------------------------------------

freeze_checks[
    "pilot_qualification_semantic_sha"
] = verify_json_semantic(
    PILOT_QUALIFICATION_PATH,
    PILOT_QUALIFICATION_SEMANTIC_SHA256,
)


if PILOT_QUALIFICATION_PATH.exists():

    observed_pilot_qualification_sha = (
        freeze_sha256_file(
            PILOT_QUALIFICATION_PATH
        )
    )

else:

    observed_pilot_qualification_sha = None


freeze_checks[
    "pilot_qualification_artifact_sha"
] = {
    "passed":
        observed_pilot_qualification_sha
        == PILOT_QUALIFICATION_ARTIFACT_SHA256,

    "detail":
        (
            f"observed={observed_pilot_qualification_sha} "
            f"expected={PILOT_QUALIFICATION_ARTIFACT_SHA256}"
        ),
}


# ------------------------------------------------------------
# Canonical manifest
# ------------------------------------------------------------

freeze_checks[
    "canonical_manifest_semantic_sha"
] = verify_json_semantic(
    CANONICAL_MANIFEST_PATH,
    CANONICAL_DATASET_SEMANTIC_SHA256,
)


if CANONICAL_MANIFEST_PATH.exists():

    observed_canonical_manifest_sha = (
        freeze_sha256_file(
            CANONICAL_MANIFEST_PATH
        )
    )

else:

    observed_canonical_manifest_sha = None


freeze_checks[
    "canonical_manifest_artifact_sha"
] = {
    "passed":
        observed_canonical_manifest_sha
        == CANONICAL_MANIFEST_ARTIFACT_SHA256,

    "detail":
        (
            f"observed={observed_canonical_manifest_sha} "
            f"expected={CANONICAL_MANIFEST_ARTIFACT_SHA256}"
        ),
}


# ------------------------------------------------------------
# Kaggle persistence
# ------------------------------------------------------------

freeze_checks[
    "kaggle_persistence_semantic_sha"
] = verify_json_semantic(
    KAGGLE_PERSISTENCE_PATH,
    KAGGLE_PERSISTENCE_SEMANTIC_SHA256,
)


if KAGGLE_PERSISTENCE_PATH.exists():

    with KAGGLE_PERSISTENCE_PATH.open(
        "r",
        encoding="utf-8",
    ) as file:

        kaggle_receipt = json.load(file)


    freeze_checks[
        "kaggle_persistence_pass"
    ] = {
        "passed":
            bool(
                kaggle_receipt.get(
                    "persistence_pass"
                )
            ),

        "detail":
            kaggle_receipt.get(
                "persistence_pass"
            ),
    }


else:

    kaggle_receipt = {}

    freeze_checks[
        "kaggle_persistence_pass"
    ] = {
        "passed":
            False,

        "detail":
            "MISSING",
    }


# ------------------------------------------------------------
# Provenance reconciliation
# ------------------------------------------------------------

freeze_checks[
    "provenance_reconciliation_semantic_sha"
] = verify_json_semantic(
    PROVENANCE_RECONCILIATION_PATH,
    PROVENANCE_RECONCILIATION_SEMANTIC_SHA256,
)


if PROVENANCE_RECONCILIATION_PATH.exists():

    with PROVENANCE_RECONCILIATION_PATH.open(
        "r",
        encoding="utf-8",
    ) as file:

        reconciliation = json.load(file)


    reconciliation_policy = (
        reconciliation.get(
            "reconciliation_policy",
            {}
        )
    )


    freeze_checks[
        "authoritative_m05_00_reference"
    ] = {
        "passed":
            (
                reconciliation
                .get(
                    "authoritative_upstream_reference",
                    {}
                )
                .get(
                    "sha256"
                )
                == M05_00_FINAL_SHA256
            ),

        "detail":
            (
                reconciliation
                .get(
                    "authoritative_upstream_reference",
                    {}
                )
                .get(
                    "sha256"
                )
            ),
    }


    freeze_checks[
        "telemetry_preserved"
    ] = {
        "passed":
            (
                reconciliation_policy.get(
                    "telemetry_regenerated"
                )
                is False

                and

                reconciliation_policy.get(
                    "parquet_files_modified"
                )
                is False

                and

                reconciliation_policy.get(
                    "kaggle_dataset_reuploaded"
                )
                is False
            ),

        "detail":
            reconciliation_policy,
    }


else:

    reconciliation = {}

    freeze_checks[
        "authoritative_m05_00_reference"
    ] = {
        "passed":
            False,

        "detail":
            "MISSING",
    }


    freeze_checks[
        "telemetry_preserved"
    ] = {
        "passed":
            False,

        "detail":
            "MISSING",
    }


# ============================================================
# 17.11 — FINAL FREEZE GATE
# ============================================================

M05_01_FREEZE_PASS = all(

    result["passed"]

    for result

    in freeze_checks.values()
)


if not M05_01_FREEZE_PASS:


    failed = [

        name

        for name, result

        in freeze_checks.items()

        if not result["passed"]
    ]


    raise RuntimeError(

        "M05_01 foundation freeze gate FAILED. "

        f"Failed checks: {failed}"
    )


# ============================================================
# 17.12 — BUILD FREEZE SEMANTIC CONTENT
# ============================================================

freeze_semantic = {

    "artifact_id":
        "module5_m05_01_foundation_freeze_manifest_v1",

    "module_id":
        MODULE_ID,

    "module_title":
        MODULE_TITLE,

    "status":
        MODULE_STATUS,

    "authoritative_m05_00_sha256":
        M05_00_FINAL_SHA256,

    "foundation_identities": {

        "cell03_schema_artifact_sha256":
            CELL03_SCHEMA_ARTIFACT_SHA256,

        "cell04_graph_v1_artifact_sha256":
            CELL04_GRAPH_V1_ARTIFACT_SHA256,

        "cell04_graph_v1_semantic_sha256":
            CELL04_GRAPH_V1_SEMANTIC_SHA256,

        "cell05_kpi_supplement_semantic_sha256":
            CELL05_KPI_SUPPLEMENT_SEMANTIC_SHA256,

        "cell06_graph_v1_1_artifact_sha256":
            CELL06_GRAPH_V1_1_ARTIFACT_SHA256,

        "cell06_graph_v1_1_semantic_sha256":
            CELL06_GRAPH_V1_1_SEMANTIC_SHA256,

        "cell07_observation_contract_semantic_sha256":
            CELL07_OBSERVATION_CONTRACT_SEMANTIC_SHA256,

        "cell08_synthetic_topology_semantic_sha256":
            CELL08_SYNTHETIC_TOPOLOGY_SEMANTIC_SHA256,

        "cell09_causal_config_semantic_sha256":
            CELL09_CAUSAL_CONFIG_SEMANTIC_SHA256,

        "cell10_diagnostic_graph_semantic_sha256":
            CELL10_DIAGNOSTIC_GRAPH_SEMANTIC_SHA256,

        "cell11_causal_topology_semantic_sha256":
            CELL11_CAUSAL_TOPOLOGY_SEMANTIC_SHA256,

        "cell12_operational_contract_semantic_sha256":
            CELL12_OPERATIONAL_CONTRACT_SEMANTIC_SHA256,
    },

    "healthy_telemetry": {

        "pilot_dataset_semantic_sha256":
            PILOT_DATASET_SEMANTIC_SHA256,

        "pilot_qualification_semantic_sha256":
            PILOT_QUALIFICATION_SEMANTIC_SHA256,

        "canonical_dataset_semantic_sha256":
            CANONICAL_DATASET_SEMANTIC_SHA256,

        "canonical_manifest_artifact_sha256":
            CANONICAL_MANIFEST_ARTIFACT_SHA256,

        "canonical_rows":
            18_829_824,

        "canonical_daily_partitions":
            92,

        "pm_interval_minutes":
            15,

        "canonical_role":
            "HEALTHY_NETWORK_OBSERVATION_BASELINE",
    },

    "external_persistence": {

        "provider":
            "KAGGLE",

        "dataset_handle":
            KAGGLE_DATASET_HANDLE,

        "visibility":
            "PRIVATE",

        "storage_map_semantic_sha256":
            KAGGLE_STORAGE_MAP_SEMANTIC_SHA256,

        "persistence_semantic_sha256":
            KAGGLE_PERSISTENCE_SEMANTIC_SHA256,

        "persistence_verified":
            True,
    },

    "provenance_reconciliation": {

        "classification":
            "PROVENANCE_ONLY",

        "semantic_sha256":
            PROVENANCE_RECONCILIATION_SEMANTIC_SHA256,

        "earlier_m05_00_reference_superseded":
            True,

        "telemetry_regenerated":
            False,

        "parquet_modified":
            False,

        "kaggle_reupload_required":
            False,
    },

    "freeze_policy": {

        "foundation_state_frozen":
            True,

        "upstream_artifacts_mutated":
            False,

        "telemetry_regenerated":
            False,

        "future_modules_must_reference_frozen_identities":
            True,

        "future_changes_require_new_artifact_versions":
            True,
    },
}


freeze_raw = json.dumps(

    freeze_semantic,

    sort_keys=True,

    separators=(",", ":"),

).encode()


freeze_semantic_sha = (

    hashlib

    .sha256(

        freeze_raw

    )

    .hexdigest()
)


freeze_artifact = {

    **freeze_semantic,

    "created_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "semantic_sha256":
        freeze_semantic_sha,
}


# ============================================================
# 17.13 — WRITE FREEZE MANIFEST
# ============================================================

FREEZE_MANIFEST_PATH = (

    ARTIFACT_ROOT

    / "manifests"

    / "module5_m05_01_foundation_freeze_manifest_v1.json"
)


FREEZE_MANIFEST_PATH.parent.mkdir(

    parents=True,

    exist_ok=True,
)


if FREEZE_MANIFEST_PATH.exists():


    with FREEZE_MANIFEST_PATH.open(

        "r",

        encoding="utf-8",

    ) as file:

        existing_freeze = json.load(file)


    if (

        existing_freeze.get(
            "semantic_sha256"
        )

        != freeze_semantic_sha

    ):

        raise FileExistsError(

            "A different M05_01 freeze manifest "

            "already exists:\n"

            f"{FREEZE_MANIFEST_PATH}"
        )


else:


    with FREEZE_MANIFEST_PATH.open(

        "w",

        encoding="utf-8",

    ) as file:

        json.dump(

            freeze_artifact,

            file,

            indent=2,

            sort_keys=True,
        )

        file.write("\n")


freeze_artifact_sha = (

    freeze_sha256_file(

        FREEZE_MANIFEST_PATH
    )
)


# ============================================================
# 17.14 — SUMMARY
# ============================================================

print("=" * 108)

print(
    "MODULE 5 — CELL 17 "
    "M05_01 FOUNDATION + HEALTHY TELEMETRY FREEZE MANIFEST V1"
)

print("=" * 108)


print("\nFreeze checks")
print("-" * 108)


for name, result in freeze_checks.items():

    print(
        f"{'PASS' if result['passed'] else 'FAIL':<7} "
        f"{name}"
    )


print("\nFrozen scope")
print("-" * 108)

print(
    f"Module                           : "
    f"{MODULE_ID}"
)

print(
    f"Title                            : "
    f"{MODULE_TITLE}"
)

print(
    f"Status                           : "
    f"{MODULE_STATUS}"
)

print(
    f"Canonical rows                   : "
    f"{18_829_824:,}"
)

print(
    f"Canonical daily partitions       : "
    f"92"
)

print(
    f"Observation interval             : "
    f"15 minutes"
)

print(
    f"Kaggle persistence               : "
    f"VERIFIED"
)


print("\nAuthoritative identities")
print("-" * 108)

print(
    f"M05_00 final SHA                 : "
    f"{M05_00_FINAL_SHA256}"
)

print(
    f"Cell 09 semantic SHA             : "
    f"{CELL09_CAUSAL_CONFIG_SEMANTIC_SHA256}"
)

print(
    f"Canonical dataset semantic SHA   : "
    f"{CANONICAL_DATASET_SEMANTIC_SHA256}"
)

print(
    f"Kaggle persistence semantic SHA  : "
    f"{KAGGLE_PERSISTENCE_SEMANTIC_SHA256}"
)

print(
    f"Reconciliation semantic SHA      : "
    f"{PROVENANCE_RECONCILIATION_SEMANTIC_SHA256}"
)


print("\nFreeze manifest")
print("-" * 108)

print(
    f"Path                             : "
    f"{FREEZE_MANIFEST_PATH}"
)

print(
    f"Artifact SHA                     : "
    f"{freeze_artifact_sha}"
)

print(
    f"Semantic SHA                     : "
    f"{freeze_semantic_sha}"
)


print("\nDecision")
print("-" * 108)

print(
    f"M05_01_FREEZE_PASS               : "
    f"{M05_01_FREEZE_PASS}"
)


print("\n" + "=" * 108)

print(
    "M05_01 FROZEN — "
    "FOUNDATION + HEALTHY TELEMETRY BASELINE READY "
    "FOR M05_02"
)

print("=" * 108)
````

##### Cell 18 — Frozen M05_01 Foundation Artifact Bundle Persistence V1

Persist the small runtime artifacts required to reconstruct the frozen M05_01 foundation after the Colab runtime is released.

Only JSON foundation, topology, contract, qualification, provenance and freeze artifacts are included. The **92-day Parquet telemetry baseline is not duplicated** because it is already independently persisted and verified in its own private Kaggle Dataset.

The cell creates the private foundation Dataset only when absent; subsequent executions perform verification only.

> **This closes the runtime-persistence gap and makes M05_01 independently restorable by M05_02.**


````python
# ============================================================
# MODULE 5 — CELL 18
# M05_01 FROZEN FOUNDATION ARTIFACT BUNDLE PERSISTENCE V1
# ============================================================
#
# Purpose
# ------------------------------------------------------------
# Persist the SMALL frozen M05_01 runtime artifacts outside
# ephemeral Colab /content storage.
#
# Included:
#   - schemas / KPI knowledge JSON
#   - reference graphs
#   - telemetry contracts
#   - topology / causal configuration
#   - pilot + canonical manifests
#   - qualification artifacts
#   - Kaggle persistence provenance
#   - Cell 16 provenance reconciliation
#   - Cell 17 freeze manifest
#
# Excluded:
#   - all Parquet telemetry
#   - Kaggle telemetry staging directories
#
# Persistence target:
#
# cliffordimaguezegie/
# module5-m05-01-foundation-artifacts-v1
#
# Idempotent behaviour:
#
#   dataset absent  → CREATE
#   dataset exists  → VERIFY_EXISTING
#
# Existing datasets are NEVER silently versioned.
#
# ============================================================


from pathlib import Path
from io import StringIO
from datetime import datetime, timezone
from importlib import metadata as importlib_metadata

import hashlib
import json
import os
import re
import shutil
import subprocess
import sys
import time

import pandas as pd


# ============================================================
# 18.1 — ROOT-OF-TRUST IDENTITY
# ============================================================

ARTIFACT_ROOT = Path(
    "/content/module5_artifacts"
)


FREEZE_MANIFEST_PATH = (
    ARTIFACT_ROOT
    / "manifests"
    / "module5_m05_01_foundation_freeze_manifest_v1.json"
)


EXPECTED_FREEZE_SEMANTIC_SHA256 = (
    "3a519602da5e211df684786441c41669"
    "fbfdaac1c0d13254d18187ad5ce874e8"
)


EXPECTED_FREEZE_ARTIFACT_SHA256 = (
    "7390295a5f1ab364f8fe914145c495c1"
    "93b2cb229e155f6bd385eefa18e8f641"
)


if not FREEZE_MANIFEST_PATH.exists():

    raise FileNotFoundError(
        FREEZE_MANIFEST_PATH
    )


# ============================================================
# 18.2 — HASH HELPER
# ============================================================

def cell18_sha256_file(
    path,
):

    digest = hashlib.sha256()


    with Path(path).open(
        "rb"
    ) as file:

        for block in iter(
            lambda: file.read(
                1024 * 1024
            ),
            b"",
        ):

            digest.update(block)


    return digest.hexdigest()


# ============================================================
# 18.3 — VALIDATE CELL 17 FREEZE MANIFEST
# ============================================================

with FREEZE_MANIFEST_PATH.open(
    "r",
    encoding="utf-8",
) as file:

    freeze_manifest = json.load(file)


observed_freeze_semantic_sha = (
    freeze_manifest.get(
        "semantic_sha256"
    )
)


observed_freeze_artifact_sha = (
    cell18_sha256_file(
        FREEZE_MANIFEST_PATH
    )
)


FREEZE_SEMANTIC_PASS = (
    observed_freeze_semantic_sha
    == EXPECTED_FREEZE_SEMANTIC_SHA256
)


FREEZE_ARTIFACT_PASS = (
    observed_freeze_artifact_sha
    == EXPECTED_FREEZE_ARTIFACT_SHA256
)


FREEZE_STATUS_PASS = (
    freeze_manifest.get(
        "status"
    )
    == "FROZEN"
)


if not all(
    [
        FREEZE_SEMANTIC_PASS,
        FREEZE_ARTIFACT_PASS,
        FREEZE_STATUS_PASS,
    ]
):

    raise RuntimeError(
        "M05_01 freeze-manifest root-of-trust "
        "verification FAILED."
    )


print(
    "✓ M05_01 freeze-manifest root of trust verified."
)


# ============================================================
# 18.4 — DISCOVER SMALL FOUNDATION ARTIFACTS
# ============================================================
#
# Collect JSON artifacts only.
#
# Parquet telemetry is deliberately excluded.
# Existing Kaggle staging directories are also excluded.
#
# ============================================================

EXCLUDED_PATH_COMPONENTS = {

    "kaggle_staging",

    "kaggle_foundation_staging",
}


foundation_files = []


for path in sorted(
    ARTIFACT_ROOT.rglob(
        "*.json"
    )
):

    relative_path = (
        path.relative_to(
            ARTIFACT_ROOT
        )
    )


    if any(
        component
        in EXCLUDED_PATH_COMPONENTS

        for component
        in relative_path.parts
    ):

        continue


    foundation_files.append(
        path
    )


if not foundation_files:

    raise RuntimeError(
        "No M05_01 JSON foundation artifacts found."
    )


if (
    FREEZE_MANIFEST_PATH
    not in foundation_files
):

    raise RuntimeError(
        "Cell 17 freeze manifest was not included "
        "in the foundation artifact set."
    )


# ============================================================
# 18.5 — KAGGLE-SAFE STORAGE ADAPTER
# ============================================================

def foundation_kaggle_name(
    relative_path,
):

    text = (
        str(relative_path)
        .replace("\\", "/")
        .replace("=", "")
        .replace("/", "__")
    )


    # Explicit safe alphabet.
    text = re.sub(
        r"[^A-Za-z0-9._-]+",
        "_",
        text,
    )


    return text


bundle_entries = []

used_names = set()


for path in foundation_files:


    relative_path = (
        path.relative_to(
            ARTIFACT_ROOT
        )
    )


    kaggle_filename = (
        foundation_kaggle_name(
            relative_path
        )
    )


    if kaggle_filename in used_names:

        raise RuntimeError(
            "Kaggle foundation filename collision: "
            f"{kaggle_filename}"
        )


    used_names.add(
        kaggle_filename
    )


    bundle_entries.append(
        {
            "relative_path":
                str(relative_path)
                .replace("\\", "/"),

            "kaggle_filename":
                kaggle_filename,

            "sha256":
                cell18_sha256_file(
                    path
                ),

            "bytes":
                int(
                    path.stat().st_size
                ),
        }
    )


# ============================================================
# 18.6 — BUILD DETERMINISTIC BUNDLE INDEX
# ============================================================

bundle_index_semantic = {

    "artifact_id":
        "module5_m05_01_foundation_bundle_index_v1",

    "module_id":
        "M05_01",

    "module_status":
        "FROZEN",

    "freeze_manifest_semantic_sha256":
        EXPECTED_FREEZE_SEMANTIC_SHA256,

    "freeze_manifest_artifact_sha256":
        EXPECTED_FREEZE_ARTIFACT_SHA256,

    "source_root":
        "/content/module5_artifacts",

    "bundle_scope":
        "JSON_FOUNDATION_ARTIFACTS_ONLY",

    "parquet_telemetry_included":
        False,

    "canonical_telemetry_storage":
        (
            "cliffordimaguezegie/"
            "module5-5g-ran-healthy-telemetry-v1"
        ),

    "artifact_count":
        len(bundle_entries),

    "files":
        bundle_entries,
}


bundle_index_raw = json.dumps(
    bundle_index_semantic,
    sort_keys=True,
    separators=(",", ":"),
).encode()


bundle_index_semantic_sha = (
    hashlib
    .sha256(
        bundle_index_raw
    )
    .hexdigest()
)


bundle_index_payload = {

    **bundle_index_semantic,

    "semantic_sha256":
        bundle_index_semantic_sha,
}


# ============================================================
# 18.7 — KAGGLE FOUNDATION DATASET IDENTITY
# ============================================================

KAGGLE_OWNER = (
    "cliffordimaguezegie"
)


KAGGLE_FOUNDATION_SLUG = (
    "module5-m05-01-foundation-artifacts-v1"
)


KAGGLE_FOUNDATION_HANDLE = (
    f"{KAGGLE_OWNER}/"
    f"{KAGGLE_FOUNDATION_SLUG}"
)


KAGGLE_FOUNDATION_TITLE = (
    "Module 5 M05_01 Frozen Foundation Artifacts V1"
)


KAGGLE_FOUNDATION_URL = (
    "https://www.kaggle.com/datasets/"
    f"{KAGGLE_FOUNDATION_HANDLE}"
)


# ============================================================
# 18.8 — KAGGLE CLI
# ============================================================

TARGET_KAGGLE_VERSION = "2.2.4"


try:

    kaggle_version = (
        importlib_metadata
        .version(
            "kaggle"
        )
    )


except importlib_metadata.PackageNotFoundError:

    kaggle_version = None


if kaggle_version != TARGET_KAGGLE_VERSION:


    print(
        "\nInstalling Kaggle CLI "
        f"{TARGET_KAGGLE_VERSION}..."
    )


    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "-q",
            f"kaggle=={TARGET_KAGGLE_VERSION}",
        ],
        check=True,
    )


kaggle_version = (
    importlib_metadata
    .version(
        "kaggle"
    )
)


# ============================================================
# 18.9 — AUTHENTICATION
# ============================================================

kaggle_token = (
    os.environ
    .get(
        "KAGGLE_API_TOKEN",
        "",
    )
    .strip()
)


try:

    from google.colab import userdata


    if not kaggle_token:

        try:

            kaggle_token = (
                userdata
                .get(
                    "KAGGLE_API_TOKEN"
                )
                .strip()
            )


        except Exception:

            pass


except Exception:

    pass


if kaggle_token:

    os.environ[
        "KAGGLE_API_TOKEN"
    ] = kaggle_token


legacy_json = (
    Path.home()
    / ".kaggle"
    / "kaggle.json"
)


if (
    not kaggle_token
    and not legacy_json.exists()
):

    raise RuntimeError(
        "No Kaggle credential found."
    )


# ============================================================
# 18.10 — COMMAND HELPERS
# ============================================================

def cell18_run_kaggle(
    args,
    *,
    check=True,
):

    result = subprocess.run(
        [
            "kaggle",
            *args,
        ],
        text=True,
        capture_output=True,
    )


    if (
        check
        and result.returncode != 0
    ):

        raise RuntimeError(
            "Kaggle command failed:\n"
            + " ".join(
                [
                    "kaggle",
                    *args,
                ]
            )
            + "\n\nSTDOUT:\n"
            + (
                result.stdout
                or "<empty>"
            )
            + "\n\nSTDERR:\n"
            + (
                result.stderr
                or "<empty>"
            )
        )


    return result


def cell18_parse_csv(
    text,
):

    if not text.strip():

        return pd.DataFrame()


    return pd.read_csv(
        StringIO(text)
    )


def cell18_owned_dataset_refs():


    result = cell18_run_kaggle(
        [
            "datasets",
            "list",
            "--mine",
            "--page-size",
            "100",
            "--format",
            "csv(ref)",
        ],
        check=False,
    )


    if result.returncode != 0:

        result = cell18_run_kaggle(
            [
                "datasets",
                "list",
                "--mine",
                "--page-size",
                "100",
                "-v",
            ]
        )


    frame = cell18_parse_csv(
        result.stdout
    )


    if "ref" not in frame.columns:

        raise RuntimeError(
            "Could not identify Kaggle dataset "
            "reference column."
        )


    return set(
        frame["ref"]
        .astype(str)
    )


# ============================================================
# 18.11 — DISCOVER EXISTING DATASET
# ============================================================

owned_refs = (
    cell18_owned_dataset_refs()
)


FOUNDATION_DATASET_EXISTS = (
    KAGGLE_FOUNDATION_HANDLE
    in owned_refs
)


print(
    "\nFoundation persistence target"
)

print(
    "-" * 108
)

print(
    f"Dataset handle          : "
    f"{KAGGLE_FOUNDATION_HANDLE}"
)

print(
    "Visibility              : PRIVATE"
)

print(
    f"Local JSON artifacts    : "
    f"{len(bundle_entries)}"
)

print(
    f"Dataset already exists  : "
    f"{FOUNDATION_DATASET_EXISTS}"
)


# ============================================================
# 18.12 — CREATE ONLY IF ABSENT
# ============================================================

FOUNDATION_CREATED_NOW = False


STAGING_DIR = (
    ARTIFACT_ROOT
    / "kaggle_foundation_staging"
    / KAGGLE_FOUNDATION_SLUG
)


if not FOUNDATION_DATASET_EXISTS:


    if STAGING_DIR.exists():

        shutil.rmtree(
            STAGING_DIR
        )


    STAGING_DIR.mkdir(
        parents=True,
        exist_ok=True,
    )


    hardlink_count = 0

    copy_count = 0


    for entry in bundle_entries:


        source = (
            ARTIFACT_ROOT
            / entry["relative_path"]
        )


        destination = (
            STAGING_DIR
            / entry["kaggle_filename"]
        )


        try:

            os.link(
                source,
                destination,
            )

            hardlink_count += 1


        except OSError:

            shutil.copy2(
                source,
                destination,
            )

            copy_count += 1


        if (
            cell18_sha256_file(
                destination
            )
            != entry["sha256"]
        ):

            raise RuntimeError(
                "Foundation staging SHA mismatch: "
                f"{destination.name}"
            )


    # --------------------------------------------------------
    # Deterministic bundle index
    # --------------------------------------------------------

    BUNDLE_INDEX_PATH = (
        STAGING_DIR
        / "foundation_bundle_index.json"
    )


    with BUNDLE_INDEX_PATH.open(
        "w",
        encoding="utf-8",
    ) as file:

        json.dump(
            bundle_index_payload,
            file,
            indent=2,
            sort_keys=True,
        )

        file.write("\n")


    bundle_index_artifact_sha = (
        cell18_sha256_file(
            BUNDLE_INDEX_PATH
        )
    )


    # --------------------------------------------------------
    # Proven minimal Kaggle metadata
    # --------------------------------------------------------

    kaggle_metadata = {

        "title":
            KAGGLE_FOUNDATION_TITLE,

        "id":
            KAGGLE_FOUNDATION_HANDLE,

        "licenses": [
            {
                "name":
                    "other"
            }
        ],
    }


    KAGGLE_METADATA_PATH = (
        STAGING_DIR
        / "dataset-metadata.json"
    )


    with KAGGLE_METADATA_PATH.open(
        "w",
        encoding="utf-8",
    ) as file:

        json.dump(
            kaggle_metadata,
            file,
            indent=2,
        )

        file.write("\n")


    print(
        "\nFoundation bundle staging ready."
    )

    print(
        f"Hard-linked artifacts   : "
        f"{hardlink_count}"
    )

    print(
        f"Copied artifacts        : "
        f"{copy_count}"
    )

    print(
        f"Bundle index SHA        : "
        f"{bundle_index_semantic_sha}"
    )

    print(
        "License                 : other"
    )


    # --------------------------------------------------------
    # Create private dataset.
    #
    # stdout/stderr are inherited intentionally so Kaggle
    # progress remains visible in the notebook.
    # --------------------------------------------------------

    print(
        "\nCreating private Kaggle foundation Dataset..."
    )

    print(
        "Upload progress will appear below.\n"
    )


    upload_result = subprocess.run(
        [
            "kaggle",
            "datasets",
            "create",
            "--path",
            str(
                STAGING_DIR
            ),
            "--dir-mode",
            "skip",
        ]
    )


    if upload_result.returncode != 0:

        raise RuntimeError(
            "Kaggle foundation Dataset creation FAILED."
        )


    FOUNDATION_CREATED_NOW = True


    # --------------------------------------------------------
    # Wait for dataset discovery.
    # --------------------------------------------------------

    FOUNDATION_DISCOVERY_PASS = False


    for _ in range(24):

        owned_refs = (
            cell18_owned_dataset_refs()
        )


        if (
            KAGGLE_FOUNDATION_HANDLE
            in owned_refs
        ):

            FOUNDATION_DISCOVERY_PASS = True

            break


        time.sleep(5)


    if not FOUNDATION_DISCOVERY_PASS:

        raise RuntimeError(
            "Foundation upload returned successfully, "
            "but the Dataset is not yet discoverable."
        )


else:


    FOUNDATION_DISCOVERY_PASS = True


    print(
        "\n✓ Foundation Dataset already exists."
    )

    print(
        "  No upload or version operation will be performed."
    )

    print(
        "  Proceeding directly to remote verification."
    )


# ============================================================
# 18.13 — EXPECTED REMOTE INVENTORY
# ============================================================

expected_remote_sizes = {

    entry["kaggle_filename"]:
        entry["bytes"]

    for entry
    in bundle_entries
}


# Rebuild deterministic bundle index locally for VERIFY_EXISTING.
VERIFY_INDEX_PATH = (
    ARTIFACT_ROOT
    / "manifests"
    / "module5_m05_01_foundation_bundle_index_v1.json"
)


with VERIFY_INDEX_PATH.open(
    "w",
    encoding="utf-8",
) as file:

    json.dump(
        bundle_index_payload,
        file,
        indent=2,
        sort_keys=True,
    )

    file.write("\n")


bundle_index_artifact_sha = (
    cell18_sha256_file(
        VERIFY_INDEX_PATH
    )
)


expected_remote_sizes[
    "foundation_bundle_index.json"
] = (
    VERIFY_INDEX_PATH
    .stat()
    .st_size
)


EXPECTED_REMOTE_COUNT = (
    len(expected_remote_sizes)
)


# ============================================================
# 18.14 — REMOTE INVENTORY VERIFICATION
# ============================================================

remote_result = (
    cell18_run_kaggle(
        [
            "datasets",
            "files",
            KAGGLE_FOUNDATION_HANDLE,
            "--page-size",
            "200",
            "--format",
            "csv(name,size)",
        ]
    )
)


remote_df = cell18_parse_csv(
    remote_result.stdout
)


if not {
    "name",
    "size",
}.issubset(
    remote_df.columns
):

    raise RuntimeError(
        "Unexpected foundation remote inventory schema: "
        f"{remote_df.columns.tolist()}"
    )


remote_df["name"] = (
    remote_df["name"]
    .astype(str)
)


remote_df["size"] = (
    pd.to_numeric(
        remote_df["size"],
        errors="raise",
    )
    .astype(int)
)


remote_sizes = dict(
    zip(
        remote_df["name"],
        remote_df["size"],
    )
)


expected_names = set(
    expected_remote_sizes
)


observed_names = set(
    remote_sizes
)


remote_missing = sorted(
    expected_names
    - observed_names
)


remote_unexpected = sorted(
    observed_names
    - expected_names
)


REMOTE_NAME_INVENTORY_PASS = (
    len(remote_missing) == 0
    and len(remote_unexpected) == 0
)


# ============================================================
# 18.15 — REMOTE BYTE-SIZE VERIFICATION
# ============================================================

size_mismatches = []


for (
    filename,
    expected_size,
) in expected_remote_sizes.items():


    observed_size = (
        remote_sizes.get(
            filename
        )
    )


    if (
        observed_size
        != expected_size
    ):

        size_mismatches.append(
            {
                "file":
                    filename,

                "expected_bytes":
                    int(
                        expected_size
                    ),

                "observed_bytes":
                    observed_size,
            }
        )


REMOTE_BYTE_SIZE_PASS = (
    len(size_mismatches) == 0
)


REMOTE_FILE_COUNT_PASS = (
    len(remote_df)
    == EXPECTED_REMOTE_COUNT
)


# ============================================================
# 18.16 — FINAL PERSISTENCE GATE
# ============================================================

FOUNDATION_PERSISTENCE_PASS = all(
    [
        FREEZE_SEMANTIC_PASS,
        FREEZE_ARTIFACT_PASS,
        FREEZE_STATUS_PASS,
        FOUNDATION_DISCOVERY_PASS,
        REMOTE_FILE_COUNT_PASS,
        REMOTE_NAME_INVENTORY_PASS,
        REMOTE_BYTE_SIZE_PASS,
    ]
)


if not FOUNDATION_PERSISTENCE_PASS:

    raise RuntimeError(
        "M05_01 foundation persistence verification FAILED.\n"
        f"Missing files    : {remote_missing}\n"
        f"Unexpected files : {remote_unexpected}\n"
        f"Size mismatches  : {size_mismatches}"
    )


# ============================================================
# 18.17 — LOCAL HANDOVER RECEIPT
# ============================================================

operation = (
    "CREATE"
    if FOUNDATION_CREATED_NOW
    else "VERIFY_EXISTING"
)


handover_semantic = {

    "artifact_id":
        "module5_m05_01_foundation_persistence_receipt_v1",

    "module_id":
        "M05_01",

    "module_status":
        "FROZEN",

    "operation":
        operation,

    "storage_provider":
        "KAGGLE_DATASET",

    "dataset_handle":
        KAGGLE_FOUNDATION_HANDLE,

    "visibility":
        "PRIVATE",

    "freeze_manifest_semantic_sha256":
        EXPECTED_FREEZE_SEMANTIC_SHA256,

    "freeze_manifest_artifact_sha256":
        EXPECTED_FREEZE_ARTIFACT_SHA256,

    "foundation_bundle_index_semantic_sha256":
        bundle_index_semantic_sha,

    "foundation_bundle_index_artifact_sha256":
        bundle_index_artifact_sha,

    "foundation_artifact_count":
        len(bundle_entries),

    "remote_file_count":
        len(remote_df),

    "remote_name_inventory_verified":
        REMOTE_NAME_INVENTORY_PASS,

    "remote_byte_sizes_verified":
        REMOTE_BYTE_SIZE_PASS,

    "parquet_telemetry_duplicated":
        False,

    "canonical_telemetry_dataset":
        (
            "cliffordimaguezegie/"
            "module5-5g-ran-healthy-telemetry-v1"
        ),

    "automatic_version_created":
        False,

    "persistence_pass":
        FOUNDATION_PERSISTENCE_PASS,
}


handover_raw = json.dumps(
    handover_semantic,
    sort_keys=True,
    separators=(",", ":"),
).encode()


handover_semantic_sha = (
    hashlib
    .sha256(
        handover_raw
    )
    .hexdigest()
)


handover_receipt = {

    **handover_semantic,

    "verified_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),

    "semantic_sha256":
        handover_semantic_sha,
}


HANDOVER_RECEIPT_PATH = (
    ARTIFACT_ROOT
    / "manifests"
    / "module5_m05_01_foundation_persistence_receipt_v1.json"
)


if HANDOVER_RECEIPT_PATH.exists():


    with HANDOVER_RECEIPT_PATH.open(
        "r",
        encoding="utf-8",
    ) as file:

        existing_receipt = (
            json.load(file)
        )


    if (
        existing_receipt.get(
            "semantic_sha256"
        )
        != handover_semantic_sha
    ):

        raise FileExistsError(
            "A different Cell 18 persistence receipt "
            "already exists:\n"
            f"{HANDOVER_RECEIPT_PATH}"
        )


else:


    with HANDOVER_RECEIPT_PATH.open(
        "w",
        encoding="utf-8",
    ) as file:

        json.dump(
            handover_receipt,
            file,
            indent=2,
            sort_keys=True,
        )

        file.write("\n")


handover_artifact_sha = (
    cell18_sha256_file(
        HANDOVER_RECEIPT_PATH
    )
)


# ============================================================
# 18.18 — FINAL SUMMARY
# ============================================================

foundation_bytes = sum(
    entry["bytes"]
    for entry
    in bundle_entries
)


print(
    "\n" + "=" * 108
)

print(
    "MODULE 5 — CELL 18 "
    "M05_01 FROZEN FOUNDATION ARTIFACT "
    "BUNDLE PERSISTENCE V1"
)

print(
    "=" * 108
)


print(
    "\nRoot of trust"
)

print(
    "-" * 108
)

print(
    f"Freeze semantic SHA              : "
    f"{EXPECTED_FREEZE_SEMANTIC_SHA256}"
)

print(
    f"Freeze artifact SHA              : "
    f"{EXPECTED_FREEZE_ARTIFACT_SHA256}"
)

print(
    f"Freeze status                    : "
    f"FROZEN"
)


print(
    "\nFoundation bundle"
)

print(
    "-" * 108
)

print(
    f"Local JSON artifacts             : "
    f"{len(bundle_entries)}"
)

print(
    f"Foundation artifact bytes        : "
    f"{foundation_bytes:,}"
)

print(
    f"Parquet telemetry included       : "
    f"NO"
)

print(
    f"Bundle index semantic SHA        : "
    f"{bundle_index_semantic_sha}"
)

print(
    f"Bundle index artifact SHA        : "
    f"{bundle_index_artifact_sha}"
)


print(
    "\nKaggle persistence"
)

print(
    "-" * 108
)

print(
    f"Kaggle CLI version               : "
    f"{kaggle_version}"
)

print(
    f"Dataset handle                   : "
    f"{KAGGLE_FOUNDATION_HANDLE}"
)

print(
    f"Visibility                       : "
    f"PRIVATE"
)

print(
    f"Operation                        : "
    f"{operation}"
)

print(
    f"Dataset discovery                : "
    f"{'PASS' if FOUNDATION_DISCOVERY_PASS else 'FAIL'}"
)

print(
    f"Remote files                     : "
    f"{len(remote_df)}"
)

print(
    f"Remote file count                : "
    f"{'PASS' if REMOTE_FILE_COUNT_PASS else 'FAIL'}"
)

print(
    f"Remote filename inventory        : "
    f"{'PASS' if REMOTE_NAME_INVENTORY_PASS else 'FAIL'}"
)

print(
    f"Remote byte-size inventory       : "
    f"{'PASS' if REMOTE_BYTE_SIZE_PASS else 'FAIL'}"
)


print(
    "\nHandover receipt"
)

print(
    "-" * 108
)

print(
    f"Path                             : "
    f"{HANDOVER_RECEIPT_PATH}"
)

print(
    f"Artifact SHA                     : "
    f"{handover_artifact_sha}"
)

print(
    f"Semantic SHA                     : "
    f"{handover_semantic_sha}"
)


print(
    "\nDecision"
)

print(
    "-" * 108
)

print(
    f"FOUNDATION_PERSISTENCE_PASS      : "
    f"{FOUNDATION_PERSISTENCE_PASS}"
)


print(
    "\n" + "=" * 108
)

print(
    "CELL 18 COMPLETE — "
    "M05_01 FROZEN FOUNDATION IS DURABLY "
    "PERSISTED AND READY FOR M05_02 RESTORE"
)

print(
    "=" * 108
)
````


###### Cell 18 — Frozen Foundation Persistence Observation

The frozen M05_01 foundation artifact set was successfully persisted to a separate private Kaggle Dataset and verified against the local bundle.

```text
Local foundation JSON artifacts : 23
Remote files                     : 24
Parquet telemetry duplicated     : NO

Dataset discovery                : PASS
Remote file count                : PASS
Remote filename inventory        : PASS
Remote byte-size inventory       : PASS

FOUNDATION_PERSISTENCE_PASS      : True
```

The additional remote file is `foundation_bundle_index.json`, which records the deterministic mapping and hashes required to verify the restored foundation.
